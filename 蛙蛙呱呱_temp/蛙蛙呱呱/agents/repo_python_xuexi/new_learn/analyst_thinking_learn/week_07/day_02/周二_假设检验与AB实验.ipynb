## 📘 区块1：核心概念与第一性原理

### 1.1 从"看到差异"到"确信差异"——为什么需要假设检验？

对照组转化率10%，实验组12%。这个2%的差异是真实的还是运气的？

你可能想："12% > 10%，实验组更好啊！"但事实是——即使策略完全无效（两组真实转化率完全相同），由于随机性，两组之间也会出现差异。就像抛硬币10次，你不会每次都恰好得到5次正面。

假设检验回答的就是：**这个差异大到"不太可能仅由随机波动解释"了吗？**

### 1.2 假设检验的标准框架

```
H₀（零假设/原假设）：实验组和对照组没有差异（策略无效）
H₁（备择假设）：实验组和对照组有差异（策略有效）

p值 = 在H₀为真的前提下，观察到当前差异（或更大差异）的概率

如果p < 0.05（显著性水平α），我们拒绝H₀，说"结果统计显著"
```

### 1.3 p值的正确理解——数据分析师最常被问的面试题

```
p值 IS:     P(数据 | H₀为真)  —— "如果策略无效，看到这数据的概率"
p值 IS NOT: P(H₀为真 | 数据) —— "策略无效的概率"
```

这是数据分析领域最常被误解的概念。p=0.03 不代表"有97%的概率策略有效"。它只代表"如果策略无效，只有3%的概率看到这种数据"。

**类比**：法官判案时，"如果被告无罪，出现这些证据的概率很小" ≠ "被告无罪的概率很小"。

### 1.4 从第一性原理推导：为什么p<0.05成为"标准"？

Ronald Fisher在1925年提出p<0.05作为"显著"的阈值。他选择0.05的理由其实很随意——"刚好是1/20，方便计算"。

但这个阈值被整个科学界和工业界继承了下来：
- α=0.05意味着：每做20次实验，即使所有策略都无效，你也会得到1次"显著"的结果
- 这就是**第一类错误（误报）**的概率
- 如果你同时跑20个实验，就期望有1个是"假阳性"

> **今天最重要的认知**：p值衡量"证据的强度"，不是"效果的大小"。大样本下，0.01%的差异也可能p<0.001；小样本下，50%的提升也可能p>0.05。永远不要只看p值——同时看效应量（effect size）和置信区间（confidence interval）。

### 1.5 面试高频考点

**面试官常问**：
- "p值是什么？用通俗语言解释"
- "p<0.05意味着什么？不能意味着什么？"
- "第一类错误和第二类错误有什么区别？"
- "统计功效是什么？为什么通常要求80%？"

**回答要点**：
- p值是"如果H₀为真，观察到当前或更极端数据的概率"
- p<0.05不是"策略有95%的概率有效"
- 第一类错误是"误报"（没效果说有效果），第二类错误是"漏报"（有效果说没效果）
- 80%功效意味着如果策略真实有效，有80%的概率能检测出来

## 🧩 区块2：知识点详细拆解

### 2.1 假设检验的两种方法：Fisher vs Neyman-Pearson

**为什么重要**：理解这两种方法的区别有助于解读实验结果。

| 维度 | Fisher方法 | Neyman-Pearson方法 |
|------|-----------|-------------------|
| 核心 | p值（证据强度） | α/β（决策规则） |
| 输出 | 连续的p值 | 二元的"接受/拒绝" |
| 需要预先指定 | 不需要 | 需要（α、β、样本量） |
| 决策方式 | "证据有多强" | "是否超过阈值" |
| 现代实践 | p值+效应量+置信区间 | 事前样本量+固定时间分析 |

**现代AB实验的融合做法**：
- 事前：用Neyman-Pearson框架（定α=0.05、β=0.20、计算样本量）
- 事后：用Fisher框架（报告p值，但同时也报告效应量和置信区间）

### 2.2 第一类错误（α）与第二类错误（β）——完整决策矩阵

**为什么重要**：这是理解统计决策的基础，面试必问。

| | H₀为真（策略无效） | H₁为真（策略有效） |
|------|--------------|--------------|
| 接受H₀（说不显著） | ✓ 正确（1-α=95%） | **β**（第二类错误：漏报） |
| 拒绝H₀（说显著） | **α**（第一类错误：误报） | ✓ 正确（功效=1-β） |

**参数解释**：
- **α（显著性水平）**：通常设为0.05。愿意承受5%的误报风险。
- **β（漏报率）**：通常设为0.20。愿意承受20%的漏报风险。
- **1-β（统计功效/Power）**：通常要求≥80%。如果策略真实有效，有80%的概率能检测出来。

**为什么Power通常设为80%而非更高？**
- Power越高→需要的样本量越大（呈非线性增长）
- 80%→90%的Power提升需要约33%的额外样本量
- 对大多数业务场景，80%是成本与收益的合理平衡点

### 2.3 效应量（Effect Size）——被忽视的真相

**为什么重要**：p值告诉你"有没有差异"，效应量告诉你"差异有多大"。在业务决策中，效应量往往比p值更重要。

**常见效应量指标**：

| 指标类型 | 效应量指标 | 适用场景 |
|---------|-----------|----------|
| 比例差异 | p₂ - p₁（绝对差值） | 转化率、留存率等比例指标 |
| 相对提升 | (p₂ - p₁) / p₁ | 直观展示提升幅度 |
| Cohen's d | (μ₂ - μ₁) / σ | 连续指标（人均GMV、时长） |
| Cohen's h | 2(arcsin(√p₂) - arcsin(√p₁)) | 比例指标的标准化差异 |

**为什么效应量重要——经典案例**：
```
实验A：n=10万，转化率从10.00%→10.05%，p<0.001（显著！）
实验B：n=1000，转化率从10%→15%，p=0.08（不显著……）
```
只看p值你会选A。但A的提升只有0.05个百分点（几乎没业务价值），而B虽然"不显著"但提升了5个百分点——如果样本量加到够大，B很可能既显著又有业务价值。

### 2.4 置信区间（Confidence Interval）——比p值更丰富的信息

**为什么重要**：置信区间同时包含了"效应量"和"不确定性"两个信息。

**95%置信区间的正确理解**：
- 如果重复实验100次，每次计算95%置信区间，大约有95个区间包含真实效应。
- 不是"真实效应有95%的概率在这个区间内"（那是贝叶斯可信区间）。

**置信区间在AB实验决策中的应用**：
```
95% CI = [1.2pp, 3.8pp] → 下限>0，可以上线
95% CI = [-0.5pp, 4.5pp] → 包含0，不能确定方向
95% CI = [0.1pp, 0.3pp] → 显著但效应太小，要评估业务价值
95% CI = [-2.0pp, -0.5pp] → 显著负效应，绝对不能上线
```

### 2.5 Z检验 vs t检验——什么时候用什么

**为什么重要**：选错检验方法会导致错误的p值。

| 维度 | Z检验 | t检验 |
|------|-------|-------|
| 适用条件 | 已知总体方差/大样本 | 未知总体方差/小样本 |
| 样本量要求 | n>30即可近似 | 无限制 |
| 分布假设 | 正态分布 | t分布（自由度=n-1） |
| AB实验中用哪个 | 大样本（n>1000）两者几乎一样 | 小样本时必须用t检验 |

**AB实验中的实践**：
- 大多数互联网AB实验样本量很大（每组>1000），Z检验和t检验结果几乎相同
- 但学术界和严谨的工业界偏好t检验（更保守）
- 如果样本量很小（<30每组），必须用t检验

### 2.6 单侧检验 vs 双侧检验

**为什么重要**：选择单侧还是双侧影响p值和结论。

| 维度 | 双侧检验 | 单侧检验 |
|------|---------|----------|
| H₁ | 实验组 ≠ 对照组 | 实验组 > 对照组（或 <） |
| 拒绝域 | 两端各α/2 | 一端α |
| p值 | 更大（更难显著） | 更小（更容易显著） |
| 适用场景 | 不确定方向时（默认选择） | 确定方向时（如"新版不可能更差"） |

**AB实验中的建议**：默认使用双侧检验。即使你认为"新版应该更好"，很多实验的实际结果是"新版更差"——如果用了单侧检验，你甚至无法检测到"显著更差"。

### 2.7 p值的分布特性——为什么随机性会产生"显著"

**为什么重要**：理解p值在H₀为真时的分布，就理解了为什么需要多重比较校正。

当H₀为真时（策略完全无效），p值服从[0,1]上的均匀分布。这意味着：
- p<0.05的概率恰好是5%
- p<0.01的概率恰好是1%
- p出现在任何区间的概率等于该区间的长度

**推导**：如果你跑20个独立实验，且所有策略都无效：
```
P(至少1个p<0.05) = 1 - P(所有p≥0.05) = 1 - 0.95^20 ≈ 64%
```
有64%的概率你会看到至少一个"显著"的结果！这就是**多重比较问题**。

**应对方法**：
- Bonferroni校正：α' = α / m（m是比较次数）
- FDR控制：允许一定比例的假阳性
- 事前注册：预定义主要指标和比较方式

### 2.8 p-hacking——数据分析中的"学术不端"

**为什么重要**：p-hacking在业界也存在，而且往往是无意识的。

**常见的p-hacking行为**：
1. **中途Peeking**：每天看p值，看到p<0.05就停（明天学）
2. **指标偷换**：主要指标不显著，换一个"恰好显著"的指标
3. **子群捞鱼**：在几十个子群中反复检验，找到"显著"的那个
4. **数据清洗自由度**：尝试不同的异常值剔除规则直到p<0.05
5. **模型选择自由度**：尝试多种模型设定直到p<0.05

**防御措施**：
- 事前注册（Pre-registration）：在实验开始前锁定假设、指标、分析方法
- 分析盲法：先不看处理组标签做数据清洗
- 独立复现：让另一个分析师独立复现结果

## 💻 区块3：动手练习

### 步骤0：数据准备——理解p值的本质

我们模拟不同场景下的假设检验，直观理解p值的行为。

````python
import numpy as np
from scipy import stats
np.random.seed(42)

print("="*60)
print("步骤0：p值在H₀为真时的分布（策略完全无效）")
print("="*60)

# 模拟10000次AA实验（两组完全相同）
p_values_h0 = []
for _ in range(10000):
    a = np.random.binomial(1, 0.10, 500)
    b = np.random.binomial(1, 0.10, 500)
    _, p = stats.ttest_ind(a, b)
    p_values_h0.append(p)

p_values_h0 = np.array(p_values_h0)
print(f"模拟10000次AA实验（两组转化率均为10%，n=500）:")
print(f"  p<0.10 的次数: {(p_values_h0 < 0.10).sum()} ({(p_values_h0 < 0.10).mean():.1%})")
print(f"  p<0.05 的次数: {(p_values_h0 < 0.05).sum()} ({(p_values_h0 < 0.05).mean():.1%})")
print(f"  p<0.01 的次数: {(p_values_h0 < 0.01).sum()} ({(p_values_h0 < 0.01).mean():.1%})")
print(f"\n→ 即使策略完全无效，也有约5%的实验p<0.05！")
print(f"→ 这就是α=0.05的含义：5%的误报率")
print(f"→ 如果公司每天跑100个实验，即使都无效，也会有约5个'显著'")
````

### 步骤1：基础练习
1.1 计算p值并理解其含义
1.2 样本量对p值的影响

````python
print("="*60)
print("步骤1.1：不同场景下的p值")
print("="*60)

np.random.seed(123)

# 场景1：大效应，大样本
c1 = np.random.binomial(1, 0.10, 5000)
e1 = np.random.binomial(1, 0.13, 5000)  # +3pp，大效应
t1, p1 = stats.ttest_ind(c1, e1)
print(f"场景1（大效应+3pp, n=5000）: p={p1:.6f} → {'显著' if p1<0.05 else '不显著'}")

# 场景2：小效应，大样本
c2 = np.random.binomial(1, 0.10, 100000)
e2 = np.random.binomial(1, 0.1005, 100000)  # +0.05pp，极小效应
t2, p2 = stats.ttest_ind(c2, e2)
print(f"场景2（小效应+0.05pp, n=100000）: p={p2:.4f} → {'显著' if p2<0.05 else '不显著'}")
print(f"  → 效应极小(0.05pp)，但因为样本超大，p值也很小")
print(f"  → 这就是'p值≠效果大小'的直观体现！")

# 场景3：大效应，小样本
c3 = np.random.binomial(1, 0.10, 100)
e3 = np.random.binomial(1, 0.18, 100)  # +8pp，大效应！
t3, p3 = stats.ttest_ind(c3, e3)
print(f"场景3（大效应+8pp, n=100）: p={p3:.4f} → {'显著' if p3<0.05 else '不显著'}")
print(f"  → 效应很大(8pp)，但因为样本小，p值仍可能>0.05")
print(f"  → 小样本可能漏掉大的真实效果（第二类错误）")
````

````python
print("="*60)
print("步骤1.2：固定效应下，样本量对p值的影响")
print("="*60)

np.random.seed(456)
true_effect = 0.02  # 固定2pp的真实效应

print(f"真实效应 = 2pp (10%→12%)")
print(f"{'样本量(每组)':<15} {'对照组':<10} {'实验组':<10} {'p值':<10} {'结论':<10}")
print("-" * 55)

for n in [50, 100, 200, 500, 1000, 2000, 5000, 10000]:
    c = np.random.binomial(1, 0.10, n)
    e = np.random.binomial(1, 0.10 + true_effect, n)
    _, p = stats.ttest_ind(c, e)
    conclusion = '显著' if p < 0.05 else '不显著'
    print(f"{n:<15} {c.mean():<10.1%} {e.mean():<10.1%} {p:<10.4f} {conclusion:<10}")

print(f"\n→ 同样的效应量，样本越大p值越小")
print(f"→ 小样本时可能'漏掉'真实效果（第二类错误）")
print(f"→ 大样本时可能'检测出'无业务价值的微小效果")
````

### 步骤2：进阶练习
2.1 计算统计功效（Power）
2.2 效应量与置信区间的计算

````python
print("="*60)
print("步骤2.1：模拟计算统计功效")
print("="*60)

def simulate_power(baseline, mde, n_per_group, alpha=0.05, n_sims=1000):
    """通过模拟计算统计功效"""
    significant_count = 0
    for _ in range(n_sims):
        c = np.random.binomial(1, baseline, n_per_group)
        e = np.random.binomial(1, baseline + mde, n_per_group)
        _, p = stats.ttest_ind(c, e)
        if p < alpha:
            significant_count += 1
    return significant_count / n_sims

baseline = 0.10
mde = 0.02  # 2pp绝对提升

print(f"基线转化率={baseline:.0%}, MDE={mde:.1%}")
print(f"\n{'每组样本量':<12} {'模拟功效':<12} {'是否达标(≥80%)':<15}")
print("-" * 40)

for n in [200, 500, 1000, 2000, 3500, 5000]:
    power = simulate_power(baseline, mde, n, n_sims=500)
    print(f"{n:<12} {power:<12.1%} {'✓' if power >= 0.80 else '✗ 样本不足'}")

print(f"\n→ 检测2pp的提升，每组需要约3500+用户才能达到80%功效")
print(f"→ 如果每组只有1000用户，功效只有约30%——大概率漏报")
````

````python
print("="*60)
print("步骤2.2：效应量与95%置信区间")
print("="*60)

np.random.seed(789)
n = 5000
c = np.random.binomial(1, 0.10, n)
e = np.random.binomial(1, 0.12, n)

# 效应量（绝对差值）
effect = e.mean() - c.mean()

# 合并标准误
se = np.sqrt(c.var()/n + e.var()/n)

# 95%置信区间
ci_lower = effect - 1.96 * se
ci_upper = effect + 1.96 * se

# Cohen's h（比例指标的标准化效应量）
cohens_h = 2 * (np.arcsin(np.sqrt(e.mean())) - np.arcsin(np.sqrt(c.mean())))

print(f"对照组转化率: {c.mean():.2%}")
print(f"实验组转化率: {e.mean():.2%}")
print(f"\n效应量分析:")
print(f"  绝对差值: {effect*100:.2f} 个百分点")
print(f"  相对提升: {(e.mean()/c.mean() - 1)*100:.1f}%")
print(f"  Cohen's h: {cohens_h:.4f}")
print(f"  95% CI: [{ci_lower*100:.2f}pp, {ci_upper*100:.2f}pp]")
print(f"\nCohen's h解读:")
print(f"  |h| < 0.2: 效应很小")
print(f"  0.2 ≤ |h| < 0.5: 效应中等")
print(f"  0.5 ≤ |h| < 0.8: 效应较大")
print(f"  |h| ≥ 0.8: 效应很大")
````

### 步骤3：挑战练习——多重比较问题

3.1 模拟多重比较导致的第一类错误膨胀
3.2 应用Bonferroni校正

````python
print("="*60)
print("步骤3：多重比较问题与Bonferroni校正")
print("="*60)

np.random.seed(999)

# 模拟：一个实验，但检查了10个指标
n_metrics = 10
n_users = 1000
n_sims = 2000

false_positive_no_correction = 0
false_positive_bonferroni = 0

for _ in range(n_sims):
    # H₀为真：所有指标上两组完全相同
    c_data = np.random.binomial(1, 0.10, (n_users, n_metrics))
    e_data = np.random.binomial(1, 0.10, (n_users, n_metrics))

    p_values = []
    for i in range(n_metrics):
        _, p = stats.ttest_ind(c_data[:, i], e_data[:, i])
        p_values.append(p)
    p_values = np.array(p_values)

    # 不校正：任一指标p<0.05即"显著"
    if (p_values < 0.05).any():
        false_positive_no_correction += 1

    # Bonferroni校正：p < 0.05/n_metrics 才算显著
    if (p_values < 0.05 / n_metrics).any():
        false_positive_bonferroni += 1

print(f"模拟{n_sims}次实验，每次检查{n_metrics}个指标（策略均无效）:")
print(f"\n无校正:")
print(f"  至少1个指标'显著'的比例: {false_positive_no_correction/n_sims:.1%}")
print(f"  远高于名义上的5%！")
print(f"\nBonferroni校正:")
print(f"  至少1个指标'显著'的比例: {false_positive_bonferroni/n_sims:.1%}")
print(f"  控制在约5%的水平")
print(f"\n→ 同时检验多个指标时，必须做多重比较校正！")
print(f"→ 这就是为什么AB实验必须预先指定一个主要指标")
````

````python
# 步骤3.2：可视化H₀为真时p值的均匀分布
print("补充验证：H₀为真时，p值服从均匀分布")
print(f"  p<0.25的比例: {(p_values_h0 < 0.25).mean():.2%} (理论25%)")
print(f"  p<0.50的比例: {(p_values_h0 < 0.50).mean():.2%} (理论50%)")
print(f"  p<0.75的比例: {(p_values_h0 < 0.75).mean():.2%} (理论75%)")
print(f"  → 完美符合均匀分布！任何'显著'都可能是纯运气")
````

## 📊 区块4：业务场景应用

### 场景1：p=0.06——"勉强不显著，但我想上线"

**问题**：实验跑完，p=0.06，刚好超过0.05。PM说"差一点点而已，效果看起来不错，上线吧"。你怎么办？

**分析**：
- p=0.06和p=0.05之间没有本质区别——0.05本身就是一个随意选的阈值
- 但你不能"因为差一点点就放松标准"——这次0.06，下次0.07也会有人说"差不多"
- p=0.06意味着如果策略真的无效，有6%的概率看到这样的数据——这个风险你能不能接受？

**决策框架**：
1. 看效应量和置信区间，而不是只看p值
2. 如果效应量大且业务急需，可以考虑延长实验收集更多数据
3. 如果效应量小，p=0.06和p=0.04都不该上线——因为没有业务价值
4. 建立"实验门禁"：预先约定决策规则，避免事后讨价还价

### 场景2：A/B/n测试的多重比较陷阱

**问题**：产品同时测试4个新版（B/C/D/E）vs旧版（A），发现C组p=0.03"显著"。能直接上线C吗？

**分析**：这是典型的多重比较问题。你做了4次比较（B vs A, C vs A, D vs A, E vs A）。

**计算**：
```
P(至少1个误报) = 1 - (1-0.05)^4 ≈ 18.5%
```
几乎1/5的概率出现假阳性！

**正确做法**：
1. 使用ANOVA先检验"各组之间是否有任何差异"
2. 如果ANOVA显著，再用Bonferroni校正做两两比较（α'=0.05/4=0.0125）
3. 或者用Dunnett检验（专门比较多个实验组 vs 一个对照组）

**决策**：C组原始p=0.03 > Bonferroni阈值0.0125 → 不显著 → 不上线。

### 场景3：大样本下的"假显著"——p值陷阱

**问题**：某大厂每天有500万DAU，一个实验跑了2周。结果是：转化率从20.00%提升到20.05%，p<0.001。要不要上线？

**分析**：
- p<0.001说明这个差异不是随机波动——0.05pp的提升是"真实"的
- 但0.05pp的提升有业务意义吗？20%的基准下，0.05pp意味着每2000个用户多1个转化
- 如果上线需要工程成本（维护新代码、增加系统复杂度），可能不划算

**决策**：不上线。不是因为"不显著"，而是因为"效应太小，没有业务价值"。

**教训**：大公司的AB实验经常遇到这个问题——样本量太大，任何微小差异都"显著"。需要引入**实用显著性（Practical Significance）**的概念：预先定义一个最小业务意义阈值，即使p<0.05，如果效应低于阈值也不上线。

### 场景4：两类错误的业务代价不对称

**问题**：一个涉及用户隐私的功能改动。第一类错误（误上线）的代价远大于第二类错误（漏掉一个好功能）。如何调整？

**分析**：α=0.05是惯例，但不是金科玉律。根据业务场景调整α。

**决策框架**：
- 高风险决策（隐私、支付、安全）：α=0.01甚至0.001（宁可漏掉，不能误上线）
- 常规决策（UI优化、文案测试）：α=0.05（标准实践）
- 低风险决策（内容推荐排序微调）：α=0.10（宁可误上线，不能漏掉好策略）

**关键原则**：α的选择不是统计问题，是业务决策问题——取决于两类错误的相对代价。

## 📝 区块7：当日小结

> **假设检验是连接"数据差异"和"统计结论"的桥梁。p值衡量证据强度而非效果大小——p<0.05意味着"如果策略无效，看到这种数据的概率小于5%"，而非"策略有95%的概率有效"。第一类错误（α，误报）和第二类错误（β，漏报）是统计决策的两种风险。永远报告三件套：p值+效应量+置信区间。多重比较和p-hacking是数据分析中最常见的自我欺骗行为。**

**三条金句带回家**：
1. p值告诉你"惊讶程度"，不是"真实概率"——P(数据|H₀) ≠ P(H₀|数据)
2. 大样本下一切皆显著——永远同时报告效应量和置信区间
3. 同时检验10个指标，假阳性概率从5%飙升到40%——多重比较校正不是可选项
