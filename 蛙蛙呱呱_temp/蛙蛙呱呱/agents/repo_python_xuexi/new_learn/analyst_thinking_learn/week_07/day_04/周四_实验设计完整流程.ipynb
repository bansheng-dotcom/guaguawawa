## 📘 区块1：核心概念与第一性原理

### 1.1 实验的完整生命周期

```
1. 定义假设 → 2. 设计实验 → 3. 计算样本 → 4. 随机分流 →
5. 运行实验 → 6. 监控数据 → 7. 停止实验 → 8. 分析结果 → 9. 做出决策
```

这9步构成了AB实验的完整生命周期。任何一步出问题，整个实验的结论都不可信。**实验设计80%的价值在"开始之前"——把假设写清楚、把样本量算清楚、把指标定义清楚。这些做好了，跑出来的结果自然可信。没做好，跑出来什么结果都不可信。**

### 1.2 为什么需要标准化流程？

在没有标准化流程的团队中，AB实验往往是这样的：
- PM："这个按钮换个颜色测一下" → 没有假设文档
- 工程师：上线分流，没算样本量
- 分析师：PM跑来说"帮我看看这个实验的数据"
- 一周后PM又问："显著了吗？"分析师说"样本不够"
- 又一周后PM："p=0.03，显著了，上线！"——但这是第14次看p值

标准化流程的价值就是让每一步都有据可查，让"凭感觉"变成"凭流程"，让结论可复现、可审计。

### 1.3 实验的"死亡"方式——为什么大多数实验不值得信？

大多数实验不是"结果不显著"死掉的，而是被以下方式"谋杀"的：
1. **SRM（样本比例不匹配）**：分流系统有bug，两组人数比例不对 → 实验无效
2. **Peeking**：每天看p值→看到p<0.05就停→假阳性率从5%飙升到20%+
3. **指标偷换**：主要指标不显著，换一个"恰好显著"的指标 → p-hacking
4. **幸存者偏差**：只报告"显著"的实验，不报告"不显著"的 → 文件抽屉问题
5. **没有护栏指标**：转化率涨了但卸载率也涨了——只看到了好处

### 1.4 实验设计文档（Experiment Design Doc）

**为什么重要**：这是防止事后诸葛亮和p-hacking的最佳工具。

一个完整的实验设计文档应该包含：
1. **实验名称和负责人**
2. **业务假设**（我们相信"做什么"会导致"什么指标"发生"什么变化"）
3. **实验设计**（随机化单元、分流比例、分组方式）
4. **指标体系**（主要指标、次要指标、护栏指标——明确定义）
5. **样本量计算**（基线、MDE、α、Power、所需n、预计天数）
6. **分析计划**（用什么检验、是否分层、如何定义离群值）
7. **决策规则**（什么条件下上线、什么条件下不上线、什么条件下延长）
8. **风险与回滚计划**（如果护栏指标恶化怎么办）

> **今天最重要的认知**：一个AB实验的可靠性，80%在"开跑之前"就决定了。把假设、样本量、指标定义、决策规则写清楚——做对了前面，结果自然可信。做错了前面，跑出来的任何数字都不值得相信。

### 1.5 面试高频考点

**面试官常问**：
- "请设计一个实验来评估XX策略的效果"（全流程设计题）
- "实验设计中最容易出错的环节是哪些？"
- "什么是护栏指标？为什么需要？"
- "SRM是什么？如何处理？"

**回答要点**：
- 9步全流程：假设→设计→样本→分流→运行→监控→停止→分析→决策
- 最容易出错：样本量不计算、Peeking、没有护栏指标
- 护栏指标是不能变差的指标（崩溃率、卸载率、核心体验指标）
- SRM用卡方检验检测，存在SRM时实验结论无效

## 🧩 区块2：知识点详细拆解——9步全流程

### 2.1 步骤1：定义假设（Day 0）

**为什么重要**：假设不清晰是实验失败的第一大原因。

**一个好的假设包含三个部分**：
```
我们相信 [做什么]
会导致 [什么指标]
发生 [什么变化]
因为我们观察到 [什么证据]
```

**坏的假设 vs 好的假设**：
- 坏："看看新版好不好"（没有明确预期）
- 好："我们相信在注册页增加社交证明（展示'已有XX万用户'），会导致注册完成率提升至少5%（从20%→21%），因为我们观察到用户在最后一步流失率最高，且用户调研显示'信任'是主要障碍"

**假设的"SMART"原则**：
- Specific（具体的）：什么改动
- Measurable（可度量的）：什么指标
- Achievable（可实现的）：MDE在可检测范围内
- Relevant（相关的）：指标与业务目标对齐
- Time-bound（有时限的）：预期多久看到效果

### 2.2 步骤2：设计实验

**为什么重要**：设计决策直接影响实验的有效性和效率。

**关键设计决策**：

| 决策 | 选项 | 考虑因素 |
|------|------|----------|
| 随机化单元 | 用户/设备/会话/城市 | 登录态、SUTVA、样本量 |
| 分流比例 | 50:50 / 90:10 / 其他 | 风险、速度、统计效率 |
| 对照组 | 旧版/空白对照 | 伦理、用户体验 |
| 实验组数 | A/B / A/B/n | 多重比较问题 |
| 分层 | 是/否 | 是否需要保证关键维度均衡 |

**分流比例的选择**：
- 50:50：统计效率最高（相同总样本下Power最大）
- 90:10：更安全，但统计效率低（实验组样本少）
- 80:20：常用于首次测试风险较高的策略

### 2.3 步骤3：计算样本量

**为什么重要**：这是从"拍脑袋"到"有依据"的关键一步。

**计算流程**（详见W7D3）：
1. 确定基线指标水平（从历史数据获取）
2. 确定MDE（与PM对齐——多小的影响值得检测？）
3. 设定α=0.05, Power=80%（常规值）
4. 选择公式（比例/连续指标）
5. 计算n
6. 转换为实验天数（n / 日活流量 / 分流比）

### 2.4 步骤4：随机分流

**为什么重要**：分流有bug → 随机性被破坏 → 实验前后两组不可比 → 结论无效。

**分流系统的技术要求**：
1. **确定性**：同一用户永远在同一组（用哈希）
2. **独立性**：不同实验之间独立分流
3. **均匀性**：各组人数比例接近预定的分流比
4. **持久性**：用户重新登录不改变分组

**分流检查清单**：
- [ ] 同一用户多次进入是否始终在同一组？
- [ ] 不同实验之间用户分配是否独立？
- [ ] 分流代码是否与业务代码解耦？
- [ ] 是否有AA实验验证过分流系统？

### 2.5 步骤5：运行实验

**为什么重要**：这是最长的阶段，也是Peeking最容易发生的阶段。

**运行期间的行为规范**：
- [ ] 不中途看p值做决策（可以看数据完整性、数据质量）
- [ ] 不修改分流比例
- [ ] 不修改实验配置（除非发现严重bug）
- [ ] 监控护栏指标（不是为了看"有没有效果"，而是为了看"有没有伤害"）
- [ ] 记录实验日志（什么时间做了什么操作）

**数据质量监控**（可以看的部分）：
- 数据上报量是否正常（有没有埋点bug）
- 两组人数比例是否正确（SRM检测）
- 用户在两组之间的特征是否均衡

### 2.6 步骤6：监控数据——与Peeking的区别

**为什么重要**：很多人分不清"合法监控"和"非法Peeking"。

| 行为 | 合法监控 | 非法Peeking |
|------|---------|------------|
| 看数据完整性 | ✓ | — |
| 看SRM | ✓ | — |
| 看护栏指标（安全） | ✓ | — |
| 看主要指标的p值 | ✗ | ✗ |
| 基于p值决定停止 | ✗ | ✗ |

**护栏指标的自动告警**：
如果护栏指标显著恶化（如崩溃率p<0.01），应该自动触发告警，甚至自动停止实验。这与"看到主要指标p<0.05就停"完全不同——护栏告警是为安全，不是为了"抢先看结果"。

### 2.7 步骤7：停止实验

**为什么重要**：什么时候停是一个需要预先定义的决策。

**停止条件（预先定义，写入实验文档）**：
1. **达成样本量**：达到预先计算的n → 正常停止
2. **安全停止**：护栏指标显著恶化 → 紧急停止
3. **时间上限**：达到预设最大天数但样本量不够 → 考虑是否延长
4. **效果极大**：效果远超预期且效应量稳定 → 考虑提前停止（需用序贯检验框架）

**不应该作为停止条件的**：
- "p<0.05了就停" —— 这是Peeking
- "老板说停了" —— 除非是安全原因
- "数据看起来差不多了" —— 主观判断

### 2.8 步骤8：分析结果

**为什么重要**：分析结果的完整性和准确性直接影响决策质量。

**标准输出（分析报告的必要组成部分）**：

1. **SRM检验**：卡方检验，确认分流比例是否正确
2. **主要指标**：p值 + 效应量 + 95%置信区间
3. **次要指标**：同上（但注意多重比较校正）
4. **护栏指标**：确认没有显著恶化
5. **异质性分析**：不同用户群的效果是否一致？
6. **数据质量**：缺失率、异常值、数据上报延迟

**分析清单**：
- [ ] SRM通过（p > 0.05 for 卡方检验）
- [ ] 主要指标：效应量+CI+p值
- [ ] 护栏指标：全部通过
- [ ] 没有指标偷换（用的是事前定义的主要指标）
- [ ] 没有子群捞鱼（子群分析标注为探索性）

### 2.9 步骤9：做出决策

**为什么重要**：实验最终的目的是决策，不是"p<0.05"。

**决策框架（3×3矩阵）**：

| | 效应量大且有业务价值 | 效应量小或无业务价值 |
|------|-------------------|---------------------|
| **显著** | ✓ 上线 | 考虑（可能上线但不值得）|
| **不显著** | 延长实验/增加样本 | ✗ 不上线 |
| **显著但护栏恶化** | ✗ 不上线（一票否决） | ✗ 不上线 |

**决策规则示例**：
```
上线条件：
  1. 主要指标 p < 0.05
  2. 效应量 > 实用显著性阈值（如+1pp）
  3. 所有护栏指标未显著恶化
  4. SRM检验通过

不上线条件：
  1. 任一护栏指标显著恶化 → 一票否决
  2. p ≥ 0.05 且效应量小 → 不上线

延长实验条件：
  1. p接近0.05（如0.05<p<0.10）
  2. 效应量有业务价值
  3. 还有流量余量
```

## 💻 区块3：动手练习——走一遍完整流程

````python
import numpy as np
from scipy import stats
np.random.seed(42)

print("="*60)
print("完整AB实验流程模拟")
print("场景：测试'新推荐算法'是否能提升下单转化率")
print("="*60)

# ===== 步骤1：定义假设 =====
print("\n步骤1：定义假设")
print("  我们相信：新推荐算法（基于协同过滤）")
print("  会导致：下单转化率提升")
print("  预期变化：从10.0%提升到10.5%（+5%相对提升）")
print("  证据：离线评估中NDCG@10提升8%，但需要在线验证")

# ===== 步骤2：设计实验 =====
print("\n步骤2：设计实验")
print("  随机化单元: 用户级（基于user_id哈希）")
print("  分流比例: 50:50")
print("  对照组: 旧推荐算法")
print("  实验组: 新推荐算法")

# ===== 步骤3：计算样本量 =====
print("\n步骤3：计算样本量")
baseline = 0.10
mde_absolute = 0.005  # 0.5pp绝对提升
alpha = 0.05
power = 0.80

z_alpha = stats.norm.ppf(1 - alpha/2)
z_beta = stats.norm.ppf(power)
p1, p2 = baseline, baseline + mde_absolute
p_bar = (p1 + p2) / 2
n_per_group = int(np.ceil(
    (z_alpha * np.sqrt(2*p_bar*(1-p_bar)) +
     z_beta * np.sqrt(p1*(1-p1) + p2*(1-p2)))**2 / mde_absolute**2
))
print(f"  基线={baseline:.0%}, MDE={mde_absolute:.1%}, α={alpha}, Power={power:.0%}")
print(f"  每组需要: {n_per_group:,} 用户")
print(f"  两组共计: {n_per_group*2:,} 用户")
print(f"  预估时长: 每天约5000用户 → {np.ceil(n_per_group/2500):.0f}天")

# ===== 步骤4-5：模拟分流和运行 =====
print("\n步骤4-5：随机分流并模拟实验运行")
n = n_per_group
control = np.random.binomial(1, baseline, n)
experiment = np.random.binomial(1, baseline + mde_absolute, n)

# ===== 步骤6：SRM检验 =====
print("\n步骤6：SRM检验（样本比例不匹配检测）")
expected_ratio = 0.5
observed_control = n
observed_experiment = n
total = observed_control + observed_experiment
expected = total * expected_ratio
chi2 = (observed_control - expected)**2 / expected + (observed_experiment - expected)**2 / expected
p_srm = 1 - stats.chi2.cdf(chi2, 1)
print(f"  对照组: {observed_control}, 实验组: {observed_experiment}")
print(f"  卡方={chi2:.4f}, p={p_srm:.4f}")
print(f"  → {'✗ SRM！分流有问题！' if p_srm < 0.05 else '✓ SRM通过'}")

# ===== 步骤7：分析结果 =====
print("\n步骤7：分析实验结果")
t_stat, p_value = stats.ttest_ind(experiment, control)
effect = experiment.mean() - control.mean()
effect_relative = (experiment.mean() / control.mean() - 1)
se = np.sqrt(control.var()/n + experiment.var()/n)
ci = stats.t.interval(0.95, 2*n-2, loc=effect, scale=se)

print(f"  对照组转化率: {control.mean():.3%}")
print(f"  实验组转化率: {experiment.mean():.3%}")
print(f"  绝对差值: {effect*100:.2f}pp")
print(f"  相对提升: {effect_relative*100:.1f}%")
print(f"  95% CI: [{ci[0]*100:.2f}pp, {ci[1]*100:.2f}pp]")
print(f"  p值: {p_value:.4f} → {'显著' if p_value < 0.05 else '不显著'}")

# Cohen's h
cohens_h = 2 * (np.arcsin(np.sqrt(experiment.mean())) - np.arcsin(np.sqrt(control.mean())))
print(f"  Cohen's h: {cohens_h:.4f} (|h|>0.2=小效应, >0.5=中效应, >0.8=大效应)")

# ===== 步骤8：护栏指标模拟 =====
print("\n步骤8：护栏指标检查")
# 模拟护栏指标：崩溃率（希望不变差）
# H₀为真：两组崩溃率相同（都是2%）
crash_control = np.random.binomial(1, 0.02, n)
crash_experiment = np.random.binomial(1, 0.02, n)
_, p_crash = stats.ttest_ind(crash_experiment, crash_control)
print(f"  崩溃率: 对照={crash_control.mean():.2%}, 实验={crash_experiment.mean():.2%}")
print(f"  p={p_crash:.4f} → {'⚠ 护栏恶化！' if p_crash < 0.05 else '✓ 护栏通过'}")

# ===== 步骤9：决策 =====
print("\n步骤9：做出决策")
print("="*40)

practical_significance = 0.003  # 实用显著性阈值：0.3pp

if p_crash < 0.05:
    print("决策: ✗ 不上线（护栏指标恶化——一票否决）")
elif p_value < 0.05 and effect > practical_significance:
    print(f"决策: ✓ 建议上线新版推荐算法")
    print(f"  预期提升: {ci[0]*100:.1f}pp ~ {ci[1]*100:.1f}pp")
    print(f"  后续: 全量上线后继续监控7天")
elif p_value < 0.05 and effect <= practical_significance:
    print(f"决策: △ 显著但效应太小（<{practical_significance:.1%}），建议不上线")
elif p_value >= 0.05 and effect > practical_significance:
    print(f"决策: △ 效应有业务价值但不显著，建议延长实验")
else:
    print(f"决策: ✗ 不上线（不显著且效应小）")

print("\n→ 注意：决策不只是看p值，而是综合效应量+护栏+业务价值")
````

### 步骤2：进阶练习——异质性分析

2.1 按平台分层分析
2.2 评估实验在不同用户群中的效果差异

````python
print("="*60)
print("进阶：异质性分析——不同平台的效果是否一致？")
print("="*60)

np.random.seed(123)

# 模拟：iOS和Android用户可能有不同的效果
n_ios = n_per_group // 3  # 1/3 iOS用户
n_android = n_per_group - n_ios  # 2/3 Android用户

# iOS上效果更大（+1pp）
control_ios = np.random.binomial(1, baseline, n_ios)
exp_ios = np.random.binomial(1, baseline + 0.01, n_ios)  # +1pp

# Android上效果较小（+0.3pp）
control_android = np.random.binomial(1, baseline, n_android)
exp_android = np.random.binomial(1, baseline + 0.003, n_android)  # +0.3pp

print(f"\n{'平台':<10} {'对照组':<10} {'实验组':<10} {'差值':<10} {'p值':<10}")
print("-" * 50)

for platform, c, e in [("iOS", control_ios, exp_ios), ("Android", control_android, exp_android)]:
    _, p = stats.ttest_ind(e, c)
    diff = e.mean() - c.mean()
    print(f"{platform:<10} {c.mean():<10.2%} {e.mean():<10.2%} {diff*100:<10.2f}pp {p:<10.4f}")

print(f"\n→ iOS效果更大且显著，Android效果较小且不显著")
print(f"→ 可能的解释：iOS用户对新算法的响应不同？还是样本量差异？")
print(f"→ 注意：子群分析标注为'探索性分析'，不能作为主要结论")
````

## 📊 区块4：业务场景应用

### 场景1：实验跑完才发现两组人数不对——SRM危机

**问题**：某实验计划50:50分流，跑了两周后发现对照组60%实验组40%。实验组转化率显著更高。PM说"效果很好，快上线"。

**分析**：
1. 先做SRM检验——几乎肯定会显著（60:40不可能是50:50的随机波动）
2. SRM存在时，不要看任何业务指标——因为分组已经不是随机的了
3. 排查原因：分流代码bug？某组用户被异常排除？

**根因**：实验组的新算法有个bug，在某些旧设备上导致加载失败，这部分用户被"自然排除"了。结果实验组剩下的都是新设备用户（转化率本身就高）——所以"实验组转化率高"不是算法的效果，而是设备的选择偏误。

**决策**：实验结果作废，修bug后重新实验。

**教训**：SRM是实验的"红灯"——SRM一票否决，不需要看任何其他结果。

### 场景2：PM想偷换指标——"能不能换一个？"

**问题**：实验跑完，主要指标（下单转化率）p=0.12，不显著。PM说"但是点击率涨了12%，p=0.001，我们把主要指标改成点击率吧"。

**分析**：这是典型的指标偷换（Metric Switcheroo）。
- 点击率可能是"中间指标"——用户点击多了但最终没有更多下单
- 也可能点击率的提升是"新奇效应"——用户因为新鲜而点击，但不会持续
- 如果事前没有定义点击率为主要指标，事后就不能改成主要指标

**正确做法**：
1. 报告：主要指标（下单转化率）不显著，次要指标（点击率）显著
2. 解读：新算法让用户更愿意点击推荐，但没有转化为更多下单
3. 下一步：优化从"点击推荐"到"下单"的转化路径，然后再做实验

**决策**：不以上线为目的扭曲分析。诚实报告结果，提炼学习，设计下一步策略。

### 场景3：护栏指标救了产品一命

**问题**：某社交App测试"更激进的推送策略"。主要指标DAU涨了8%，p<0.001。看起来大获成功。

**但是**：护栏指标显示——推送通知关闭率从3%涨到15%（p<0.001）。

**分析**：
- DAU涨了是因为推送把用户拉回来了
- 但大量用户选择关闭推送——这是长期价值的损失
- 这些关闭推送的用户，以后所有推送策略都触达不了他们

**决策**：不上线。虽然短期DAU涨了，但推送关闭率的暴涨是不可逆的长期伤害。需要重新设计推送策略（减少频率、提高精准度）。

**教训**：护栏指标的一票否决权是防止"指标近视"的最后防线。只看主要指标不看护栏 = 只看速度表不看油箱。

### 场景4：实验设计的"黄金法则"被打破后

**问题**：某公司CEO说"我们上季度跑了50个实验，30个显著，25个上线了"。CEO觉得很好，但你觉得有问题。

**分析**：
1. 50个实验，30个显著 = 60%的"胜率"——远高于业界10-20%的平均水平
2. 高"胜率"的原因可能是：
   - Peeking和指标偷换（最可能）
   - 只报告成功的实验，不报告失败的
   - 选择性记忆（成功的被记住了，失败的被忘记了）
3. 业界的"10-20%胜率"是指真正有效果的实验比例。大部分想法不会work——这是正常的

**正确理解实验文化**：
- 实验的目标不是"证明自己是对的"，而是"找出什么是有效的"
- 一个"不显著"的实验不是失败——它帮你排除了一条路
- 高"胜率"往往意味着你的假设不够大胆（只在安全区域做实验）

**决策**：建立实验数据库，记录所有实验（包括不显著的和失败的），定期复盘胜率。推动"学习型"而非"证明型"的实验文化。
