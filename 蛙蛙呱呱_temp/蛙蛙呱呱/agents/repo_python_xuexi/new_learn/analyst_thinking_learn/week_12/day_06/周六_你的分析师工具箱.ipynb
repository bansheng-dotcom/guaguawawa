## 🧩 区块2：工具清单详解

### 2.1 思维框架类（"用什么想"）

#### ISS框架——面对陌生问题的万能钥匙
- **I**(Identify)：界定问题——谁的问题、什么时候开始的、影响了什么
- **S**(Segment)：拆解定位——用指标拆解树找到问题出在哪一层
- **S**(Solve)：方案——验证假设、给出行动建议
- **适用场景**：任何你不确定"该从哪入手"的问题
- **来源**：W3D2
- **常见错误**：跳过I直接拉数据——错了方向再多的数据也没用

#### OSM框架——从目标推导度量
- **O**(Objective)：业务目标（如"提升用户粘性"）
- **S**(Strategy)：策略（如"优化新用户引导"）
- **M**(Measurement)：度量（如"新用户次日留存率"）
- **适用场景**：制定指标体系、对齐目标
- **来源**：W1D2

#### AARRR漏斗——增长全链路诊断
- **A**cquisition(获取) → **A**ctivation(激活) → **R**etention(留存) → **R**evenue(变现) → **R**eferral(传播)
- **适用场景**：诊断增长瓶颈——卡在获取、激活还是留存？
- **来源**：W2D1
- **常见错误**：只看总量不看转化率——1000人进来500人激活=50%，和10000人进来2000人激活=20%——总量在涨但漏斗在变宽

#### 指标拆解——定位指标变化的底层原因
- **乘法拆解**：GMV = DAU × 付费率 × ARPU
- **加法拆解**：DAU = 新增 + 留存 + 回流
- **适用场景**：任何指标发生变化时
- **来源**：W1D3
- **黄金法则**：至少拆2层——第一层看"总量"，第二层看"结构"

#### PDS拆解——DAU内部结构诊断
- **P**(Primary new)：新增用户 → 获客能力
- **D**(Deferred)：留存用户 → 产品粘性
- **S**(Salvage)：回流用户 → 召回能力
- **适用场景**：DAU波动、诊断增长质量
- **来源**：W2D5

#### 同期群分析（Cohort）——不同批次用户的质量对比
- 按注册日期/首次使用日期分组，纵向看每组的留存/付费变化
- **适用场景**：新用户质量评估、产品改版效果、渠道质量对比
- **来源**：W2D2
- **核心洞察**：同期群能回答"是最近来的用户质量差了，还是一直都这样"——单日数据无法回答这个问题

#### 漏斗分析——找出转化最大的"漏水点"
- 从曝光→点击→加入购物车→下单→支付
- **适用场景**：转化率分析、用户行为路径分析
- **来源**：W2D3
- **关键方法**：看"绝对流失量"最大的环节（不是"转化率"最低的——1%到0.8%的差异可能比10%到5%影响更大）

#### RFM分群——用户价值分层
- **R**(Recency)：最近一次消费时间
- **F**(Frequency)：消费频率
- **M**(Monetary)：消费金额
- **适用场景**：用户分群、差异化运营、召回策略
- **来源**：W2D6

#### 归因模型——渠道贡献分配
- 末次点击 / 首次点击 / 线性 / 时间衰减 / 位置模型 / 数据驱动
- **适用场景**：广告渠道效果评估、预算分配
- **来源**：W3D1
- **核心认知**：末次点击归因会高估"收割渠道"(搜索)，低估"播种渠道"(信息流/社媒)

#### 金字塔原理——分析报告结构
- 结论先行 → 以上统下 → 归类分组 → 逻辑递进
- **适用场景**：所有分析报告和汇报
- **来源**：W10D6

### 2.2 技术方法类（"用什么做"）

#### SQL窗口函数家族
| 函数 | 用途 | 关键字 |
|------|------|--------|
| ROW_NUMBER | 严格排名（不管并列） | 取Top-N |
| RANK | 跳跃排名（并列跳号） | 排行榜 |
| DENSE_RANK | 连续排名（并列不跳号） | 等级划分 |
| LAG/LEAD | 取前/后行数据 | 环比/同比/时间间隔 |
| SUM OVER | 累计值计算 | 累计GMV/累计用户 |
| AVG OVER + ROWS | 移动平均 | 滚动指标 |
| FIRST_VALUE/LAST_VALUE | 窗口首尾值 | 首次/末次购买 |

#### Gap-and-Island算法
- 三步法：去重 → 编号(ROW_NUMBER) → 日期减编号 → GROUP BY
- **适用场景**：连续行为分析——连续登录、连续购买、连续活跃
- **来源**：W5D2

#### CTE分层写法
- 每个CTE代表一个逻辑步骤，命名清晰
- **适用场景**：超过2层嵌套的复杂SQL
- **来源**：W5D4

#### SQL优化工具箱
- EXPLAIN分析执行计划
- 创建合适的索引（单列/复合/覆盖）
- 避免索引失效（LIKE '%x'/函数/类型不匹配/OR）
- 游标分页代替OFFSET分页
- 大表JOIN先过滤再连接
- **来源**：W5D7

#### AB实验全流程
1. 样本量计算（基线/MDE/α/Power）
2. 随机分流 + SRM检查
3. 护栏指标配置
4. 预设停止规则（跑多久/多少样本/序贯检验）
5. 一次性分析（非Peeking）
6. 报告效应量+显著性
- **来源**：W7-W8

#### 因果推断四种武器
| 方法 | 一句话总结 | 前提假设 | 典型场景 |
|------|----------|---------|----------|
| DID | 看处理组相对对照组在"前后"的变化差异 | 平行趋势 | 政策评估 |
| PSM | 找特征相似的"对子"比 | 条件独立 | 用户行为 |
| IV | 用工具变量消除内生性 | 相关+排他 | 价格弹性 |
| RDD | 断点附近的"准随机" | 局部随机化 | 分数线 |

#### 数据质量嗅探
- 5分钟快速检查：空值率/重复值/异常值/数据分布/数据新鲜度
- **来源**：W3D4

### 2.3 业务敏感度类（"警惕什么"）

| 警惕什么 | 为什么警惕 | 典型表现 | 怎么发现 |
|---------|----------|---------|----------|
| 虚荣指标 | 看起来在涨但对业务没帮助 | DAU涨了但付费不涨 | 和业务结果指标对比 |
| Peeking使α失控 | 提前停实验让假阳性率飙升 | p值刚小于0.05就停 | 看实验日志中的检查次数 |
| JOIN行数膨胀 | 一对多JOIN让统计结果翻倍 | COUNT后发现总数不对 | 检查JOIN前后的行数 |
| 数据倾斜 | 大数据量下某些节点被打垮 | 任务跑到99%卡住 | 看执行日志的时间分布 |
| 归因模型偏差 | 模型选错→预算分错了 | 信息流ROI被低估 | 用2种模型重算看差异 |
| 补贴侵蚀利润 | 花钱买GMV但利润在降 | 促销期间GMV涨了但毛利跌了 | 看GMV和毛利的时间差 |
| 相关≠因果 | 把同时发生的当因果关系 | 冰淇淋销量和溺水人数都涨 | 问"还有别的解释吗？" |
| 新老用户混淆 | 把老用户质量当新产品效果 | "新功能上线后留存提升"——可能只是老用户在活跃 | 分开看新老用户 |
| 幸存者偏差 | 只看到了留下的，没看到走掉的 | "活跃用户都很满意"——因为不满意的已经走了 | 看流失用户的最后行为 |
| 辛普森悖论 | 分组趋势和总体趋势相反 | 每个渠道转化率都在涨但总体跌了 | 分维度看数据 |

你的回答："先用RFM给用户分层——R是Recency(FROM最后一次购买天数)，F是Frequency(购买次数)，M是Monetary(累计消费金额)。分成2×2×2=8组后，每一组有不同的运营策略。比如'高R高F高M'是Core用户——别打扰他们。'低R低F高M'是高潜用户——推他们复购。你先把这个做出来，有问题随时问我。"

你不是给了ta一个答案——你是给了ta一个**工具**。ta下次遇到"用户分群"的场景，就知道用RFM了。

### 场景3：面试——在30秒内展现你的"工具箱广度"

面试官问："你用什么方法评估渠道效果？"

差的回答："我看ROI。哪个渠道ROI高就投哪个。"

好的回答："我一般从三个角度评估：①归因模型——至少用首次点击和末次点击两种模型算，看同一渠道在不同模型下的ROI差异有多大 ②同期群——不只比ROI，还比不同渠道来的用户的留存和LTV ③增量实验——如果条件允许，在设计时加入'暂停测试'来判断渠道的边际贡献。这样做过几次后发现，末次点击往往会严重低估信息流这类'播种渠道'的价值。"

### 横向连接：工具的"组合使用"模式

单个工具的威力有限，真正的分析能力来自工具的**组合使用**：
- 指标拆解 + 同期群 = 不仅知道GMV跌了，还知道是"新用户的付费率在跌"
- AARRR + 漏斗 = 不仅知道用户在流失，还知道"从激活到付费是最大断层"
- PDS + 归因 = 不仅知道DAU以新增为主，还知道"信息流贡献了大部分新增但质量最差"
- ISS + 指标拆解 + AB实验 = 完整的"问题定义→假设→验证"科学方法链

**核心认知**：工具之间不是独立的——它们是为同一个分析目标服务的不同武器。就像外科手术——你不会只用手术刀，你会根据手术类型选择不同的器械组合。
