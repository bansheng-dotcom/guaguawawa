## 📘 区块1：核心概念与第一性原理

### 1.1 为什么需要"知识盲区排查"？

学完11周的课程后，每位学员的知识结构是不同的。同样的课程内容，有人对AB实验烂熟于心但对因果推断一知半解，有人SQL能力突出但商业分析框架薄弱。课程能给你的是"接触过"所有模块，但"掌握"的程度取决于你的练习量和背景。

知识盲区排查的本质是**结构化自省**——用系统化的方式识别"我以为我懂了"和"我真的懂了"之间的差距。

### 1.2 从第一性原理推导：为什么要用"能独立输出"作为检验标准？

学习科学中有一个概念叫"流畅度错觉"（Fluency Illusion）——你读完笔记觉得"这我懂了"，但合上笔记发现写不出来。这是因为"识别"和"回忆"是两种不同的认知过程：

- **识别**：看到答案时能判断这是对的——这是被动能力
- **回忆**：没有提示的情况下独立产出答案——这是主动能力

面试、工作中的实际问题不会给你四个选项让你选——你需要从零开始构建解决方案。所以我们的检验标准必须是"能独立输出"，而非"能看懂笔记"。

### 1.3 盲区排查的三层模型

我们把知识掌握程度分为三个层次：

| 层次 | 描述 | 典型表现 | 检验方法 |
|------|------|----------|----------|
| 🟢 能独立输出 | 闭卷能写出代码/框架/分析方案 | 被问到"怎么做留存分析"能直接说出步骤 | 不看笔记独立完成练习 |
| 🟡 知道但做不出来 | 看到笔记能回忆，但独立写会卡住 | 知道窗口函数有ROW_NUMBER，但写不出gap-and-island | 看笔记能做，不看不行 |
| 🔴 完全不会 | 学的时候就没理解 | 不记得有这个知识点或完全无法解释 | 看笔记也不理解 |

### 1.4 为什么大部分人高估自己的掌握程度？

心理学中的"达克效应"（Dunning-Kruger Effect）告诉我们：能力越低的人越容易高估自己。在数据分析学习中，这个效应同样存在——因为学习材料是结构化的、有参考答案的，你在学习时面对的永远是"有提示的环境"。但真实工作中没有提示——你需要自己判断该用什么方法。

**典型的高估信号**：
- "这个我学过了"——但能独立做出来吗？
- "笔记里都有"——但你能不看笔记说出来吗？
- "逻辑我懂了"——但能把逻辑翻译成SQL/Python吗？

### 1.5 面试高频考点

**面试官常问**（关于自我认知类）：
- "你觉得你最擅长的分析领域是什么？最薄弱的是什么？"
- "说一个你曾经做错的分析，你是怎么发现的？"
- "如果让你给自己在AB实验方面的能力打分（1-10），你打几分？为什么？"

**回答要点**：
- 坦诚面对弱点比假装全能更可信
- 展现"知道自己不知道什么"的元认知能力
- 描述具体的学习和补漏方法而不仅仅是承认不足

### 2.3 最容易被高估的五个模块

**为什么重要**：这五个模块是"流畅度错觉"最高发的地方。

| 模块 | 为什么容易被高估 | 真正的检验标准 |
|------|----------------|---------------|
| 窗口函数 | 看笔记觉得简单，但实际场景千变万化 | 不看笔记写出gap-and-island |
| AB实验样本量 | 知道有公式，但不会手算 | 给定基线、MDE、α、Power，能算出所需样本量 |
| 归因模型 | 会用末次点击，但不会选模型 | 能说出至少3种归因模型的适用场景和局限性 |
| 因果推断 | 知道DID/PSM的名字，但不会实现 | 能说出每种方法的前提假设和检验方法 |
| 数据分析报告 | 觉得"把数据放PPT就行" | 能写出"结论优先、有行动建议"的分析报告 |

### 2.4 如何进行有效的自我评估

**为什么重要**：自评不准确是最大的问题——你需要客观的检验标准。

**三档自评标准**：

**🟢 熟练（能独立输出）**：
- 不看任何参考资料，能独立写出代码/框架
- 能向非技术同事解释这个概念的直觉
- 能识别别人在这个模块上犯的典型错误
- 能在30秒内判断一个业务问题该用什么方法

**🟡 了解（知道但不能独立输出）**：
- 能看懂相关的代码和分析
- 对着笔记能复现
- 知道大概在说什么，但无法独立写出完整方案
- 知道概念但解释不清楚为什么

**🔴 不会（完全陌生或学的时候就卡住了）**：
- 听到术语需要查才知道是什么
- 学的时候就没跟上
- 看笔记也理解不了
- 无法判断什么场景该用这个方法

### 2.5 盲区排查的方法论

**为什么重要**：没有方法论的盲区排查=凭感觉，结果不准确。

**四步排查法**：

**步骤1：知识地图覆盖检查**
对着12周的知识点清单逐项打勾——"我能独立做出来"。诚实是第一原则。

**步骤2：输出检验**
随机抽取3个"我觉得我熟练"的模块，关掉所有资料，独立写一份完整的分析方案或代码。如果卡住了——这个模块就是盲区。

**步骤3：交叉检验**
找同事或学习伙伴问："你觉得我在哪个模块上最弱？"他人的观察往往比自我认知更准确。

**步骤4：难度梯度测试**
对每个模块，用"初级→中级→高级"三级测试。很多人初级能过，中级就卡了——这说明该模块需要补漏。

### 2.6 制定补漏计划的SMART原则

**为什么重要**："我要补SQL"不是计划——太模糊了，最后什么也不会发生。

| SMART要素 | 坏的例子 | 好的例子 |
|-----------|---------|----------|
| Specific | "补SQL" | "补窗口函数中的gap-and-island" |
| Measurable | "多练" | "独立完成3道连续登录SQL题" |
| Achievable | "一周内精通所有因果推断" | "本周掌握DID的基本用法" |
| Relevant | 补一个工作中不用的模块 | 优先补工作中最常用的模块 |
| Time-bound | "有空的时候" | "周三下午2-5点" |

### 2.7 优先级排序框架

**为什么重要**：时间有限，不能每个盲区都深度补。需要优先级排序。

**优先级矩阵**：

| | 高频使用 | 低频使用 |
|------|---------|----------|
| **能力弱** | 🔴 最高优先级 | 🟡 中等优先级 |
| **能力强** | 🟢 维持即可 | ⚪ 低优先级 |

**判断"高频"的标准**：
- 日常工作中每周都会遇到的——高频
- 面试中90%会问到的——高频
- 只在特定场景用到的——低频

| JD关键词 | 检验题 |

- "给你一个场景，你会从哪些维度分析？"——考查框架性思维

### 场景2：接手新项目时的盲区排查

**分析步骤**：

3. 制定学习计划：先补最紧急的（项目需要的），再补长期的（职业发展需要的）

**常见新业务的盲区**：

- 从C端转到B端 → 销售漏斗、客户成功是盲区

### 场景3：升职答辩前的盲区排查

**背景**：你准备申请从初级升中级或从中级升高级，需要证明你的分析能力达到了下一级别的标准。

**不同级别的分析能力要求**：

| 级别 | 分析能力要求 | 典型盲区 |
|------|------------|----------|

| 中级→高级 | 能定义分析框架、能指导初级分析师、能推动分析结果落地 | 跨部门沟通、项目推动力、业务影响力 |
| 高级→专家 | 能定义团队的分析方法论、能预判业务趋势、能培养团队 | 更偏领导力和行业洞察 |

**答辩前自查**：
1. 我能讲清楚至少3个完整项目的故事吗？（不是技术细节，是业务影响）

3. 我能被问到"为什么用这个方法"时说出可信的理由吗？

### 场景4：团队能力评估

**背景**：你是一个分析团队的Leader，需要评估团队每个成员的能力并制定培养计划。

**团队盲区常见模式**：
- 全员弱于AB实验 → 整个团队的实验结论可能都有问题
- 全员弱于SQL优化 → 报表越来越慢，但没有人在意
- 全员弱于表达 → 做了很多好分析但没人知道
- 个别成员有强项 → 可以让ta做内部培训

**Leader的排查方法**：
1. 给每个人发一份知识地图，自评+360评估

4. 制定团队整体培训计划和个人发展计划

> **关键认知**：一个人的盲区是盲区，一群人的盲区是系统性风险。
