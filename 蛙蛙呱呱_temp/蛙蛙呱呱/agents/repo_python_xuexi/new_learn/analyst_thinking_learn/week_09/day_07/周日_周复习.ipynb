# 第9周 · 周日：周复习——因果推断方法体系总览\n\n> 难度：⭐⭐⭐⭐ | 定位：整合W9所有因果推断方法，建立选择决策框架\n> 前置依赖：W9D1-D6全部内容 | 后续用途：W10-W12综合案例中方法选择的基础\n\n---\n\n## 📘 区块1：因果推断方法选择决策树\n\n### 1.1 一图胜千言——你的因果推断工具箱\n\n```\n你能做随机化实验吗？\n  ├── 是 → AB实验/RCT（W7-W8）\n  │        └── 有non-compliance? → IV（W9D6）：分组=IV, 实际处理=内生变量\n  │\n  └── 否 → 有前后数据+自然对照组+平行趋势?\n            ├── 是 → DID（W9D2-3）\n            │        ├── 处理前趋势不平行? → DDD(三重差分)/加入趋势项\n            │        └── 个别城市/个体处理时间不同? → 交错DID(异质性稳健估计量)\n            │\n            └── 否 → 有丰富的协变量+共同支撑?\n                      ├── 是 → PSM（W9D4-5）\n                      │        ├── 担心未观测混淆? → Rosenbaum敏感性分析\n                      │        ├── 有处理前数据? → PSM+DID\n                      │        └── 共同支撑差? → 修剪/IPTW/换方法\n                      │\n                      └── 否 → 有工具变量(外生冲击)?\n                                ├── 是 → IV（W9D6）\n                                │        ├── 弱工具变量(F<10)? → LIML/弱IV稳健方法\n                                │        └── 多个IV? → 过度识别检验(J test)\n                                │\n                                └── 否 → 有明确断点?\n                                          ├── 是 → RDD（W9D6）\n                                          │        ├── 有操纵嫌疑? → McCrary检验\n                                          │        └── 带宽敏感? → 多带宽敏感性分析\n                                          │\n                                          └── 否 → 诚实地说"现有数据无法可靠推断因果"\n```\n\n### 1.2 五种方法的核心对比\n\n| 方法 | 识别策略 | 反事实来源 | 核心假设 | 估计目标 | 何时用 |\n|------|---------|-----------|---------|---------|--------|\n| **AB实验** | 随机化 | 随机分配的对照组 | SUTVA(无干扰) | ATE | 能做实验时 |\n| **DID** | 双重差分 | 对照组的趋势外推 | 平行趋势 | ATT | 有前后+对照+趋势平行 |\n| **PSM** | 倾向得分匹配 | 特征相似的对照组 | 可忽略性(无遗漏混淆) | ATT | 有丰富协变量+共同支撑 |\n| **IV** | 工具变量 | 工具变量的局部随机 | 外生性+排他性 | LATE | 有好的工具变量 |\n| **RDD** | 断点回归 | 断点附近的局部随机 | 断点处连续 | LATE | 有明确断点 |\n\n### 1.3 "假设不可检验"是常态——诚实是最高原则\n\n所有观测因果推断方法都有一个"不可直接检验"的核心假设：\n- DID：平行趋势（只能说处理前平行，不能证明处理后平行）\n- PSM：可忽略性（只能说可观测的平衡了，不能证明没有未观测混淆）\n- IV：外生性（完全不可检验！只能靠理论和制度知识论证）\n- RDD：断点处连续性（McCrary检验只能测密度是否连续，不是所有维度）\n\n**这意味着**：你永远无法从观测数据中"证明"因果——你只能"逼近"因果。你的任务不是"找到真相"，而是"用最可靠的方法给出最可信的估计，同时诚实地说明局限"。\n\n**面试中的加分回答**：当被问到"你怎么确定你的结论是因果而不是相关"时，最好的回答不是"我用了PSM/DID/IV"——而是"我用了X方法，它依赖于Y假设。为了检验这个假设，我做了Z检验。但需要说明的是，Y假设无法被完全验证——如果Y假设不成立，我的估计可能偏X方向。为应对这一担忧，我还做了ABC敏感性分析..."\n\n---\n\n## 📘 区块2：本周知识点体系化回顾

### 周一：因果推断基础\n\n**核心概念**：\n- 相关≠因果（混淆/反向因果/选择偏差/巧合）\n- 反事实框架：因果效应 = 事实 - 反事实\n- 混淆变量三条件：影响处理 + 影响结果 + 不在因果路径上\n- DAG(有向无环图)：三种结构（链式/分叉/对撞）\n- 辛普森悖论：分组趋势和总体趋势相反\n- 所有因果推断方法本质上是"构建反事实"的不同策略\n\n**关键检验**：\n- 用DAG判断该控制什么变量\n- 控制混淆变量后假相关消失\n- 分层分析（控制混淆）优先于总体分析\n\n**一句话总结**：\n> 因果推断的核心是回答"如果没被处理会怎样"——这个反事实永远观测不到，所有方法都在"构建一个可信的反事实"。\n\n### 周二：DID双重差分\n\n**核心概念**：\n- DID = (实验组后-前) - (对照组后-前) = 双重差分\n- 回归形式：Y = β₀ + β₁·T + β₂·P + β₃·(T×P) + ε，β₃是DID\n- 固定效应版本(TWFE)：Y_it = α_i + λ_t + β·D_it + ε\n- 平行趋势假设：如果没有处理，两组趋势相同\n- 交错处理下的TWFE偏误（负权重问题）\n\n**关键检验**：\n- 画趋势图——处理前两组趋势是否平行\n- 聚类标准误（处理分配级别）\n- 事件研究图（多期时的标准输出）\n\n**一句话总结**：\n> DID用对照组的时间趋势来"模拟"实验组的反事实——核心假设是平行趋势。\n\n### 周三：DID实现与检验\n\n**核心概念**：\n- 事件研究法：Y = α_i + λ_t + Σβ_k·D_k + ε（处理前后各期分别估计）\n- 时间安慰剂：把处理时间提前→DID应不显著\n- 组别安慰剂：随机打乱处理组1000次→真实DID应在分布尾部\n- PSM+DID：先匹配再差分（取两家长处）\n\n**关键检验**：\n- 事件研究图：处理前系数≈0，处理后系数"跳起"\n- 联合检验：F test检验所有处理前系数=0\n- 聚类标准误的选择（处理分配级别）\n- 多重安慰剂检验\n\n**一句话总结**：\n> 事件研究图是DID"最重要的一张图"——同时展示假设检验和效应估计。\n\n### 周四：PSM倾向得分匹配\n\n**核心概念**：\n- 四个步骤：估计ps→匹配→诊断→估计效应\n- 倾向得分 = P(T=1|X)——一个降维工具\n- 匹配方法：最近邻(1:1/1:K)/卡尺/半径/核/分层\n- 可忽略性假设（不可检验！）\n- 共同支撑假设\n\n**关键检验**：\n- caliper = 0.2 × SD(ps)（Austin 2011标准）\n- AUC 0.65-0.85为理想（太高→共同支撑差）\n- PSM+DID组合增强可信度\n\n**一句话总结**：\n> PSM用倾向得分降维——在1D上匹配等价于在p维上匹配（前提：倾向得分模型正确）。\n\n### 周五：PSM平衡性检验\n\n**核心概念**：\n- SMD = (两组均值差) / 合并标准差（<0.1良好）\n- Love Plot：所有变量匹配前后SMD对比\n- QQ图：检查分布（非仅均值）的匹配质量\n- 共同支撑检查：倾向得分分布的重叠\n- Rosenbaum敏感性分析：多大的遗漏混淆能推翻结论\n\n**关键检验**：\n- SMD：每变量匹配前后对比（>0.2需要重新建模）\n- 共同支撑覆盖率应>80%\n- Rosenbaum Γ：Γ>2时结论仍显著→稳健\n- 多方法(1:1/卡尺/核匹配)做稳健性检验\n\n**一句话总结**：\n> PSM的"匹配"只是第一步——匹配质量的诊断才是核心。Love Plot和Rosenbaum分析一个都不能少。\n\n### 周六：IV与RDD\n\n**核心概念**：\n- IV：通过工具变量Z识别因果——Z→T→Y，Z与ε无关\n- IV估计LATE（只对Compliers有效），不是ATE\n- 第一阶段F>10：强工具变量 ✓\n- RDD：利用断点附近的局部随机\n- Sharp RDD（确定性跳跃）vs Fuzzy RDD（概率跳跃=IV）\n- McCrary检验：检查forcing variable密度在断点处是否连续\n\n**关键检验**：\n- 第一阶段F统计量>10（Stock-Yogo准则）\n- McCrary密度检验（RDD）\n- RDD带宽敏感性分析（0.5×/1×/2×最优带宽）\n- RDD假断点安慰剂检验\n\n**一句话总结**：\n> IV和RDD通过"自然界的随机"来识别因果——IV靠工具变量的外生性，RDD靠断点附近的局部随机。好工具变量极度稀缺。

## 💻 区块3：动手练习——本周知识整合实践\n\n### 综合练习：同一个问题，用三种方法解决

````python
import numpy as np; import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression, LinearRegression
import warnings; warnings.filterwarnings('ignore')
np.random.seed(42)
plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

print('='*60)
print('W9综合练习：同一问题→三种因果推断方法对比')
print('='*60)

# 场景：某APP在特定时间点(第0天)对"高活跃用户"上线了新功能
# 我们有处理前后各7天的面板数据
# 问题：新功能到底带来了多少使用时长增量？

n_users = 500
periods = list(range(-7, 8))  # -7到+7天

# 用户特征
activeness_base = np.random.normal(0, 1, n_users)  # 基准活跃度
age = np.random.uniform(18, 55, n_users)

# 处理分配：最活跃的前50%用户获得新功能（选择偏差！）
treated = (activeness_base > np.median(activeness_base)).astype(int)

# 生成面板数据
data_all = []
for i in range(n_users):
    for day in periods:
        post = 1 if day >= 0 else 0
        time_trend = 0.5 * day  # 共同时间趋势
        weekend = 3 if abs(day) % 7 in [5, 6] else 0  # 周末效应
        feature_effect = 8 if (post and treated[i]) else 0  # 新功能效应=8分钟
        y = (45 + 10*activeness_base[i] + 0.2*age[i] + time_trend 
             + weekend + feature_effect + np.random.normal(0, 5))
        data_all.append({'user': i, 'day': day, 'post': post, 
                         'treated': treated[i], 'minutes': y,
                         'activeness': activeness_base[i], 'age': age[i]})
df_all = pd.DataFrame(data_all)

true_effect = 8
print(f'\n真实处理效应: {true_effect}分钟/天')
print(f'处理组(新功能): {(treated==1).sum()}人')
print(f'对照组(无新功能): {(treated==0).sum()}人')
print()
````

````python
# --- 方法1：朴素对比（作为基准——展示它多糟糕）---
post_data = df_all[df_all.post == 1]
naive = post_data[post_data.treated==1]['minutes'].mean() - post_data[post_data.treated==0]['minutes'].mean()
print(f'方法1-朴素对比(处理后): {naive:.1f}分钟 (偏误: {naive-true_effect:+.1f})')
print('  解释：直接比较两组→把"本来就活跃"的差异归因为功能效应')
print()

# --- 方法2：DID ---
import statsmodels.formula.api as smf
df_all['treated_post'] = df_all['treated'] * df_all['post']
model_did = smf.ols('minutes ~ treated + post + treated_post', data=df_all).fit()
did_est = model_did.params['treated_post']
print(f'方法2-DID: {did_est:.1f}分钟 (偏误: {did_est-true_effect:+.1f})')
print('  解释：用对照组的时间趋势剥离了自然变化，但依赖平行趋势')
print()

# --- 方法3：PSM ---
# 只用处理前的数据估计倾向得分
df_pre = df_all[df_all.post == 0].groupby('user').agg({
    'minutes': 'mean', 'activeness': 'first', 'age': 'first', 'treated': 'first'
}).reset_index()
df_pre.columns = ['user', 'pre_minutes', 'activeness', 'age', 'treated']

lr_psm = LogisticRegression(max_iter=1000)
lr_psm.fit(df_pre[['pre_minutes', 'age']], df_pre['treated'])
df_pre['ps'] = lr_psm.predict_proba(df_pre[['pre_minutes', 'age']])[:, 1]

# 匹配
treat_pre = df_pre[df_pre.treated==1].copy()
ctrl_pre = df_pre[df_pre.treated==0].copy()
matched_users = []
for _, row in treat_pre.iterrows():
    idx = (ctrl_pre['ps'] - row['ps']).abs().idxmin()
    matched_users.append(ctrl_pre.loc[idx, 'user'])
    ctrl_pre = ctrl_pre.drop(idx)

# 用匹配后的用户，比较处理后的均值
post_treat = df_all[(df_all.post==1) & (df_all.treated==1)]
post_match = df_all[(df_all.post==1) & (df_all.user.isin(matched_users))]
psm_est = post_treat['minutes'].mean() - post_match['minutes'].mean()
print(f'方法3-PSM: {psm_est:.1f}分钟 (偏误: {psm_est-true_effect:+.1f})')
print('  解释：匹配了处理前特征相似的用户，但仍可能遗漏未观测混淆')
print()

# 汇总
print('='*60)
print(f'三种方法对比 (真实效应={true_effect}):')
print(f'  朴素对比: {naive:>6.1f}分钟 (偏误 {naive-true_effect:>+.1f})')
print(f'  DID:      {did_est:>6.1f}分钟 (偏误 {did_est-true_effect:>+.1f})')
print(f'  PSM:      {psm_est:>6.1f}分钟 (偏误 {psm_est-true_effect:>+.1f})')
print(f'\n结论：DID和PSM都大幅改善了朴素对比的偏误')
print(f'在本模拟中，DID表现更好——因为处理分配与activeness(未观测)相关，')
print(f'而DID通过固定效应消除了不随时间变化的activeness差异')
````

````python
# 可视化：三种方法的估计量和真实效应的对比
fig, ax = plt.subplots(figsize=(10, 6))
methods = ['朴素对比', 'DID', 'PSM']
estimates = [naive, did_est, psm_est]
colors = ['red', 'green', 'blue']

x_pos = range(len(methods))
ax.bar(x_pos, estimates, color=colors, alpha=0.7, width=0.5)
ax.axhline(y=true_effect, color='black', linestyle='--', linewidth=2, label=f'真实效应={true_effect}')
ax.set_xticks(x_pos)
ax.set_xticklabels(methods)
ax.set_ylabel('估计效应(分钟/天)')
ax.set_title('W9方法对比：同一问题，三种估计')
ax.legend()

for i, (est, color) in enumerate(zip(estimates, colors)):
    ax.text(i, est + 0.3, f'{est:.1f}', ha='center', fontweight='bold', color=color)
plt.tight_layout(); plt.show()
````
