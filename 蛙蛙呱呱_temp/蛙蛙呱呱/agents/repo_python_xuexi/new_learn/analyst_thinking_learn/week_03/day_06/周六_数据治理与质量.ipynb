## 📘 区块1：核心概念与第一性原理

### 1.1 GIGO原则——数据分析的第一性原理

**Garbage In, Garbage Out。** 这是数据分析领域的第一条纪律——比任何方法论、任何框架都基础。

模型再精妙（深度学习/大模型/XGBoost）、可视化再炫（那些花哨的仪表盘）、分析框架再完善（ISS/AARRR）——如果数据源是脏的，一切分析都是垃圾。

**为什么GIGO如此重要但常被忽略**：因为数据质量问题往往发生在"上游"（埋点/ETL/业务系统），而分析发生在上游的下游。分析师接触数据时，数据已经"完成"了——如果没有人告诉你数据有问题，你会默认它是干净的。这种"默认信任"是数据分析中最危险的假设。

### 1.2 数据质量的六个维度——全面评估框架

| 维度 | 含义 | 检验问题 | 举例 |
|------|------|---------|------|
| **准确性** | 数据和真实情况一致吗？ | "这个数字反映了客观事实吗？" | 用户实际年龄18，系统里显示180 |
| **完整性** | 该有的数据都有吗？ | "有没有缺失的记录/字段？" | 交易表里30%的订单缺支付方式字段 |
| **一致性** | 不同表里同样的数据一致吗？ | "同一个事物在不同地方描述一样吗？" | 用户表性别="男"，订单表性别="M" |
| **及时性** | 数据的延迟在可接受范围内吗？ | "这个数据'新鲜'吗？" | 昨天的DAU到今天中午还没出来 |
| **唯一性** | 没有重复记录？ | "有没有不应该重复的数据重复了？" | 同一个用户注册了3个账号（非预期） |
| **有效性** | 数据符合定义的格式和范围吗？ | "这个值在合理的范围内吗？" | 手机号字段里有"abcdefg" |

**六维度的优先级**：不是所有维度同等重要。对于不同的分析场景，核心维度不同：
- 财务报表 → 准确性第一（差一分钱都不行）
- 实时监控 → 及时性第一（晚1小时可能意味着损失）
- 用户画像 → 完整性第一（缺字段导致画像片面）
- 跨系统报表 → 一致性第一（口径不统一=决策依据矛盾）

### 1.3 数据质量问题的来源——上游的"蝴蝶效应"

```
埋点bug → 数据没记上或记错了（前端改了页面结构但没通知埋点团队）
ETL异常 → 数据管道中途断了（上游表结构变了→下游任务失败→数据缺失）
人为操作 → 运营手动导入Excel时覆盖了历史数据（"我只是想改一个数字..."）
业务变更 → 某字段含义变了但没有通知数据团队（"我们上周改了订单状态的定义"）
系统迁移 → 老系统和新系统的数据格式不一致（老系统用"1/0"，新系统用"true/false"）
```

**核心洞察**：数据质量问题的根因往往不在"数据"本身——而在产生数据的上游流程。光靠清洗数据解决不了根因——你需要追溯到上游流程的漏洞。

### 1.4 从第一性原理推导：为什么"5分钟嗅探"比"2小时分析"更重要

假设你拿到一份数据，直接开始分析。你花了2小时，发现了一个"有趣的洞察"——"手机上号全是abcdefg"。你兴奋地准备汇报，但再一看——这只是数据bug。2小时白费了。

如果分析前花5分钟做嗅探：行数合理吗？空值率多少？有极值吗？时间范围对吗？→你会发现手机号字段全是脏数据→先解决数据质量→再做分析。

**5分钟嗅探的期望价值**：
- 如果数据是干净的：你花了5分钟，获得了"数据可用的信心"
- 如果数据是脏的：你花了5分钟，避免了2小时的"假分析"
- 如果数据是脏的但你没嗅探：你花了2小时，得出了一个"假洞察"，更危险的是——你可能基于这个假洞察做了一个错误决策

**结论**：5分钟嗅探的期望价值 >> 5分钟的时间成本。这是数据分析中ROI最高的投入之一。

### 1.5 面试高频考点

**面试官常问**：
- "你拿到一份新数据，第一件事做什么？"
- "你怎么确保你的分析结果是可信的？"
- "如果两个表的数据对不上，你怎么排查？"
- "数据质量的六个维度是什么？"

**回答要点**：
- 第一件事永远是"嗅探"——不是"分析"
- 可信=数据质量验证+口径明确+交叉验证
- 对不上→先确认口径是否一致→再查ETL是否有异常→最后查上游数据源
- 六维度：准确性/完整性/一致性/及时性/唯一性/有效性

> **今天最重要的认知**：数据分析师的第一条纪律——拿到数据后，先用5分钟做"嗅探"（行数、空值率、极值、时间范围）。如果跳过这步直接开始分析，你可能花了2小时得出了一个"手机上号全部是abcdefg"的深刻洞察。数据质量不只是数据工程师的事——分析师是数据质量的最后一道防线。

## 🧩 区块2：知识点详细拆解

### 2.1 数据质量的"5分钟嗅探"——标准操作流程

**为什么重要**：这是每个分析师拿到新数据后必须执行的标准程序。

**嗅探清单（每次拿到数据，按顺序过一遍）**：
1. **行数合理吗？** 今天的数据行数是昨天的5倍？是正常促销还是数据bug？对比近7天均值±3σ。
2. **空值率多少？** 核心字段（user_id、order_id、关键金额字段）空值>5%要追查原因。非核心字段空值>20%建议忽略该字段。
3. **极值有吗？** 客单价显示999999？可能是测试数据或系统默认值。看max/min、P1/P99，检查是否在业务合理范围内。
4. **时间范围对吗？** 你需要最近30天的数据，怎么只有15天？看min/max日期，确认分区是否完整。
5. **关键指标跟昨天差多少？** DAU突然翻倍或腰斩？先怀疑数据bug而不是"业务突变了"。
6. **枚举字段的值对吗？** 状态字段预计有3个值(0/1/2)，出现了"X"？性别字段出现了"未知"+""+"M"+"F"(应该是统一的枚举值)。
7. **关联键能关联上吗？** 订单表的user_id能在用户表里找到吗？关联不上的比例是多少？

**嗅探的产出**：一个"数据质量评估"——✓数据可用 / ⚠️ 数据有瑕疵但可用（说明瑕疵） / ✗ 数据不可用（说明原因）。

### 2.2 常见数据质量问题的"急救"方法

**为什么重要**：分析不能等数据工程师修好数据再做——有时候你需要临时方案。

| 问题 | 临时方案（分析师自己做） | 长期方案（推动上游修） |
|------|---------------------|-------------------|
| 核心字段有空值 | 空值排除/用中位数填充/用"未知"标记 | 在ETL中加非空校验 |
| 两表口径不一致 | 每次写SQL时加注释说明口径差异 | 统一指标字典(One Definition) |
| 埋点漏了/错了 | 用代理指标替代/手动补数据 | 埋点变更走审批流程 |
| 极值/异常值 | 用P99替代max/排除分析/标记说明 | 在数据入库时加范围校验 |
| 重复数据 | DISTINCT去重/GROUP BY去重 | 在主键上加强制唯一约束 |
| 延迟数据 | 标注"数据截止到T-2天，T-1天数据不完整" | 加SLA监控，延迟>阈值告警 |

**关键原则**：临时方案要标注——在分析报告里明确说明"本分析中XX字段有X%空值，已做XX处理"。不要让读者以为你的数据是完美的。

### 2.3 指标口径管理——One Definition原则

**为什么重要**：当市场部说"DAU=100万"，产品部说"DAU=95万"，两个人在会上吵架——通常不是数据错了，而是"DAU的定义不同"。

**常见的口径差异**：
- "活跃用户"=打开App就算？还是有有效行为（浏览/点击）才算？
- "GMV"=下单就算？还是支付成功才算？还是确认收货才算？退款的算不算？
- "新用户"=首次打开App？还是首次注册？还是首次完成核心行为？
- "转化率"的分母是"访问用户"还是"注册用户"？分子是"下单"还是"支付"？

**One Definition原则**：
1. 每个核心指标只有一个"官方定义"——写在指标字典里
2. 所有看板/报表/分析都必须使用这个定义（或用注释明确说明差异）
3. 指标定义变更必须走审批+通知流程（不能某人某天偷偷改了）
4. 指标字典包含：指标名、业务口径（人话）、技术口径（SQL/代码）、分子、分母、去重逻辑、时间窗口、更新频率、责任人

### 2.4 数据血缘——当上游出了问题，影响哪些下游？

**为什么重要**：上游一张表改了一个字段→下游100张看板和50个分析受影响。如果你不知道数据血缘，你就不知道问题的爆炸半径。

**数据血缘（Data Lineage）** = 数据从产生到被使用的完整链路图。

```
埋点上报 → Kafka → ETL清洗 → ODS层 → DWD层 → DWS层 → ADS层 → 看板/分析
  ↑                         ↑                    ↑
问题发生点              问题被发现          影响扩散到所有下游
```

**作为分析师，你不需要画出完整的数据血缘图（那是数据工程师的工作）**，但你需要知道：
- 你分析用的数据来自哪张表？（上游是什么）
- 如果这张表今天没更新，你的分析能等多久？（SLA）
- 如果这张表的数据有问题，你会是第几个发现的？（你的嗅觉灵敏吗？）

### 2.5 数据质量的"治理"vs"管理"

**为什么重要**：很多公司把"数据治理"等同于"建规范、写文档"——然后发现没人遵守。

| | 数据管理 | 数据治理 |
|------|---------|--------|
| 做什么 | 制定规范、写文档 | 确保规范被执行 |
| 产出物 | 数据字典、命名规范、SLA文档 | 质量监控告警、合规审计、根因追踪 |
| 典型角色 | 数据架构师 | 数据治理委员会 + 数据 steward |
| 风险 | 规范束之高阁，没人执行 | 执行成本高，团队抵触 |

**适合分析师的治理策略**：
1. 用"告警"代替"规范"——与其写一条"订单金额不能为负"的规范，不如建一个监控：订单金额<0时自动告警
2. 用"工具"代替"流程"——与其要求每个人都检查数据质量，不如做一个自动化质量检测脚本
3. 用"复盘"代替"追责"——数据出了质量问题，不是"谁的责任"，而是"怎么防止下次再发生"

### 2.6 分析师的数据质量"责任心"——最后一道防线

**为什么重要**：数据质量不只是数据工程师的事。分析师是数据质量的最后一道防线——你最经常看数据、最了解数据"应该长什么样"。

**你的"数据质量直觉"**：
- 当你发现DAU突然翻倍——你的第一反应是"数据是不是有问题"而不是"业务突飞猛进了"
- 当你发现某个渠道的转化率是0——你不会觉得"这个渠道真差"，而是会想"是不是埋点没上报"
- 当两个表JOIN后行数变少了——你不会直接开始分析，而是会检查"是不是关联键有NULL导致了数据丢失"

**这种"直觉"不是天生的——是通过无数次"分析做完后发现数据有问题"的惨痛教训培养出来的。**

### 2.7 数据质量的"可接受标准"——过度追求质量也是一种浪费

**为什么重要**：数据质量不是越高越好——追求"零缺陷"的成本是无穷大的。你需要知道"够用"的标准是什么。

| 分析场景 | 可接受的质量标准 | 为什么 |
|---------|---------------|--------|
| 财务报表/对账 | 99.99%准确 | 差一分钱都是问题 |
| AB实验决策 | 95%置信 + 数据完整度>99% | 统计推断本身就容忍一定的误差 |
| 运营活动效果评估 | 90%准确 + 趋势方向正确 | 运营决策更依赖方向性判断而非精确值 |
| 探索性分析 | 80%准确 + 明确说明局限 | 快速验证假设，如果假设成立再追求精确 |

**原则**：数据质量投入应匹配决策的风险。决策风险越高→对数据质量的要求越高。

### 2.8 数据治理的"文化"——从"这是我的数据"到"这是公司的数据"

**为什么重要**：数据治理的最大障碍不是技术，是文化——每个团队都把"自己的数据"当成私有财产。

**从"我的数据"到"公司的数据"**：
1. 数据不应属于"产生它的团队"——应属于公司
2. 任何团队需要其他团队的数据时，应该有获取途径（数据目录/数据API）
3. 数据的使用应该有记录（谁、什么时候、用了什么数据、做了什么决策）
4. 数据质量的反馈应该有闭环（"我发现这个数据有问题"→"感谢反馈，已修复"）

## 💻 区块3：动手练习

````python
import numpy as np
import pandas as pd

np.random.seed(42)

# ============================================================
# 练习1：数据嗅探——拿到数据后的标准程序（基础）
# ============================================================

# 创建一份"脏数据"（模拟真实世界中常见的数据质量问题）
n = 1000
df = pd.DataFrame({
    "user_id": range(1, n+1),
    "age": np.random.choice([25, 30, 35, 40, 999, -1], n, p=[0.3, 0.3, 0.2, 0.15, 0.03, 0.02]),
    "gender": np.random.choice(["M", "F", "", "未知"], n, p=[0.45, 0.45, 0.05, 0.05]),
    "order_amount": np.random.choice([100, 200, 500, 1000, 999999], n, p=[0.3, 0.3, 0.25, 0.14, 0.01]),
    "order_date": ["2026-01-01"] * n,
    "phone": np.random.choice(["13812345678", "13987654321", "", "abcdefg", None], n, p=[0.35, 0.35, 0.15, 0.10, 0.05]),
})
df.loc[np.random.choice(n, 50), "age"] = np.nan
df.loc[np.random.choice(n, 30), "order_amount"] = np.nan

print("=" * 60)
print("数据嗅探报告 —— 5分钟标准程序")
print("=" * 60)

# 1. 行数
print(f"\n① 行数: {len(df)}")
print(f"  预期: ≈1000 ✓ (符合)")

# 2. 空值
print(f"\n② 空值情况:")
null_counts = df.isnull().sum()
for col in df.columns:
    if null_counts[col] > 0:
        print(f"  {col}: {null_counts[col]}个空值 ({null_counts[col]/len(df)*100:.1f}%) {'⚠️ 需关注' if null_counts[col]/len(df)>0.05 else '✓ 可接受'}")

# 3. 极值
print(f"\n③ 极值检查:")
print(f"  age: min={df['age'].min()}, max={df['age'].max()}")
print(f"    → 异常值: age=-1 ({(df['age']==-1).sum()}个), age=999 ({(df['age']==999).sum()}个)")
print(f"  order_amount: min={df['order_amount'].min():.0f}, max={df['order_amount'].max():.0f}")
print(f"    → 极值: 999999 ({(df['order_amount']==999999).sum()}个, 可能是测试数据)")

# 4. 时间范围
print(f"\n④ 时间范围: 所有订单日期={df['order_date'].iloc[0]}")
print(f"    → ⚠️ 所有1000行数据都是同一天！可能是导入错误或测试数据")

# 5. 枚举字段
print(f"\n⑤ 枚举字段检查:")
print(f"  gender值分布: {df['gender'].value_counts().to_dict()}")
print(f"    → ⚠️  包含空字符串和'未知'(口径不统一)")
print(f"  phone值示例: {df['phone'].value_counts().head().to_dict()}")
print(f"    → ⚠️  包含空值/空字符串/'abcdefg'(不是有效手机号)")

# 6. 统计摘要
print(f"\n⑥ 关键指标统计:")
print(f"  order_amount (排除极值999999):")
valid_amounts = df[df['order_amount'] < 999999]['order_amount'].dropna()
print(f"    mean={valid_amounts.mean():.0f}, median={valid_amounts.median():.0f}")
print(f"    P1={valid_amounts.quantile(0.01):.0f}, P99={valid_amounts.quantile(0.99):.0f}")

print(f"\n{'='*60}")
print(f"数据质量总评: ⚠️ 有严重质量问题，分析前必须先清洗")
print(f"  必做: 排除age=-1和999, 排除order_amount=999999, 标注手机号空值")
print(f"  建议: 确认日期为什么全是同一天, 统一性别字段口径")
````

````python
# ============================================================
# 练习2：口径校验——同一个指标不同算法对比（进阶）
# ============================================================

print("口径校验：同一个'DAU'的三种算法")
print("=" * 60)

# 模拟场景：市场部、产品部、数据部对"DAU"的三种定义
np.random.seed(123)
n_users = 10000

# 模拟用户行为
users = pd.DataFrame({
    'user_id': range(n_users),
    'app_open': np.random.choice([0, 1], n_users, p=[0.6, 0.4]),          # 40%打开App
    'has_action': np.random.choice([0, 1], n_users, p=[0.65, 0.35]),      # 35%有行为
    'has_valid_action': np.random.choice([0, 1], n_users, p=[0.72, 0.28]), # 28%有有效行为
    'is_registered': np.random.choice([0, 1], n_users, p=[0.1, 0.9]),     # 90%已注册
})

# 三种DAU算法
dau_market = users[users['app_open'] == 1].shape[0]       # 市场部: 打开App就算
dau_product = users[(users['app_open'] == 1) & (users['has_action'] == 1)].shape[0]  # 产品部: 有行为才算
dau_data = users[(users['app_open'] == 1) & (users['has_valid_action'] == 1) & (users['is_registered'] == 1)].shape[0]  # 数据部: 有有效行为的注册用户

print(f"\n定义对比:")
print(f"  市场部定义(打开App): DAU = {dau_market}")
print(f"  产品部定义(有行为):   DAU = {dau_product} (差异: {(dau_product-dau_market)/dau_market*100:+.1f}%)")
print(f"  数据部定义(有效行为+注册): DAU = {dau_data} (差异: {(dau_data-dau_market)/dau_market*100:+.1f}%)")

print(f"\n三个部门报给CEO三个不同的DAU → CEO: '你们到底谁的数据是准的？'")
print(f"→ 问题不在于'数据不准'——在于'口径不一致'")
print(f"→ 解决方案: One Definition——全公司使用统一的'DAU'定义")
print(f"→ 指标字典: DAU = 当日至少完成1次有效行为(浏览/点击/搜索/下单)的注册用户数")

# 展示指标字典的格式
print(f"\n【指标字典示例】")
print(f"  指标名称: DAU (Daily Active Users)")
print(f"  业务口径: 当日至少完成1次有效行为的注册用户数")
print(f"  有效行为定义: 浏览商品详情页/点击加购/搜索/下单/支付")
print(f"  技术口径: SELECT COUNT(DISTINCT user_id) FROM user_actions WHERE action_date = '{date}' AND action_type IN ('view','cart','search','order','pay') AND user_id IN (SELECT user_id FROM registered_users)")
print(f"  去重逻辑: 按user_id去重")
print(f"  时间窗口: 自然日(00:00-23:59)")
print(f"  更新频率: T+1h")
print(f"  责任人: 数据部-张三")
````

````python
# ============================================================
# 练习3：模拟数据泄漏导致的分析灾难（挑战）
# ============================================================

print("数据泄漏如何毁掉一个机器学习模型——"假数据"的真实案例")
print("=" * 60)

np.random.seed(42)

# 模拟用户流失预测的数据
n_users = 5000

# 真正的特征
true_days_since_last_active = np.random.exponential(30, n_users)
true_order_count = np.random.poisson(5, n_users)
true_avg_order_value = np.random.lognormal(4, 0.5, n_users)

# 真实的流失标签（基于真实特征）
churn_score = -0.05 * true_days_since_last_active + 0.3 * true_order_count + 0.001 * true_avg_order_value
true_churn_prob = 1 / (1 + np.exp(churn_score - 1))
true_label = (np.random.random(n_users) < true_churn_prob).astype(int)

# 但是！"最后活跃日期"字段有一个ETL bug——流失用户被填入了"2099-01-01"
last_active_date = np.where(
    true_label == 1,
    '2099-01-01',  # bug: 流失用户的最后活跃日期是2099年
    pd.Timestamp('2026-05-01') - pd.Timedelta(days=true_days_since_last_active.astype(int))
)

# 从日期转换为距今天数（分析师看到的"特征"）
days_since_last_active_model = np.where(
    last_active_date == '2099-01-01',
    -99999,  # 2099年→距今天数是一个巨大的负数（看起来"非常活跃"）
    true_days_since_last_active
)

print(f"\n真实世界:")
print(f"  活跃用户平均距上次活跃: {true_days_since_last_active[true_label==0].mean():.0f}天")
print(f"  流失用户平均距上次活跃: {true_days_since_last_active[true_label==1].mean():.0f}天")
print(f"  → 距上次活跃越久→越可能流失 (符合直觉)")

print(f"\n分析师看到的数据(含bug):")
print(f"  活跃用户平均距上次活跃: {days_since_last_active_model[true_label==0].mean():.0f}天")
print(f"  流失用户平均距上次活跃: {days_since_last_active_model[true_label==1].mean():.0f}天")
print(f"  → 流失用户看起来'极其活跃'(距上次活跃是负数)!")
print(f"  → 模型会学到: '距上次活跃天数越少(负数) → 越可能流失'")
print(f"  → 这完全反了! 但AUC会非常高(因为数据泄漏→模型记住了bug)")

# 计算"假模型"的AUC（简化模拟）
from sklearn.metrics import roc_auc_score
fake_pred = 1 / (1 + np.exp(-days_since_last_active_model / 1000))
auc = roc_auc_score(true_label, fake_pred)
print(f"\n假模型的AUC: {auc:.3f}")
print(f"  → AUC极高! 团队以为做出了一个极好的模型")
print(f"  → 部署到生产→发券给'预测会流失的用户'")
print(f"  → 实际发给了活跃用户(因为bug让他们看起来'即将流失')")
print(f"  → 发了100万券, 流失率没变化, 还浪费了钱")
print(f"\n教训: 数据质量不仅影响'报表准不准'——更影响'模型和决策是否有效'")
````

## 📊 区块4：业务场景应用

### 场景1：一次"假数据"引发的灾难——流失预测模型的AUC骗局

某电商分析师做了一个"用户流失预测模型"，AUC=0.95（极好！）。团队部署到生产，对"预测会流失的用户"发优惠券挽留。发了100万的券——流失率没变化。

**排查发现**：
1. 模型学到的是"order_date=2099-01-01→一定流失"——这是一个ETL bug，流失用户的最后活跃日期被填入了默认值2099-01-01
2. 模型不是在学习"什么行为预示流失"，而是在学"哪个字段=2099-01-01"
3. 部署后，模型预测"会流失的用户"其实是"已经有2099标记的用户"——但这批用户已经流失了，发券也没用

**如果做了数据嗅探**：发现"最后活跃日期"的max=2099-01-01→这就暴露了bug。但没人做嗅探——所有人都被AUC=0.95的好消息冲昏了头。

### 场景2：两个部门的GMV对不上——口径战争的终结

财务部："本月GMV=5200万"
运营部："本月GMV=5500万"
→ 300万差在哪？

**排查**：
- 财务的GMV=支付成功+已结算-退款（严格口径）
- 运营的GMV=下单金额（不论支付成功与否，不论是否退款）

**300万差异=支付失败订单+退款订单+未结算订单**

**解决**：不是"哪个对哪个错"——而是"需要统一口径"。以后所有对外汇报用财务口径（最保守），内部运营分析可以用下单口径（但必须标注"含未支付和退款"）。

### 场景3：数据质量"文化"的培养——从"追责"到"学习"

**错误做法**：数据分析出错了→追责"谁用了脏数据"→下次大家都不敢用数据→数据文化变成"多一事不如少一事"。

**正确做法**：数据出错了→复盘"这个错误怎么被发现的？为什么没有更早发现？"→在流程中加一道防线（比如数据嗅探SOP）→把这个案例加入团队的"数据质量案例库"→团队从错误中学习，而不是从错误中恐惧。

**答案提示**：监控规则=范围校验+空值率+枚举值+波动检测。大小区分=核心字段vs非核心字段。权衡=分析用途决定质量要求。

### 🔴 高级检验

**答案提示**：监控体系=收集→检测→告警→修复→反馈闭环。文化=从追责到学习，建立案例库。治理上游=问题的根因不在数据，在产生数据的流程。

## 📝 区块7：当日小结

> **分析师的第一个动作永远是"嗅探数据"，而不是"开始分析"。花5分钟检查行数、空值、极值、时间范围、枚举值，能帮你避免花2小时得出一个"数据bug产生的假洞察"。当你发现数据异常时，追查到底——不要觉得"这是数仓的问题"。分析师是数据质量的最后一道防线：你最经常看数据、最了解数据"应该长什么样"。GIGO原则是数据分析的第一性原理——垃圾进，垃圾出，无例外。**

**三条金句带回家**：
1. GIGO——模型再精妙、可视化再炫，数据源是脏的，一切分析都是垃圾
2. 5分钟嗅探 > 2小时假分析。数据嗅探是数据分析中ROI最高的投入
3. 口径不一致是两个部门为数据吵架的头号原因——不是数据错了，是定义不同
