# 拓展：分析思维的里子——从「知道五步」到「每一步能站稳」

> **本拓展讲什么**：主课教了五步循环（提问→拆解→取数→分析→落地）。本拓展告诉你**每一步里面最容易翻车的地方**——为什么大多数分析死在第一步（问题没问对），为什么最后一步最容易说「建议继续观察」。
> **适合谁**：刚学完周一主课，知道五步循环的名字，但真拿到一个业务问题还是不知道从哪开始。
> **预计时长**：约 1.5h（全文本，无需代码）

# 第一部分：沙盘设定

> 本拓展第一次引入「快达出行」沙盘。后面周二到周六的拓展都会在这个沙盘里继续。

## 快达出行——分析师的第一天

**今天是你入职快达出行的第一天。**

**快达出行**是一个网约车平台（和滴滴类似），B 轮，覆盖 30 个城市。每天约 120 万人打开 App 叫车，平台上活跃着约 8 万个司机。

**上午 10:00**，你的直属领导——产品 VP 李哥——走过来，撂下一句话：

> 「最近用户好像不太满意。帮我看看怎么回事。」

然后他就走了。

你打开电脑，盯着这句话。**「最近」是多近？「不太满意」是从什么数据看出来的？「怎么回事」——你想让我看什么？**

这就是数据分析的日常。没人会给你一个「写好了的问题」——他们给你一个**模糊的感觉**。你的工作，是从这个模糊的感觉出发，走到一个具体的、可验证的、能落地的答案。

**本拓展的沙盘任务**：把李哥这句「最近用户好像不太满意」，变成一个你能用数据回答的问题。

# 第二部分：入口自诊

**Q1**：五步循环是哪五步？每一步的核心任务用一句话写。

**Q2**：李哥说「最近用户好像不太满意」。用主课学的「翻译官」思路，把这个模糊的感觉改写成 1-2 个**可量化的问题**。

**Q3**：主课说「分析始于问题，终于行动」。如果分析做完了，你的建议是「建议继续观察」——这在主课里被称为什么？为什么它不好？

---

### 自诊反馈

| 题号 | ✅ 参考答案 | ❌ 如果答偏了 |
|------|----------|----------|
| Q1 | 提问（把模糊需求变可量化问题）→拆解（把大问题拆成小问题）→取数（拿需要的数据）→分析（计算对比找模式）→落地（翻译成业务听得懂的建议） | 混淆拆解和取数→「先拉数据再拆」vs「先拆再决定拉什么数据」是完全不同的效率 |
| Q2 | 「过去 30 天，乘客评分（或等待时长/取消率）的趋势是什么？和 30-60 天前比变化了多少？哪个城市/时段的变化最大？」— 注意：问题里有时间、有指标、有对比基准 | 只说了「看看乘客评分」→没有时间范围、没有对比基准、没有拆解方向。这本质上还是「帮我看看数据」——只是换了个指标名 |
| Q3 | 「建议继续观察」= 没有结论。它意味着你的分析没有产生任何价值——你没有告诉业务方「问题在哪」或「应该做什么」。它被主课称为「分析师最贵的五个字」 | 认为「有时候确实需要继续观察」→如果一个分析没有结论，你应该说「当前数据不足以判断，建议收集 XX 数据后再分析」，而不是「继续观察」——后者是在假装有结论 |

# 第三部分：概念网络展开

## 🔴 核心深化：五步循环——每一步在消除什么不确定性

> 主课告诉你五步循环的顺序。但你有没有想过——**为什么是这个顺序？为什么不能跳过提问直接拆解？为什么不能跳过拆解直接取数？**

### 🔴 认知冲突

回到快达。李哥说「最近用户好像不太满意」。

**分析师 A（跳过提问，直接取数）**：打开后台，把最近能拉的数据全拉出来——DAU、完单量、等待时长、评分、取消率、投诉量……拉了几十兆数据，开始一个个看哪个跌了。两小时后，他发现「等待时长从 5 分钟涨到了 5.3 分钟」。于是汇报：「等待时长涨了 6%，可能是用户不满意的原因。」

**分析师 B（先提问再动手）**：在李哥走之前多问了三个问题——
- 「你说的『最近』是指这个月还是这个季度？」→ 「这个月。」
- 「『不满意』——是投诉变多了，还是评分跌了，还是完单量跌了？」→ 「评分跌了，而且客服那边说投诉也在涨。」
- 「你怀疑是什么原因？」→ 「上个月我们改版了 App，我怀疑跟改版有关。」

然后 B 花了 5 分钟想清楚：要拉的数据 = 改版前后各 30 天的评分趋势 + 按城市/版本拆开 + 改版涉及的功能页面投诉分类。15 分钟后，他定位到「改版后『取消订单』按钮被移到了不易发现的位置→用户误触后找不到取消→直接投诉」。

**A 花了两小时，B 花了 20 分钟。A 只找到了一个模糊的「等待时长涨了」，B 找到了具体的根因和修复方向。**

差别不在技术能力——在「提问」这一步。

> 五步循环的顺序不是教条。每一步都在消除上一层的模糊性——跳过一步，就是把模糊性带到了下一步，代价是**时间**。

### [必修] 每一步消除什么——以及跳过它的代价

**① 一句话白话**：五步循环就像剥洋葱——每剥一层，不确定性缩小一个数量级。跳到下一层之前，确认当前层已经「剥干净了」。

**② 五步的不确定性地图**：

| 步骤 | 消除什么不确定性 | 这一步的输出 | 跳过它的代价 | 快达案例 |
|------|---------------|-----------|-----------|--------|
| **提问** | 「我到底要回答什么问题？」 | 一个带时间范围+指标+对比基准的**可量化问题** | 你不知道该拉什么数据——于是什么都拉，浪费几小时 | 李哥说「用户不满意」→ 你翻译成：「过去 30 天，乘客评分均值是否低于前 30 天？如果是，哪个城市/场景的跌幅最大？」 |
| **拆解** | 「这个问题可以从哪几个角度去回答？」 | 一个**分析框架**（如按城市拆、按时段拆、按用户类型拆） | 你拉了一堆数据但不知道怎么看——「等待时长涨了，然后呢？」 | B 拆解：「评分跌了 = 评分降低的订单变多了。这些订单集中在哪个环节——下单时？等车时？完单后？」 → 三个假设，分别验证 |
| **取数** | 「需要哪些数据来验证我的拆解假设？」 | 一组**针对性数据**（不是全部数据，而是刚好够验证假设的数据） | 你被无关数据淹没，或者漏了关键数据 | B 取数：改版前后各 30 天的评分数据 + 按页面/功能分类的投诉数据（不是全部 35 张表的全部字段） |
| **分析** | 「数据告诉了我什么——是哪个假设成立？」 | 一个**验证过的根因**（至少排除了 2-3 个备选假设） | 你报告「等待时长涨了 6%」——这是描述，不是分析。分析 = 「等待时长涨了 6%，但这 6% 里 80% 是因为成都派单系统故障，跟改版无关」 | B 分析：投诉集中在「取消订单」功能 → 对比改版前后的投诉量 → 改版后投诉翻了 3 倍 → 确认根因是「取消按钮位置变更」 |
| **落地** | 「所以呢？谁该做什么？」 | 一个**有负责人+有时间线+有验证标准的建议** | 你的分析再好也没人用——「我知道了，然后呢？」 | B 落地：「建议回滚取消按钮到原位置，或者在新位置加气泡引导。产品团队 1 周内上线修复。上线后 2 周对比投诉量。」 |

**③ 核心洞察**：每一步的输出，就是下一步的输入。提问不清→拆解方向偏→取了错误的数据→分析不出有用的结论→给不出可行动的建议。五步循环之所以是**循环**——是因为落地之后很可能发现新的问题，又回到第一步。

**④ 与核心概念的关系**：这不是新概念——而是理解五步循环的「为什么是这个顺序」。以后你拿到任何分析需求，第一反应不是「我需要什么数据」，而是「我要回答什么问题」——这个顺序本身就是分析师的职业本能。

### 练习：诊断一个「死在第一步」的分析

快达运营王姐找到你：「最近投放效果好像变差了，帮我看看。」

以下是一个分析师接到这个需求后的第一反应。请诊断——他在「提问」这一步漏了什么？

> 「好的。我拉一下最近三个月的投放数据——各渠道的花费、曝光、点击、转化。然后看哪个渠道的 ROI 在跌。」

---

### 参考诊断

他漏了三个关键追问——每漏一个，后面的分析就多一分跑偏的风险：

| 漏了什么 | 为什么重要 | 应该怎么问 |
|--------|---------|----------|
| **没问「什么叫变差」** | 「投放效果」可能指：ROI 降了？CPA 涨了？带来的用户次留低了？还是「花的钱没变但完单量涨得慢了」？不同的「差」对应完全不同的分析方向 | 「你觉得哪个数字最能反映『效果变差』——是 ROI、CPA、还是新用户的留存？」 |
| **没问「和什么比」** | 「变差」= 和上个月比？和去年同期比？还是「没达到预期」？如果 ROI 从 3.0 跌到 2.8，但目标就是 2.5——那其实不差 | 「你是跟什么对比觉得变差了——上个月？去年 Q2？还是 KPI 目标？」 |
| **没问「你怀疑是什么原因」** | 王姐管投放，她每天盯着渠道数据。她可能已经有猜测（「我觉得是信息流渠道的质量在下降」）——不问就浪费了她的领域知识 | 「你有没有初步的猜测——是哪个渠道出的问题？」 |

**修正后的第一回应**：「好的。我先确认一下——你说的『投放效果变差』是指 ROI 跌了还是 CPA 涨了？跟什么时间段对比？你有怀疑是哪个渠道的问题吗？我拉这三个方向的数据，下午给你初步结论。」

## 概念群 A：[前置] 翻译的实操模板——从「老板说」到「数据能回答的问题」

> **与核心概念的关系**：主课说分析师是「翻译官」。但怎么翻？本概念给你一个具体的模板——三个追问 + 一个转译句式。这是五步循环中「提问」步的 [前置] 技能——不学会翻译，提问那步就永远是瞎猜。

### [必修] 三追问 + 一个转译句式

**① 一句话白话**：老板说的是人话（「用户不满意」），你需要把它翻译成数据能听懂的话（「过去 30 天评分是否低于前 30 天」）。模板就是三句追问 + 一句转译。

**② 模板**：

| 步骤 | 做什么 | 问什么 | 快达案例 |
|------|--------|--------|--------|
| **追问 1** | 锁定指标 | 「你说的『不满意』，是用什么数字来衡量的？评分跌了？投诉多了？完单量跌了？还是取消率涨了？」 | 李哥说「评分跌了，客服投诉也在涨」→ 锁定两个指标：乘客评分 + 投诉量 |
| **追问 2** | 锁定时间 | 「『最近』是多近？这个月？最近两周？这个季度？跟什么时间对比？」 | 李哥说「这个月，跟上个月比」→ 锁定：过去 30 天 vs 前 30 天 |
| **追问 3** | 锁定方向 | 「你有怀疑的原因吗？」（这一步是可选的——老板可能没有猜测。但如果有，省你两小时的拆解时间） | 李哥说「怀疑是上个月改版导致的」→ 拆解方向从「全量排查」缩小到「改版相关功能」 |
| **转译** | 写成一句话 | 「过去 [时间]，[指标] 比 [对比基准] [变化了多少]，[怀疑方向]」 | 「过去 30 天乘客评分比前 30 天跌了 0.3 分，投诉量涨了 40%。初步怀疑与改版有关。」 |

**③ 关键**：追问 1 和 2 是**必须的**——不加时间范围和指标的问题，本质上还是「帮我看看数据」。追问 3 是加分项——不是每次都有答案，但养成问的习惯，你会在老板的直觉里发现很多有价值的线索。

**④ 与核心概念的关系**：这是「提问」步的 [前置] 技能。不学会翻译，你连一个可量化的问题都写不出来——五步循环在第一步就卡住了。

### 练习：翻译下面的模糊需求

将以下 3 条模糊需求翻译成可量化的问题（用转译句式）：

**需求 1**：「最近我们的增长好像慢下来了。」
**需求 2**：「新上的推荐算法有没有效果？」
**需求 3**：「我们发的优惠券到底有没有用？」

### 参考翻译

| 需求 | 翻译 |
|------|------|
| 1 | 「过去 3 个月，月新增乘客数的环比增速是否在下降？如果是，下降最快的是哪个渠道？」（追问 1 锁定了「新增乘客数」作为增长的度量，追问 2 锁定了「过去 3 个月」和「环比增速」作为对比维度） |

## 概念群 B：[并行] 假设驱动 vs 数据驱动——不是二选一，是两种武器

> **与核心概念的关系**：主课说「先猜再验证」（假设驱动）比「先拉数据再找规律」（数据驱动）更高效。但现实里不是所有场景都适合先猜。本概念告诉你——什么时候先猜，什么时候先看。

### [必修] 两种模式的选择标准

**① 一句话白话**：如果你对这个业务很熟（知道什么因素通常会导致什么问题）→ 先猜再验。如果你对这个业务不熟（刚接手、新业务、没历史经验）→ 先看再猜。

**② 选择标准**：

| 场景 | 用什么模式 | 为什么 | 快达案例 |
|------|----------|--------|--------|
| **你对业务很熟**，拿到的需求和过去类似 | 假设驱动（先猜再验） | 你的经验告诉你「完单量跌了，80% 的概率是接单率的问题」→ 直接验证接单率，5 分钟定位 | 老分析师接到「完单量跌了」→ 第一反应「查接单率和取消率」→ 而不是「把全部 O 层指标拉一遍」 |
| **你对业务不熟**，刚接手一个新领域 | 数据驱动（先看再猜） | 你没有经验可以依赖，先拉数据做探索性分析，建立对数据的直觉，再形成假设 | 你刚入职快达，接到「用户不满意」→ 你不知道快达的「不满意」通常来自等待时长还是价格还是 App bug。先拉评分/投诉/取消率的整体趋势看一下，形成初步判断后再深挖 |
| **数据量太大**，探索性分析太费时间 | 假设驱动 | 几百张表、几千个指标——不可能扫一遍。必须靠业务经验缩到 3-5 个最可能的假设 | 快达有 6 张核心表、35 个看板指标。靠经验缩到 3 个假设：接单率跌了？取消率涨了？高峰运力不足？逐一验证 |
| **业务发生了结构性变化**（新产品、新市场、新策略） | 数据驱动 | 老经验可能失效——过去「涨价→完单量跌 10%」的规律在「竞品也涨价」的环境下不再成立。需要重新探索数据 | 快达新开了拼车业务——老分析师对快车的经验在拼车上不一定适用。先拉拼车的完整数据做探索 |

**③ 核心原则**：假设驱动的效率优势来自「缩小的搜索空间」。但如果你的假设是错的——你缩小的方向就是错的。**所以假设驱动的关键是「假设的质量」，而假设的质量来自你对此类问题的经验积累。**没有经验的人用假设驱动 = 在错的路上加速。

**④ 与核心概念的关系**：这是「提问→拆解」之间的一道工序——在拆解之前，先判断「我该先猜还是先看」。它决定了拆解的粒度和取数的范围。

### 参考分析

| 场景 | 模式 | 为什么 |
|------|:--:|------|
| 2 年老分析师 + 常见问题 | 假设驱动 | 你对完单量波动的常见原因有肌肉记忆——直接验证「接单率/取消率/运力供给」三个假设 |
| 刚入职 + 陌生业务 | 数据驱动 | 你不知道快达司机的典型留存曲线长什么样、通常什么因素影响留存。先拉数据建立直觉，再形成假设 |
| 新业务 + 无经验 | 数据驱动 | 老经验在新业务上可能失效——快车的价格弹性 ≠ 顺风车的价格弹性。先探索 |
| 老问题 + 经验丰富 | 假设驱动 | 你知道大促 ROI 的核心是「增量识别」（多少 GMV 是真实的增量 vs 时间转移 vs 本来就要消费）。直接套用你写过 5 次的框架 |

## 概念群 C：[防护] 分析的三种死法——为什么大多数分析死在最后一公里

> **与核心概念的关系**：主课说「分析不落地 = 没做」。本概念告诉你落地为什么这么难——三种常见的「假落地」，以及怎么让建议真的被采纳。

### [必修] 三种假落地——以及真落地长什么样

**① 一句话白话**：分析做完了，交了一个报告。但报告被打开了一次，然后躺在邮箱里再也没人看过——这就是假落地。真落地 = 有人因为你的分析，做了一个**和之前不同的决策**。

**② 三种假落地 vs 真落地**：

| 假落地 | 表现 | 为什么没用 | 怎么改成真落地 |
|--------|------|----------|-------------|
| **「建议继续观察」** | 分析做了 3 天，结论是「数据还不够，需要继续观察」 | 你在把决策推回给老板——「我不知道该怎么办，你再等等看」。老板不需要你告诉他「再等等」——他自己也会等 | 如果数据真的不够，你应该说：「当前数据只够判断『问题确实存在』但不够判断『根因是什么』。建议做 XX 动作（如限时 AB 实验/特定城市定向调研），2 周后可以获得足够数据做决策。」——你给了他一个**具体的下一步** |
| **「建议进一步分析」** | 报告末尾写了「需要进一步分析 XX 方向」 | 你在暗示「这次分析没做完」。如果分析没做完，在先不要交报告——先把「进一步分析」做了再交 | 如果你真的没做完——诚实说「还在分析中，预计周五一稿」。不要用「建议进一步分析」来凑报告结尾。这跟「建议继续观察」一样，都是「没有结论」的马甲 |
| **「数据表明存在相关性」** | 你发现了「等待时长和用户评分呈负相关」，但不下结论 | 你在给一个学术论文的结论，不是业务决策的依据。老板不需要知道「r=-0.6」，他需要知道「所以应该缩短等待时长吗？预期能提升多少评分？」 | 从相关性走到建议：「等待时长每增加 1 分钟，评分降低约 0.15 分。当前等待时长 6.5 分钟→如果能通过 XX 策略降到 5 分钟，预计评分回升约 0.2 分。成本预估 XX 元，建议试点成都。」——**把统计发现翻译成了「要不要做 + 做了预期什么效果」** |

**③ 真落地的三个要素**：

| 要素 | 什么意思 | 例子 |
|------|--------|------|
| **可执行** | 运营/产品拿到就能做，不需要再翻译 | 「建议回滚取消按钮到原位置」——李哥今天就能安排。而不是「建议改善用户体验」——这跟没说一样 |
| **有预期** | 做了之后，预计什么指标会怎么变化 | 「预计投诉量回落到改版前水平（降约 60%），2 周后验证」 |
| **有负责人** | 明确谁来做——不是「运营应该」，而是「@王姐」 | 「@李哥（产品）1 周内修复，@你（数据）2 周后验证效果」 |

**④ 与核心概念的关系**：这是五步循环「落地」步的 [防护] 版——防止你以为「交了报告 = 完成了分析」。真正的分析终点不是「报告发出」，而是「有人因为你的分析做了一个不同的决策」。

### 练习：把假落地改成真落地

以下 3 条是快达分析师常见的报告结尾。请把每一条改成真落地。

**假落地 1**：「Q2 成都完单量同比下降 15%。建议继续观察后续趋势。」
**假落地 2**：「数据表明司机收入水平与留存率存在显著正相关。建议进一步分析不同收入段的留存差异。」
**假落地 3**：「本次分析发现峰值时段（17-19点）运力不足是等待时长恶化的主因。建议考虑增加高峰时段运力。」

---

### 参考改进

| # | 真落地版 | 改了什么 |
|----|--------|--------|
| 1 | 「成都完单量同比下降 15%——主因是竞品 5 月在成都上线并做了 3 个月的新客补贴。快达在成都的用户正在被竞品蚕食。建议：(1)@王姐，在成都做定向补贴（预计成本 25 万/月），重点挽回过去 3 个月流失的老用户；(2) 同步在成都上线『邀请好友』功能。2 周后对比完单量恢复情况。」 | 加了根因（竞品）、加了谁做（@王姐）、加了成本预估、加了验证时间 |
| 2 | 「司机收入每降低 10%，留存率降低约 8pp。当前成都司机月均收入 4200 元（全国 5300 元）——成都的司机留存率（55%）显著低于全国（68%）。建议：(1)@李哥，在成都试点『高峰保底』——高峰期接满 10 单的司机保底收入 200 元/天，预计月均收入提升至 4800 元；(2) 试行 1 个月后对比成都司机留存率变化。」 | 把「相关」翻译成了具体的数字关系、定位到了具体城市、给出了具体的干预方案和预期 |
| 3 | 「峰值运力不足的主因是『特惠快车』司机在高峰时段下线率高（42%，快车司机仅 18%）。特惠司机觉得高峰堵车不划算。建议：(1)@李哥，对特惠司机上线高峰时段『堵车补贴』（每堵 10 分钟补 2 元）；(2) 在成都试点 2 周，目标将特惠司机高峰下线率从 42% 降到 25%。2 周后对比等待时长变化。」 | 加了根因（不是笼统的「运力不足」，而是「特惠司机高峰下线率高」）、加了具体策略、加了量化目标 |

## 综合练习 2：沙盘变体——如果李哥是一个「不回答追问」的老板

现实中有些老板不会耐心回答你的三个追问——「你别问了，帮我看看数据就行」。

**问题**：这时候你怎么办？直接拉数据吗？

---

### 参考策略

**不要直接拉数据。**「帮我看看数据」= 从 6 张表的几千个字段里「看看」——你会浪费一整天。

**换一种方式追问**：把开放式问题变成选择题。「李哥，我理解你想了解用户不满意的原因。我准备从三个方向查——(A)评分趋势 (B)投诉类别变化 (C)取消率变化。你看先从哪个方向开始？如果你选 A，我 1 小时内给你初步结论。」

**为什么这样有效？** 选择题的认知负荷远低于开放题——老板只需要点个头，而不是「重新思考自己的需求」。同时你在展示你的专业判断——你已经在脑子里拆解了方向，只是让他选优先级。

**最差情况**：三个方向他都不选。「都看看吧」——那你就按自己判断的优先级（A→B→C），做完一个汇报一个。不要三个方向同时拉数据——那样你会同时卡在三个方向上。

| 场景 | 当前层次 | 怎么升一层 |
|------|:--:|------|
| 运营：「帮我拉一下上周各城市的完单量。」分析师：「好的。」发了一张表。 | ? | ? |
| 运营：「完单量跌了，帮我看看。」分析师：「完单量跌了 10%，主因是成都接单率从 91% 跌到 53%。」 | ? | ? |
| 运营：「完单量跌了，帮我看看。」分析师：「成都接单率暴跌因为特惠车司机高峰下线率高。建议高峰时段给特惠司机堵车补贴，预计成都完单量回升 8%。要不要在成都试点 2 周？」 | ? | ? |

### 参考诊断

| 场景 | 层次 | 升一层的话 |
|------|:--:|------|
| 1 | **第一层（取数工具人）** | 「你是想对比各城市的完单量变化，还是想看哪个城市对整体跌幅贡献最大？我帮你把贡献度也算出来。」——从「你要什么我给什么」变成「我帮你把需求翻译成分析」 |
| 2 | **第二层（分析执行者）** | 「我建议对成都的特惠司机做高峰堵车补贴——因为下线的 70% 是特惠司机。预计上线 2 周后接单率能回到 85%。成本约 3 万/月。」——从「告诉你发生了什么」变成「告诉你应该做什么」 |
| 3 | **第三层（业务伙伴）** | 已经在第三层。继续升的话→「上个月的高峰堵车补贴在成都试行了 2 周，接单率从 53% 回到了 83%。建议推广到合肥和武汉——这两个城市的特惠车占比和成都类似。预计推广后全公司完单量提升 3%。」——从不只解决当前问题，到把一次分析变成可复制的策略 |
