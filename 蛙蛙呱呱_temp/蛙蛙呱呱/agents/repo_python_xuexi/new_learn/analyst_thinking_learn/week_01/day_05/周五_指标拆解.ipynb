## 一、核心概念与第一性原理

### 1.1 为什么需要拆解

老板问："这个月GMV为什么跌了？"——"GMV跌了"四个字是没法直接回答的。

你需要拆解它：
```
GMV = 订单数 × 客单价。是哪个跌了？
订单数 = 流量 × 转化率。是哪个跌了？
流量 = 自然流量 + 付费流量。是哪个跌了？
```

**拆解的本质：把一个你无法回答的大问题，变成一系列你可以直接回答的小问题。**

### 1.2 从第一性原理推导：拆解的逻辑基础

一个指标可以被拆解，当且仅当满足以下之一：

**乘法拆解**（按转化链路）：存在一条从"入口"到"出口"的转化链，每一步有明确的"通过率"。
- 公式形式：Y = A × B × C（每个因子是一个转化率或比率）
- 适用场景：GMV、收入、转化率等有明确"漏斗"关系的指标

**加法拆解**（按维度分类）：存在互斥且穷尽(MECE)的分类维度。
- 公式形式：Y = A + B + C（每个加数是某个维度的贡献）
- 适用场景：DAU（新+留+回）、GMV（各品类/渠道之和）、收入（各业务线之和）

### 1.3 拆解的终极目的：找到杠杆点

拆解不是为了"把公式写好看"，而是为了找到**哪个因子/维度对整体变化贡献最大**。

找到了，你就知道该在哪里用力——这就是"杠杆点"。没找到，你就会"所有环节都优化一点"——资源分散，效果微乎其微。

> **今天最重要的认知**：拆解的目的不是把数字变小，而是把"回答不了的问题"变成"回答得了的问题"。一个好的拆解应该有"可行动性"——拆到最底层的时候，你看到一个因子变了，就知道应该哪个团队去做什么事。

## 二、知识点详细拆解

### 2.1 乘法拆解：转化漏斗法

**为什么重要**：最经典的分析方法，适用于所有"有转化链路"的指标。

电商GMV的完整乘法拆解：
```
GMV = UV × (浏览→商品页转化率) × (商品页→加购率) × (加购→下单率)
       × (下单→支付率) × 客单价
```

拆解之后你可以精确回答：
- 是"来的人少了"(UV)还是"来了不买"(转化率)？
- 如果转化率跌了，是哪个环节的"筛子"变大了？

**乘法拆解的关键约束**：每个因子之间必须是"串联"关系——上一步的输出是下一步的输入。如果你拆出的因子之间没有这种串联关系，乘法拆解不成立。

### 2.2 加法拆解：维度下钻法

**为什么重要**：GMV跌了可能是"整体跌"也可能是"某个品类暴跌其他正常"——加法拆解帮你定位到具体维度。

```
GMV = 服装GMV + 数码GMV + 食品GMV + ...
GMV = 新用户GMV + 老用户GMV
GMV = 各渠道GMV之和
```

加法拆解的核心原则是**MECE（互斥且穷尽）**：
- 互斥(Mutually Exclusive)：每个用户/订单只属于一个分类。"新用户"和"老用户"互斥。
- 穷尽(Collectively Exhaustive)：所有分类加起来=全量。所有分类的GMV加起来=总GMV。

### 2.3 贡献度分析——拆解之后最重要的计算

**为什么重要**：拆解告诉你"哪些因子变了"，贡献度告诉你"每个因子的变化对整体的影响有多大"。

**乘法拆解的贡献度**（逐一替换法）：
```
GMV = 流量 × 转化率 × 客单价
上月: 100k × 5% × 200 = 1,000k
本月: 120k × 4% × 210 = 1,008k

仅流量变: 120k × 5% × 200 = 1,200k → 贡献 +200k
仅转化率变: 100k × 4% × 200 = 800k → 贡献 -200k
仅客单价变: 100k × 5% × 210 = 1,050k → 贡献 +50k
→ 流量+20%和转化率-20%的效果几乎完全抵消
```

**加法拆解的贡献度**：
```
每个分类的贡献度 = (该分类的变化量) / (整体变化量) × 100%
注意：看贡献度而非变化率！C分类涨了50%但如果只占总体的2%，影响微乎其微。
```

### 2.4 拆解的常见错误

| 错误 | 为什么错 | 正确做法 |
|------|---------|--------|
| 因子之间有重叠 | "DAU = 新增 + 活跃 + 付费"——活跃可能包含付费 | 维度互斥 |
| 只看变化率 | "C渠道涨了50%！"但只占总量的2% | 看贡献度 |
| 拆到底层无法行动 | "GMV = 用户数 × 人均订单 × 均价 × 天数"——但"天数"谁能控制？ | 拆到可行动层级 |
| 忽略因子之间的相关性 | 流量涨了转化率通常会跌（新流量质量差） | 考虑联动效应 |

## 三、动手练习

### 练习1：贡献度计算——逐一替换法（基础）

````python
# ===== 练习1：乘法拆解 + 贡献度 =====

traffic_last, cvr_last, asp_last = 100000, 0.05, 200
traffic_this, cvr_this, asp_this = 120000, 0.04, 210

gmv_last = traffic_last * cvr_last * asp_last
gmv_this = traffic_this * cvr_this * asp_this

print("GMV 拆解与贡献度分析")
print("=" * 50)
print(f"上月: {traffic_last:,} × {cvr_last:.0%} × {asp_last} = {gmv_last:,.0f}")
print(f"本月: {traffic_this:,} × {cvr_this:.0%} × {asp_this} = {gmv_this:,.0f}")
print(f"总变化: {gmv_this - gmv_last:+,.0f} ({(gmv_this/gmv_last-1)*100:+.1f}%)")
print()

# 逐一替换
gmv_t_only = traffic_this * cvr_last * asp_last
gmv_c_only = traffic_last * cvr_this * asp_last
gmv_a_only = traffic_last * cvr_last * asp_this

print("逐一替换贡献度:")
print(f"  仅流量:  {gmv_t_only:,.0f} (贡献 {gmv_t_only-gmv_last:+,.0f})")
print(f"  仅转化率:{gmv_c_only:,.0f} (贡献 {gmv_c_only-gmv_last:+,.0f})")
print(f"  仅客单价:{gmv_a_only:,.0f} (贡献 {gmv_a_only-gmv_last:+,.0f})")
print()
print("结论: 流量+20%和转化率-20%几乎完全抵消，GMV基本没变")
print("      → 转化率是杠杆点——把它从4%拉回到5%，能在不增加流量的情况下增长25%")
print()
print("【预期输出】流量贡献+200k, 转化率贡献-200k, 客单价贡献+50k, 净值+50k")
````

### 练习2：加法拆解——找"最大贡献者"（进阶）

````python
# ===== 练习2：加法拆解 + 贡献度 =====

channels = {
    "自然搜索": {"last": 500, "this": 480},
    "付费搜索": {"last": 300, "this": 250},
    "社交媒体": {"last": 120, "this": 100},
    "直接访问": {"last": 200, "this": 210},
    "邮件营销": {"last": 80,  "this": 85},
}

total_last = sum(d["last"] for d in channels.values())
total_this = sum(d["this"] for d in channels.values())
total_change = total_this - total_last

print("各渠道收入拆解与贡献度")
print("=" * 55)
print(f"总收入: {total_last} → {total_this} (变化 {total_change:+.0f})")
print()

print(f"{'渠道':<12} {'上月':>8} {'本月':>8} {'变化':>8} {'变化率':>8} {'贡献度':>8}")
print("-" * 55)
for name, d in channels.items():
    change = d["this"] - d["last"]
    change_rate = change / d["last"] * 100
    contribution = change / abs(total_change) * 100 if total_change != 0 else 0
    print(f"{name:<12} {d['last']:>8.0f} {d['this']:>8.0f} {change:>+8.0f} {change_rate:>+7.1f}% {contribution:>7.1f}%")

print()
print("关键发现:")
print(f"  最大负贡献: 付费搜索(-50)，占总变化的{abs(-50)/abs(total_change)*100:.0f}%")
print(f"  注意: 社交媒体变化率(-17%)比付费搜索(-17%)接近，")
print(f"        但贡献度更小（因为基数小）→ 优先排查付费搜索")
print()
print("【预期输出】付费搜索贡献最大负面变化，应优先排查")
````

## 四、业务场景应用

### 场景1：DAU涨了30%，但CEO说"没感觉产品变好"

某社交App的DAU从100万涨到130万。CEO说："DAU涨了但我没感觉产品变好了。"

你拆解DAU：
```
DAU = 新增用户 + 留存老用户 + 回流用户
拆开: 新增从10万→35万(+250%)，留存从70万→68万(-3%)，回流正常
再拆新增: 信息流从3万→25万(+733%)，自然新增从7万→10万(+43%)
```

结论: DAU增长≈全部来自信息流投放砸钱。产品的自然增长和留存都没改善。投放一停→DAU跌回原点。

**用vs不用**：不拆解→"DAU涨了30%，我们做得很好！"；拆解后→"我们只是在花钱买DAU，产品本身没有进步"。

### 场景2：GMV涨了但毛利率跌了——怎么判断"该不该继续"

运营做了一个"满200减50"活动，GMV涨了40%。但你需要拆解：
```
增量GMV = 券带来的GMV × 增量比例
增量毛利 = 增量GMV × 毛利率
补贴ROI = 增量毛利 / 补贴成本
```

如果40%的GMV增长中只有25%是真正的增量（其余是"本来就要买"和"提前消费"），ROI可能<1。

**用vs不用**：不拆解→"ROI=5，券有效！"；拆解后→"增量ROI=0.8，券在亏钱，需要优化"。

### 场景3：App内"搜索功能使用率"突然涨了50%——好事还是坏事？

直觉觉得"用户更爱用搜索了"。但拆解后：
- 搜索使用率 = 搜索用户数 / DAU
- 搜索用户数没变，是DAU跌了（分母变小了）
- DAU跌了是因为导航栏改版，很多用户找不到入口

真相：不是搜索变好了，是导航变差了，用户被迫用搜索。

**用vs不用**：不拆解→"搜索功能优化成功了！"；拆解后→"导航改版出了问题，用户被逼到搜索"。

## 五、分级检验清单

### 初级检验
- [ ] 能区分乘法拆解和加法拆解的适用场景
- [ ] 能拆解GMV（乘）和DAU（加）
- [ ] 能用逐一替换法在代码中计算各因子的贡献度

**答案提示**：乘法=有转化链路(GMV=流量×转化×客单价)；加法=按维度分类(DAU=新+留+回)；逐一替换=固定其他因子，只变一个，看GMV变化。

### 中级检验
- [ ] 能拆解一个陌生指标（给出至少两层拆解，乘法和加法都用到）
- [ ] 计算贡献度后能正确识别"主要矛盾"（是哪个因子驱动了大部分变化）
- [ ] 能判断拆解出来的因子是"可行动的"还是"纯数学的"（天数不可行动，转化率可行动）

**答案提示**：主要矛盾=贡献度绝对值最大的因子；可行动=有人能对这个数字直接负责。

### 高级检验
- [ ] 对互相依赖的因子（如流量↑通常导致转化率↓），能判断先动哪个
- [ ] 能在拆解结果中找到"惊喜"——即数据中违反直觉的部分（如搜索使用率涨了但可能是DAU跌了）
- [ ] 能用拆解结果说服业务方：不是"所有地方都优化一点"，而是"集中资源改善贡献度最大的那个因子"

**答案提示**：互相依赖时先动"上游"因子（UV影响所有下游）；惊喜=看起来好但拆解后发现坏（或反之）。
