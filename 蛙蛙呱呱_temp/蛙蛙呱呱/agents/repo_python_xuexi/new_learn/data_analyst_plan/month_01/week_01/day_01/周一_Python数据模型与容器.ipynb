## 概念1：变量与赋值 + 核心数据类型

### 🤔 变量——你确定真的理解 `=` 吗？

你写`x = 299`，Python做了什么？大多数人会说"把299放进x里"——**但你错了**。

**代码A**：
```python
a = [1, 2, 3]
b = a
b[0] = 999
print(a)  # 打印什么？
```

**代码B**：
```python
a = 100
b = a
b = 999
print(a)  # 打印什么？
```

正确答案：**A打印[999,2,3]，B打印100**。为什么同样是`b = a`，一个影响了a，另一个不影响？这就是"变量到底是什么"的问题。

### 变量与赋值

**① 说白了就是**：Python的变量不是"盒子"（把东西装进去），而是**便利贴**（贴在东西上）。`x = 299`的意思是"给299这个数字对象贴一张叫x的标签"。你可以随时把标签撕下来贴到别的东西上——`x = "hello"`就是把x标签从299身上撕下来，贴到"hello"上。

**② 在沙盘里怎么用**：小王给你的每笔订单的金额、商品名、城市——每个都是一个标签贴在一个值上。`amount = 299`就是给299贴个amount标签。用`id(amount)`查看"299这个对象在内存里的身份证号"——两个变量指向同一个对象时，它们id相同。变量名必须字母/下划线开头（不能数字开头！），避开`class/if/for/import`等关键字。多重赋值`a, b = 1, 2`一行搞定两个变量。

**③ 跟你已知的Excel的关系**：一个变量就像Excel里的一个单元格——但更灵活：不需要先"建表"，随时可用，同一变量可以先指向数字再指向文本（动态类型）。

**④ 为什么这样设计**：Python选择"标签贴在对象上"而非"盒子装东西"，是为了支持**动态类型**——你不需要像Java那样声明`int x = 5`。代价是：拼写错误的变量名要到运行时才发现，而且`lst2 = lst1`这种"贴第二张标签"的操作很容易被误解为"复制了一份"——这是Python生涯中最高频的bug之一。

### 核心数据类型

**① 说白了就是**：就像超市商品分"蔬菜""水果""日用品"——不同类型的东西能做不同的事。数字可以加减乘除，文本可以拼接，布尔值回答"是或否"。

**② 在沙盘里怎么用**：
- `int`：订单金额`299`——Python的int**不限制大小**（不像C语言int最大约21亿会溢出），`2**1000`是一个300多位的数字
- `float`：带小数的金额`299.99`——注意：**`0.1 + 0.2 != 0.3`**！IEEE 754双精度浮点数，二进制无法精确表示0.1（就像十进制无法精确表示1/3）
- `str`：商品名`"蓝牙耳机"`——字符串是**不可变的**：你不能修改字符串里的单个字符
- `bool`：是否大单`True`/`False`——真值测试：`bool(0)`=False、`bool([])`=False、`bool("")`=False
- `None`：表示"还没填"——**None不是0、不是空字符串、不是False**。判断None用`is None`而不是`== None`
- `type(x)`：随时查看"x现在是什么类型"——数据分析中探索陌生数据时最常用的工具

**③ 跟变量的关系**：变量是"标签"，类型是"标签指向的那个东西是什么种类"。同一个标签可以先贴在int上、再撕下来贴在str上——这就是动态类型。隐式转换：`1 + 2.0`自动变成`3.0`。

**④ 为什么这样设计**：动态类型让你在数据分析中写代码更快（你经常不知道一个变量最终会是什么类型），代价是类型错误要到运行时才发现——养成用`type()`检查的习惯。**int的任意精度**是Python区别于C/Java的重要设计——自动处理大整数。**float的IEEE 754**是硬件标准——理解`0.1+0.2!=0.3`是你迈入"真正理解计算机如何表示数据"的第一步。

````python
# 变量与数据类型——它们长这样：

# === 变量是"贴标签"，不是"装盒子" ===
x = 299          # 给299贴张标签叫x
print(id(x))     # 查看299在内存里的"身份证号"
y = x            # 给同一个299再贴张标签叫y（不是复制！）
print(id(y))     # 和x的id一样——它们指向同一个对象
x = "hello"      # 把x标签撕下来，贴到"hello"上
print(id(x))     # id变了——x现在指向全新的对象

# === 核心数据类型 ===
a = 2**1000      # int任意精度——300多位也不会溢出！
b = 0.1 + 0.2    # 看好了：这不等于0.3！
print(b)         # 0.30000000000000004
print(b == 0.3)  # False！
print(type(b))   # <class 'float'>
print(bool([]))  # False——空list是False
print(bool(0))   # False——0是False
c = None         # None只表示"没有值"
print(c == 0)    # False——None不是0
print(c is None) # True——用is判断是不是None
````

````python
# 🔨 变量与类型小练
# 1. 创建 name="速卖通", daily_orders=1523, avg_price=299.99, is_active=True, note=None
# 2. 用type()打印每个变量的类型
# 3. 用id()打印name的id，然后改成"AliExpress"，再打印id——变了吗？为什么？
# 4. 验证 0.1+0.2==0.3 是True还是False
# 5. 验证 bool("")、bool(0)、bool(None) 的值
# 6. 在注释里回答：为什么代码A改了b[0]影响a，代码B改了b不影响a？
#因为本质上a是一个变量，而b是一个标签，所以代码A改了b[0]影响a，而代码B改了b不影响a。
# 👇 在下面写你的代码：
name = "速卖通"
daily_orders = 1523
avg_price = 299.99
is_active = True
note = None
print(id(name))
name = "AliExpress"
print(id(name))
print(0.1+0.2==0.3)
bool("")
bool(0)
bool(None)
````

## 概念2：list —— 一排格子，有顺序，可修改

### 🤔 5笔订单金额散落一地——你怎么"一把抓"？

你有5笔订单金额：299、1599、89、499、2599。用变量只能`amount_1=299; amount_2=1599...`创建5个变量。

小王问你"第3笔多少钱"——你得记住变量名叫amount_3。"一共几笔"——手动数。"来了一笔新的"——手动创建amount_6。"哪笔最大"——用眼睛看。

500笔订单呢？你不可能创建500个变量。你需要**一个东西，能把一堆数据按顺序装在一起，通过"第几个"来找，随时加新的进去**。

你刚才撞到的这个墙，`list`就是专门解决它的。

### list

**① 说白了就是**：list就像你的购物清单——从上到下一项一项写，每项占一个位置，有先后顺序。你可以加新的到末尾、删掉不要的、改某一项、一眼看到"第3样是什么"。

**② 在沙盘里怎么用**：5笔订单金额全部放进一个list：`[299, 1599, 89, 499, 2599]`。小王问"第3笔多少钱"→`amounts[2]`（注意！Python从0开始数，第3笔的位置是2！）。小王说"又来了一笔699的"→`amounts.append(699)`一行搞定。想知道共几笔→`len(amounts)`。最高金额→`max(amounts)`。有没有1599元的订单→`1599 in amounts`。

**③ 跟你刚学的变量的关系**：变量一次贴一个值，list一次贴一堆值。`lst = [a, b, c]`——把三个变量的值放进一个list统一管理。`lst[0]`取第一个——你不需要记住"那个值叫什么名字"，只需要知道"它是第几个"。

**④ 为什么这样设计**：Python的list底层是**动态数组**——在内存里是一段连续空间，每个格子存一个元素的"内存地址"（所以list可以同时存不同类型的东西）。`append()`加到最后**均摊O(1)**——大部分时候直接放末尾空格子，偶尔满了才"搬家"（申请更大空间→复制过去）。但`insert(0, val)`在开头插入需要把后面所有元素往后移一格→**O(n)**。理解这个=你知道什么时候该用list：**频繁在末尾操作→list高效；频繁在中间或开头插入删除→考虑其他结构**。

**最最重要的陷阱**：`lst2 = lst1`就是给同一个list对象再贴一张标签！改`lst2[0]`就是改同一个list→`lst1`也"被改了"。要真正复制一份独立list→用`lst1.copy()`或`lst1[:]`。

````python
# list长这样——注意：Python从0开始数！
amounts = [299, 1599, 89, 499, 2599]  # 5笔速卖通订单金额
print(amounts[0])     # 第1笔 → 299（不是1！是0！）
print(amounts[-1])    # 最后1笔 → 2599（-1=倒数第一）
print(amounts[1:4])   # 第2到第4笔 → [1599, 89, 499]（含左不含右）
amounts.append(699)   # 新订单699元，加到末尾
print(len(amounts))   # 现在共6笔
print(2599 in amounts) # 有2599元的订单吗？→ True

# ⚠️ 陷阱演示：lst2=lst1不是复制！
lst1 = [1, 2, 3]
lst2 = lst1          # 两张标签贴在同一个list上
lst2[0] = 999
print(lst1)          # [999, 2, 3]——lst1也被改了！
lst2 = lst1.copy()   # ✅ 这才是真正的复制
````

````python
# 🔨 list小练：一周销售额管理
# 速卖通(电商) | 武器：list
# 小王给了你本周5天成交额（万元）：周一12、周二8、周三15、周四10、周五22
# 1. 放进 weekly_sales list
# 2. 取出周三（第3个，位置2）→ 打印
# 3. 周六加了18万，append → 打印共几天
# 4. sum算总成交额、max找最高 → 打印
# 预期输出：周三15 / 共6天 / 总额85 / 最高22

# 👇 在下面写你的代码：
weekly_sales = [12, 8, 15, 10, 22]
print(weekly_sales[2])
weekly_sales.append(18)
print(weekly_sales)
print(len(weekly_sales))
print(sum(weekly_sales))
print(max(weekly_sales))
````

## 概念3：tuple —— 有顺序，但不可修改

### 🤔 如果有个list你永远不想让人改——怎么办？

你用list存了北京分公司经纬度：`bj_location = [39.9, 116.4]`。

然后你不小心敲了`bj_location[0] = 40.0`——把纬度改了。代码不报错、不警告——它就默默地让你的北京分公司飘到了内蒙古上空。

有些数据**天然不该被修改**：坐标、身份证号、出生日期、数学常数。list允许你改——但"允许改"本身就是风险。

更关键的是：dict（马上要学）要求"钥匙"必须永远不变。list是可变的→不能做钥匙。那谁来做钥匙？

你刚才撞到的这两个问题——"不可修改的数据组"和"能做钥匙的序列"——`tuple`就是专门解决它们的。

### tuple

**① 说白了就是**：tuple就像身份证上的信息——印好了不能涂改。list是"可修改的购物清单"，tuple是"固定下来的记录"。

**② 在沙盘里怎么用**：GPS坐标用tuple存——`(39.9, 116.4)`，不应被修改。函数返回多个值时，背后就是tuple：`return total, count`实际返回`(total, count)`。

**③ 跟list的关系**：list和tuple都是"有顺序的一排格子"。list可修改，tuple不可修改。**为什么有list还要tuple？**三个原因：①不可变=可哈希→能做dict的key（list不能！面试高频题）②不可变=安全（约束"这东西不该被改"）③性能略好（Python知道tuple不会变，可做优化）。

**④ 为什么这样设计**：Python设计tuple的初衷不是"不可变的list"——而是**轻量级的、用于打包一组相关数据的结构**。注意：**逗号才是tuple的本质，括号只是辅助**。`1, 2, 3`不加括号也是tuple。这就是为什么`(1)`不是tuple而是int——括号被数学表达式占用了语义。必须`(1,)`加逗号！这个设计细节让无数新手——包括有经验的程序员——踩坑。

````python
# tuple长这样——逗号是本质，括号是辅助
coord = (39.9, 116.4)  # 北京经纬度
print(coord[0])         # 可以取值，但不能coord[0]=40.0（报错！）

# ⚠️ 经典陷阱：(1)不是tuple！
not_a_tuple = (1)
print(type(not_a_tuple))  # <class 'int'>！！
real_tuple = (1,)          # ✅ 逗号告诉Python这是tuple
print(type(real_tuple))    # <class 'tuple'>

# tuple能做dict的key（不可变→可哈希），list不能！
city_coords = {(39.9, 116.4): "北京", (31.2, 121.5): "上海"}
print(city_coords[(39.9, 116.4)])  # 北京
````

## 概念4：dict —— 标签贴纸，按名字找，O(1)快

### 🤔 list只知道"第几个"，但你想知道"叫什么"——怎么办？

你用list存了一笔订单：`order = ["ORD001", "蓝牙耳机", 299, "北京"]`。

小王问你"这笔订单多少钱"——你得记住"金额在位置2"（因为第3个，从0开始）。六个月后回来看代码：`order[2]`——这是什么？翻到前面找注释……哦，是金额。

这叫做**隐式知识**："位置2=金额"这个信息只存在于你脑子里，没有写在代码里。更严重的是：如果订单结构变了——"运费"字段插在中间——所有位置都往后移！`order[2]`可能就变成了"运费"。这种bug极难排查（代码不报错，只是取错的值）。

你需要一种**"按名字取，而不是按位置取"**的容器。"金额"就叫"金额"，不管它在前面还是后面。

你刚才撞到的这个墙，`dict`就是专门解决它的。

### dict

**① 说白了就是**：dict就像手机通讯录——要找"张三"的电话，不是一页一页翻到第37个（list做法），而是直接搜"张三"，瞬间定位。dict就是**给每个数据贴一张标签（key）**，通过标签来找数据（value），而不是通过"第几个"。

**② 在沙盘里怎么用**：一笔订单有三个信息——订单编号、商品名、金额。list只能`["ORD001", "蓝牙耳机", 299]`，你得记住"第3个是金额"。dict用标签来存：`{"订单号": "ORD001", "商品": "蓝牙耳机", "金额": 299}`，直接查`order["金额"]`——**自文档化**：代码本身就告诉你"我要的是金额"。

**③ 跟list的关系**：list是"按位置找人"（第3个是谁），dict是"按名字找人"（叫"金额"的那个是谁）。**什么时候用list？**数据是同一种东西（一堆金额，按时间排列）。**什么时候用dict？**一个东西有多种属性（一笔订单有编号、商品、金额、城市——属性名各不相同）。list套dict=数据分析中最常用的数据结构。

**④ 为什么这样设计**：dict底层是**哈希表**——通过哈希函数把key映射到存储位置→查找/插入/删除都是**O(1)**（不管dict有100条还是100万条，查找几乎瞬间完成）。而list的查找是O(n)。代价：①**key必须是不可变类型**（str/int/tuple）——因为可变对象（list）的内容变了→哈希值也变了→dict在原来位置找不到这个key→数据结构乱套。这就是为什么list不能做dict key而tuple能——今天最重要的知识点，面试高频题。②dict比list占用更多内存（需要维护哈希表）。

**额外提醒**：`d.get("key", default)`比`d["key"]`更安全——key不存在时返回默认值而不是直接崩溃。数据分析中处理不完整数据时，get是你的好朋友。

````python
# dict长这样——花括号，key:value对，按名字查不是按位置查
order = {"订单号": "ORD001", "商品": "蓝牙耳机", "金额": 299, "城市": "北京"}
print(order["金额"])         # 299——直接喊"金额"，不用记位置
print(order.get("备注", "无")) # "无"——安全查，key不存在不报错
order["城市"] = "上海"       # 修改
order["是否大单"] = True     # 新增
del order["是否大单"]        # 删除


# 遍历：三种方式
for k in order:              # 只拿标签
    print(k)
for k, v in order.items():   # 标签+值（最常用）
    print(f"{k}: {v}")

# ⚠️ key必须不可变——list不能做key！
# d = {[1,2]: "a"}  → TypeError: unhashable type: 'list'
d = {(1,2): "a"}     # ✅ tuple可以
````

## 概念5：set —— 不重复的袋子，无顺序，判断"在不在"极快

### 🤔 500笔订单来自哪些城市？用list做这件事有多慢？

小王问"上周有哪些城市下过订单"。你的list里有500条记录，很多城市重复出现。

**用list的做法**：遍历500条，碰到没见过的城市就加到另一个list。每次都要检查"这个城市在不在新list里"——新list越长，检查越慢。100万条数据呢？

**用set的做法**：`unique_cities = set(cities)`——一行，自动去重，瞬间完成。

更关键的是"判断在不在"：`"杭州" in city_list`（list）→一个一个比较，最坏500次。`"杭州" in city_set`（set）→算哈希值→直接定位，1次。500条差500倍，100万条差100万倍。

你刚才撞到的——"去重"和"极速判断在不在"——`set`就是专门解决它们的。

### set

**① 说白了就是**：set就像你去超市买的"所有商品种类"——苹果买了3次但在"种类集合"里只出现1次。set自动去重，且能极快回答"XX在不在里面"。

**② 在沙盘里怎么用**：小王问"哪些城市下过订单"——直接把所有城市放进set→自动去重。想知道"北京有没有订单"→`"北京" in cities`（O(1)）。想做"在北京和上海都下过单的用户"→`北京用户 & 上海用户`（交集运算，一行搞定）。

**③ 跟list/dict的关系**：set像"只有key没有value的dict"（底层也是哈希表→O(1)）。**和list的区别**：set无序+不重复+判断存在O(1)，list有序+可重复+判断存在O(n)。**和dict的区别**：dict是key→value映射，set只是"有或没有"的集合。

**④ 为什么这样设计**：set解决"成员判断"和"去重"——数据分析中两个最高频需求。只需要知道"A在不在这个集合里"→用set（O(1)），不要用list（O(n)）。数据量越大差异越夸张。**注意**：`{}`是空dict不是空set——这是Python设计的历史遗留问题（dict比set先出现）。空set必须写`set()`。

````python
# set长这样——花括号但没有冒号，自动去重
cities = {"北京", "上海", "深圳", "北京", "上海"}  # 有重复
print(cities)         # {'北京', '上海', '深圳'} → 自动去重！
print("北京" in cities)  # True——O(1)极速判断

# ⚠️ 经典陷阱：{}是空dict不是空set！
empty_dict = {}
print(type(empty_dict))   # <class 'dict'>！！
empty_set = set()         # ✅ 空set必须用set()
print(type(empty_set))    # <class 'set'>

# 集合运算
a = {1, 2, 3}
b = {2, 3, 4}
print(a & b)  # {2, 3}——交集
print(a | b)  # {1, 2, 3, 4}——并集
print(a - b)  # {1}——差集（在a但不在b）
````

````python
# 🔨 set小练：去重和快速判断
# 速卖通(电商) | 武器：set
city_list = ["北京", "上海", "北京", "深圳", "上海", "北京", "广州", "深圳"]
# 1. 用set去重→打印不重复的城市
# 2. 用in判断"杭州"有没有订单→打印True/False
# 3. 打印"共有几个不重复的城市"
# 4. 创建空set——注意不能用{}！
# 5. 在注释里回答：为什么判断"在不在"用set比list快？

# 👇 在下面写你的代码：
print(set(city_list))
print("杭州" in set(city_list))
print(len(set(city_list)))
empty_set = set()
print(empty_set)
# 因为set是无序的，所以不能用索引来访问元素。
````

## 概念6：容器选择决策树 —— 今天最重要的思维框架

### 🤔 四种容器都会了——但给你一个新问题，你知道用哪个吗？

学完5种容器，你现在手上有了一把瑞士军刀。但真正的考验是：
- "存每个城市订单数"——list？dict？set？
- "判断一个用户ID在不在这批用户里"——list还是set？
- "存商品名称/价格/库存"——tuple还是dict？

**会用一个工具 ≠ 知道什么时候该用什么工具。** 后者才是从L2（能用）到L3（能讲）的关键跨越。

### 🔥 容器选择决策树

面对任何数据存储需求，按这个顺序自问：

```
Q1: 只是一个值（一个数字/一段文本/一个布尔）？
    → YES：用普通变量。x = 299
    → NO：继续 ↓

Q2: 需要按标签/名字来找（"给我金额"而不是"给我第3个"）？
    → YES：用dict。{"金额": 299, "商品": "耳机"}
    → NO：继续 ↓

Q3: 需要去重？或核心操作是"判断XX在不在里面"？
    → YES：用set。{"北京", "上海", "深圳"}
    → NO：继续 ↓（说明你需要有顺序的一排东西）

Q4: 数据需要被修改（增/删/改）吗？
    → YES：用list。[299, 1599, 89, 499]
    → NO（数据固定不变/需要做dict的key）：用tuple。(39.9, 116.4)
```

### 四对关键对比（L3检测点——你要能脱稿讲清楚）

| 对比 | 核心差异 | 什么时候选A不选B |
|------|---------|----------------|
| **list vs dict** | list按位置(`[0]`)，dict按标签(`["金额"]`) | 同一种数据→list；一个东西多种属性→dict |
| **list vs tuple** | list可修改，tuple不可修改 | 需要增删改→list；保证不变/做dict key→tuple |
| **list vs set** | list有序可重复O(n)，set无序不重复O(1) | 需要顺序/允许重复→list；去重/判断存在→set |
| **tuple能做dict key** | tuple不可变→可哈希→能做key，list可变→不能 | 理解这个=掌握Python类型系统的核心约束 |

### 常见组合模式

| 模式 | 示例 | 场景 |
|------|------|------|
| **list套dict** | `[{"姓名":"张三","评分":4.8}, {...}]` | 多条记录组成的表（数据分析最高频） |
| **dict套list** | `{"北京":[299,89,1200], "上海":[1599,2599]}` | 按某个维度分组后的数据 |
| **dict的key用tuple** | `{(39.9,116.4):"北京"}` | 复合键（多条件联合查找） |

## ⚠️ 新手最容易踩的坑

以下是教学规格列出的**全部5个陷阱**，每一个都曾让无数新手困惑数小时。

### 坑1：Python从0开始计数——"第一个"是`[0]`不是`[1]`
```python
amounts = [12, 8, 15]
# ❌ 想取第一个12，写了amounts[1]
print(amounts[1])  # → 8！不是你想要的12！
# ✅ 第一个的位置是0
print(amounts[0])  # → 12
```
**为什么错**：Python（以及几乎所有编程语言）从0开始计数。底层内存地址计算：`地址 = 起始地址 + 索引 x 元素大小`，从0开始最自然。**记住：位置编号 = 第几个 - 1**。

### 坑2：`lst2 = lst1`不拷贝——改一个另一个也变
```python
lst1 = [1, 2, 3]
# ❌ lst2和lst1贴了同一张标签！
lst2 = lst1
lst2[0] = 999
print(lst1)  # [999, 2, 3] → lst1也被改了！
# ✅ 创建独立副本
lst2 = lst1.copy()  # 或 lst1[:]
```
**为什么错**：Python变量是"标签"不是"盒子"。`=`是"再贴一张标签"而非"复制一份内容"。这是**数据分析中最高频bug**。

### 坑3：`{}`是空dict不是空set
```python
# ❌ 想创建空set，写了{}
s = {}
print(type(s))  # <class 'dict'>！！
# ✅ 空set必须用set()
s = set()
print(type(s))  # <class 'set'>
```
**为什么错**：`{}`在Python设计之初被分配给了dict（dict更常用）。set后来加入，只能用`set()`构造函数。这是历史遗留问题，永远不会被修复（兼容性）。

### 坑4：dict的key必须不可变——list不能做key
```python
# ❌ list做dict的key
d = {[1, 2]: "a"}  # TypeError: unhashable type: 'list'
# ✅ 用tuple（不可变→可哈希）
d = {(1, 2): "a"}  # OK
```
**为什么错**：dict底层是哈希表。如果key可变（如list）→内容变了→哈希值也变了→dict在原来的位置找不到这个key→数据结构乱套。所以Python强制key必须是"可哈希"的（=不可变的）。**面试高频题**。

### 坑5：`(1)`不是tuple——是int！
```python
# ❌ 想创建单元素tuple
t = (1)
print(type(t))  # <class 'int'>！！不是tuple！
# ✅ 单元素tuple必须加逗号
t = (1,)
print(type(t))  # <class 'tuple'>
```
**为什么错**：括号`()`被数学表达式优先占用了语义——`(1)`就是数学里的"括号1"=1。**逗号才是tuple的本质标记**。`1, 2, 3`不加括号也是tuple！
