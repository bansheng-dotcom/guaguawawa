### if/elif/else —— 程序的分岔路口

**① 说白了就是**：if就是你每天做的判断。"**如果**下雨，带伞；**否则如果**阴天，带外套；**否则**啥也不带。"程序里的if完全一样——只不过条件用代码写。

**② 在沙盘里怎么用**：每笔订单过来，你让程序看一眼金额——大于500标记"大单"，200到500之间标记"中单"，不到200标记"小单"。3000条数据，程序0.1秒跑完。

**③ 跟你昨天学的容器的关系**：昨天你把订单存在list和dict里——今天if判断的对象就是容器里的元素。`if orders[0]["金额"] > 500:`——**list/dict负责"装"，if负责"判断"。容器+if = 数据有了脑子。**

**④ 为什么这样设计**：
- **`if` 和 `elif` 是从上到下检查的，第一个条件成立就停，后面全跳过。** 这就是为什么必须"先窄后宽"——如果把宽的条件写前面，窄的永远轮不到。
- **`and` 和 `or` 是"短路"的**：`a and b`，如果a已经是False，Python连b都不看——因为不管b是什么，结果都是False。同样，`a or b`，如果a已经是True，b被直接跳过。这个设计既省计算又安全——比如 `x is not None and x > 5`，如果x是None，后面的`x > 5`根本不会执行，不会报错。
- **`in` 操作符**：list是O(n)逐个比较，dict和set是O(1)哈希查找——这也是你昨天学的容器底层的直接体现。
- **`is` vs `==`**：`is`比较两个变量是否指向**同一个对象**（用`id()`验证），`==`比较**值**是否相等。判断None永远用`x is None`，不要用`x == None`。
- **真值测试**：Python里`0`、`""`、`[]`、`{}`、`None`都算False，其他算True。所以你可以直接写`if lst:`来判断list非空，不用写`if len(lst) > 0:`。

````python
# if/elif/else 长这样 —— 速卖通订单金额分类
amount = 680

if amount > 500:
    label = "大单"
elif amount >= 200:           # 200~500之间
    label = "中单"
else:                          # <200
    label = "小单"

print(f"订单金额{amount}元 → {label}")

# and/or 短路求值实验
city = "深圳"
if amount > 500 and city == "深圳":  # 两个条件同时满足
    print("⚠️ 深圳大单，需人工审核")

# in 成员检查
cities = ["深圳", "北京", "上海"]
print("深圳" in cities)   # True
print("广州" in cities)   # False

# is 判断None
x = None
print(x is None)          # True —— 推荐写法
print(x == None)          # 也能用，但不推荐

# 真值测试
print(bool([]))           # False
print(bool([1, 2, 3]))    # True
````

````python
# if小练：下面有三个金额，用if/elif/else判断每笔订单的等级
# >1000 → "超大单"  |  500~1000 → "大单"  |  100~500 → "中单"  |  <100 → "小单"
amounts = [1200, 350, 80]

# 写出你的代码（提示：对每个amount做判断）
for amount in amounts:
    # 在这里写if/elif/else
    if amount > 1000:
        label = "超大单"
    elif amount >= 500:
        label = "大单"
    elif amount >= 100:
        label = "中单"
    else:
        label = "小单"
    print(f"金额{amount}元 → {label}")
````

### 🤔 for循环——你能对3000条数据逐一操作吗？

刚学会的if可以做判断了，但问题来了——你有100条订单，每条都要判断：

```python
# 这样做？？
if orders[0]["金额"] > 500: ...
if orders[1]["金额"] > 500: ...
if orders[2]["金额"] > 500: ...
# ...写到第100条？
```

更糟的是这个：

```python
amounts = [680, 150, 2500, 90, 420]
# 要算总金额？你得写：
total = amounts[0] + amounts[1] + amounts[2] + amounts[3] + amounts[4]
# 3000条呢？写到明天早上，还容易写错。
```

if能判断一次，但**没法自动对一堆数据逐一操作**。你需要一个"替身"——它帮你把每条数据取出来，套上if，自动重复。

> 你刚才撞到的这个墙，`for循环` 就是专门解决它的。

### for循环 —— 自动遍历一切

**① 说白了就是**：for循环像一个"流水线质检员"。传送带上的包裹一个个过来，质检员拿起一个→检查→放下→拿起下一个→检查→放下……你不用管传送带怎么转，只需要告诉质检员"每个包裹到你手里该做什么"。

**② 在沙盘里怎么用**：`for order in orders:`——把orders列表里的每条订单逐一取出来，交给缩进里的代码块（循环体）去处理。3000条订单，你写一次if判断，for帮你套3000次。

**③ 跟你昨天学的容器的关系**：D1学的list、dict、tuple、set——全都是`for`可以遍历的对象（专业术语叫"可迭代对象"）。**容器是弹药库，for是机关枪。** 这是你第一次能把"存数据"和"处理数据"串成完整链条：

```python
for order in orders:               # ← D1的list是D2 for的"靶子"
    if order["金额"] > 500:        # ← D2的if做判断
        big_orders.append(order)   # ← D1的list存结果
```

**④ 为什么这样设计**：
- Python的for循环底层用**迭代器协议（Iterator Protocol）**：容器返回一个迭代器，每次`next()`取一个元素，取完抛出`StopIteration`。你的`for x in lst`，Python自动帮你做了这套繁琐操作。
- `enumerate()`让你同时拿索引和值（不用自己维护计数器），`zip()`让你并行遍历多个list（不用自己管理下标）。这两个不是语法但和for高度绑定，是数据分析极高频的工具。
- `range(n)`不生成list，它返回一个惰性的range对象——`range(1000000)`不占内存，只有用到的时候才生成下一个数字。
- **嵌套for的执行逻辑**：外层走一步，内层走一整轮。外层3个元素，内层5个元素，循环体执行3×5=15次。
- **三大基本模式**是数据分析的"肌肉记忆"：累加（`total += x`→总和）、计数（`count += 1`→频率）、找极值（`if x > max_val: max_val = x`→范围）。后面W2学pandas，`.sum()`、`.count()`、`.max()`的底层逻辑和你现在手写的一模一样。

````python
# for循环长这样 —— 速卖通订单批量处理
orders = [
    {"订单号": "ORD001", "金额": 680, "城市": "深圳"},
    {"订单号": "ORD002", "金额": 150, "城市": "北京"},
    {"订单号": "ORD003", "金额": 2500, "城市": "深圳"},
    {"订单号": "ORD004", "金额": 90, "城市": "上海"},
    {"订单号": "ORD005", "金额": 420, "城市": "北京"},
]

# 三大基本模式：累加、计数、找极值
total = 0           # 累加器
big_count = 0       # 计数器
max_amount = 0      # 极值追踪器

for order in orders:
    amount = order["金额"]
    total += amount                     # 累加
    if amount > 500:
        big_count += 1               # 计数
    if amount > max_amount:
        max_amount = amount             # 找最大值

print(f"总金额: {total}, 大单数: {big_count}, 最大单: {max_amount}")

# enumerate: 带索引遍历
for i, order in enumerate(orders):
    print(f"第{i+1}笔: {order['订单号']}")

# zip: 并行遍历
cities = ["深圳", "北京", "上海"]
amounts = [680, 150, 420]
for city, amt in zip(cities, amounts):
    print(f"{city}: {amt}元")
````

### 🤔 while循环——"不知道要做多少次"怎么办？

for循环很好，但它有个前提：**你得知道要遍历的东西是什么**——一个list、一个dict、一个range(100)。

但现实中常有这种情况：

> mentor小王说："我的基金每天涨8%。你从1000块开始算，**多少天**能涨到10000块？"

你不知道要算多少天——10000块是"停止条件"，但天数正是你要算的答案。for循环不知道range里该写几（因为你不知道天数！）。你需要一种**"不知道走多少步，但知道走到什么时候停"**的循环。

> 你刚才撞到的这个墙，`while循环` 就是专门解决它的。

### while循环 —— 直到条件满足才停

**① 说白了就是**：while像"健身增肌"——"**当**体重还没到80公斤，继续吃继续练。"今天称一下→没到→继续。你不知道要练多少天，但你知道**什么时候可以停**。

**② 在沙盘里怎么用**：初始金额=1000，每天乘以1.08——当金额还不到10000，继续算下一天。程序自动跑出天数。

**③ 跟for的关系——什么时候用哪个**：核心判断标准就一条——**知道循环次数用for，知道停止条件但不知道次数用while。**
- for："把这3000条订单都检查一遍"——你知道是3000次
- while："存到10000块为止"——你知道停止条件是10000，但不知道要多少天

**④ 为什么这样设计**：
- `while`的条件在**每次循环开始前**检查。如果一开始条件就是False，循环体一次都不执行。
- `break`和`continue`是两个控制开关：`break`是"立刻终止整个循环，跳出去"，`continue`是"跳过本轮剩余代码，直接开始下一轮"。在数据分析里，`continue`常用于跳过脏数据。
- `while True:` + `if 条件: break` 是一个经典模式——先进入无限循环，然后在循环体内部判断什么时候跳出去。用在你需要先做一些操作、再判断是否继续的场景。
- **死循环不是bug，是设计特性**：while循环天然有死循环的风险——设计上就是让程序员自己负责确保条件最终变成False。所以每次写while之前，先问自己："什么变量在每次循环后往退出条件靠近了一步？"忘掉那一步→死循环。

````python
# while循环长这样 —— 计算投资增长到目标需要多少天
money = 1000          # 初始金额
target = 10000        # 目标金额
days = 0              # 天数计数器
rate = 0.08           # 每天涨8%

while money < target:
    money = money * (1 + rate)   # 金额增长
    days += 1                     # 天数+1（忘了这行→死循环！）

print(f"从1000元到{money:.0f}元需要{days}天")

# break和continue的区别
print("\nbreak vs continue:")
for i in range(10):
    if i == 3:
        continue      # i=3跳过，不执行下面的print
    if i == 7:
        break         # i=7直接终止整个循环
    print(i, end=" ")  # 输出: 0 1 2 4 5 6

# while True + break 经典模式
print("\n\nwhile True模式:")
count = 0
while True:
    count += 1
    print(f"第{count}次尝试...")
    if count >= 3:    # 满足条件就跳出
        print("找到了！")
        break
````

````python
# while小练：一张100元的电话卡，每分钟通话扣3.5元
# 用while算最多能打多少分钟（余额不够扣3.5元就停）
balance = 100   # 初始余额
cost = 3.5      # 每分钟扣费
minutes = 0

# 写出你的代码（提示：当balance >= cost时继续）
while balance >= cost:
    balance -= cost
    minutes += 1
print(f"最多能打{minutes}分钟, 剩余{balance:.1f}元")
# 参考答案逻辑（写完后对照）：
# while balance >= cost:
#     balance -= cost
#     minutes += 1
# print(f"最多通话{minutes}分钟，剩余{balance:.1f}元")
````

### 异常处理 (try/except)

**① 说白了就是**：try/except像"过马路"——**尝试**走过去（`try`），如果突然有车（出错），你不是躺在马路上等撞，而是立刻收脚（跳到`except`），然后继续走（程序不崩）。

**② 在沙盘里怎么用**：`try: amount = int(order["金额"])` → 如果"金额"是"待定"这种脏数据，`except ValueError`接住 → 标记为"异常"，跳过，继续处理下一条。

**③ 跟你今天学的for/if怎么组合**：
```python
for order in orders:                     # for遍历
    try:
        amount = int(order["金额"])       # 尝试转换
        if amount > 500:                 # if判断
            big_orders.append(order)
    except (ValueError, KeyError):
        error_count += 1
        continue                         # 跳过脏数据
```

**④ 为什么这样设计**：
- Python用异常机制而不是"错误码"——正常逻辑（try块）和错误处理（except块）分开，代码清晰。
- `except ValueError`指定捕获一种异常类型。**越具体越好**——你不小心把变量名写错了（NameError），希望程序报错告诉你，而不是被笼统的`except`吞掉让你永远发现不了bug。
- `except Exception as e:`可以拿到错误信息`e`，便于调试和记录日志。
- `finally:`里的代码无论报不报错都会执行——适合做"收尾工作"（关文件、关数据库）。
- **异常处理不是万能的**：只包裹你知道怎么恢复的错误。不知道怎么恢复？让异常抛出去——吞掉所有错误会让bug永远藏在地下。

### 三元表达式

**① 说白了就是**：一行写完if/else。"天气好？去公园：宅家里。"只适合简单二选一，不要嵌套。

**② 写法**：`结果A if 条件 else 结果B`（条件为True→A，False→B）

````python
# 异常处理长这样 —— 清洗脏数据
orders = [
    {"订单号": "ORD001", "金额": "680"},
    {"订单号": "ORD002", "金额": "待定"},     # 脏数据！
    {"订单号": "ORD003", "金额": "2500"},
]

clean = []
errors = 0

for order in orders:
    try:
        amount = int(order["金额"])          # 尝试转整数
        clean.append({"订单号": order["订单号"], "金额": amount})
    except ValueError:
        errors += 1                           # 记录脏数据条数
        continue                               # 跳过，继续下一条

print(f"清洗完成: {len(clean)}条有效, 跳过{errors}条脏数据")

# 三元表达式长这样
amount = 680
# 不用三元（4行）:
# if amount > 500:
#     label = "大单"
# else:
#     label = "非大单"
# 用三元（1行）:
label = "大单" if amount > 500 else "非大单"
print(f"{amount}元 → {label}")
````

## ⚠️ 新手最容易踩的坑

---

### 坑1：`=` 和 `==` 搞混

❌ 错误写法：
```python
if amount = 500:   # =是赋值，不是比较！
    print("成立")
# SyntaxError —— 因为 amount = 500 是一个赋值语句，不是条件表达式
```

✅ 正确写法：
```python
if amount == 500:  # == 才是比较
    print("成立")
```

**为什么错**：`=`是赋值（给变量贴标签），`==`是比较（问"相等吗？"）。if后面必须是一个能算出True/False的表达式，赋值语句不算。

---

### 坑2：elif条件顺序——先宽后窄

❌ 错误写法（宽条件写前面）：
```python
score = 95

if score >= 60:
    grade = "及格"
elif score >= 80:
    grade = "良好"
elif score >= 90:
    grade = "优秀"      # 95分永远走不到这里！

print(grade)  # 输出"及格"——明明是95分！
```

✅ 正确写法（先窄后宽）：
```python
score = 95

if score >= 90:
    grade = "优秀"      # 最窄的条件放最前面
elif score >= 80:
    grade = "良好"
elif score >= 60:
    grade = "及格"

print(grade)  # 输出"优秀"——正确！
```

**为什么错**：`if/elif`从上到下检查，**第一个成立的就执行、后面全跳过**。如果宽条件在前面，把窄条件的数据也"截胡"了。

---

### 坑3：`if a == 3 or 4` —— 永远为True

❌ 错误写法：
```python
x = 10

if x == 3 or 4:   # 你以为在判断"x等于3或x等于4"
    print("命中")
# 永远输出"命中"！不管x是多少！
```

✅ 正确写法：
```python
x = 10

if x == 3 or x == 4:   # 必须写成两个完整的比较
    print("命中")

# 或者用 in：
if x in (3, 4):
    print("命中")
```

**为什么错**：Python会把`x == 3 or 4`解读为`(x == 3) or (4)`。而`bool(4)`是True（非零即真），所以`or 4`永远让整个条件为True。

---

### 坑4：遍历list时删除元素

❌ 错误写法：
```python
nums = [1, 2, 3, 4, 5]
for n in nums:
    if n % 2 == 0:
        nums.remove(n)  # 遍历时删除！
print(nums)  # [1, 3, 5] —— 居然对了？试试[2, 4, 6, 8]再看看
```

更明显的例子：
```python
nums = [1, 2, 3, 4, 5]
for n in nums:
    nums.remove(n)
print(nums)  # [2, 4] —— 为什么不是空的？！
# 因为删除元素后索引全乱了，for循环会跳过元素
```

✅ 正确写法：
```python
nums = [1, 2, 3, 4, 5]
nums = [n for n in nums if n % 2 != 0]  # 新建一个list（列表推导式）
# 或者
nums = [n for n in nums if condition]   # 保留满足条件的
```

**为什么错**：for循环内部靠索引遍历。你删掉一个元素→后面的元素往前移→下一个索引指向的其实是"下下个"元素→跳过了一个。

---

### 坑5：while忘记更新循环变量 → 死循环

❌ 错误写法：
```python
money = 1000
days = 0
while money < 10000:
    money = money * 1.08
    # 忘了写 days += 1 ！！！
    # 程序永远跑下去（按Ctrl+C强制停止）
```

✅ 正确写法：
```python
money = 1000
days = 0
while money < 10000:
    money = money * 1.08
    days += 1            # 每次循环天数+1，确保循环在推进

print(f"{days}天")       # 30天
```

**为什么错**：while的条件里有`money`。如果循环体内不修改`money`（或任何条件变量），条件永远不会变成False，循环就停不下来。写while前先问自己：**"什么在每次循环后往退出条件靠近一步？"**

### 迁移练习：成绩分析系统

```python
students = [
    {"姓名": "张三", "语文": 85, "数学": 92, "英语": 78, "编程": 90},
    {"姓名": "李四", "语文": 72, "数学": 45, "英语": 88, "编程": 65},
    {"姓名": "王五", "语文": 90, "数学": 85, "英语": 92, "编程": 95},
    {"姓名": "赵六", "语文": 55, "数学": 60, "英语": 58, "编程": 70},
    {"姓名": "钱七", "语文": 78, "数学": 95, "英语": 82, "编程": 88},
]

# ① 查张三的数学成绩
for s in students:
    if s["姓名"] == "张三":
        print(f"张三的数学成绩: {s['数学']}")

# ② 每人的总分和平均分
subjects = ["语文", "数学", "英语", "编程"]
for s in students:
    total = sum(s[subj] for subj in subjects)
    avg = total / len(subjects)
    print(f"{s['姓名']} 总分:{total} 平均:{avg:.1f}")

# ③ 总分最高的人
top_name = ""
top_total = 0
for s in students:
    total = sum(s[subj] for subj in subjects)
    if total > top_total:
        top_total = total
        top_name = s["姓名"]
print(f"总分最高: {top_name} ({top_total}分)")

# ④ 班级数学平均分
math_sum = sum(s["数学"] for s in students)
print(f"班级数学平均分: {math_sum / len(students):.1f}")

# ⑤ 哪个科目挂科（<60分）最多
fail_count = {}
for subj in subjects:
    fail_count[subj] = 0
    for s in students:
        if s[subj] < 60:
            fail_count[subj] += 1

worst_subj = max(fail_count, key=fail_count.get)
print(f"挂科人数最多: {worst_subj} ({fail_count[worst_subj]}人)")
```

| 如果你的输出/报错是... | 你可能的误解是... | 补救 |
