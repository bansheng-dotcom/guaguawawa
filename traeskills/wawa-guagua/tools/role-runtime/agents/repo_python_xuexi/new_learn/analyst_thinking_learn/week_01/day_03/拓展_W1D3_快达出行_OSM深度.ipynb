# 第三部分：概念网络展开

## 🔴 核心深化：OSM 为什么是这三个字母——在它被发明之前

> 主课告诉你 OSM 的三个字母代表什么。但你有没有想过——为什么是这三个字母？多一个会怎样？少一个会怎样？在 OSM 流行之前，公司是怎么把目标变成行动的？

### 🔴 认知冲突

回到快达。CEO 说：「Q3 毛利率从 9% 提升到 15%。」

如果你是三个不同时代的分析师，你会怎么回应？

**时代 A（2005 年，只有 KPI）**：你把「毛利率≥15%」设为全公司 KPI，然后发下去。三个月后——毛利率到了 15%，但你发现运营是通过「砍掉所有不赚钱的路线 + 把短途单价格翻倍」达成的。完单量暴跌 40%，乘客大量流失。**KPI 达成了，但公司更糟了。**

**时代 B（2015 年，用 OKR）**：你写了 Objective「成为最健康的出行平台」，Key Results「毛利率≥15% + 乘客 NPS≥40」。比时代 A 好——至少你知道不能只盯毛利率。但 KR 写出来了，**团队还是不知道该做什么**——「毛利率到 15%」是一个结果，不是一个动作。

**时代 C（2020 年后，用 OSM）**：你不仅写了 O（毛利率 9%→15%），还写了 S（怎么做——涨价？控成本？优化高毛利路线？），每条 S 配了 M（怎么知道这条策略有没有用）。

**问题**：时代 A 和 B 的方案各自缺了什么？为什么 OSM 比前两者更完整？

> OSM 不是在「发明新东西」——它是在补前两个时代的窟窿。

### [必修] OSM 的历史脉络——每一个字母在补什么窟窿

**① 一句话白话**：OSM 不是凭空发明的。它的三个字母，每一个都是在解决前一代工具的已知问题。

**② 三代演变**：

| 时代 | 工具 | 有什么 | 缺什么 | 导致什么问题 | 快达案例 |
|------|------|--------|--------|-----------|--------|
| 2000s | **KPI 分解** | 只有 M（度量）。「公司要毛利率 15%→每个部门领一个数字」 | 没有 O（为什么），没有 S（怎么做到） | KPI 达成但业务恶化——因为各部门不择手段完成自己的数字 | 运营砍路线提价→毛利率达标但完单量暴跌 |
| 2010s | **OKR** | O（目标）+ KR（关键结果）。比 KPI 多了「为什么要做」 | 没有 S（策略）。O 和 KR 之间有一个真空——团队知道要达成什么，但不知道该做什么 | KR 写得好但执行不了——因为从「目标」到「结果」之间没有策略层 | 团队知道要毛利率 15%，但不知道是涨价还是控成本还是改路线结构 |
| 2020s | **OSM** | O + S + M。**在目标(O)和度量(M)之间，显式插入策略(S)层** | — | 三个字母合在一起，补上了从「为什么」到「怎么做」到「怎么判断」的完整链条 | CEO 说毛利率 15%→你写出 3 条策略→每条策略配度量→执行→检验 |

**OSM 的真正贡献**：不是发明了「目标-策略-度量」这三个概念（它们各自早就存在），而是**强制你在 O 和 M 之间显式地写出 S**。

没有 S 的 OSM = OKR = KPI——本质上只是「把大目标拆成小目标」。有 S 的 OSM，才是「把目标变成行动」。

**③ 与核心概念的关系**：这不是新概念，而是理解 OSM **为什么长这样而不是其他样子**。懂了历史，你就不会写出「O=提升体验，S=优化产品，M=看DAU」这种空洞的 OSM——因为你会意识到那本质上只是重命名版的 KPI。

### 练习：判断一个 OSM 是「真 OSM」还是「KPI 换皮」

以下 3 个 OSM 分别来自不同的团队。请判断它们是真 OSM 还是 KPI 换皮。

| # | O | S | M | 你的判断 |
|----|----|----|-----|--------|
| 1 | 提升用户活跃度 | 优化产品体验 | DAU 到 50 万 | ? |
| 2 | Q3 新用户次留从 15%→20% | ①注册后引导关注 3 个话题 ②首次发布后 1 分钟内给反馈 ③次日 push 召回 | ①引导完成率 ②首次发布率 ③push 点击率 | ? |
| 3 | 成为用户首选的学习平台 | ①降低内容发布门槛 ②冷启动内容供给 ③发布即反馈 | ①新用户发帖率 ②推荐话题点击率 ③首次发帖后次留 | ? |

---

### 参考分析（做完再看）

| # | 判断 | 为什么 |
|----|------|------|
| 1 | 🚨 **KPI 换皮** | O 太模糊（什么叫活跃度？）；S 太空洞（「优化产品体验」谁都能写，等于没写）；M 只有一个结果数字，没有过程度量。三个框都填了，但填的内容跟没填一样。**诊断方法**：遮住 OSM 的标签，只看内容——你能分得清哪个是 O、哪个是 S、哪个是 M 吗？ |
| 2 | ✅ **真 OSM（但 O 有坑）** | O 是一个 SMART 目标。但注意——「新用户次留 15%→20%」其实是一个 M（度量），不是 O（目标）。真正的 O 应该是「让新用户第一次来就觉得有用」。不过这个「伪 O」在实践中可以接受——因为它是 SMART 的，而且三条策略都有清晰的因果假设。 |
| 3 | ✅ **真 OSM** | O 是定性的方向（「成为首选」），S 是三条有因果假设的策略，M 同时覆盖了过程度量和结果度量。每条 S 都能独立检验。 |

**核心检验**：判断一个 OSM 是不是「真」的，问三句话——
1. O 说清楚了「为什么」吗？（不是「DAU 到 50 万」，而是「为什么要到 50 万」）
2. S 能独立检验吗？（每条 S 都有自己的 M，可以判断「这条策略有没有用」）
3. O→S→M 的因果链能讲通吗？（做了 S，真的会导致 M 变好吗？M 变好真的等于 O 在达成吗？）

## 概念群 A：[并行] ICE 优先级——有 5 条策略，只能做 2 条，先做哪个？

> **与核心概念的关系**：OSM 的 S 层写出来后，通常会有 3-8 条策略。但资源（人、钱、时间）永远不够全做。ICE 帮你在多条策略中选出「最可能有效」的 1-2 条先启动。这是 [并行] 关系——OSM 定策略，ICE 排优先级。

### 🔴 认知冲突

快达 CEO 的目标是「Q3 毛利率从 9% 提升到 15%，完单量跌幅控制在 5% 以内」。你作为分析师，和团队一起头脑风暴出了 5 条策略：

| # | 策略 | 预期效果 |
|----|------|--------|
| A | 高峰时段涨价 15%（抑制需求，提高单均收入） | 毛利率 +4pp，完单量 -8% |
| B | 优化派单算法，减少司机空驶（降低司机流失→降低拉新成本） | 毛利率 +1pp，完单量 +3% |
| C | 对短途单取消补贴（短途单本来就不赚钱） | 毛利率 +3pp，完单量 -10% |
| D | 推出「高峰保底」——高峰期接单满 10 单的司机保底收入 200 元 | 毛利率 -1pp（短期），完单量 +5% |
| E | 把低毛利的企业客户合同重新谈判（提高大客户单价） | 毛利率 +2pp，完单量无影响 |

**问题**：CEO 说「Q3 只能重点推 2 条策略，其他先放一放。你帮我排个序。」

你会怎么选？是选「毛利率提升最多」的 A？还是「完单量影响最小」的 E？还是「两个维度都兼顾」的 B？

> 凭感觉选 = 扔骰子。ICE 给你一个结构化的选择框架。

### [必修] ICE 优先级排序法

**① 一句话白话**：ICE 就是给每条策略在三个维度上打分（1-5 分），然后乘起来。总分最高的先做。

**② 三个维度**：

| 字母 | 代表什么 | 白话解释 | 打分标准 |
|------|---------|--------|--------|
| **I** | Impact（预期影响） | 「这条策略如果做成了，对目标的推动有多大？」 | 5 分 = 直接达标；3 分 = 显著改善；1 分 = 锦上添花 |
| **C** | Confidence（信心） | 「你有多确定这条策略会有效？」 | 5 分 = 有 AB 实验数据支撑；3 分 = 有竞品案例或历史类似经验；1 分 = 纯猜测 |
| **E** | Ease（实现难度） | 「这条策略做起来有多麻烦？」 | 5 分 = 1 周内可上线；3 分 = 需要跨部门协作，约 1 个月；1 分 = 需要大改底层系统，3 个月+ |

**注意**：E 的打分方向是反直觉的——**越容易做，分数越高**。很多新手会搞反——把「最难做」的打 5 分。

**③ 回到快达的 5 条策略，打分**：

| # | 策略 | I（影响） | C（信心） | E（难度） | ICE 总分 | 排名 |
|----|------|:--:|:--:|:--:|:--:|:--:|
| A | 高峰涨价 15% | 4（毛利率+4pp） | 3（经济学原理支持，但快达没试过） | 4（改价格参数即可，1 周上线） | 4×3×4=**48** | 🥇 1 |
| B | 优化派单算法 | 2（毛利率只+1pp） | 2（算法效果不确定，需要调参迭代） | 2（算法团队排期 1 个月+） | 2×2×2=**8** | 5 |
| C | 取消短途补贴 | 3（毛利率+3pp） | 4（不花钱就能提毛利，确定性强） | 5（改补贴规则，1 天上线） | 3×4×5=**60** | 🥇 实际上最高！ |
| D | 高峰保底 | 2（短期还亏钱） | 3（司机激励常见手段） | 3（需要和财务对齐预算） | 2×3×3=**18** | 4 |
| E | 大客户合同重谈 | 2（毛利率+2pp） | 3（合同重谈有不确定性） | 1（需要销售团队逐个谈，3 个月+） | 2×3×1=**6** | 6 |

**出乎意料的结果**：绝大多数人会直觉选 A（涨价），因为它的毛利率提升最大。但 ICE 告诉你——**C（取消短途补贴）才是最优选择**。因为它的信心高（确定有效）、难度极低（1 天就能上线）、影响也不小（+3pp）。

**ICE 的核心价值**：它强迫你把「直觉上最重要」和「实际上最划算」分开。A 看起来很美，但如果涨价后完单量跌超预期（信心只有 3 分），效果会大打折扣。C 没有那么性感，但它**确定有效且零成本**——先做 C，A 可以作为备选。

**④ 与核心概念的关系**：ICE 是 OSM 的 [并行] 工具——OSM 帮你「写出策略」，ICE 帮你「排出优先级」。两者在真实工作中同时使用。

| 1 | | | | | | |

### 常见偏差

| ❓ 你可能的误解 | 🔧 纠正 |
|--------------|--------|
| E 打反了——给最难的策略打了 5 分 | E = Ease（容易度），不是 Effort（努力度）。越容易做，分数越高 |
| I 只看「对目标的影响」，忘了看「对护栏的影响」 | 比如 A 策略毛利率 +4pp 但完单量 -8%——需要在 I 中反映这个副作用。你可以给 A 的 I 打折（从 4 分降到 3 分） |
| C 全打 3 分（「都不太确定」） | 如果所有策略的 C 都低，说明你的策略是拍脑袋想出来的。先停下来——去找数据或案例提高 C，而不是在低信心下硬选 |

## 概念群 B：[防护] 因果链断裂——你的 O→S→M 真的能走通吗？

> **与核心概念的关系**：主课说「OSM 三个字母之间有因果链」。但怎么检查这条链是真的通的——还是你**以为它通了、实际上中间断了一截**？这是 [防护] 概念——防止你写出一个「看上去完整、其实不成立」的 OSM。

### 🔴 认知冲突（错误暴露型）

以下是快达产品 VP 李哥为「提升乘客体验」写的一个 OSM。看起来三个框都填了。但**你能找到它的问题吗**？

```
O（目标）: Q3 乘客平均等待时长从 6.5 分钟降到 5 分钟
S（策略）:
  1. 优化 App 首页 UI——把「叫车」按钮做得更大更显眼
  2. 在乘客等待超过 3 分钟时，自动发一张 2 元优惠券作为补偿
  3. 增加司机供给——加大司机拉新投放
M（度量）:
  策略1: 首页「叫车」按钮点击率
  策略2: 优惠券领取率 + 领取后 7 日内复购率
  策略3: 新增司机数 + 司机首周完单率
```

**你的任务**：逐条检查 O→S→M 的因果链。哪条策略的因果链是通的？哪条是断的？

（先自己想，再往下翻看诊断。）

### [必修] 因果链断裂的三种模式

**① 一句话白话**：OSM 写完后，不是「填完了」就完了。你要像侦探一样去查——做了 S，**真的**会导致 M 变好吗？M 变好，**真的**等于 O 在达成吗？

**② 三种断裂模式**：

| 断裂模式 | 什么意思 | 李哥 OSM 中的例子 | 怎么修 |
|---------|--------|----------------|--------|
| **跳跃式断裂** | S → M 之间跳过了关键中间步骤。你以为 A 会导致 B，其实 A 到 B 中间还有 C、D、E 三个变量，而这些变量不受你控制。 | **策略 1**：「优化首页 UI → 等待时长下降」。实际上：首页 UI → 用户更容易点叫车按钮 → 叫车请求量增加 → 但等待时长取决于司机供给，不是按钮大小。按钮变大让用户更容易**发出请求**，但不等于司机更快**响应请求**。 | 把 S 修正为「在用户点击叫车后，优化派单算法缩短司机匹配时间」——这样因果链才连得上 |
| **错位式断裂** | S → M 之间方向反了。你期望 S 导致 M 上升，但实际上 S 可能导致 M **下降**。 | **策略 2**：「发优惠券补偿等待 → 乘客体验变好」。实际上：发券可能让乘客形成「等待=有钱拿」的预期，反而降低了他们对等待的容忍度。更重要的是——O 是「降低等待时长」，发券不解决等待时长，只是在等待发生后做补救。 | 策略 2 的 M 是对的，但 S 和 O 之间没有因果关系——发券不能降低等待时长。这条策略应该砍掉，或移到另一个 O（如「提升乘客满意度」）下面 |
| **空转式断裂** | S 被正确执行了（过程度量达标），但 O 纹丝不动（结果度量不变）。说明 S 本身无效——你做了正确的事，但正确的事对目标没有帮助。 | **策略 3**：「加大司机拉新 → 等待时长下降」。实际上：新司机不熟悉路线、接单意愿低、活跃率低——拉新**量**涨了但**有效供给**没涨。过程度量（新增司机数）达标，结果度量（等待时长）纹丝不动。 | 把 S 修正为「激活沉睡司机」而非「拉新司机」——1 个活跃老司机比 3 个新司机对等待时长的改善更大 |

**③ 诊断方法——三句自问**（写完后逐条 S 问自己）：

1. **「做了 S，M 一定会动吗？」** ——如果答案不是「一定会」而是「可能会」「希望会」→ 跳跃式断裂风险
2. **「有没有一种可能，做了 S 反而让 M 往反方向走？」** ——如果答案是「有」→ 错位式断裂风险
3. **「如果过程度量达标但结果度量不动——我能下结论说策略无效吗？还是说执行本身有问题？」** ——如果分不清 → 你缺了一个中间度量，卡在过程和结果之间

**④ 与核心概念的关系**：主课说 OSM 有因果链，但没告诉你因果链具体怎么断。本概念把「断裂」分成三种可识别的模式——以后检查 OSM，就是在排查这三种断裂。

### 练习：诊断两个 OSM 的因果链

以下两个 OSM 都来自其他团队。请检查每条 S 的因果链是否成立。

**场景 A**：某电商的 GMV 提升 OSM——
- O：Q3 月 GMV 从 3000 万提升到 3500 万
- S1：首页推荐算法从「热度排序」改为「个性化排序」
- M1：推荐点击率 + 推荐后加购率

你的诊断：S1→M1 的因果链通的吗？如果推荐点击率涨了但 GMV 没涨——可能是什么断裂？

**场景 B**：某内容社区的创作者激励 OSM——
- O：提升社区 UGC 内容质量
- S1：对发布内容被点赞 ≥ 10 次的作者发放现金奖励
- M1：每日发文量 + 点赞 ≥ 10 的内容占比

你的诊断：S1→M1 的因果链通的吗？有没有错位式断裂的风险？

---

### 参考分析（做完再看）

| 场景 | 诊断 |
|------|------|
| A | **空转式断裂风险**。推荐点击率涨 → 只能说明推荐的东西用户更爱点。但「爱点」≠「爱买」。你推荐了一个用户爱看的商品（点击率↑），但如果这个商品价格太高、评论太少、或者用户只是好奇点一下——GMV 不会涨。需要加一个中间度量：「推荐后实际下单转化率」。如果点击率涨但下单率不涨 → 推荐变得更「吸引眼球」了但没变得更「促进成交」。 |
| B | **错位式断裂风险**。「发文量」可能暴涨（因为有钱拿），但「内容质量」可能暴跌——因为创作者会发大量凑数内容去博点赞。现金激励「被点赞≥10的内容」实际上在激励创作者「发更容易被点赞的内容」——比如段子、热点蹭流量——而不是「更有价值的内容」。现金激励在提升量的同时很可能**降低**了质。这就是错位——S 的方向和 O 的方向相反。 |

## 概念群 C：[落地] OSM 一页纸——怎么把你的 OSM 讲给 CEO 听

> **与核心概念的关系**：写好了 OSM，下一步是让别人理解它、认可它、执行它。这需要一种和「自己分析」完全不同的技能——把复杂的因果链压缩到一页纸，让非数据同事 30 秒内看懂。

### 🔴 认知冲突（对比冲击型）

**方式 A（常见的「分析师汇报」）**：
> 「我们的 O 是 Q3 毛利率从 9% 提升到 15%。基于对 Q2 亏损单的分析，我们发现短途单的补贴 ROI 仅为 0.4，是毛利率的主要拖累。因此 S1 是取消短途单一刀切补贴，改为按距离阶梯补贴。M1 是短途单毛利率变化。同时考虑到完单量可能受冲击，S2 是……」

CEO 听到第三句就开始看手机了。

**方式 B（一页纸 OSM）**：

```
┌─────────────────────────────────────────────┐
│  快达 Q3 增长策略 — 一页纸                      │
├─────────────────────────────────────────────┤
│  O: 毛利率 9%→15%，完单量控制在 -5% 以内       │
│                                              │
│  策略              度量            负责人 优先级│
│  S1 取消短途补贴   短途毛利率      王姐   P0   │
│  S2 高峰涨价15%   高峰完单+毛利    李哥   P1   │
│  S3 大客户重谈     大客户毛利率    陈姐   P2   │
│                                              │
│  护栏: 完单量≥9.5万 | 等待时长≤5分钟           │
│  先行: 司机周活跃率（领先完单 1-2周）           │
└─────────────────────────────────────────────┘
```

CEO 30 秒看完，问：「S1 和 S2 同时做，完单量会不会跌超 5%？」——这个问题，说明他看懂了。

> 差距不在于分析深度，在于信息密度。一页纸 OSM 和长篇分析的关系 = 电影海报和剧本的关系。

### [必修] OSM 一页纸的五个要素

**① 一句话白话**：把一整套分析压缩到一页纸——CEO 30 秒扫完知道方向对不对，有问题再往下问。

**② 五个要素**：

| 要素 | 占多少空间 | 写什么 | 常见错误 |
|------|----------|--------|--------|
| **O（目标）** | 1 行 | SMART 目标 + 当前基线。「毛利率 9%→15%，完单量≥9.5万」 | 写得太长（「在考虑市场环境和竞争格局的前提下……」→ 砍掉） |
| **S+M 对照表** | 主体的 60% | 每条策略右边紧挨着它的度量。让人一眼看到「策略-度量」的对应关系。标注优先级（P0/P1/P2）和负责人 | S 和 M 分开写（读者需要来回对照——认知负荷翻倍） |
| **护栏** | 1 行 | 哪几个数字亮了红灯就停。「完单量<9.5万」或「等待时长>5分钟」→ 暂停 S1/S2 | 护栏太模糊（「用户体验不恶化」→ 不可测量，等于没设） |
| **先行指标** | 1 行 | 哪几个数字能提前 1-2 周预警。「司机周活跃率」跌了 → 完单量大概率下周跌 | 放太多先行指标（CEO 只需要 1-2 个最关键的） |
| **负责人** | 每行 S 后面 | 谁对这条策略的执行和度量负责。只有一个名字——不写「运营+产品」 | 写「相关部门」或者空着 → 没人认领，等于没分派 |

**③ 与核心概念的关系**：这是 OSM 的 [落地] 版——把你自己笔记本里的分析，变成管理层能决策的依据。

### 练习：把下面的 OSM 压缩成一页纸

> S3：对连续 7 天未上线的司机发 push 召回 + 首单奖励 10 元
> M1：首月留存率（保底组 vs 对照组）+ 保底成本/留存司机
> M2：等级体系上线后，各等级司机的人均完单量 + 高等级司机流失率

## 概念群 D：[替代] OSM 的适用边界——什么场景不该用 OSM

> **与核心概念的关系**：学了锤子，看什么都像钉子。OSM 是好工具，但不是所有场景都适合。知道「什么时候不该用」和知道「怎么用」同样重要。

### [必修] 三个不该用 OSM 的场景

| 场景 | 为什么不该用 | 该用什么 | 快达里的例子 |
|------|-----------|--------|----------|
| **探索期（产品还没 PMF）** | OSM 假设你**知道目标是什么**。探索期的核心问题恰恰是「目标应该是什么」——你还没找到用户需求，定 O 是瞎定。 | 用「假设-验证」循环：每周提一个假设，用最快的方式验证，不写 OSM | 快达如果在「要不要做顺风车」的探索阶段——先跑 2 周小范围测试看用户需不需要，而不是写一个「顺风车 OSM」 |
| **因果链太长（超过 6 个月才见效）** | OSM 的 M 层需要能及时反馈——你做了 S，1-2 周内应该能从 M 看到信号。如果 M 要 6 个月才能观测，OSM 的反馈循环就断了。 | 用「里程碑管理」：把长期目标拆成阶段性里程碑，每个里程碑有自己的短期度量 | CEO 说「3 年内成为行业第一」——这是一个愿景，不是 OSM 的 O。把它拆成年度 O，每年度有 OSM |
| **变量太多、互相影响（多团队博弈场景）** | OSM 是线性思维——O→S→M。但在双边市场（如打车、外卖）中，运营、算法、产品三个团队的策略会互相干扰——运营补贴司机→算法匹配效率下降→产品体验恶化。单独的 OSM 看不到跨团队交互。 | 用「策略组合测试」：多个策略同时上线，用 AB 实验测交互效应。这是 W7-W8 的领域 | 快达的「涨价」和「补贴司机」同时做——涨价让完单量跌，补贴让完单量涨，两者互相抵消。单独看每个 OSM 都「达标」，合在一起效果为零 |

**核心原则**：OSM 适合「目标清晰、反馈周期短、变量可控」的场景。如果这三个条件任何一个不满足——先确认条件，再写 OSM。

## 综合练习 1：从零写一份快达 Q3 的完整 OSM + ICE 排序

CEO 说：「Q3 的核心目标是把毛利率从 9% 拉到 15%，但完单量不能跌超 5%。你帮我写一个执行方案。」

请完成以下 4 步：

**第 1 步**：写出 3-5 条可能的策略（S）。每条策略一句话，要有因果假设。

**第 2 步**：用 ICE 给这些策略打分排序。挑出 P0（最高分）和 P1（第二高分）。

**第 3 步**：为 P0 和 P1 各写一个过程度量和一个结果度量。

**第 4 步**：检查每条策略的因果链——用三种断裂模式逐条排查。

---

### 参考方案（做完再看）

| # | 策略 | I | C | E | ICE | 排序 |
|----|------|:--:|:--:|:--:|:--:|:--:|
| S1 | 取消短途单一刀切补贴，改为按距离阶梯补贴（短途补 0 元，中途补 1 元，长途补 3 元） | 4 | 4 | 5 | 80 | 🥇 P0 |
| S2 | 高峰时段（7-9点、17-19点）涨价 10%，非高峰降价 5%（平衡供需） | 3 | 3 | 4 | 36 | 🥈 P1 |
| S3 | 把 20% 的司机拉新预算转移到「激活沉睡司机」 | 3 | 3 | 3 | 27 | P2 |

**因果链检查**：
- S1：取消短途补贴 → 短途单毛利直接提升（无中间变量）→ 因果链通 ✅。风险：短途单完单量可能跌。护栏：短途单完单量 < 当前 70% 则回滚。
- S2：高峰涨价 → 部分价格敏感用户放弃打车 → 完单量跌但单均收入涨 → 因果链通但需监控平衡点 ✅
- S3：激活沉睡司机 → 司机供给增加 → 接单率提升 → 因果链中间有「接单率」这一步，需设中间度量（司机激活后首周接单率）

## 综合练习 2：沙盘变体——快达要探索「拼车」新业务

CEO 说：「我们想试试拼车——顺路的乘客共享一辆车，价格是快车的 6 折。你写个 OSM。」

**问题**：这个场景适合用 OSM 吗？如果不适合，该用什么？

---

### 参考分析（做完再看）

**不适合用 OSM**。理由是——
- **探索期**：快达不知道用户会不会接受拼车（跟陌生人同坐一辆车）。O 应该是「验证拼车是否有 PMF」，而不是「拼车 GMV 到 XX」。在 PMF 验证之前定 SMART 目标是自欺欺人。
- **该用什么**：小范围 MVP 测试。选 1 个城市、100 个司机、上线 2 周。核心度量不是 GMV，而是——(1) 有多少人愿意选拼车（接受率），(2) 选了拼车的人次留率，以及 (3) 司机的拼车接单意愿。如果这三个数字达标，再写 OSM。

**原则**：OSM 是「执行工具」，不是「探索工具」。先确认方向，再写 OSM。

## 综合练习 3：诊断一个「看起来很完整」的 OSM

以下是快达运营王姐为「提升乘客满意度」写的 OSM。请逐条检查——有哪些因果链断裂？

```
O: Q3 乘客满意度（App 内评分）从 4.2 提升到 4.5
S1: 增加「一键投诉」按钮——乘客遇到问题可以一键投诉司机
M1: 投诉按钮点击率 + 投诉后 24h 内解决率
S2: 乘客等车超过 5 分钟，自动推送一篇有趣的文章（帮乘客打发时间）
M2: 文章打开率 + 文章阅读时长
S3: 给评分 5 星的乘客发放 5 元优惠券
M3: 5 星评价占比 + 优惠券核销率
```

---

### 参考诊断

| S | 因果链诊断 |
|----|----------|
| S1 | **部分通，但方向有疑问**。投诉按钮→投诉量增加→如果投诉被解决→满意度可能提升。但「投诉量增加」本身可能**降低**满意度——因为用户在投诉之前已经不满意了。投诉按钮让原本沉默的不满意用户「表达了不满意」→M1（投诉率）涨了可能说明体验在**变差**而非变好。这条 M 的方向不确定。 |
| S2 | **跳跃式断裂**。推送文章→用户读了文章→但等待时长没有变短。O 是「提升满意度」，但等车时读文章不能替代「少等一分钟」。这是一个典型的「转移注意力」策略——它不能解决根本问题（等待太长）。 |
| S3 | **错位式断裂（严重）**。发券奖励 5 星 → 5 星占比涨了。但这是「买来的 5 星」——用户因为券给了 5 星，而不是因为体验好给了 5 星。更糟的是，用户可能形成「给 5 星 = 有钱拿」的预期——以后不给券就不评 5 星。O 是「提升满意度」，但发券提升的是「评分」而非「满意度」——O 和 M 之间概念偷换了。 |
