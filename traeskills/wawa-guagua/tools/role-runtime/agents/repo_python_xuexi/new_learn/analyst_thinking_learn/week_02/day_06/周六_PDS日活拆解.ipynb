## 一、核心概念与第一性原理

### 1.1 DAU是一个"混合物"——同样100万DAU，内部结构天差地别

DAU（Daily Active Users）= 今天打开App的所有人。但这群人的"用户状态"完全不同。PDS把DAU拆成三个互斥且穷尽的部分：

```
DAU = New（今日新增）+ Retained（昨日活跃且今天也活跃）+ Resurrected（之前流失，今天又回来）
```

三者的需求和行为模式完全不同，因此需要完全不同的产品策略：
- **New（首次打开App）**：对产品一无所知。质量完全取决于"从哪个渠道来的"——信息流标题党来的可能打开3秒就关，SEO搜素来的可能认真体验。New需要：引导和教育，尽快触发"啊哈时刻"。
- **Retained（连续活跃）**：产品最重要的用户资产。昨天用了，今天又用——说明产品对ta有持续价值。Retained不需要教育，需要持续的新内容和个性化体验。Retained的流失是最昂贵的损失。
- **Resurrected（回流用户）**：曾经流失但今天又回来了。需要搞清楚两个问题：当初为什么走？（产品问题？竞品抢走？自然遗忘？）今天为什么回来？（被push催回来的？被活动吸引？突然想起来？）。如果是被push催回来的——大概率明天又走了。

**同样DAU=100万的两个产品**：产品A的Retained占70%（底盘稳，用户有粘性），产品B的New占40%（虚胖，靠砸钱买量）。虽然DAU数字相同，但健康度天差地别——B的投放预算一砍，DAU立即断崖。

### 1.2 从第一性原理推导：DAU的本质是一个累积的留存积分

DAU_t = New_t + New_{t-1}×次留率 + New_{t-2}×2日留存率 + ... + New_{t-N}×N日留存率

**DAU = 历史上每一天的新增 × 那一批用户留存到今天的比例。** 这是一个累积的"留存积分"——积分值越大，DAU越大。

这个公式揭示了提升DAU的两条根本路径，它们的成本和可持续性完全不同：
1. **增加New（分子变大）**：短期见效（投广告就有量），但要持续花钱。如果留存低→新增很快流失→DAU对投放产生"依赖性"——投就有、不投就跌。
2. **提升留存率（积分权重变大）**：长期效果，要改进产品。但如果成功→每批新增的"积分贡献"都变大→DAU自然增长→不需要持续加投放预算。

**最优策略**：留存底盘稳的基础上拉新。如果留存很差就猛拉新——就像往漏水的桶里倒水——你倒进去的水（新增用户）当天就从桶底的洞（低留存）漏走了。更糟的是，你花的拉新成本也一起漏走了。

### 1.3 PDS的三种DAU增长模式——你的DAU属于哪一种

| 模式 | DAU怎么涨的 | 占比特征 | 可持续性 | 崩盘风险 | 典型案例 |
|------|-----------|--------|--------|--------|--------|
| **留存驱动型** | Retained占比>60%且绝对值在涨 | Retained>60%, New<20% | 高——产品本身有粘性，不需要依赖投放 | 低——除非产品体验严重恶化或竞品颠覆 | 微信、抖音成熟期 |
| **拉新驱动型** | New占比持续>30%且在涨 | New>30%, Retained<50% | 低——花钱买的，投放预算一砍就跌 | 极高——边际获客成本递增，迟早扛不住 | 很多"烧钱换增长"的产品 |
| **召回驱动型** | Resurrected占比>20% | Resurrected>20% | 低——靠push/短信/活动催活 | 高——用户对push脱敏，每次召回效果递减 | 工具型产品、电商大促后 |

**好的增长**：留存驱动为主、拉新为辅——Retained底盘稳，New增量往上垒，Resurrected只是锦上添花。**坏的增长**：纯拉新或纯召回——表面DAU在涨，底盘在萎缩。这就是"虚胖DAU"——看起来胖了，其实是肿了。

### 1.4 PDS的两大核心用途
1. **诊断**：DAU跌了→30秒看New/Retained/Resurrected哪个跌了→快速把40个可能原因缩小到5个→大幅提升排查效率
2. **评估增长质量**：DAU涨了→看涨在哪部分→是"Retained底盘变大"（真增长）还是"New/Resurrected堆量"（虚胖）

### 1.5 面试高频考点
**面试官**："DAU涨了但你觉得不对——怎么判断？"
**标准答案**："用PDS拆解看Retained占比和绝对值。如果Retained绝对值和占比都在涨——产品健康、增长扎实。如果Retained在跌但New在猛涨——增长靠砸钱维持，不可持续。如果Retained和New都在跌但Resurrected在涨——正在靠push/促销透支用户。"

> **今天最重要的认知**：DAU的总数会撒谎。PDS拆解让你看到DAU的"骨架"——Retained绝对值和占比才是产品健康度最诚实的信号。Retained在涨=真增长；Retained在跌但New/Resurrected在涨=虚胖；三者都在跌=危机。

## 二、知识点详细拆解

### 2.1 Retained的深层拆解——N日留存矩阵告诉你"底盘在怎么变"

**为什么重要**：Retained是一个"黑箱"——你只知道总数在变，但不知道为什么变。进一步拆解：Retained_t = Σ(New_{t-i} × day_i_retention)。

如果Retained在降，拆开看：
- **短期（1-7天）留存在降** → 新用户激活体验出问题了（或渠道质量恶化） → 优化新用户引导+渠道质量门槛
- **中期（7-30天）留存在降** → 产品"习惯养成"能力出问题了 → 加强push/内容更新/社交功能
- **长期（30-90天）留存在降** → 产品核心价值在下降或竞品在抢用户 → 需要产品迭代+竞品分析

同样的Retained降幅，不同时间段→不同原因→不同解法。这就是为什么要深拆Retained。

### 2.2 虚胖DAU的三种典型画像——学会"看脸识病"

**纯拉新虚胖**：New占比>30%且在涨，Retained绝对值在降。DAU在涨但全是新用户的"泡沫"。特征：DAU和MAU的比值在降（MAU也在涨但DAU涨更快——多为"一日游"用户）。投放一停→DAU断崖下跌→管理层质问"为什么DAU跌了"→没有人能回答因为没有人知道DAU是靠投放撑着的。

**纯召回虚胖**：Resurrected占比>20%。靠频繁push/短信/活动"催活"——用户没有主动使用习惯。每次push的点击率在递减（从5%→3%→1.5%）——这是用户对push"脱敏"的信号。当脱敏到一定程度，push效果趋近于0——那时候DAU会突然跌一大截。

**虚假拉新**：New在涨但次留暴跌（如投放素材从"精准卖点"改成"标题党"→CTR涨了但次留从20%→5%）。New数量涨了但几乎没转化为Retained——这些"新用户"打开App就关了，第二天不会再回来。但DAU统计中他们被算作了"New"——所以DAU看起来在涨，但实际上是在吸入"僵尸用户"。

### 2.3 PDS贡献度分析——"DAU涨了20万，到底是谁的功劳"

**为什么重要**：贡献度让你精准量化每个部分对DAU变化的贡献——不是"感觉"，是"数字"。

DAU从100万→120万(+20万)：New从10万→35万(+25万贡献)，Retained从70万→65万(-5万贡献)，Resurrected从20万→20万(0贡献)。DAU涨了20万全靠New贡献+25万——但Retained贡献了-5万！**整体在涨，但最核心的部分在跌——不拆贡献度你永远发现不了这个"内伤"。**

贡献度公式：i部分的贡献度 = (i部分的变化量) ÷ (DAU总变化量) × 100%

### 2.4 PDS + 同期群 + 漏斗 = 增长三维诊断体系

三者各看一个维度，拼在一起才是完整画面：
- **PDS看"今天DAU结构"**（横截面）：现在DAU由什么构成？健康吗？
- **同期群看"新批次质量趋势"**（纵向）：新用户是在变好还是变差？
- **漏斗看"转化步骤泄漏"**（深度）：用户在哪个环节流失？

**组合诊断示例**：PDS告诉你DAU靠拉新撑着→同期群告诉你新批次W1留存在恶化→漏斗告诉你恶化发生在浏览→加购环节→综合结论：投放素材出了问题——吸引来的人跟产品不匹配→打开App→浏览商品→发现"不是想要的"→不加购→不回来。

### 2.5 用PDS做DAU预测——比"去年同月×增长率"靠谱

DAU_pred = 预测New + 现有Retained×预测留存率 + 预测Resurrected。给三组数字（乐观/基准/悲观）→三个DAU预测值→一个预测区间。这比"去年同月×120%"的简单外推准确得多——因为它基于DAU的"骨架"（三个构成部分各自预测），而非只基于"总数"。

### 2.6 PDS日常监控看板——每周5分钟看透DAU健康度

设计一个周度PDS看板，包含：
1. 上周日均DAU的PDS三元拆解（New/Retained/Resurrected占比）
2. Retained绝对值12周趋势线（最重要的一张图）
3. New占比12周趋势线（如果持续>30%→黄色告警）
4. Retained的N日留存矩阵（短期/中期/长期分别的趋势）

自动告警规则：Retained绝对值连续3周↓→黄色（关注）；连续5周↓→红色（行动）；New占比连续3周↑→提醒检查渠道质量。

### 2.7 PDS + 维度下钻的组合诊断法——构建"下钻树"

**为什么重要**：PDS告诉你"Retained跌了"，但Retained是一个聚合指标——它背后是几千万用户在多个平台、多个版本上的行为总和。需要结合维度下钻，把"Retained跌了"这个粗粒度信号，拆解到可以行动的具体维度。

**下钻树的标准路径**（以Retained为例）：
```
第0层：Retained总量下降（-8%，从70万→64.4万）
  ↓ 按平台下钻（iOS / Android / Web）
第1层：iOS Retained -2%（正常波动），Android Retained -3%（正常），Web Retained -61%（异常！）
  ↓ 按版本下钻（Web端有2个版本正在运行）
第2层：Web v3.2 Retained 正常，Web v3.3 Retained -96%（异常！v3.3是昨天刚发的版本）
  ↓ 按用户注册时段下钻（确认是新用户还是老用户受影响）
第3层：所有注册时段均受影响——不是特定批次问题，是版本级bug。
  ↓ 结论+行动
行动：Web v3.3发版引入bug→立即回滚版本。
```

**下钻树的三个原则**：
1. **从粗到细**：先平台→再版本→再用户群——每一步缩小搜索范围
2. **找突变点**：不是看"哪个维度最低"，而是看"哪个维度变化最大"——一个从80%突然跌到5%的维度比一个一直20%的维度更值得关注
3. **优先排查最近的变更**：新版本、新渠道、新投放素材——80%的突降跟最近72小时内的变更有关

### 2.8 PDS的周级/月级变体——WAU和MAU的三分法

**为什么重要**：PDS不只能用在DAU上，同样的三分法可以推广到任何时间窗口的活跃指标。但时间窗口越长，各部分的业务含义变化越大。

**WAU（周活）的三分法**：
```
WAU = Weekly New（本周首次打开App的新用户）
     + Weekly Retained（上周活跃且本周也活跃）
     + Weekly Resurrected（上周未活跃但本周活跃了）
```
关键区别：在周级窗口下，"Resurrected"包含两类完全不同的用户——（a）真的流失后回流（停了>2周）；（b）只是上周太忙没打开（停了1周）。后者本质上是"轻度的Retained"，只是被7天窗口切割成了Resurrected。因此WAU的Resurrected占比天然比DAU高——正常现象，不用惊慌。

**MAU（月活）的三分法**：
```
MAU = Monthly New（本月首次打开的新用户）
     + Monthly Retained（上月活跃且本月也活跃）
     + Monthly Resurrected（上月未活跃但本月活跃了）
```
在月级窗口下，Resurrected的含义进一步变化——包含了"季节性用户"（如只在月初查工资条的银行App用户）、"活动唤回用户"（被月度大促吸引回来）、"真正的回流用户"。月级Resurrected如果超过35%——说明产品的"月度习惯"没有建立，用户每月来1-2次就消失。

**三个时间窗口的PDS对比**——暴露不同层次的问题：
| 指标 | Retained健康占比 | Resurrected正常占比 | 最擅长发现的问题 |
|------|----------------|-------------------|----------------|
| DAU | >55% | <15% | 产品日常粘性、昨日体验影响 |
| WAU | >50% | <25% | 产品"周习惯"、周末效应 |
| MAU | >45% | <30% | 产品长期价值、用户生命周期 |

**实战用法**：如果DAU的Retained比>55%（健康）但MAU的Retained比<40%（不健康）——说明产品"每天用的人很忠诚，但大部分用户没有形成月度习惯"，即"核心用户极高粘性，但长尾用户流失严重"。这是一个需要"提升长尾用户的月度留存"的信号。

### 2.9 PDS与净增(Net Add)的关系——"谁来了、谁走了、谁留下了"

**为什么重要**：PDS告诉你DAU的"存量结构"，净增告诉你DAU的"增减动态"。两者结合才是最完整的DAU故事。

**基础关系**：
```
DAU_t = DAU_{t-1} + New_t - Churned_t + Resurrected_t
即：DAU的变化 = New（进来的新人）+ Resurrected（回来的老人）- Churned（走了的人）
```

而PDS告诉你：在这DAU_t里面，谁是新来的、谁是留下的、谁是回来的。

**净增（Net Add）** = New - Churned。如果Net Add为正，DAU在涨。但PDS会告诉你这个"涨"的质量：
- 情况A：New +10万，Churned -3万 → Net Add +7万 → DAU涨了7万。Retained绝对值保持稳定。这种增长是"**净增长**"——真实的增量。
- 情况B：New +15万，Churned -13万 → Net Add +2万 → DAU涨了2万。但花15万获客成本换来了2万净增——获客效率极低。Retained绝对值可能在降（因为Churned -13万包含大量老用户）。这种增长是"**虚假增长**"——表面DAU在涨，但用户换了一拨人。

**核心洞察**：PDS让你看到"留下来的用户是谁"——New补充了新血，Retained是基本盘。如果Retained下降的量 > New增加的量 → 净增为负 → DAU在跌。很多公司只盯着DAU的总数变化，但不知道DAU内部的"用户换血"有多严重——80%的用户上个月还在、这个月已经换了——DAU数字相同，但已经不是同一群人了。

### 2.10 PDS的预测能力——比线性外推准确得多

**为什么重要**：CEO总问"下个月DAU能到多少？"如果只用过去12个月DAU做线性外推——季节性、投放变化、留存变化全部被忽略，误差极大。PDS框架让你把DAU拆成三部分各自预测，每部分有不同的预测逻辑。

**PDS预测三步法**：

**第1步——预测New**：基于投放计划、自然量趋势、渠道预算。New是最容易预测的部分——有多少预算就有多少量。同时输入乐观（预算加20%）、基准（预算持平）、悲观（预算减10%）三组数字。

**第2步——预测Retained**：基于历史留存率曲线+产品改版计划。用过去90天的留存率衰减曲线作为基准，如果有产品改版→用AB实验修正留存率。这是最难预测的部分——但也是最重要的部分，因为Retained通常占DAU的50%以上。

**第3步——预测Resurrected**：基于运营日历（大促、push计划、活动节奏）。Resurrected跟运营强度强相关——如果下个月有大促→Resurrected预测上调；如果计划减少push频率→下调。

**三组预测合成**：乐观(New↑, Retained↑, Resurrected↑) / 基准(New→, Retained→, Resurrected→) / 悲观(New↓, Retained↓, Resurrected↓) → 三个DAU预测值 → 一个预测区间。

**跟简单外推的对比**：假设DAU 100万，过去6个月月均增长+3%。简单外推：下月DAU=103万。PDS预测：New可能从10万→8万（投放预算要砍），Retained从60万→57万（留存曲线在缓慢下行），Resurrected从30万→30万（不变）→下月DAU=95万。**简单外推给你103万的幻觉，PDS揭露95万的真相。**

### 2.11 不同产品类型的PDS健康基准——一套数字不能适用所有产品

**为什么重要**：你不能拿微信的Retained比（70%+）来要求一个工具App——不同产品类型的用户行为模式完全不同。使用对标的基准才能正确判断健康度。

| 产品类型 | Retained健康占比 | 告警线 | New正常占比 | 代表产品 | Retained高的原因 |
|------|--------------|------|-----------|---------|---------------|
| **社交/通讯** | >65% | <55% | <15% | 微信、WhatsApp | 高频刚需，人际网络锁定效应 |
| **内容/信息流** | >55% | <45% | <25% | 抖音、小红书、头条 | 内容新鲜感驱动，但习惯形成后粘性高 |
| **工具** | >40% | <30% | <35% | 计算器、WPS、扫描App | 低频刚需，用后即走是正常行为 |
| **电商** | >30% | <20% | >30% | 淘宝、拼多多、京东 | 非每日刚需，但有购物需求时高频 |
| **游戏** | >25% | <18% | >25% | 王者荣耀、原神 | 内容消耗型，新版本/新活动刺激关键 |

**关键认知**：
1. **不要跨品类比较**——拿电商的Retained比（30%）跟社交比（65%）毫无意义。要跟自己的历史趋势比、跟直接竞品比。
2. **Retained告警线 !== 目标线**——告警线是"低于这个值就要行动了"，目标应该是持续提升——哪怕从41%提升到43%，对于工具产品来说也是实质性的改善。
3. **New占比的解释因品类而异**——电商New占比30%是正常的（大促期间甚至会到40%+），但社交New占比30%就是异常——你不可能每天都有一大群新人用微信。
4. **Resurrected的健康标准与产品"可替代性"成反比**——工具产品可替代性高，Resurrected达到20%也正常（用户尝试了竞品又回来）；社交产品可替代性低（关系链锁定），Resurrected超过10%反而说明产品出了大问题（为什么有这么多人走了？）。

## 三、动手练习
### 步骤0：模拟90天DAU数据
模拟一个产品的90天DAU数据，包含New、Retained、Resurrected三个部分。使用40%的简化留存率模拟Retained。

````python
import numpy as np; np.random.seed(42)
days=90; new_users=np.random.randint(8000,12000,days); r1d=0.40
dau=np.zeros(days,dtype=int); ret_u=np.zeros(days,dtype=int); res_u=np.zeros(days,dtype=int)
for t in range(days):
    if t>0:
        # Retained: 昨日DAU × 保留率（简化，实际应用应按同期群分别算）
        ret_u[t]=int(dau[t-1]*r1d)
        # Resurrected: 随机波动，模拟自然回流
        res_u[t]=np.random.randint(1000,3000)
    dau[t]=new_users[t]+ret_u[t]+res_u[t]

print("="*55)
print("  90天DAU模拟数据完成！")
print("="*55)
print(f"  日均DAU:        {dau.mean():>10,.0f}")
print(f"  日均New:         {new_users.mean():>10,.0f}  ({new_users.mean()/dau.mean()*100:.1f}%)")
print(f"  日均Retained:    {ret_u.mean():>10,.0f}  ({ret_u.mean()/dau.mean()*100:.1f}%)")
print(f"  日均Resurrected: {res_u.mean():>10,.0f}  ({res_u.mean()/dau.mean()*100:.1f}%)")
print("-"*55)
print(f"  DAU变化趋势: 第1天{dau[0]:,} → 第90天{dau[-1]:,}")
print(f"  模拟参数: r1d(简化留存率)={r1d}, 随机种子=42")
# 【预期输出】日均DAU≈13000, New≈10000, Retained≈5000, Resurrected≈2000
````

### 步骤1：PDS拆解+增长模式诊断（基础）
基于最近30天的均值，完成PDS三元拆解，自动诊断增长模式并给出Retained趋势判断。

````python
last30=slice(-30,None)
ad=dau[last30].mean(); an=new_users[last30].mean()
ar=ret_u[last30].mean(); are=res_u[last30].mean()

print("DAU PDS拆解（最近30天均值）"); print("="*55)
print(f"日均DAU:        {ad:>8,.0f}  (100%)")
print(f"  New(新增):     {an:>8,.0f}  ({an/ad*100:.1f}%)")
print(f"  Retained(留存):{ar:>8,.0f}  ({ar/ad*100:.1f}%)")
print(f"  Resurrected:   {are:>8,.0f}  ({are/ad*100:.1f}%)")
print()

if ar/ad>0.55:
    print("诊断：留存驱动型 → 产品健康，增长扎实。可适度加大拉新。")
elif an/ad>0.35:
    print("诊断：拉新驱动型 → 注意Retained是否在恶化！检查渠道质量和留存趋势。")
elif are/ad>0.20:
    print("诊断：召回驱动型 → 用户主动习惯未建立！减少对push的依赖。")
else:
    print("诊断：混合型 → 结合趋势判断增长质量。")
print()

# Retained趋势分析
f30=ret_u[:30].mean(); l30=ret_u[last30].mean()
print(f"Retained趋势: 前30天 {f30:,.0f} → 后30天 {l30:,.0f} ({(l30/f30-1)*100:+.1f}%)")

# 趋势+模式联合诊断
if l30>f30 and ar/ad>0.55:
    print("↑ Retained底盘在变大 + 占比健康 → 真增长！产品粘性在提升。")
elif l30>f30 and ar/ad<=0.55:
    print("↑ Retained在涨但占比不高 → 增长在改善，继续努力。")
elif l30<f30 and an/ad>0.35:
    print("↓ Retained在跌 + New占比过高 → 典型的虚胖DAU！立即检查。")
elif l30<f30 and ar/ad>0.55:
    print("↓ Retained在跌但占比仍健康 → 警惕信号，深拆Retained找原因。")
else:
    print("↓ Retained底盘在萎缩——即使DAU在涨也要警惕！")
# 【预期输出】PDS三部分占比+增长模式标签+Retained趋势判断
````

### 步骤2：虚胖DAU识别——四维PDS诊断（进阶）
通过一个真实案例，展示如何用PDS框架在30秒内识别"DAU在涨但产品在变差"的危险信号。

````python
print("案例：DAU从80万涨到120万（+50%）——是好是坏？"); print("="*60)
print("半年前PDS: New=8万(10%)  Retained=60万(75%)  Resurrected=12万(15%)")
print("现在PDS:    New=36万(30%) Retained=55万(46%)  Resurrected=29万(24%)")
print()
print("━ 四维PDS诊断 ━")
print(f"  ① Retained绝对值: 60万→55万(-8.3%, -5万)  ← 产品粘性变差！")
print(f"  ② Retained占比:   75%→46%(-29pp)        ← 从留存驱动降级为混合型")
print(f"  ③ New贡献:        8万→36万(+350%)       ← 大量砸钱买量")
print(f"  ④ Resurrected:    12万→29万(+142%)      ← 频繁push催活")
print()
print("结论：典型的【虚胖DAU】——表面+50%但底盘在萎缩")
print("  增长驱动力：拉新(350%↑) + 召回(142%↑)——两者都要花钱且不可持续")
print("  恶化信号：留存(-8.3%↓)——唯一'免费'且可持续的增长动力在减弱")
print("  崩盘预测：如果今天停投放+停push→DAU可能1-2周跌回80万甚至更低")
print()

# 额外分析：计算虚胖指数
ret_old, ret_new = 60, 55
new_old, new_new = 8, 36
res_old, res_new = 12, 29
dau_old, dau_new = 80, 120

# 虚胖指数 = New增长速度 / Retained增长速度
new_growth = (new_new - new_old) / new_old
ret_growth = (ret_new - ret_old) / ret_old
vanity_index = new_growth / abs(ret_growth) if ret_growth != 0 else float('inf')
print(f"━ 虚胖指数(New增速/Retained增速绝对值) = {vanity_index:.1f} ━")
print(f"  虚胖指数>3 → 高度依赖拉新，增长质量极差")
print(f"  虚胖指数<1 → 增长以留存为主，健康")
print()

print("建议：")
print("  ① 立即暂停次留<15%的低质渠道")
print("  ② 深入分析Retained为什么降——产品改版？竞品？自然波动？")
print("  ③ 等Retained绝对值止跌回升→再逐步加大拉新")
print("  ④ 建立PDS周监控——不让'虚胖'再次发生")
# 【预期输出】四维PDS诊断+虚胖指数+三级行动建议
````

### 步骤3：PDS贡献度分析——精确量化每部分的功劳（进阶）
DAU变了，每个部分各贡献了多少？用贡献度公式精确量化，找出"谁在推动增长、谁在拖后腿"。
**关键公式**：贡献度 = 该部分变化量 ÷ DAU总变化量 × 100%

````python
import numpy as np

# 比较前后30天的PDS变化并精确计算贡献度
first30 = slice(0, 30)
last30 = slice(-30, None)

# 前后30天均值
new_f30 = new_users[first30].mean()
new_l30 = new_users[last30].mean()
ret_f30 = ret_u[first30].mean()
ret_l30 = ret_u[last30].mean()
res_f30 = res_u[first30].mean()
res_l30 = res_u[last30].mean()
dau_f30 = dau[first30].mean()
dau_l30 = dau[last30].mean()

# 变化量
d_new = new_l30 - new_f30
d_ret = ret_l30 - ret_f30
d_res = res_l30 - res_f30
d_dau = dau_l30 - dau_f30

print("\n" + "="*65)
print("  PDS 贡献度分析：量化每部分对DAU变化的贡献")
print("="*65)
print(f"  DAU 变化:  {dau_f30:>10,.0f} → {dau_l30:>10,.0f}  (变化: {d_dau:>+10,.0f})")
print("-"*65)

# 贡献度计算
if d_dau != 0:
    c_new = d_new / d_dau * 100
    c_ret = d_ret / d_dau * 100
    c_res = d_res / d_dau * 100
else:
    c_new = c_ret = c_res = 0

print(f"  {'部分':<18} {'前30天均值':>10} {'后30天均值':>10} {'变化量':>10} {'贡献度':>10}")
print("-"*65)
print(f"  {'New(新增)':<18} {new_f30:>10,.0f} {new_l30:>10,.0f} {d_new:>+10,.0f} {c_new:>+9.1f}%")
print(f"  {'Retained(留存)':<18} {ret_f30:>10,.0f} {ret_l30:>10,.0f} {d_ret:>+10,.0f} {c_ret:>+9.1f}%")
print(f"  {'Resurrected(回流)':<18} {res_f30:>10,.0f} {res_l30:>10,.0f} {d_res:>+10,.0f} {c_res:>+9.1f}%")
print("-"*65)

# 贡献度解读
print()
print("━ 贡献度解读规则 ━")
print("  贡献度>100% → 该部分增长超过了DAU总增长，其他部分在拖后腿")
print("  贡献度<0%   → 该部分在萎缩，抵消了其他部分的增长")
print("  贡献度≈0%   → 该部分基本无变化，对DAU变化无贡献")
print()

# 质量判断
if d_dau > 0:
    if c_ret > 50:
        print(f"质量判断：Retained贡献度 {c_ret:.0f}% > 50% → 健康的留存驱动型增长")
    elif c_ret < 0:
        print(f"质量判断：Retained贡献度 {c_ret:.0f}% < 0%  → 危险！DAU在涨但Retained在跌！虚胖增长！")
    elif c_new > 80:
        print(f"质量判断：New贡献度 {c_new:.0f}% > 80% → 拉新依赖型增长，Retained贡献不足")
    else:
        print(f"质量判断：混合型增长，持续监控Retained趋势")
elif d_dau < 0:
    if c_ret < 0 and abs(c_ret) > 50:
        print(f"质量判断：DAU下跌主要由Retained萎缩驱动 → 产品粘性恶化！紧急！")
    elif c_new < 0:
        print(f"质量判断：DAU下跌主要由拉新减少驱动 → 可能是投放预算被砍")
    else:
        print(f"质量判断：综合分析各因素，结合业务上下文判断")

# 计算健康得分 (Retained贡献度占比越高越好)
if d_dau != 0:
    pos_contrib = sum([max(0, c_new), max(0, c_ret), max(0, c_res)])
    health_score = max(0, c_ret) / pos_contrib * 100 if pos_contrib > 0 else 0
    print(f"\n增长健康得分: {health_score:.0f}/100 (Retained正向贡献占比)")
    if health_score >= 60:
        print("  ≥60分 → 增长质量良好，Retained是主要增长引擎")
    elif health_score >= 30:
        print("  30-60分 → 增长质量一般，Retained不够强势")
    else:
        print("  <30分 → 增长质量差，过度依赖拉新或召回")

# 【预期输出】贡献度表格+质量判断+健康得分
````

### 步骤4：挑战练习——基于PDS的DAU预测模型（高级）
用过去30天的PDS数据和留存率基准，预测未来7天DAU。输出三种情景（乐观/基准/悲观）的预测区间。这是分析师面试中常见的"建模题"。

````python
import numpy as np; np.random.seed(42)

# ━━━ 1. 数据准备：使用过去30天作为预测基准 ━━━
recent_new = new_users[-30:]
recent_ret = ret_u[-30:]
recent_res = res_u[-30:]
recent_dau = dau[-30:]

# 计算经验留存率（每日Retained_{t} / DAU_{t-1}）
daily_retention_rates = []
for i in range(1, len(recent_dau)):
    if recent_dau[i-1] > 0:
        daily_retention_rates.append(recent_ret[i] / recent_dau[i-1])
mean_retention = np.mean(daily_retention_rates)
std_retention = np.std(daily_retention_rates)

print("="*60)
print("  基于PDS的DAU预测模型（未来7天）")
print("="*60)
print(f"  模型参数（基于近30天数据）:")
print(f"    基准留存率:        {mean_retention:.2%} (±{std_retention:.2%})")
print(f"    日均New:           {recent_new.mean():.0f}")
print(f"    日均Resurrected:   {recent_res.mean():.0f}")
print(f"    当前DAU(第90天):   {dau[-1]:,}")
print("-"*60)

# ━━━ 2. 三种情景的预测参数 ━━━
scenarios = {
    "悲观": {
        "new_mult": 0.85,        # New减少15%
        "ret_adj": -0.03,         # 留存率下降3pp
        "res_mult": 0.90,         # Resurrected减少10%
        "desc": "投放预算削减 + 留存继续恶化"
    },
    "基准": {
        "new_mult": 1.00,         # New持平
        "ret_adj": 0.00,          # 留存率不变
        "res_mult": 1.00,         # Resurrected持平
        "desc": "一切维持现状"
    },
    "乐观": {
        "new_mult": 1.15,         # New增加15%
        "ret_adj": +0.02,         # 留存率提升2pp
        "res_mult": 1.10,         # Resurrected增加10%
        "desc": "投放加量 + 留存改善"
    }
}

# ━━━ 3. 执行预测 ━━━
pred_days = 7
last_dau = dau[-1]
predictions = {}

for scenario_name, params in scenarios.items():
    pred = np.zeros(pred_days, dtype=int)
    pred_new = np.zeros(pred_days, dtype=int)
    pred_ret = np.zeros(pred_days, dtype=int)
    pred_res = np.zeros(pred_days, dtype=int)
    prev_dau = last_dau
    for t in range(pred_days):
        n = int(recent_new.mean() * params["new_mult"])
        r = int(prev_dau * max(0, mean_retention + params["ret_adj"]))
        rs = int(recent_res.mean() * params["res_mult"])
        pred[t] = n + r + rs
        pred_new[t] = n
        pred_ret[t] = r
        pred_res[t] = rs
        prev_dau = pred[t]
    predictions[scenario_name] = {
        "dau": pred, "new": pred_new, "ret": pred_ret, "res": pred_res
    }

# ━━━ 4. 结果展示 ━━━
print("\n未来7天DAU预测（三种情景）:")
print("-"*75)
header = f"{'Day':<8} | {'悲观':>12} | {'基准':>12} | {'乐观':>12} | {'预测区间':>16}"
print(header)
print("-"*75)
for d in range(pred_days):
    p_pes = predictions["悲观"]["dau"][d]
    p_base = predictions["基准"]["dau"][d]
    p_opt = predictions["乐观"]["dau"][d]
    range_str = f"[{p_pes}, {p_opt}]"
    print(f"第{d+1}天{'':>4} | {p_pes:>10,}  | {p_base:>10,}  | {p_opt:>10,}  | {range_str:>16}")
print("-"*75)

# 日均值汇总
for name in ["悲观", "基准", "乐观"]:
    arr = predictions[name]["dau"]
    print(f"  {name} 7日均值: {arr.mean():>10,.0f}  (当前: {last_dau:,})")

# 置信区间
base_end = predictions["基准"]["dau"][-1]
lower_end = predictions["悲观"]["dau"][-1]
upper_end = predictions["乐观"]["dau"][-1]
interval_width = (upper_end - lower_end) / base_end * 100 if base_end > 0 else 0
print(f"\n━ 预测质量评估 ━")
print(f"  第7天基准预测: {base_end:,}")
print(f"  第7天预测区间: [{lower_end:,}, {upper_end:,}]")
print(f"  区间宽度/基准:  {interval_width:.1f}%")
print(f"  区间越窄 → 预测越稳定（<20%为高可信度）")

# ━━━ 5. 对比简单外推 ━━━
simple_trend = (dau[-30:].mean() - dau[:30].mean()) / 30  # 日均变化量
simple_pred_d7 = last_dau + simple_trend * 7
print(f"\n━ 对比：简单线性外推 vs PDS预测 ━")
print(f"  简单外推第7天DAU: {simple_pred_d7:.0f}")
print(f"  PDS基准 第7天DAU: {base_end:,}")
print(f"  差异: {base_end - simple_pred_d7:+,.0f}")
print(f"  简单外推忽略了DAU内部结构变化 → 可能严重高估或低估")
# 【预期输出】三情景预测表+预测区间+与简单外推的对比
````

## 四、业务场景应用

### 场景1：DAU涨了30%但CEO"没感觉产品变好"——PDS瞬间破案

**背景**：某社交产品，过去3个月DAU从200万涨到260万（+30%）。团队在全员会上庆祝，但CEO皱着眉说："数据在涨，但我直觉产品没变好。"

**PDS拆解发现**：
- New：5万→22万（+340%），信息流投放从每天3万预算加到每天25万
- Retained：150万→146万（-2.7%），绝对值在跌
- Resurrected：45万→92万（+104%），push频率从每周2次加到每天1次

**真相**：DAU增长≈100%来自"花钱"——要么是投放（New），要么是push催活（Resurrected）。产品本身的留存（Retained）在下降——150万→146万，少了4万老用户。CEO的直觉完全正确：产品确实没有变好，只是投放和push在撑场面。

**分析师的行动**：向CEO展示PDS三元占比变化趋势图——Retained占比从75%→56%（红线在下滑），Resurrected从23%→35%（反超New）。用一句话总结："您的直觉是正确的——260万DAU里只有146万人是因为产品好才来的，剩下的都是花钱买来的。"

**不用PDS的后果**："DAU+30%！我们做得很好！"→继续砸钱→3个月后留存进一步恶化→某天投资人要求缩减预算→投放砍半→DAU从260万断崖跌到160万→CEO质问"不是涨得好好的吗怎么突然跌了"→信任崩塌、团队士气崩盘。分析师因为"没有提前预警"被追责。

### 场景2：DAU跌了15%，5分钟从告警到结论——分析师"救火"的标准操作

**背景**：周五下午4:55，运营VP在群里@你："今天DAU比昨天少了15%！快看怎么回事！"你只有5分钟给出初步判断——这是分析师的"应急响应"能力考验。

**你的操作流程（精确到分钟）**：

**T+0min**：确认数据完整性——ETL链路是否完整？数据是否已全部入库？排除数据延迟导致的"假下降"。确认结论：数据完整，非数据问题。

**T+1min**：PDS快速拆解——DAU从昨天80万→今68万（-12万, -15%）。New 8万→8万（正常），Resurrected 12万→12万（正常），**Retained 60万→48万（-12万, -20%）** ← 问题100%在Retained。

**T+2min**：维度下钻——按平台拆分Retained：iOS 25万→24万（正常，-4%），Android 30万→29万（正常，-3%），**Web端 5万→0.5万（-90%）** ← 找到问题！只有Web端崩溃。

**T+3min**：问Web团队："今天上午Web端有没有发版？"→回复："上午10点发了v4.7。"→你立即回："v4.7的登录cookie处理可能有bug，影响了约5万Web Retained用户。能看一下吗？"→Web团队确认：登录cookie过期时间设置错误，大量用户被当作未登录。

**T+5min**：在群里回复VP："Web端今天上午10点发版v4.7导致登录态丢失，影响约5万Web用户。Web团队已确认bug，正在修复，预计30分钟内恢复。Retained的下降全部来自Web端，iOS和Android正常。修复后DAU应恢复到接近昨天水平。"

**关键能力复盘**：不慌（保持冷静是分析师的专业素养）；PDS快速缩小范围（把"DAU跌了"缩小到"Retained跌了"然后缩小到"Web Retained跌了"）；知道问谁（不是自己盲目排查，而是精准定位后找对应团队）；用量化数据汇报（"影响约5万用户"而非"好像有点问题"）。

**不用PDS的后果**：花2小时拉了50张表——看了渠道数据、看了页面访问、看了地域分布、看了用户画像……最后通过排除法才定位到Web端问题。期间DAU一直跌，VP在群里催了5次。最终找到问题但已经是晚上7点——应急响应变成了加班噩梦。更糟的是，VP以后对你的信任降低——"拉个数据要2小时"。

### 场景3：用PDS数据说服CEO"现在不能加大投放，要先修产品"

**背景**：CEO在周会上说："既然拉新能涨DAU，下个月投放预算加50%！我们要快速做大规模！"房间里都在点头，只有你知道这个决策很危险。

**你的反直觉论证**（用数据而非观点说话）：

你站起来说："我理解增长的紧迫性。但请先看三组数据，然后我们再决定。"

**数据1——Retained趋势图（12个月）**："我们的Retained绝对值在连续下降——从年初的70万→65万（-7%）。Retained占比从75%→52%(-23pp)。这意味着每100个DAU中，愿意主动回来的人从75个变成了52个——产品粘性在持续恶化。"

**数据2——新增用户的次留趋势**："今年1月新增用户次留率38%，现在只有24%（-37%）。我们在花更多钱获得更差的用户——信息流投放素材从精准定向变成了泛投，CTR涨了但留存跌了。每一块钱投放的'用户质量回报'在下降。"

**数据3——停投模拟**："基于当前PDS数据，如果今天完全停止投放→DAU会在2周内从130万跌到约55万（=Resurrected 20万 + Retained 35万）。因为New会消失，Resurrected少了push也会跌30%+。这说明130万DAU中有约75万是靠投放和催活'撑'着的——真正的自然活跃用户只有55万。"

**你的建议**："我的建议不是不投——而是换一种投法。将50%的增量投放预算（约200万/月）重新分配：100万用于产品体验优化（留存专项），50万用于高质渠道的精准投放（次留>30%的渠道才投），50万储备。目标：6个月内把Retained拉回到70万+，次留拉回到35%+。等底盘稳了，再逐步加大投放——那时候每一分投放的钱都能沉淀为Retained，而不是像现在这样投了就漏。"

**不用PDS的后果**：CEO拍板"加投50%"→半年后DAU从130万涨到180万（大家庆祝）→投资人要求盈利→砍投放→DAU从180万断崖跌到60万（因为Retained底盘只剩40万）→公司估值暴跌→融资困难→裁员。这就是经典的"烧钱陷阱"——用投放掩盖产品问题，最终被现实清算。PDS是你规避这个陷阱的数据工具。

### 场景4：PDS烧钱陷阱深度论证——向CEO证明"我们正在走向悬崖"

**背景**：某电商产品DAU连续12个月在涨，从200万→250万（+25%），所有人都在庆祝"持续增长"。但你用PDS拆解后发现了可怕的结构变化：

| 指标 | 12个月前 | 现在 | 变化 | 信号 |
|------|--------|------|------|------|
| DAU总量 | 200万 | 250万 | +25% | 表面看很好 |
| New | 15万(7.5%) | 45万(18%) | +200% | 投放量翻了三倍 |
| Retained | 80万(40%) | 62万(24.8%) | **-22.5%** | 核心用户资产在蒸发 |
| Resurrected | 105万(52.5%) | 143万(57.2%) | +36% | 靠push和大促催回来的 |

**核心发现**：DAU涨了50万，但Retained跌了18万——每涨1个DAU，就同时失去0.36个老用户。这不是增长，这是"用户置换"——用高成本的新用户替换低成本的老用户。

**停投放冲击推演**（向CEO展示的定量模拟）：
```
假设今天停止所有投放（New归零）且push频率减半（Resurrected跌40%）：

DAU(停投第1天) = 0(New) + 62万(Retained) + 143万×0.6(Resurrected) = 0 + 62万 + 86万 = 148万
  → 一天之内从250万跌到148万（-41%）

DAU(停投第7天) = 0(New) + 148万×0.35(留存率) + 86万×0.7(Resurrected再降) ≈ 52万 + 60万 = 112万
  → 一周后跌到112万（-55%）

DAU(停投第14天) = 0(New) + 112万×0.3 + 60万×0.5 ≈ 34万 + 30万 = 64万
  → 两周后跌到64万（-74%）
```

**结论呈现**："CEO，我们发现了一个严重的问题——250万DAU中只有62万人（24.8%）是因为产品好而主动回来的。剩下75%的DAU靠两样东西撑着：投放（45万New）和push/大促催活（143万Resurrected）。这两样都要花钱，且不可持续。如果今天投资人要求盈利、砍掉投放和push预算——我们的DAU会在2周内跌到60万级别——比12个月前的200万还低。这不是增长，这是烧钱陷阱——DAU在涨，但地基在塌。"

**行动方案**（给出有建设性的出路，而非只提出问题）：
1. **立即止血**：暂停次留<20%的投放渠道——这些渠道的New几乎不转化为Retained，纯浪费
2. **留存专项**：成立跨部门"Retained提升专项"——目标6个月内Retained从62万回到75万
3. **渠道分级**：按"次留×渠道量级"对渠道分级，A级渠道（次留>35%）保留并加量，B级（25-35%）优化后保留，C级（<25%）暂停
4. **主动汇报**：在下一期月报里主动披露PDS结构变化，展示"停止投放后的DAU推演"——让管理层看到全貌，而非等到出事才解释

**不用PDS的后果（真实灾难）**：继续加投放→半年后DAU冲到300万→投资人A轮→尽调发现Retained占比只有20%→估值从2亿砍到5000万→公司大裁员→产品停运。这不是假设——过去5年有至少30家互联网公司死于"DAU在涨心里美，停了投放就掉队"。

## 五、分级检验清单

### 初级检验（必须能独立完成的）
- [ ] 能写出DAU的PDS公式（DAU = New + Retained + Resurrected）
- [ ] 能从PDS占比判断增长模式（Retained>60%=留存驱动健康型）
- [ ] 能区分三类用户的业务含义和行为特征
- [ ] 能用Python（numpy）完成基础的PDS拆解计算

**答案提示**：
- PDS公式：DAU = New(今日新增) + Retained(昨日活跃且今日也活跃) + Resurrected(之前流失今日回来)
- 增长模式判断：Retained>60%=留存驱动（健康）、New>30%=拉新驱动（警惕）、Resurrected>20%=召回驱动（不可持续）
- 三类用户特征：New=需要引导教育（对产品一无所知）、Retained=需要持续价值和新鲜内容（已有使用习惯）、Resurrected=需要搞清楚"为什么走+为什么回来"（曾经流失过）
- Python基础代码：dau = new_users + retained_users + resurrected_users

### 中级检验（能解释给同事听的原理）
- [ ] 能计算各部分对DAU变化的贡献度（DAU涨跌了多少→各部分各贡献了多少）
- [ ] 能识别"虚胖DAU"的三类典型画像（纯拉新虚胖/纯召回虚胖/虚假拉新）
- [ ] 能结合维度下钻（平台/渠道/版本）定位PDS问题的具体责任群体
- [ ] 能向非技术同事解释"为什么DAU涨了但产品可能在变差"

**答案提示**：
- 贡献度公式：该部分变化量 ÷ DAU总变化量 × 100%。如New贡献120% → New的增长超过了DAU总增长 → 其他部分在拖后腿
- 虚胖DAU三类画像：①纯拉新虚胖（New>30%且Retained在跌）→投放依赖型；②纯召回虚胖（Resurrected>20%且Retained在跌）→push依赖型；③虚假拉新（New↑但次留暴跌）→渠道质量恶化型
- 下钻标准路径：DAU跌→PDS锁定Retained→按平台下钻→按版本下钻→按注册时段下钻→找到"哪个维度变化最大"（不是哪个维度最低）
- 向非技术同事解释："DAU涨了20万，但其中25万是花钱买的新用户，老用户反而少了5万。就像餐馆——门口排队的人多了（因为发了优惠券），但老顾客在变少。"

### 高级检验（能举一反三的迁移能力）
- [ ] 能深拆Retained到N日留存矩阵——判断是短期/中期/长期留存哪个在恶化
- [ ] 能基于PDS构建DAU预测模型（预测New+预测留存率→预测DAU区间）
- [ ] 能用PDS结果说服管理层调整增长策略：从"砸钱拉新"转向"先修产品再拉新"
- [ ] 能设计完整的PDS监控周报体系——告警规则+趋势看板+行动建议

**答案提示**：
- N日留存矩阵拆解：Retained_t = Σ(New_{t-i} × day_i_retention)。短期(1-7天)恶化→激活/渠道问题；中期(7-30天)恶化→习惯养成问题；长期(30-90天)恶化→核心价值问题/竞品冲击
- DAU预测三步法：①预测New（基于投放计划）；②预测Retained（基于历史留存曲线×产品改版修正）；③预测Resurrected（基于运营日历）。三组情景（乐观/基准/悲观）→预测区间
- 说服管理层五步法：①展示Retained绝对值下降趋势图（情感冲击）；②展示停投模拟结果（"如果今天停投,DAU会跌到XX"）；③展示渠道次留对比（"有些渠道的次留只有18%"）；④给出建设性替代方案（不是"不要投"而是"换种方式投"）；⑤建议用AB实验验证（"我们先拿10%预算做实验验证'修产品'的效果"）
- PDS周报体系：核心指标+三元结构+Retained深度分析+New质量分析+告警建议。告警规则：红(连5周↓或降幅>10%)、黄(连3周↓或New>35%)、绿(Retained↑且New<25%)

### 横向连接
**与LTV的关系(W2D4)**：Retained用户的N日留存矩阵直接决定了LTV的大小。DAU里Retained占比越高→整体LTV越高→可以承受的CAC越高→拉新空间越大。如果Retained占比从60%降到40%→同样的DAU，LTV总量可能下降了25%+——因为New用户留存低、贡献的LTV远低于Retained用户。这就是为什么"不修留存就猛拉新"是亏损放大器。

**与同期群的关系(W2D2)**：Retained降了→同期群纵向看"哪批用户的留存变差了"→定位是"新批次质量问题"还是"老批次加速流失"。前者→优化渠道和激活；后者→产品体验恶化或竞品冲击。

**与漏斗的关系(W2D3)**：Retained恶化→看用户在漏斗哪个环节卡住了→是新用户激活环节（短期留存降）还是日常使用环节（中期留存降）还是长期价值环节（长期留存降）。

**与AB实验的关系(W7-W8)**：改进产品后Retained是否真的提升了？——需要AB实验排除自然波动和季节性后才能下结论。PDS提供的结构分解能大幅提升AB实验的统计灵敏度——因为你可以只看Retained组的变化，而不是被New和Resurrected的噪声干扰。

## 七、当日小结

> **今天最重要的一个认知**：DAU的总数会撒谎。PDS拆解让你看到DAU的"骨架"——Retained绝对值和占比是产品健康度最诚实的信号。记住三种模式：Retained在涨=真增长（可以加大拉新）；Retained在跌但New/Resurrected在涨=虚胖（先止血修产品）；Retained在跌且New也在跌=双输（紧急状态）。

**三条金句带回家**：
1. DAU = New + Retained + Resurrected——三者的占比就是增长质量的"体检报告"
2. Retained绝对值是产品健康度最诚实的信号——比DAU总量可靠10倍，因为它是"免费"的增长（用户主动回来）
3. New涨但Retained跌=往漏水的桶里倒水——先修桶（提升留存），再倒水（加大拉新），顺序不能反

### 面试特训——高频PDS面试题及完整答案模板

**Q1**："DAU涨了但你觉得增长不健康——怎么判断并向老板说明？"（大厂数据分析师面试70%概率出现）

**回答框架**（3分钟标准答案）：
（第1分钟）"用PDS框架拆解DAU。DAU = New + Retained + Resurrected。我会先看Retained的绝对值和占比。如果DAU涨了20%但Retained跌了5%——说明增长全靠New和Resurrected在堆量，产品本身的粘性在变差。这是虚胖型增长。"
（第2分钟）"向老板说明时，我不会只说'有问题'——我会展示两个东西：①Retained绝对值在过去12周的趋势图（曲线在往下走）；②停投模拟——如果今天停止投放，DAU会在2周内跌到多少。这两个图一看就懂——老板不用理解PDS公式也能明白'增长在透支未来'。"
（第3分钟）"最后我不会只提问题——我会给出有建设性的替代方案。比如'建议将30%的投放预算转为产品优化预算，目标6个月内把Retained拉回到XX万。等底盘稳了再恢复投放——那时候每花1块钱投放，有更好的留存率兜底，ROI会更高。'"

**Q2**："New、Retained、Resurrected的业务区别是什么？"

**回答**："三者需要完全不同的产品策略。New需要引导和激活——他们第一次来，不知道产品能干什么，需要尽快触发'啊哈时刻'。Retained需要持续的新鲜感和个性化——他们已经养成了使用习惯，最怕'无聊'和'内容枯竭'。Resurrected需要回答两个问题——'当初为什么走'和'今天为什么回来'——如果是被push催回来的，大概率明天又走；如果是产品更新吸引回来的，可能重新成为Retained。三者的流失成本和挽回概率完全不同——Retained流失是最贵的，因为他们是已经证明过'产品对ta有价值'的人。"

**Q3**："如何用PDS做DAU预测？比线性外推好在哪？"

**回答**："PDS预测比线性外推好因为它基于DAU的'骨架'而非'总数'。第一步预测New（基于投放计划——最容易预测）；第二步预测Retained（基于历史留存曲线——最难但最重要）；第三步预测Resurrected（基于运营日历）。三组参数（乐观/基准/悲观）合成三个预测值→一个预测区间。线性外推的致命缺陷：它假设DAU的'内部结构不变'——但如果下个月投放预算被砍、New暴跌，线性外推完全预测不到。PDS预测能捕捉到结构变化——比如'预算砍了但留存改善了'对不同部分的不同影响。"

### 课后拓展

1. **练习——PDS实战**：如果你有你们产品的DAU数据，做一次完整的PDS拆解：
   - Retained占比多少？过去12周趋势是改善还是恶化？
   - 如果恶化→深拆到N日留存矩阵，看是短期(1-7天)、中期(7-30天)还是长期(30-90天)在恶化
   - 计算New各渠道的次留率——哪些渠道的New质量在下降？
   - 输出一份PDS健康度一页纸报告（参考场景5的模板）

2. **思考——如何在周报中写"DAU涨但Retained跌"这个问题**：如果你的DAU在涨但Retained在跌——短期报告里DAU是涨的（老板看到会开心）。你该怎么在周报里写这个问题，既不隐瞒真相也不制造恐慌？参考做法：在"亮点"板块写DAU增长，在"风险提示"板块写Retained趋势+推演如果止损后的影响。用"机会"和"风险"而非"功"和"过"的框架。

3. **预习——ISS系统化分析(W3)**：PDS发现问题（如"Retained连续3月降"）后，需要一个系统化的诊断框架来回答"为什么降"和"怎么办"。W3的ISS框架就是做这个的——Issue（问题定义）→ Segment（分层分析）→ Solution（方案设计）。提前思考：如果你的产品Retained在降，你会从哪些维度去诊断根因？

4. **挑战——设计一份PDS自动化监控系统**：如果一个产品有500万DAU，如何自动化地每天运行PDS拆解并在异常时自动告警？需要考虑：数据管道怎么搭建（ETL频率？延迟容忍度？）、告警阈值怎么设（不能太灵敏导致告警疲劳，也不能太迟钝导致漏报）、告警路由怎么设计（什么级别的告警发给谁）、如何避免"数据波动误报"（用连续N天而非单日作为触发条件）。这是一个分析师进阶为数据产品经理的典型路径——从"手动分析"到"系统化监控"。
