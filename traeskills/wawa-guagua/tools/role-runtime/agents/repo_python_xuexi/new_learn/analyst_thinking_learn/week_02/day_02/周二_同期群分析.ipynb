## 一、核心概念与第一性原理

### 1.1 为什么需要同期群——一个平均数能骗你多久

看一个数："我们的7日留存率是25%。"这个数听起来还行。但拆开看：
- 1月注册的用户7日留存是35%
- 6月注册的用户7日留存只有15%

平均25%让你觉得"还行"，但实际上新用户质量在持续恶化——而你被平均数骗了。这就是"辛普森悖论"的典型表现：整体趋势看起来稳定，但内部结构已经发生了根本性变化。

**同期群分析（Cohort Analysis）的核心思想**：把用户按"首次行为时间"分成一组一组（同期群），然后单独追踪每一组在后续每一天/每周/每月的表现。不做同期群，你看不到"新用户质量在变好还是变差"这个最基本的问题。

### 1.2 从第一性原理推导：为什么平均数会撒谎

平均数 = 各组的总和 / 总数。如果各组的人数权重变了，平均数就会变——即使每组内部没有变。这是一个简单的数学事实：

```
1月：100个用户，留存35% → 35个留存
6月：1000个用户，留存15% → 150个留存
整体留存 = (35+150)/(100+1000) = 16.8%

你看：1月的用户质量很好(35%)，但人少(100)；6月的用户质量差(15%)，但人多(1000)。
整体16.8%被6月主导——你看到的"整体变化"其实是"权重变化"而非"质量变化"。
```

同期群的价值：把"时间趋势"和"用户质量"两个维度解耦。横向看一行——同一批用户随时间衰减，这是产品的"自然规律"。纵向看一列——不同批次在同一时间节点的质量变化，这才是"产品在变好还是变差"的信号。

### 1.3 同期群表格——三角矩阵

同期群的经典呈现方式是一个三角矩阵（越近的批次数据越少）：

```
            Week0   Week1   Week2   Week3   Week4
1月新用户    100%    40%    28%    22%    18%
2月新用户    100%    35%    24%    19%    —
3月新用户    100%    32%    20%    —      —
4月新用户    100%    28%    —      —      —
```

- **横向看一行**：某一批用户随时间的留存衰减——这是产品的"自然衰减规律"
- **纵向看一列**：同样过了N周，不同批次用户的留存差异——这就是"用户质量在变好还是变差"
- W1列从40%→35%→32%→28%，持续下降——新用户质量在恶化！

### 1.4 同期群的四种定义方式

一个同期群 = 在同一个时间段内，因同一个事件第一次进入产品的一组用户。但什么是"进入产品"？不同定义适用于不同场景：

| 定义方式 | 同期群分组依据 | 适用场景 | 例子 |
|---------|-------------|--------|------|
| 按注册周/月 | 用户第一次注册的时间 | 追踪新用户质量变化趋势 | "3月注册的所有用户" |
| 按首次购买日期 | 用户第一次下单的时间 | 追踪付费用户的复购行为 | "1月首次下单的所有用户" |
| 按渠道 | 用户来源渠道(有机/付费/社媒) | 对比不同渠道的用户质量 | "信息流渠道来的所有新用户" |
| 按行为 | 用户是否完成某个关键行为 | 对比"激活"vs"未激活"用户的后续表现 | "7天内完成3次搜索的用户" |

### 1.5 同期群分析的四个关键时机

**为什么重要**：不是任何时候都需要做同期群——在以下四个时机做，收益最大。

| 时机 | 触发信号 | 同期群回答的问题 | 典型动作 |
|------|---------|---------------|--------|
| 投放策略调整后 | 换素材/放宽定向/开新渠道后2-4周 | 新渠道来的用户质量比旧渠道好吗？ | 对比渠道×同期群W1留存矩阵 |
| 产品大改版后 | 新版本上线后4-8周 | 改版后新用户留存是否高于改版前？ | 对比改版前后批次的W1/W4列 |
| 整体留存出现异常波动 | 某天整体次留突降/突升>5pp | 是所有批次都在变，还是某一批拖了后腿？ | 拆开同期群看哪一批出了问题 |
| 季度/年度复盘 | 定期(每月/每季度) | 这半年的新用户质量是在变好还是变差？ | 纵向对比近6个月W1/W4列趋势 |

### 1.6 面试高频考点

**面试官常问**：
- "整体留存率没变，你能说产品没问题吗？"
- "怎么判断新用户质量是在变好还是变差？"
- "什么是辛普森悖论？在数据分析中怎么发现？"
- "同期群分析和漏斗分析有什么区别？各回答什么问题？"

**回答要点**：
- "不能只看整体——要用同期群纵向对比，看新批次的留存是否在恶化"
- "纵向对比同期群的W1/W2/W4列——如果新批次各周留存都高于旧批次，质量在变好"
- "辛普森悖论=各组内部没有变化但整体趋势变了，因为各组权重变了——同期群就是用来发现这个的"
- "漏斗看单步转化(一个时刻的快照)，同期群看长期趋势(多期追踪)——两者互补"

> **今天最重要的认知**：不看同期群的留存率=不看K线图看股价。纵向对比一列，你就能判断产品在变好还是变坏。同期群把"时间衰减"和"批次质量"两个维度拆开，让你看到趋势而非平均。

## 二、知识点详细拆解

### 2.1 同期群的两个分析维度

**为什么重要**：横向和纵向回答完全不同的问题，混淆了会得出完全错误的结论。

| 维度 | 看什么 | 怎么判断好坏 | 回答的问题 |
|------|--------|------------|----------|
| **横向（时间衰减）** | 同一批用户随时间推移的留存变化 | 曲线是否在某个时间点"拉平"？拉平高度=核心用户占比 | "产品有粘性吗？有一批固定用户吗？" |
| **纵向（批次对比）** | 不同批次在同一时间节点的表现差异 | 新批次比旧批次好→产品在改善；变差→需要警惕 | "产品在变好还是变差？" |

**举例**：W1列从40%降到28%→新用户质量在恶化。但如果只横向看1月（40%→28%→22%→18%），你觉得"衰减很正常"——你错过了"恶化"的信号。

### 2.2 同期群的三大核心洞察

**为什么重要**：这三个洞察直接指导产品和运营的实际决策，不是"分析完了就完了"。

1. **产品改进有没有用？** 3月做了产品优化→看3月新用户的留存曲线是否高于1-2月的新用户。如果3月W1/W2/W4都更高→改版有效。如果W1涨了但W4没涨→改版只改善了"第一印象"，没有改善长期粘性。

2. **渠道质量有没有下降？** 最近加大了投放→看新批次W1留存是否低于老批次。如果显著下降→花钱买垃圾流量→要限制渠道质量门槛，不能只看"CPA便宜"。

3. **季节性效应**：12月新用户留存可能比6月高——不是因为产品在变好，而是因为节日效应（用户有更多时间）。正确做法：跟去年同期比（同比），不是跟上个月比（环比）。

### 2.3 同期群的常见陷阱

**为什么重要**：同期群分析本身很简单，但如果不注意这些陷阱，你会得出错误的结论。

- **样本量太小**：某一周只有50个新用户→留存率波动可能±20%→没有统计意义。通常要求每个同期群至少100-200人。
- **只看一列**：只看"W1留存"忽略了长期留存曲线的拉平效应。一个产品W1可能只有30%，但如果W4-W12稳定在20%→说明有一批核心用户真的留下来了。
- **不看绝对数**：留存率涨了但新用户绝对量跌了90%→虽然"质量"在提升但"数量"在暴跌→整体DAU可能在大幅下降。
- **忽略权重变化（辛普森悖论）**：整体留存率没变但内部各组的留存率都变了——因为各组的人数权重也变了。这是最常见的"被平均数骗"的场景。

### 2.4 留存曲线拉平的三种形态

**为什么重要**：留存曲线的形状比任何一个单点的留存率都重要。形状决定了你的产品策略。

| 形态 | 曲线特征 | 含义 | 策略 |
|------|---------|------|------|
| 陡降+拉平在高位(>20%) | 前14天快速衰减，之后稳定 | 产品有筛选作用，留下了一批强需求用户 | 聚焦服务核心用户，按他们画像精准获客 |
| 陡降+永不拉平 | 一直衰减到趋近0 | 产品未满足用户需求，或目标用户不匹配 | 先搞清楚用户为什么走（用户调研），别拉新 |
| 缓降+拉平在高位(>25%) | 衰减缓慢，稳定在高比例 | 产品粘性强，有高频使用场景 | 可以加大拉新，重点是找到相似用户 |

### 2.5 同期群 vs 其他分析方法

| 方法 | 看什么 | 时间维度 | 对比维度 |
|------|--------|---------|--------|
| AARRR | 全链路转化 | 单期 | 不同环节 |
| 漏斗 | 单步骤转化率 | 单期 | 不同步骤 |
| 同期群 | 长期留存趋势 | 多期（横+纵） | 时间+批次 |
| PDS | DAU内部结构 | 单日 | New/Retained/Resurrected |

三者互补：AARRR看全局→漏斗找泄漏点→同期群看长期趋势→PDS做日常监控。

### 2.7 同期群的"纵向深度"分析——不止看W1列

**为什么重要**：W1、W2、W4、W8列恶化分别意味着完全不同的问题。只盯W1就像只看当天天气不看7天预报。

| 恶化的列 | 可能原因 | 诊断方向 | 例子 |
|---------|---------|--------|------|
| W1列恶化 | 获客质量下降/第一印象变差 | 检查渠道定向变化、注册引导流程、新用户落地页 | "投放从精准定向改为通投→W1从35%→22%" |
| W2列恶化但W1正常 | 激活SOP有问题/用户来了但没被引导 | 检查新手任务完成率、首次核心行为引导、推送策略 | "用户注册后48小时内没任何引导→W2从25%→15%" |
| W4列恶化 | 中期粘性问题/产品核心价值不够强 | 检查功能使用频次、内容新鲜度、社交关系链 | "首页推荐算法改了→内容重复率高→W4从20%→12%" |
| W8列恶化 | 长期留存问题/缺乏持续回来的理由 | 检查push策略、用户生命周期管理、老用户激励 | "长期用户没有等级/积分体系→W8从15%→8%" |

**实战案例**：某社交产品的同期群分析发现W1列稳定在38%，但W4列从22%持续降到14%→诊断方向不是"新用户引导"而是"产品中期粘性"——引导没问题，但用户使用2-3周后没有足够理由留下来。最终发现是内容Feed流的新鲜度下降太快。

**为什么必须看多列**：如果只看W1你会觉得"38%还不错"，会错过"14%的W4正在杀死你的DAU"。

### 2.8 同期群+渠道交叉——最精准的增长诊断工具

**为什么重要**：只看"同期群恶化"→只知道"用户质量在下降"但不知道为什么。加上渠道维度→能精确到"哪个渠道的哪批用户质量在变差"。

二维分组的诊断逻辑：
- 第一步：按渠道×同期群月份做交叉分组
- 第二步：对于每个(渠道, 月份)组合，计算W1留存率
- 第三步：锁定W1留存最低的(渠道, 月份)组合
- 第四步：针对该组合深入分析——该渠道的定向参数、投放素材、落地页最近是否发生了变化

示例矩阵：
```
            1月新用户  2月新用户  3月新用户  4月新用户  5月新用户  6月新用户
自然流量      42%      44%      43%      45%      44%      46%  ← 稳定且高
信息流投放    35%      32%      28%      25%      22%      19%  ← 持续恶化！
搜索投放      38%      37%      36%      36%      35%      35%  ← 小幅下降
社媒投放      30%      31%      33%      32%      34%      35%  ← 小幅改善
```

**最关键的诊断发现**：信息流渠道的W1留存从1月的35%持续降到6月的19%，而自然流量一直在42-46%之间稳定。结论：产品本身没问题（自然流量证明），信息流渠道的定向在持续松懈（可能是为了追求更低的CPA而不断放宽定向条件）。

**决策建议**：不是让产品团队"改善产品"，而是让投放团队收紧信息流定向——这不是产品问题，是渠道策略问题。如果误判为"产品需要改进"，产品团队迭代3个月也不会改善这个数字。

### 2.9 同期群的统计显著性——当样本量太小时

**为什么重要**：同期群每批人数可能只有50-200人。这么小的样本下，±20%的留存波动是纯随机的——你今天看到的"恶化"可能是运气不好，下周的"改善"也可能是侥幸。

**留存率的置信区间公式**（基于二项分布的正态近似）：

```
95% CI = p ± 1.96 × sqrt(p × (1-p) / n)

其中: p = 观察到的留存率, n = 同期群人数
```

**实际计算示例**：
- n=50人, 观察留存率p=40%: CI = [26.4%, 53.6%]——区间宽度27pp，几乎无法得出任何结论
- n=200人, 观察留存率p=40%: CI = [33.2%, 46.8%]——区间宽度约14pp，有一定参考价值
- n=1000人, 观察留存率p=40%: CI = [37.0%, 43.0%]——区间宽度约6pp，统计可靠

**判断标准**：
- 同期群人数<100人：留存率仅供参考，不做重要决策依据
- n=100-300人：可以判断大致趋势，但对于±5pp的变化需要谨慎解读
- n>1000人：可以进行精细的对比分析和严格决策

**当样本量不满足正态近似的条件时**（如np<5或n(1-p)<5），应该使用Bootstrap方法替代正态近似——这将在步骤4中实践。

### 2.10 同期群在财务报表中的应用——谁是最值钱的那批用户

**为什么重要**：同期群不只是"留存分析"的工具，它也是财务分析的核心框架——按"首次购买月份"分同期群，追踪每一批用户的累计消费金额，判断哪批用户最值钱。

**同期群累计GMV表**：
```
同期群(首购月)    M1累计GMV    M3累计GMV    M6累计GMV    M12累计GMV    用户人均贡献
   2024-01        50,000       120,000      180,000      220,000        220元
   2024-03        45,000       105,000      155,000        —              —
   2024-06        40,000        95,000         —           —              —
```

**关键洞察**：
- 1月用户在12个月内人均贡献220元，3月用户6个月人均155元——3月用户可能质量不如1月，但也可能是"还没到12个月"
- 如果"历史数据"显示M6到M12的增量通常只有约20%→3月用户的M12预测值=155×1.2=186元→仍然低于1月用户的220元
- 找到高质量用户所属的渠道和时段特征，把预算往那边倾斜

**公式**：
```
用户终身价值(LTV) = 同期群每用户累计贡献 = Σ(每个用户在时间t的付费) / 同期群总人数
同期群付费衰减率 = M(n+1)累计 / Mn累计 —— 用于预测未来的累计贡献
```

这与场景5的LTV预测直接关联——用历史同期群的衰减规律预测新批次的未来价值。

## 三、动手练习

### 步骤0：数据准备

用NumPy生成6个月×6个时间节点的模拟留存数据。

````python
import numpy as np
np.random.seed(42)

cohorts = ["1月", "2月", "3月", "4月", "5月", "6月"]
weeks = ["W0", "W1", "W2", "W3", "W4", "W8"]
# 典型衰减曲线：W0=100%, W1=40%, W2=28%, W3=22%, W4=18%, W8=12%
base_retention = np.array([1.0, 0.40, 0.28, 0.22, 0.18, 0.12])
# 产品在持续优化——新批次质量逐步提升（1.0→1.20）
quality_factor = np.array([1.0, 1.05, 1.10, 1.12, 1.15, 1.20])

print("数据准备完成！")
print(f"同期群: {cohorts}")
print(f"观察节点: {weeks}")
print(f"基础留存曲线: {base_retention}")
print(f"质量系数(新产品在改善): {quality_factor}")

# 【预期输出】
# 数据准备完成！
# 同期群: ['1月', '2月', '3月', '4月', '5月', '6月']
# 观察节点: ['W0', 'W1', 'W2', 'W3', 'W4', 'W8']
# 基础留存曲线: [1. 0.4 0.28 0.22 0.18 0.12]
# 质量系数(新产品在改善): [1. 1.05 1.1 1.12 1.15 1.2]
````

### 步骤1：构建同期群留存矩阵（基础）

````python
retention_matrix = np.zeros((len(cohorts), len(weeks)))
for i in range(len(cohorts)):
    for j in range(len(weeks)):
        if j <= len(cohorts) - i - 1:  # 三角矩阵约束
            r = min(1.0, base_retention[j] * quality_factor[i] * np.random.uniform(0.95, 1.05))
            retention_matrix[i, j] = r
        else:
            retention_matrix[i, j] = np.nan  # 未来数据不可得

print("同期群留存矩阵（纵向对比看列！）")
print("=" * 70)
header = f"{'同期群':<10}"
for w in weeks:
    header += f"{w:>10}"
print(header)
print("-" * 70)
for i, cohort in enumerate(cohorts):
    row = f"{cohort:<10}"
    for j in range(len(weeks)):
        if not np.isnan(retention_matrix[i, j]):
            row += f"{retention_matrix[i, j]:>9.1%}"
        else:
            row += f"{'':>10}"
    print(row)

print()
print("【横向分析】1月新用户的留存衰减:")
for j, w in enumerate(weeks):
    print(f"  {w}: {retention_matrix[0, j]:.1%}", end="")
print("\n  → 从100%衰减到W8≈14%，但W4-W8趋于稳定（有核心用户留存）")

print()
print("【纵向分析】W1列——所有批次过了一周的留存率:")
for i, cohort in enumerate(cohorts):
    print(f"  {cohort}: {retention_matrix[i, 1]:.1%}")
print(f"  → W1留存从{retention_matrix[0, 1]:.1%}涨到{retention_matrix[-1, 1]:.1%}——产品在持续改善！")

# 【预期输出】
# 6×6三角矩阵
# 横向1月: W0=100% → W1≈42% → W4≈22% → W8≈14%
# 纵向W1: 从42%逐步涨到48%——产品改善明显
````

### 步骤2：对比两种同期群场景——产品改善 vs 渠道恶化（进阶）

````python
print("场景A：产品在改善（改版后新批次留存越来越好）")
print("-" * 55)
for i, c in enumerate(["1月(改版前)", "3月(改版后1月)", "6月(改版后4月)"]):
    curve = base_retention[:4] * (1.0 + i * 0.10)
    print(f"  {c}: W0={curve[0]:.0%}  W1={curve[1]:.0%}  W2={curve[2]:.0%}  W3={curve[3]:.0%}")
print("  → W1列: 40%→44%→48%，产品持续改善")
print("  → 纵向对比确认：改版确实提升了用户粘性")

print()
print("场景B：渠道质量在恶化（投放扩大，低质用户占比越来越高）")
print("-" * 55)
for i, c in enumerate(["1月(精准投放)", "3月(放宽定向)", "6月(通投)"]):
    curve = base_retention[:4] * (1.0 - i * 0.12)
    print(f"  {c}: W0={curve[0]:.0%}  W1={curve[1]:.0%}  W2={curve[2]:.0%}  W3={curve[3]:.0%}")
print("  → W1列: 40%→35%→30%，渠道质量持续恶化")
print("  → 纵向对比确认：投放策略变化导致用户质量下降")

print()
print("=" * 55)
print("关键启示：")
print("  如果1月新用户100人、6月新用户1000人——场景B的整体平均留存")
print("  可能看起来和之前一样（因为6月权重大了但留存低了）。")
print("  → 不拆同期群，你完全看不到恶化趋势。")

# 【预期输出】
# 场景A: W1列 40%→44%→48% 改善
# 场景B: W1列 40%→35%→30% 恶化
# 但如果权重变了，整体平均可能看起来一样
````

### 步骤3：同期群+渠道交叉分析——定位最差的(渠道, 同期群)组合（进阶）

这是本章最重要的实战练习。当整体留存下降时，你需要知道"是哪个渠道的哪批用户在出问题"——交叉分析给出最精确的答案。

用NumPy构建"渠道×同期群"二维留存矩阵，纵向对比每个渠道的质量趋势。

````python
# 步骤3：同期群+渠道交叉分析——定位最差的(渠道, 同期群)组合

import numpy as np
np.random.seed(123)

# 定义渠道和同期群
channels = ["自然流量", "信息流投放", "搜索投放", "社媒投放", "达人推荐"]
cohorts_cross = ["1月", "2月", "3月", "4月", "5月", "6月"]

# 每个渠道的基础W1质量系数（自然流量最高，信息流投放最差）
channel_base = np.array([1.00, 0.65, 0.85, 0.72, 0.88])
# 每个渠道的月度变化趋势（<1.0=恶化, >1.0=改善）
channel_trend = np.array([1.00, 0.90, 0.97, 1.03, 1.05])

# 构造渠道×同期群 W1留存矩阵
base_w1 = 0.40
w1_by_channel_cohort = np.zeros((len(channels), len(cohorts_cross)))
for ch in range(len(channels)):
    for co in range(len(cohorts_cross)):
        base_val = base_w1 * channel_base[ch] * (channel_trend[ch] ** co)
        noise = np.random.normal(0, 0.015)
        w1_by_channel_cohort[ch, co] = np.clip(base_val + noise, 0.05, 1.0)

print("渠道×同期群 W1留存矩阵")
print("=" * 75)
header = f"{'渠道':<12}"
for co in cohorts_cross:
    header += f"{co:>10}"
print(header)
print("-" * 75)
for ch_idx, ch in enumerate(channels):
    row = f"{ch:<12}"
    for co_idx in range(len(cohorts_cross)):
        row += f"{w1_by_channel_cohort[ch_idx, co_idx]:>9.1%}"
    print(row)

print()
print("【关键发现】")
# 找出最差的(渠道, 同期群)组合
min_idx = np.unravel_index(np.argmin(w1_by_channel_cohort), w1_by_channel_cohort.shape)
print(f"  最差组合: {channels[min_idx[0]]} × {cohorts_cross[min_idx[1]]} (W1={w1_by_channel_cohort[min_idx]:.1%})")
max_idx = np.unravel_index(np.argmax(w1_by_channel_cohort), w1_by_channel_cohort.shape)
print(f"  最佳组合: {channels[max_idx[0]]} × {cohorts_cross[max_idx[1]]} (W1={w1_by_channel_cohort[max_idx]:.1%})")
print(f"  质量差距: {w1_by_channel_cohort[max_idx]:.1%} - {w1_by_channel_cohort[min_idx]:.1%} = {(w1_by_channel_cohort[max_idx]-w1_by_channel_cohort[min_idx]):.1%} (最佳是最差的{w1_by_channel_cohort[max_idx]/w1_by_channel_cohort[min_idx]:.1f}倍!)")

print()
print("【趋势诊断】各渠道W1留存的变化方向:")
for ch_idx, ch in enumerate(channels):
    first_val = w1_by_channel_cohort[ch_idx, 0]
    last_val = w1_by_channel_cohort[ch_idx, -1]
    delta = last_val - first_val
    direction = "↑改善" if delta > 0.01 else ("↓恶化" if delta < -0.01 else "→持平")
    print(f"  {ch}: {first_val:.1%} → {last_val:.1%} ({delta:+.1%}) {direction}")

print()
print("【行动建议】")
# 找到恶化最严重的渠道
trends = w1_by_channel_cohort[:, -1] - w1_by_channel_cohort[:, 0]
worst_ch = np.argmin(trends)
best_ch = np.argmax(trends)
print(f"  恶化最严重: {channels[worst_ch]}（变化幅度: {trends[worst_ch]:+.1%}）")
print(f"  → 建议: 检查该渠道的定向参数、投放素材和落地页是否近期发生了变化")
print(f"  改善最明显: {channels[best_ch]}（变化幅度: {trends[best_ch]:+.1%}）")
print(f"  → 自然流量W1始终稳定→产品本身没问题，是投放策略需要调整")
print(f"  → 核心启示: 做渠道×同期群交叉分析，避免让产品团队背渠道的锅")

# 【预期输出】
# 渠道×同期群矩阵显示信息流投放持续恶化，自然流量稳定
# 精准定位到具体渠道+月份组合的问题
# 最佳组合W1可能是最差组合的2-3倍
````

### 步骤4：挑战练习——Bootstrap方法评估小样本同期群留存率的置信区间

当同期群人数较少时（如n<100），单点留存率的波动可能是纯随机的。Bootstrap方法（自助法）通过"有放回重采样"来估计留存率的经验分布，从而计算置信区间——不依赖正态近似的假设。

这个练习也是面试中的加分项：面试官问你"样本量50人，留存率从40%涨到55%，能说改版有效吗？"——你用Bootstrap回答。

````python
# 步骤4：挑战练习——Bootstrap方法评估小样本同期群留存率的置信区间

import numpy as np
np.random.seed(456)

def bootstrap_retention_ci(retained, total, n_bootstrap=10000, ci=0.95):
    """
    用Bootstrap方法估计留存率的置信区间
    
    参数:
        retained: 留存用户数
        total: 总用户数
        n_bootstrap: bootstrap重采样次数(默认10000)
        ci: 置信水平(默认95%)
    
    返回:
        (lower_bound, observed_rate, upper_bound)
    """
    # 构造原始样本: 1=留存, 0=未留存
    observed = np.array([1] * retained + [0] * (total - retained))
    observed_rate = retained / total
    
    # Bootstrap重采样
    bootstrap_rates = np.zeros(n_bootstrap)
    for i in range(n_bootstrap):
        sample = np.random.choice(observed, size=total, replace=True)
        bootstrap_rates[i] = sample.mean()
    
    # 计算置信区间
    alpha = (1 - ci) / 2
    lower = np.percentile(bootstrap_rates, alpha * 100)
    upper = np.percentile(bootstrap_rates, (1 - alpha) * 100)
    
    return lower, observed_rate, upper


print("=" * 75)
print("小样本同期群的留存率——是真实改善还是随机波动?")
print("=" * 75)

# 模拟不同样本量下的置信区间
test_cases = [
    ("大样本(1000人)", 1000, 400),
    ("中样本(200人)",  200,   84),
    ("小样本(80人)",    80,   38),
    ("微样本(35人)",    35,   22),
]

print()
print(f"{'场景':<20} {'n':<8} {'留存数':<8} {'观察率':<10} {'95%CI下限':<12} {'95%CI上限':<12} {'区间宽度':<10} {'结论'}")
print("-" * 95)
for label, total, retained in test_cases:
    lower, obs, upper = bootstrap_retention_ci(retained, total)
    width = upper - lower
    if width > 0.30:
        conclusion = "波动太大,不做决策依据"
    elif width > 0.15:
        conclusion = "可参考方向,需积累样本"
    else:
        conclusion = "统计可靠,可以决策"
    print(f"{label:<20} {total:<8} {retained:<8} {obs:<10.1%} {lower:<12.1%} {upper:<12.1%} {width:<10.0%} {conclusion}")

print()
print("=" * 75)
print("【实际案例】产品改版前后W1留存对比(Bootstrap法)")
print("=" * 75)

# 改版前: 3月80人, W1留存38人 → 47.5%
# 改版后: 4月55人, W1留存35人 → 63.6%
lower_before, obs_before, upper_before = bootstrap_retention_ci(38, 80)
print(f"\n改版前(3月): n=80, 留存38人, 观察率={obs_before:.1%}")
print(f"  95% Bootstrap CI: [{lower_before:.1%}, {upper_before:.1%}] (宽度={upper_before-lower_before:.0%})")

lower_after, obs_after, upper_after = bootstrap_retention_ci(35, 55)
print(f"\n改版后(4月): n=55, 留存35人, 观察率={obs_after:.1%}")
print(f"  95% Bootstrap CI: [{lower_after:.1%}, {upper_after:.1%}] (宽度={upper_after-lower_after:.0%})")

print()
print("【统计判断】")
if upper_before < lower_after:
    print(f"  改版前CI上限({upper_before:.1%}) < 改版后CI下限({lower_after:.1%})")
    print(f"  → CI完全不重叠! 改版效果统计显著, 留存率确实提升了。")
elif lower_before > upper_after:
    print(f"  改版前CI下限({lower_before:.1%}) > 改版后CI上限({upper_after:.1%})")
    print(f"  → 改版后显著变差!")
else:
    overlap_start = max(lower_before, lower_after)
    overlap_end = min(upper_before, upper_after)
    if overlap_end > overlap_start:
        print(f"  两个CI有重叠区域≈{overlap_end-overlap_start:.0%}")
    if obs_after > obs_before:
        print(f"  改版后点估计({obs_after:.1%})高于改版前({obs_before:.1%})")
        print(f"  → 趋势向好但样本量不够(两个CI有重叠), 建议再观察1-2个月积累更多数据再下结论")
    else:
        print(f"  → 趋势不明显, 不建议下结论")

print()
print("【面试话术】")
print("  '小样本同期群(n<100)的留存率波动±20%是统计正常的。'")
print("  '我会用Bootstrap方法计算95%置信区间——如果'")
print("  '两个同期群的CI完全不重叠，说明差异是显著的。'")
print("  '如果重叠但趋势一致，我会建议继续观察，'")
print("  '积累更多样本后再下结论。'")

# 【预期输出】
# 微样本(35人)的CI宽度可能达到40%以上——几乎无法判断
# 大样本(1000人)的CI宽度约6%——可以精细对比
# 实际案例中两个CI有一定重叠但改版后点估计更高——建议继续观察
````

## 四、业务场景应用

### 场景1：CEO说"我们留存没变"，但同期群说在恶化

**问题**：公司每周发一份"上周新用户次留"报告，数字一直在22%左右，看起来很稳定。CEO说"留存还行，继续保持投放策略"。

**分析**：你做了同期群分析，按注册月份拆开看每个月的次留：
- 1月份新用户（自然流量为主，1000人）：次留35%
- 6月份新用户（投放占比80%，5000人）：次留15%
- 整体22% = (1000×35% + 5000×15%) / 6000 = 没错，确实22%

**真相**：产品的用户体验没有变差——老渠道（自然流量）的用户次留依然是35%。但投放策略变了——从精准投放变成了通投，大量买入低质量流量（次留15%）。整体22%看起来没变，是因为低质用户的数量（权重）暴增了。这就像把一杯热水和一杯冷水倒在一起——温度刚好，但水量已经不是原来的比例了。

**量化分析**：假设1月自然用户LTV=200元、投放用户LTV=60元；6月自然用户LTV=210元(产品略好)、投放用户LTV=40元(质量更差)。整体平均LTV从(1000×200+200×60)/1200=177元降到(1000×210+5000×40)/6000=68元——暴跌了62%！而整体次留只从22%微降到20%——两个信号完全背道而驰。

**决策**：
1. 限制投放渠道的质量门槛——任何渠道的W1留存低于20%立即停投
2. 建立"按渠道+按同期群"的留存监控看板，不再只看一个整体平均数
3. 对已投的低质渠道用户做补救分析——有没有可能通过激活引导提升这批用户的留存？

**不用同期群的后果**：只看整体留存22%"很稳定"→继续加大低质渠道投放→低质用户权重从50%涨到80%→整体留存被持续拉低到18%→发现时已经晚了，投放预算也花光了。

### 场景2：产品改版后的效果评估——同期群比你想象的更精确

**问题**：3月份对产品做了一次大改版（重新设计了新用户引导流程）。产品经理说"改版后DAU涨了，改版成功"。但你知道DAU涨了可能是因为其他原因（自然增长、加大投放、季节性的开学季）。

**分析**：用同期群来剥离其他因素——只看"3月及之后注册的新用户"，对比"1-2月注册的新用户"：
- 如果3月新用户的W1/W2/W4留存都高于1-2月→改版确实提升了用户粘性
- 如果W1涨了(+8pp)但W4没涨(+1pp)→改版改善了"第一印象"但没改善长期粘性（可能需要进一步优化）
- 如果全部没变→改版没效果，DAU涨是其他因素导致的（自然增长或加大投放）

实际数据：W1从38%→45%(+7pp)，W2从25%→32%(+7pp)，W4从18%→23%(+5pp)。结论：改版全面有效，短期和中期留存都有提升。W4提升幅度略小但仍然是正的——说明长期粘性还需要持续优化。

**量化影响测算**：假设每月新增5万用户，W1留存+7pp意味着每月多了3500个用户在Day7仍然活跃；W4留存+5pp意味着每月多了2500个用户在Day30仍然活跃。按每个留存用户日均贡献0.5元的广告收入算→改版每月带来约(3500+2500)×0.5×30/2≈45,000元增量收入。

**决策**：全量上线改版，并对W4/W8进行持续监控以评估是否需要进一步优化长期留存。

**不用同期群的后果**：只看"改版后DAU涨了"→无法判断是改版效果还是季节性和自然增长→可能把无效改版当成宝（浪费研发资源继续迭代），或把有效改版的效果归因到"年末自然增长"（错失推广良机）。

### 场景3：12月用户留存飙升——别高兴，可能是"圣诞老人效应"

**问题**：12月的新用户W1留存突然从平时的25%飙升到38%。运营团队开心地说"我们的年末活动太成功了"。但作为分析师，你需要保持怀疑——12月本身就是一个特殊的月份。

**分析**：做同期群同比对比——不能只跟11月比（12月天然的节假日会让数字好看），要跟去年12月对比：
- 今年12月新用户W1留存：38%
- 去年12月新用户W1留存：35%（去年12月也有"异常"涨幅）
- 今年6月vs去年6月：25% vs 24%（基本持平，非节日月没有"异常"）
- 结论：12月的飙升中，只有3pp（38%-35%）可能是活动或产品改善带来的，其余约10pp（35%-25%）来自每年都有的节日效应。

**更严谨的做法**：计算"季节性指数"——每个月的留存÷年均留存。如果12月的指数连续3年都>1.3，这确实是规律而非偶然。
- 假设三年的12月季节性指数分别为1.38、1.35、1.42→均值1.38，标准差0.035→规律稳定
- 今年的活动额外贡献=实际值-季节性预测值=38%-(25%×1.38)=38%-34.5%=3.5pp→活动确实创造了一些增量，但远没有运营说的"+13pp"那么夸张

**决策**：活动确实有一定效果（+3pp），但不要因为12月的38%就盲目乐观、制定激进的Q1目标。建议用"经季节性调整的留存"作为评估标准。

**不用同期群的后果**：把季节性当成产品和活动改善→年初基于12月的数据制定激进目标→1月数据回落到25%→CEO质问"为什么跌了这么多？你们做了什么？"→你说"12月本来就会涨..."但已经失去了信任。

### 场景4：产品大改版后——W1涨了但W4跌了，这意味着什么？

**问题**：产品做了一次大改版（重新设计了首页和新手引导流程）。改版后数据出来了，但信号是矛盾的：
- W1留存从30%涨到38%（+8pp）——改版明显改善了第一印象
- W4留存却从20%跌到18%（-2pp）——长期粘性反而下降了

产品经理说"W1涨了8个点，改版大获成功！"但你知道W4在跌——这里藏着更深的真相。

**诊断思路**：
1. W1涨说明"新用户引导"和"首页体验"确实变好了——用户第一次来的感觉更好了
2. W4跌说明"长期粘性"变差了——用户在第4周回来的比例反而降低了
3. 这可能的原因：
   - 新首页虽然好看，但推荐算法被动了——内容的新鲜度和个性化程度下降了（用户2-3周后发现推送的都是重复内容）
   - 新引导流程虽然顺畅，但"过度承诺"了——用户带着过高期望进来，2-3周后发现产品满足不了，反而走了
   - 改版引入了新的"痒点"——比如加载速度变慢了、某个常用功能被挪到了三级菜单

**分析方法**：
- 做同期群的"纵向深度"分析（2.7节）：把多个时间节点(W1/W2/W4/W8)都拉出来对比
- 同时拆开看改版后不同渠道的表现——如果自然流量用户W4也跌了，说明是产品问题；如果只有投放渠道W4跌了，说明渠道质量在变差
- 做用户访谈或问卷——问"第3-4周流失的用户，为什么不再来了？"

**决策**：
1. W1的改善值得肯定——新手引导的优化保留
2. 对W4的下降必须重视——进一步排查推荐算法和内容新鲜度的变化
3. 下次改版要同时监控W1和W4——W1是"第一印象"，W4是"长期关系"，两者缺一不可

**不用同期群的后果**：如果只看整体"改版后DAU涨了"→"改版成功"→继续在这个方向上迭代→W4持续下降到10%→3个月后DAU反而比改版前更低，但已经不知道是哪个改动导致的了。

### 场景5：用同期群数据做新用户LTV预测——"这批用户值多少钱？"

**问题**：财务VP问你："这个月新增了2万用户，他们未来6个月能贡献多少收入？"你需要用一个有数据支撑的方法回答，而不是拍脑袋。

**分析方法——同期群LTV预测三步法**：

**第一步：拉历史同期群的"留存×付费"双曲线**
- 过去12个月，每个月的同期群数据：W1/W2/W4/W8/W12/W26留存率 + 各周人均付费金额
- 关键发现：虽然不同批次用户的留存率有差异，但"留存用户的付费率"大致稳定——40%的活跃用户在Week1内会付费，且后续每周约8-12%的活跃用户有付费行为

**第二步：用历史数据计算"每新增用户的预期累计收入曲线"**
```
累计LTV(第t周) = Σ[留存率(第i周) × 活跃用户付费率 × 人均付费金额]
                   i=1到t

例子（基于历史均值）：
Week1: 40%留存 × 40%付费率 × 15元 = 2.4元/新增用户
Week2: 28%留存 × 10%付费率 × 12元 = 0.34元/新增用户
Week4: 20%留存 × 9%付费率 × 20元 = 0.36元/新增用户
Week8: 15%留存 × 8%付费率 × 25元 = 0.30元/新增用户
...
26周累计LTV ≈ 8.2元/新增用户
```

**第三步：用新用户的实际W1留存来修正LTV预测**
- 如果这批新用户的W1留存是35%（低于历史均值40%），那么按比例下调：
  - 预测26周LTV = 8.2 × (35%/40%) = 7.2元/新增用户
- 2万新用户 × 7.2元 = 14.4万元（预测6个月累计收入）

**关键前提和限制**：
- 这个方法假设"留存曲线的形状大致稳定"——即W4/W1的比值不变。如果产品改版改变了曲线的形状（比如W1跌但W4涨），需要重新校准
- 付费率和客单价可能随季节变化（大促期间付费率翻倍）——需要用同比数据调整
- 这不是精确预测（±15-20%的误差是正常的），但比"凭感觉拍一个数"强100倍

**面试价值**："你做过LTV预测吗？"→把上面三步讲出来，面试官就知道你真的理解同期群不只是看留存，而是可以直接指导财务决策。

## 五、分级检验清单

### 初级检验（必须能独立完成的）
- [ ] 能解释什么是同期群、为什么需要它（平均数掩盖趋势）
- [ ] 能看懂同期群留存矩阵（横向=时间衰减，纵向=批次对比）
- [ ] 能区分"横向看"和"纵向看"各自回答什么业务问题
- [ ] 能列举同期群的四种定义方式（按注册/按首购/按渠道/按行为）
**答案提示**：横向看=同一批用户随时间衰减（"产品有粘性吗？"）；纵向看=不同批次在同一节点的质量对比（"产品在变好还是变差？"）。四种定义方式取决于你把什么作为"进入产品"的标志。

### 中级检验（能解释给同事听的原理）
- [ ] 能用代码构建同期群表格并判断趋势（纵向变好还是变差）
- [ ] 能通过同期群判断"产品在变好还是变差"
- [ ] 能识别"辛普森悖论"（整体不变但内部结构巨变）
- [ ] 能用Bootstrap方法评估小样本留存率的置信区间
- [ ] 能解释W1/W2/W4/W8列恶化分别意味着什么不同的问题
**答案提示**：产品变好=新批次各周留存都高于旧批次；辛普森悖论=各组留存变了但权重也变了导致整体看起来没变——用同期群纵向对比发现。W1恶化→获客/第一印象；W2恶化→激活SOP；W4恶化→中期粘性；W8恶化→长期留存。

### 高级检验（能举一反三的迁移能力）
- [ ] 能结合"用户量"和"留存率"两个维度判断增长健康度（光留存涨但新增暴跌也不行）
- [ ] 能设计同期群的多维度分组（按渠道×行为×属性交叉），找到"什么样的用户留存好"
- [ ] 能用同期群结果说服管理层："我们不是产品不好，是渠道策略问题"或"我们不是渠道问题，是激活体验需要优化"
- [ ] 能用同期群数据做新用户LTV预测——用历史留存曲线+新批次W1留存修正预测值
- [ ] 能判断"W1涨但W4跌"这种矛盾信号的真正含义，不会只看一个数字就下结论
**答案提示**：增长健康度=新增量↑+留存↑=真增长；新增量↑+留存↓=买来的增长。多维度分组同期群=渠道×时段×行为特征交叉分析。LTV预测=历史留存曲线 × 付费率 × 客单价，再用新批次W1做比例修正。W1涨但W4跌=第一印象好了但长期粘性差了——不能只看一个时间节点。

### 横向连接：与其他分析方法的关联

**与漏斗的关系**：漏斗看"单步转化"，同期群看"多期留存"。一个用户可能漏斗走到最后一步（支付成功），但30天后流失——这只有同期群能看到。两者结合：漏斗告诉你"哪一步泄漏最严重"，同期群告诉你"泄漏在变好还是变差"。

**与PDS的关系(W2D6)**：PDS里的Retained（留存用户）是一个总量。同期群告诉你这个总量由"哪些批次"构成——是1月的高质量老用户还在撑，还是新批次也在持续贡献？如果Retained总量靠老用户在撑，而新批次留存很差——DAU很快会崩塌。

**与RFM的关系(W2D4)**：RFM的R（最近一次消费时间）和F（消费频率）本质上可以用同期群的思想来分析——"最近一次消费"是纵向对比（不同批次用户的活跃度），"消费频率"是横向趋势（同一批用户随时间的消费行为变化）。两者的交叉可以回答"哪个月的付费用户复购率最高"。

**与AB实验的关系(W7-W8)**：同期群告诉你"新用户质量在恶化"，AB实验帮你测试"改版能不能改善它"。两者是"发现问题→验证方案"的关系。如果不用同期群先发现问题就直接做AB实验——你可能在测试一个错误的方向。

**与财务报表的关系**：同期群按首购月份分组追踪累计GMV，本质上是"收入确认的同期群视角"。SaaS公司的NDR(Net Dollar Retention)指标就是用同期群计算的——"去年获得的那批客户，今年贡献的收入是去年的多少%？"

**与用户增长模型的关系**：增长模型（如Logistic增长）需要输入留存率参数。如果你输入的是"整体平均留存"而非"分期群留存"，模型会高估增长——因为整体平均留存被老用户的高留存"美化"了。

## 七、当日小结

> **今天最重要的一个认知**：不拆同期群看留存=不看K线图看股价。同期群把"时间衰减"和"批次质量"两个维度拆开，让你分别看到：(1)产品的自然留存规律是怎样的（横向），(2)新批次用户是在变好还是变差（纵向）。两者缺一不可。更进一步——把渠道也加进来做交叉分析，你能精确到"哪个渠道的哪批用户在恶化"，这才是增长分析师的真正价值。

**三条金句带回家**：
1. "整体留存25%"可能掩盖了"1月35%→6月15%"的持续恶化——同期群让隐藏的趋势现形
2. 横向看一行=看衰减规律（产品粘性），纵向看一列=看质量变化（产品改善还是恶化）
3. 不要把"季节性"当成"产品改善"——跟去年同期比（同比），别跟上个月比（环比）

### 面试准备：同期群分析三连问

**Q1: "如何判断一个产品是在变好还是变差？"**
> 标准答案：用同期群的纵向对比——看最近几批新用户的W1/W2/W4留存是否持续高于更早的批次。如果是，产品在改善。同时要排除季节性因素——跟去年同期对比而非跟上个月对比。如果不同时间节点的变化方向不一致（如W1涨但W4跌），要进一步分析改版的具体改动对短期和长期留存的不同影响。

**Q2: "整体留存没变但你说产品在恶化——怎么证明？"**
> 标准答案：拆同期群。1月新用户留存35%（1000人），6月新用户留存15%（5000人），整体平均22%。整体没变是因为6月低质用户的权重暴增了。但如果你看自然流量渠道（权重没变）——W1留存从42%稳定在43%，产品本身没变。恶化的是投放渠道——从精准投放变成通投，拉来了大量低质用户。所以不是产品问题，是渠道策略问题——这是交叉分析的价值。

**Q3: "小样本同期群的留存率涨了10pp——能说产品变好了吗？"**
> 标准答案：不能直接下结论。样本量50人时，留存率波动±20%是正常的。我会用Bootstrap方法计算95%置信区间——如果改版前后的CI不重叠，说明变化统计显著。如果CI有重叠但趋势向好，建议继续观察1-2个月积累更多数据。另外要检查其他同期群的趋势是否一致——如果只有一个批次"表现好"而其他批次没有，很可能是随机波动。

### 课后拓展

1. **练习**：如果你有你们产品的数据权限，拉近6个月新用户的同期群数据。纵向对比W1列——在变好还是变差？如果不看同期群，只看整体留存——你会被平均数骗吗？进阶：按渠道拆开看，有没有某个渠道在恶化但被其他渠道"平均"掉了？

2. **思考**：如果你发现W1列在变差但W4列在变好——这意味着什么？（提示：新用户激活门槛变高了，但留下来的用户质量比之前更高——这可能不是坏事，而是产品"筛选"能力变强了。但要注意新增用户的绝对数——如果门槛提高导致新增暴跌，整体DAU可能会出问题。）

3. **思考**：假设你是电商App的分析师，你发现"1月首购用户"的M6累计GMV是220元，"6月首购用户"的M1累计GMV是40元——6月的用户更差吗？不一定——他们还没有6个月的消费时间。用历史同期群的"累计GMV增长曲线"来预测6月用户的M6值，应该怎么做？（提示：计算历史同期群M1→M6的累计增长倍数，用这个倍数推断新同期群的未来值。）

4. **预习**：明天要学的漏斗分析——同期群看"长期留存"，漏斗看"单步转化"，两者结合才能完整诊断一个产品的增长健康度。面试中经常被问到"漏斗分析和同期群分析有什么区别？各在什么场景下使用？"——明天会给出完整答案。
