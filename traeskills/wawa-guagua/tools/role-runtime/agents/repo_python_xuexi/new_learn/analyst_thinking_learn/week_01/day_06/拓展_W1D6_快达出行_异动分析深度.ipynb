# 拓展：异动分析的里子——从「发现异常」到「讲清原因」

> **本拓展讲什么**：主课教你异动分析三步法（定位→拆解→解释）。本拓展教你**每一步里面最容易踩的坑**——定位不准就开工、拆开发现是虚惊一场、找到了原因但说不清「所以呢」。

# 第三部分：概念网络展开

## 🔴 核心深化：定位——不画边界就开工，是异动分析最常见的错误

> 主课说「定位」是第一步。但怎么定位？不是「哦完单量跌了 22%」就冲进去拆——你得先画清楚四个边界。少画一个，后面全白做。

### 🔴 认知冲突

早上 9:05，CEO 说「完单量跌了 22%」。你的第一反应是什么？

**分析师 A**：直接打开 CMOP 看板，按城市下钻——「成都跌了 60%，合肥跌了 5%，苏州涨了 3%……成都是主因！」——然后开始查成都发生了什么。

**分析师 B**：先停下来，问了四个问题——
1. 「比昨天跌了 22%」——是跟上周同一天比（环比），还是跟昨天比（日环比）？因为昨天是周日，今天是周一——周日的完单量本来就比周一高。跟上周一比，只跌了 3%。
2. 「这个 22% 的跌幅正常吗？」过去三个月，日环比的波动范围是 ±15%。22% 确实超出了——但这个 22% 里，有多少是因为「周日→周一」的正常回落？
3. 「是所有城市都在跌，还是个别城市？」成都跌了 60%，但成都昨天下暴雨——这种天气事件导致的完单量骤降，不是你产品的问题。
4. 「数据本身有没有问题？」成都的完单量暴跌 60%——但成都的「打开 App 次数」没跌。下单量正常，完单量暴跌→不是没人打车，是打了车没完成——可能是成都的支付系统出了故障。

**问题**：分析师 B 比 A 多花了 3 分钟在「定位」上。这 3 分钟帮他省掉了后面可能浪费的 30 分钟——因为如果他不做第 4 个检查，他会花半小时分析「成都用户为什么突然不打车了」——结果根本不是用户的问题，是支付系统挂了。

> 定位不是「看一眼数字」——是「画四个边界，确认你真的在追一个值得追的问题」。

### [必修] 定位的四边界检查

**① 一句话白话**：看到异常数字，别急——先画四道线把它框住。框好了，你才知道你在追什么。框歪了，后面全是白费。

**② 四边界**：

| 边界 | 问什么 | 快达「完单量跌 22%」的检查 | 这一步排除了什么 |
|------|--------|--------------------------|---------------|
| **时间边界** | 「这个异常是从什么时候开始的？是突然跌还是慢慢跌？跟什么比跌的——环比？同比？还是预期？」 | 跟昨天比跌 22% → 但昨天是周日、今天是周一。周日的完单量本来就比周一高约 15%。跟上周一比→只跌了 4%。真正异常的其实是「周日→周一的跌幅比平时大了一倍」——平时跌约 15%，今天跌了 22%，多跌了 7 个点 | 排除了「周期波动」——22% 里有 15% 是正常的周日→周一回落 |
| **空间边界** | 「是所有城市/所有品类/所有用户群都在跌，还是局部的？」 | 全国跌 22% → 但拆开：成都跌 60%、合肥跌 5%、苏州涨 3%。异常集中在成都。其他城市正常。→ 不是「快达的产品出了问题」，是「成都出了什么事」 | 排除了「全公司级别的问题」——问题范围锁定在成都 |
| **口径边界** | 「这个数字本身有没有问题？是不是口径被改了？是不是埋点坏了？是不是数据延迟了？」 | 成都完单量暴跌 60%，但成都的「打开 App 次数」没跌、下单量正常。→ 用户正常打开 App、正常下单——但在「完单」这个环节断了。可能原因：支付系统故障、司机端 App 推送失败导致司机没看到新订单 | 排除了「数据口径问题」和「用户需求下降」——问题锁定在「下单→完单」的转化环节 |
| **幅度边界** | 「跌 22% 在历史上是什么水平？是每周都会有一次的波动，还是过去三年只出现过两次的极端事件？」 | 过去三个月，日环比波动范围 ±15%。今天的 22%（剔除周日效应后约 7%）在正常范围内？7% 刚好卡在边界线上——过去三个月有 8% 的日子日环比超过 ±7%。→ 不算极端，但值得关注 | 排除了「需要拉响全公司警报的紧急程度」——不是「产品崩了」，是「需要查一下但不至于恐慌」 |

**③ 关键认知**：90% 的「异常」在时间边界和口径边界这两步就被排除了——不是数据错了，就是正常的周期波动，要不就是已知的外部事件（天气/节假日/竞品动作）。剩下的 10%，才是真正需要拆解和解释的。**定位的目的不是「确认有异常」，而是「确认这个异常值得你花时间去追」。**

| 幅度 | 38% 在什么水平？过去一年的最低点是多少？ | 过去一年最低点是 49%（春节期间，司机回乡）。38% 是历史最低——但原因是已知的（大促拉新），不是未知的恶化 | **不是紧急事件**——不需要 9:30 会汇报。但需要在周五的周报里写清楚：「6/21 大促拉新导致留存率技术性下跌约 15pp，剔除大促新司机后留存率为 56%（+1pp）。结论：存量司机留存稳定，大促新司机需要激活方案」 |

## 概念群 A：[防护] 虚惊一场——看起来是异常，其实不是

> **与核心概念的关系**：定位的四个边界检查做完后，你可能会发现——根本不需要拆解。因为这个「异常」属于三种**不需要追**的情况。学会识别它们，能帮你省下大量时间。

### [必修] 三种「不需要追的异常」

| 类型 | 表现 | 为什么不需要追 | 快达案例 | 但什么时候需要追？ |
|------|------|-------------|--------|-----------------|
| **周期波动** | 每个周一都跌、每个周日都跌、每年春节都跌——跌幅在历史同期的正常范围内 | 它是一个规律，不是一次事故。你需要的是「把周期预测模型建好」，不是「每次跌了都从头分析一遍」 | 完单量周日→周一跌 22%→但过去三个月每周一都跌 15%-20%。这次多跌了 7 个点 | 如果跌幅**超出**历史同期的正常范围（比如平时跌 15%±5%，今天跌了 22%——超出了上界 20%）→ 那多出来的 7 个点值得追 |
| **已知事件** | 你知道原因——昨天成都下暴雨、上周竞品做了大促、今天是节假日。这些事件的影响是可预期的 | 你不需要「分析为什么跌」——原因已知。你需要的是「量化这个事件的影响有多大」和「事件过后指标恢复了吗」 | 成都昨天下暴雨→完单量跌 60%。不需要分析为什么——需要判断：暴雨导致的损失是否需要做「雨天运力预案」 | 如果暴雨过后第三天完单量还没恢复→说明有些用户被暴雨赶去了竞品，回不来了。这时候需要追——但不是追「为什么跌」（原因已知），而是追「为什么没回来」 |
| **口径变更** | 指标的计算方式被改了——分母扩大、分子收窄、排除条件增加。数字变了不是业务变了，是尺子变了 | 你需要追的是「谁改了口径、为什么没通知」，不是「业务为什么变」 | 周二学的指标身份证就是用来防这种情况的——口径锁死，变更需要审批 | 如果口径没变但数字变了→才是业务异常。但每次看到异常，第一反应应该是「口径变了吗」——这个习惯比任何分析框架都重要 |

**核心原则**：每一次看到异常，先问自己——「这个异常，值不值得我花时间追？」如果答案是周期波动、已知事件、或口径变更→不值得。把你的时间留给那些「未知原因、超出正常范围、可能产生持续影响」的异常。

### 练习：判断哪些不需要追

快达本周出现了以下 4 个「异常」。哪些值得花时间深度追查？哪些是虚惊一场？

1. DAU 比昨天跌了 30%——但今天是正月初一（春节）
2. 乘客等待时长从 5 分钟涨到 5.3 分钟——涨幅 6%，过去三个月波动范围是 4.5~5.5
3. 成都毛利率从 12% 跌到 -2%——历史波动范围是 8%~15%
4. 完单量比昨天涨了 40%——但昨天是中秋节，今天是节后第一天复工

### 参考分析

| # | 判断 | 为什么 |
|----|------|------|
| 1 | 🟢 虚惊（已知事件） | 春节 DAU 跌是每年都会发生的。不需要「分析为什么」——需要的是「今年的跌幅和去年春节比怎么样？春节后第几天恢复？」 |
| 2 | 🟢 虚惊（正常波动范围） | 5.0→5.3（+6%）。历史范围 4.5-5.5——还在正常波动带内。不需要追，除非连续 5 天方向一致（持续恶化） |

| 4 | 🟢 虚惊（周期波动） | 中秋节→节后复工。和周日→周一的道理一样——节日的出行需求爆发（回家/出游），节后回落。你需要看的是「跟去年中秋后的第一天的恢复幅度比」——如果去年恢复了 85%，今年只恢复了 60%→那就有问题了 |

## 概念群 B：[防护] 过早收敛——找到一个原因就停，是异动分析第二常见的错误

> **与核心概念的关系**：拆解过程中最常见的陷阱——你顺着一条线索往下追，找到了一个「合理的解释」，就停下来了。但你不知道的是，这个解释可能只覆盖了异常的 30%，剩下 70% 的原因你还没找到。

### 🔴 认知冲突

回到快达——完单量跌了 22%（正常化后约跌 7%）。你顺着 CMOP 下钻：

**第 1 步**：M 层——按城市拆。成都跌了 60%，贡献了异常的 85%。「成都是主因。」
**第 2 步**：O 层——成都的接单率从 91% 暴跌到 53%。「司机不接单。」
**第 3 步**：P 层——成都昨天下暴雨。「暴雨→司机不接单→完单量跌。分析完毕。」

你给 CEO 汇报：「成都暴雨导致完单量跌了 22%（正常化后约 7%）。预计明天恢复。」

CEO 问：「其他城市也在下雨吗？」你愣住了——你没查。回头一看——合肥昨天也下了中雨，但合肥的完单量只跌了 5%。同样的天气，为什么成都跌 60%、合肥跌 5%？

你只找到了第一个原因（天气），但漏了更深的原因（为什么成都在雨天比合肥脆弱这么多？）。

> **「天气」不是根因——它是扳机。根因是成都的运力结构让它在雨天特别脆弱。**

### [必修] 防止过早收敛——三问追根法

**① 一句话白话**：找到一个原因后，别停——问自己三句话。如果任何一句回答不了，说明你还没挖到真正的根因。

**② 三问**：

| 追问 | 什么意思 | 快达「天气」案例 |
|------|--------|---------------|
| **第一问：「这个原因能解释多少？」** | 异常跌了 7%（正常化后）。成都的天气能解释多少？如果成都暴雨历史上平均导致完单量跌 20%，但这次跌了 60%——那天气只解释了 1/3，还有 2/3 是未知的 | 天气能解释约 20pp 的跌幅。剩下的 40pp——需要别的解释。你再查：成都的「特惠快车」司机占比是 45%（全国 25%）。特惠快车单价低，暴雨天司机觉得「冒险出来接单不划算」→特惠司机大量下线。**天气是扳机，特惠占比过高是根因** |
| **第二问：「为什么同样的原因，在不同地方/时间的效果不同？」** | 合肥也下雨了，为什么跌 5% 而不是 60%？去年成都也遇到过暴雨，当时跌了 30%，为什么今年跌了 60%？ | 合肥的特惠快车占比只有 18%。去年成都的特惠快车占比只有 22%。→ 成都的问题不是天气，是「特惠快车的占比从 22% 涨到了 45%——导致运力结构变得对天气极度敏感」 |
| **第三问：「如果这个原因不再存在，问题会恢复吗？」** | 雨停了，成都的完单量能恢复到跌之前的水平吗？——不一定。因为暴雨期间，大量用户打不到车→转向了竞品→暴雨后也未必回来。 | 雨停后，完单量恢复到了平时水平的 85%。剩下的 15% 去哪了？你查：暴雨当天快达的竞品在成都推出了「雨天免起步价」活动——竞品抢走了 15% 的用户。**这个 15% 才是真正需要追的——它关系到成都的用户在持续流失** |

**③ 核心原则**：第一个找到的原因（天气）通常不是根因——它是**显因**。根因藏在「为什么同样的显因，在不同条件下产生不同的结果」这个追问里。

**④ 与核心概念的关系**：这是拆解→解释过程中的 [防护] 概念——防止你把第一个找到的合理解释当成最终答案。主课教你怎么拆，拓展教你怎么不满足于第一个答案。

## 概念群 C：[落地] 汇报的陷阱——从「发现了什么」到「所以呢」

> **与核心概念的关系**：解释之后要汇报。但大多数分析的汇报死在最后一公里——你讲了 10 分钟你的分析过程，CEO 只想知道「所以呢」。

### [必修] 异动分析的汇报结构——三句话原则

**① 一句话白话**：CEO 不需要知道你的分析过程。他只需要知道三件事：(1) 发生了什么、(2) 为什么、(3) 我该做什么。

**② 两版汇报对比**：

**分析师 A 的汇报（常见但无效）**：
> 「早上发现完单量比昨天跌了 22%。我先按城市拆开——成都跌了 60%，苏州涨了 3%，合肥……然后我查了成都的 O 层指标——发现接单率从 91% 跌到 53%。然后我看天气——成都昨天下暴雨。所以是暴雨导致的。」

CEO 在你说到「先按城市拆开」的时候已经开始看手机了。

**分析师 B 的汇报（三句话结构）**：
> 「完单量跌了 22%——但其中 15% 是正常的周日→周一回落，实际异常跌幅约 7%。原因是成都暴雨导致特惠快车司机大量下线——特惠司机在成都占了 45%，大雨天不上线导致供给断崖。不过这是短期波动，预计明天恢复。但有一个长期信号需要注意——暴雨期间竞品抢走了成都约 15% 的用户。建议明天跟进——如果 48 小时内这些用户没回来，需要在成都做一次定向召回活动。」

三句话：**是什么（7% 的异常）+ 为什么（特惠占比过高导致暴雨天脆弱）+ 做什么（明天监控回流，48h 不恢复就做召回）**。

**③ 三句话模板**：

| 句子 | 内容 | 字数 | 快达案例 |
|------|------|------|--------|
| **第 1 句：是什么** | 异常的真实幅度（剔除周期/口径/已知事件后）。不是「跌了 22%」——是「正常化后跌了约 7%，超出正常波动上限 2 个点」 | ≤ 30 字 | 「完单量异常跌幅约 7%，主因是成都。」 |
| **第 2 句：为什么** | 根因——不是「天气」，是「为什么这次天气的影响比以往大」。如果根因未知，诚实说「还在排查，已经锁定到 XX 环节，预计 XX 时间有结论」 | ≤ 60 字 | 「成都特惠快车占比 45%→大雨天司机下线→供给断崖。竞品同期做活动抢走约 15% 用户——这个可能持续。」 |
| **第 3 句：做什么** | 行动建议 + 时间点 + 谁来做。如果不需要行动（虚惊一场），也要明确说「不需要行动，明天再观察一次确认恢复」——CEO 需要的是一个决策，不是一个故事 | ≤ 40 字 | 「明天监控回流率。如 48h 内不回→成都定向召回活动。@王姐准备方案。」 |

### 练习：把下面的分析过程改写为三句话汇报

你的分析过程（给同事看的版本）：

> 司机 30 日留存率从 55% 跌到 38%。我按注册日期拆了同期群——发现 6/21 注册的司机留存率只有 25%。6/21 是快达的「司机拉新大促」——当天通过渠道 A 拉来了 3000 个新司机。渠道 A 以前没投放过，这次试投放发现用户质量很差——这批司机注册后首周完单量只有 3 单（正常新司机 8 单），而且 70% 在第三天就流失了。但好消息是存量司机（6/20 之前注册）的留存率是 56%——比 Q1 的 55% 还涨了 1pp。

### 参考答案

**第 1 句（是什么）**：「司机 30 日留存率技术性下跌到 38%——原因是 6/21 大促拉新导致分母结构变化，存量司机留存率实际为 56%（+1pp），没有恶化。」

## 综合练习 2：沙盘变体——异动分析的「假阳性」

快达产品 VP 李哥非常紧张地跑来找你：「我们上周改了 App 的首页——把『叫车』按钮从底部移到了中间。结果这周完单量跌了 10%！改版肯定有问题，赶紧回滚！」

你会怎么回应？在做任何分析之前，你需要先确认什么？

---

### 参考回应

**先确认四件事（四边界）**：
1. 时间边界——改版是上周四上线的。完单量从周五开始跌。但——上周四也是「竞品大促」的第一天。你怎么区分「改版导致跌」和「竞品大促导致跌」？
2. 空间边界——如果改版导致跌，应该是所有城市都跌（因为所有城市的 App 都更新了）。但如果只有几个城市跌了、其他正常→改版不是主因。
3. 口径边界——改版后完单量的口径有没有变？比如「完单」的埋点有没有因为改版被影响？
4. 幅度边界——10% 的跌幅在改版历史上正常吗？快达过去每次改版后完单量波动约 ±3%-5%。10% 确实超过了正常范围——但如果竞品同期做了大促，那 10% 里可能只有 2-3% 是改版的影响。

**在你回答「改版有没有问题」之前，先回答「改版是唯一变量吗」——大多数被归咎于「改版」的异常，其实和改版关系不大。**

## 综合练习 3：识别「过度归因」

快达运营王姐做了一个实验——在成都给司机发「高峰时段完单奖励」。实验后，成都高峰时段的完单量涨了 15%。王姐说：「奖励有效！应该全量上线。」

**问题**：在宣布「有效」之前，你需要排除哪 3 种可能性？

---

### 参考分析

| 需要排除的 | 什么意思 | 快达案例 |
|----------|--------|--------|
| **巧合（同期有其他因素）** | 完单量涨了 15%——但实验期间成都刚好开学季，出行需求自然上涨。你要对比：同期没做实验的合肥，完单量涨了多少？如果合肥也涨了 10%→那奖励的效果大概只有 5%（15%-10%） |
| **时间转移** | 司机看到「高峰奖励」→把平峰的订单推迟到高峰接→高峰完单量涨了，但平峰完单量跌了。总完单量没涨——只是时间分布变了 |
| **新奇效应** | 第一次做奖励，司机觉得新鲜→积极参与。但如果每个月都做→效果递减。第一周 15%，第四周可能只有 3%——因为司机习惯了。要看连续 4 周的衰减曲线才能判断长期效果 |
