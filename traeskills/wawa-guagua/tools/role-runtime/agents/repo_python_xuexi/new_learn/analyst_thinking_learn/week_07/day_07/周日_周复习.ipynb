## 🧩 区块2：知识点深度复盘

### 2.1 D1复盘：AB实验是什么——随机化是因果推断的免费午餐

**核心概念**：
- 随机分组 → 两组实验前完全可比 → 实验后差异=因果效应
- 随机化消除了所有混淆变量（已知的和未知的）
- AB实验 = 互联网版的RCT（随机对照试验）

**三个基本要素**：
1. 随机化单元（用户/设备/城市）
2. 处理（什么干预）
3. 指标（主要/次要/护栏）

**常见误区**：
- 灰度发布 ≠ AB实验（没有同期对照组）
- 前后对比 ≠ AB实验（无法排除时间因素）
- "使用X的用户表现更好" ≠ "X导致表现更好"（混淆偏误）

**SUTVA假设**：
- 无干扰：用户的处理不影响其他用户
- 一致性：同一处理效果稳定
- 违反场景：社交产品、双边市场、广告竞价

**面试重点**：为什么随机化能推断因果？AB实验和观察性分析的区别？

### 2.2 D2复盘：假设检验——从"看到差异"到"确信差异"

**核心概念**：
- H₀：策略无效；H₁：策略有效
- p值 = P(数据 | H₀为真)，不是P(H₀为真 | 数据)
- α（第一类错误，误报）= 0.05；β（第二类错误，漏报）= 0.20
- Power = 1-β = 80%（业界标准）

**p值的三个常见误解**：
1. p<0.05 ≠ 策略有95%概率有效
2. p值不随效应量单调变化（大样本下微小效应也显著）
3. p=0.051和p=0.049没有本质区别

**必报三件套**：
- p值（统计显著性）
- 效应量（实际意义）
- 95%置信区间（不确定性的范围）

**多重比较问题**：检验10个指标 → 至少1个假阳性的概率≈40%
- 解决方案：Bonferroni校正、预注册主要指标

**面试重点**：p值是什么（通俗解释）？第一类错误和第二类错误的区别？

### 2.3 D3复盘：样本量与MDE——实验需要多少用户？

**核心概念**：
- n = f(α, β, σ, δ)
- 最重要的关系：n ∝ 1/δ²（MDE减半 → 样本量×4）
- MDE = 在给定参数下能检测到的最小效应

**样本量公式**：
- 比例指标：n = (Zα√[2p̄(1-p̄)] + Zβ√[p₁(1-p₁)+p₂(1-p₂)])² / (p₂-p₁)²
- 连续指标：n = 2(Zα+Zβ)²σ² / δ²

**MDE的双向使用**：
- 正向：给定MDE → 计算需要多少样本 → 需要多少天
- 反向：给定流量 → 计算可检测的MDE → 和PM对齐期望

**实验天数 = n / (DAU × 分流比) + 指标窗口**

**流量不足时的选择**：
1. 接受更大的MDE
2. 放宽α（高风险）或降低Power（高风险）
3. 用代理指标减少窗口期
4. 诚实地说"我们检测不到那么小的效果"

**面试重点**：MDE是什么？为什么MDE减半样本量×4？怎么和PM沟通MDE？

### 2.4 D4复盘：实验设计完整流程——9步从假设到决策

**9步流程**：
```
假设 → 设计 → 样本 → 分流 → 运行 → 监控 → 停止 → 分析 → 决策
```

**实验设计文档（8要素）**：
1. 实验名称和负责人
2. 业务假设（三要素：做什么→影响什么→变化多少）
3. 实验设计（单元、分流、分组）
4. 指标体系（主要1个+次要+护栏）
5. 样本量计算
6. 分析计划
7. 决策规则（上线/不上线/延长条件）
8. 风险与回滚计划

**决策矩阵（3×3）**：
```
              效应量足够          效应量不足
显著          ✓ 上线             △ 考虑成本
不显著        △ 延长实验          ✗ 不上线
护栏恶化      ✗ 一票否决          ✗ 一票否决
```

**实验的"死亡"方式**：SRM、Peeking、指标偷换、幸存者偏差、没有护栏

**面试重点**：请设计一个完整的AB实验。护栏指标为什么需要一票否决权？

### 2.5 D5复盘：AA实验与Peeking——质检与防作弊

**AA实验**：
- 两组配置完全相同 → 不应有差异
- 跑100次 → p<0.05比例应≈5%
- 偏离5% → 分流系统有bug
- = 实验系统的"单元测试"

**SRM（样本比例不匹配）**：
- 用卡方检验检测
- p<0.05 → 分流有问题
- SRM时不要看任何其他指标

**Peeking**：
- 每天看p值 → p<0.05就停 → α从5%飙升到20%+
- 原因：在"挑选"偶然的低p值
- 正确做法：固定样本量，到终点一次性分析

**序贯检验**：
- 合法的"中途查看"
- 每次用调整后的阈值（早期极严格，后期接近0.05）
- 总α控制在5%
- 但实践中最安全的还是"不Peeking"

**面试重点**：Peeking为什么不行？AA实验怎么做？SRM怎么检测？

````python
print("\n二、代码实战：完成一个AB实验分析（75分）")
print("="*60)

import numpy as np
from scipy import stats
np.random.seed(777)

# 题目：某电商测试新搜索算法，主要指标是搜索转化率
# 基线转化率12%，期望检测到15%以上的相对提升
# 请完成完整的分析

print("\n场景：搜索算法AB实验")
print("  主要指标：搜索转化率（搜索后有下单的比例）")
print("  基线：12%")
print("  期望检测：+15%相对提升（12%→13.8%）")

# (1) 计算所需样本量（15分）
print("\n(1) 样本量计算:")
baseline = 0.12
mde_rel = 0.15

za = stats.norm.ppf(0.975)  # α=0.05双侧
zb = stats.norm.ppf(0.80)   # Power=80%
p1 = baseline
p2 = baseline * (1 + mde_rel)
p_bar = (p1 + p2) / 2
n_needed = int(np.ceil(
    (za * np.sqrt(2*p_bar*(1-p_bar)) +
     zb * np.sqrt(p1*(1-p1) + p2*(1-p2)))**2 / (p2 - p1)**2
))
print(f"  每组需要: {n_needed:,} 用户")

# (2) 模拟数据并做SRM检验（15分）
print("\n(2) SRM检验:")
n = n_needed
true_effect = 0.018  # 真实+1.8pp

control = np.random.binomial(1, baseline, n)
experiment = np.random.binomial(1, baseline + true_effect, n)

chi2 = (n - n/2)**2 / (n/2) * 2  # 两组相等，简化计算
p_srm = 1 - stats.chi2.cdf(0, 1)  # 这里故意简化
# 实际SRM
chi2_real = ((n - n/2)**2 / (n/2) + (n - n/2)**2 / (n/2))
p_srm_real = 1 - stats.chi2.cdf(chi2_real, 1)
print(f"  对照组={n:,}, 实验组={n:,} → 50:50 ✓")
print(f"  SRM: p≈1.000 → ✓ 通过")

# (3) 主要指标分析（20分）
print("\n(3) 主要指标分析:")
t_stat, p_val = stats.ttest_ind(experiment, control)
effect = experiment.mean() - control.mean()
effect_rel = experiment.mean() / control.mean() - 1
se = np.sqrt(control.var()/n + experiment.var()/n)
ci = stats.t.interval(0.95, 2*n-2, loc=effect, scale=se)

print(f"  对照: {control.mean():.3%}, 实验: {experiment.mean():.3%}")
print(f"  效应: {effect*100:.2f}pp (相对{effect_rel*100:.1f}%)")
print(f"  95%CI: [{ci[0]*100:.2f}pp, {ci[1]*100:.2f}pp]")
print(f"  p值: {p_val:.4f} → {'显著' if p_val<0.05 else '不显著'}")

# (4) 护栏指标检查（10分）
print("\n(4) 护栏指标:")
# 搜索结果加载时间（不希望显著变慢）
ctrl_latency = np.random.normal(200, 50, n)
exp_latency = np.random.normal(205, 50, n)  # 慢了5ms
_, p_latency = stats.ttest_ind(exp_latency, ctrl_latency)
print(f"  页面加载时间: 对照={ctrl_latency.mean():.0f}ms, 实验={exp_latency.mean():.0f}ms")
print(f"    p={p_latency:.4f} → {'⚠ 显著变慢！' if p_latency<0.05 else '✓ 通过'}")

# (5) 做出决策（15分）
print("\n(5) 决策:")
print("="*40)
practical_threshold = 0.01  # 1pp的业务价值

if p_latency < 0.05:
    print("✗ 不上线（页面加载时间显著恶化——护栏一票否决）")
    print("  建议: 优化新算法的性能后再实验")
elif p_val < 0.05 and effect > practical_threshold:
    print("✓ 建议上线")
elif p_val < 0.05 and effect <= practical_threshold:
    print("△ 显著但效应太小，不建议上线")
elif p_val >= 0.05 and effect > practical_threshold:
    print("△ 效应有价值但不显著，延长实验")
else:
    print("✗ 不上线")

print(f"\n→ 自检清单:")
print(f"  □ 样本量是否事前计算？")
print(f"  □ SRM是否检验？")
print(f"  □ 是否报告了p值+效应量+CI？")
print(f"  □ 是否检查了护栏指标？")
print(f"  □ 决策是否基于多维度判断（而非只看p值）？")
````

## 📊 区块4：核心认知总结——10条金句

### 本周最重要的10个认知

1. **随机化是因果推断的免费午餐**：不需要模型，随机分组就能消除所有混淆变量。这就是AB实验成为黄金标准的根本原因。

2. **p值 ≠ 策略有效的概率**：p值 = P(数据|H₀为真)，不是P(H₀为真|数据)。p<0.05不是"策略有95%概率有效"。

3. **大样本下一切皆显著**：永远同时报告p值+效应量+置信区间。只看p值会被大样本的"虚假精确"欺骗。

4. **MDE是你的诚实工具**：n ∝ 1/δ²——MDE减半，样本量4倍。MDE让业务方理解"小实验只能检测到大效果"。

5. **实验设计80%的价值在开跑之前**：假设写清楚、样本量算清楚、指标定义清楚——前面做好了，结论自然可信。

6. **护栏指标需要一票否决权**：只看主要指标不看护栏 = 只看速度表不看油箱。转化率涨了但卸载率也涨了=不可上线。

7. **Peeking让α从5%飙升到20%+**：你不是在"监控实验"，你是在"等待p值降到0.05以下"。正确的做法是固定样本量，到终点一次性分析。

8. **AA实验是实验系统的单元测试**：没通过AA验证的平台，跑出来的任何AB结论都不值得信。

9. **SRM是实验的红灯**：卡方检验p<0.05 → 分流有问题 → 不要看任何其他指标 → 修bug后重新实验。

10. **实验不是为了"证明自己是对的"**：大部分实验不会成功（业界胜率10-20%）。一个"不显著"的实验帮你排除了一条路，这不是失败。
