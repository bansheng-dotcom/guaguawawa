### 1.2 从第一性原理推导：为什么"分析思维"比"分析技能"更重要？

技能会过时——SQL语法会变、AB实验工具会更新、新的因果推断方法会出现。但"分析思维"不会过时：
- 界定问题的方法（ISS）——换什么工具处理数据，这一步都不变
- 拆解指标的思维（乘法/加法）——不管用什么语言写，逻辑是一样的
- 假设驱动的验证——这本质上是科学方法
- 区分相关和因果——这是所有数据工作的核心
- 结论优先的表达——这是所有沟通的根本

你在这12周里学的不是"怎么用SQL写留存率"，而是"为什么留存率这么重要、怎么分析留存率变化、怎么根据留存率做决策、怎么跟别人讲清楚留存率的故事"。前者会在2年后被新版工具替代，后者会陪伴你整个职业生涯。

## 🧩 区块2：你现在的"出厂配置"——面对业务问题的标准反应流程

### 2.1 你的大脑现在应该自动执行的7步

学完12周，面对一个业务问题，你的大脑应该自动执行以下流程：

```
业务方/老板抛来一个问题
    ↓
步骤1：【先界定问题】（不要拉数据！）
    ├── ISS：这到底是什么问题？什么时候开始的？影响了什么？
    └── 输出：一句清晰的问题定义
    ↓
步骤2：【选分析框架】
    ├── 指标异动 → 指标拆解 + PDS
    ├── 用户增长 → AARRR + 同期群
    ├── 渠道评估 → 归因模型 + CAC/LTV
    ├── 效果评估 → AB实验 或 因果推断(DID/PSM)
    ├── 用户分群 → RFM
    └── 陌生问题 → ISS
    ↓
步骤3：【设计分析路径】
    ├── 指标拆解（至少2层）
    ├── 假设驱动（3个最可能的假设，按优先级排列）
    └── 验证方案（每个假设对应的检验方法）
    ↓
步骤4：【写SQL/Python获取数据】
    ├── CTE分层写法
    ├── 窗口函数（按需选择）
    └── 每一条查询都在回答一个具体的假设
    ↓
步骤5：【检查数据质量】
    ├── 5分钟嗅探：空值/重复/异常/分布/新鲜度
    └── 如果数据有问题→回到步骤4修正
    ↓
步骤6：【区分相关和因果】
    ├── 问自己："还有别的解释吗？"
    ├── 如果能随机化 → AB实验
    ├── 如果不能 → 因果推断(DID/PSM/IV/RDD)
    └── 如果都不行 → 诚实标注"这是相关性，不是因果性"
    ↓
步骤7：【给出结论+行动建议】
    ├── 结论优先："建议_____，因为_____"
    ├── 给选项+推荐："方案A/B，推荐A，因为_____"
    ├── 量化预期："预计带来_____的变化"
    └── 风险兜底："如果判断错了怎么办？"
```

### 2.2 "分析思维"的七个核心信念

这些不是你"知道"的——是你"相信"的。如果你真的相信了，它们会改变你面对每一个数据问题的方式：

| # | 信念 | 解释 | 反例 |
|---|------|------|------|
| 1 | **先界定问题，再拉数据** | 方向错了，再好的数据也没用 | "DAU跌了，我先拉最近30天数据看看" |
| 2 | **看结构不看总量** | 好的分析一定是分层下钻的 | "DAU涨了20%，很好！"不看新老用户结构 |
| 3 | **假设驱动而非数据驱动** | 带着猜想去验证，不要大海捞针 | "我把所有维度的数据都拉出来'看看'" |
| 4 | **相关不等于因果** | 任何两个一起变化的变量都要追问第三变量 | "信息流曝光量和GMV高度相关→信息流很有效" |
| 5 | **结论优先、行动导向** | 分析的价值=别人因为你的分析做了什么 | 报告最后写"建议继续观察" |
| 6 | **不确定也是答案** | 诚实地说"数据不足以判断"比编一个结论好 | 为了给结论而忽略数据的不确定性 |
| 7 | **分析不落地 = 没做** | 没人懂、没人信、没人动的分析=0 | 做了一份50页的报告发出去没有回音 |

### 4.1 分析不是"报数"，是"翻译"

这是这门课最核心的一句话：

> **分析不是"报数"，是"翻译"——把业务问题翻译成数据问题，把数据结果翻译成业务决策。**

两个翻译方向：

**方向1：业务→数据**
- CEO说："我们用户增长好像停滞了"
- 你翻译成："需要做DAU的PDS拆解，看是新增跌了/留存跌了/回流跌了。然后分渠道看同期群，对比不同批次用户的质量变化。"

**方向2：数据→业务**
- 你发现："信息流渠道新增用户的7日付费率从3.2%跌到1.8%"
- 你翻译成："建议立即回滚信息流投放素材——虽然DAU在涨，但这些'标题党'吸引来的用户根本不付费。预计回滚后DAU可能下降10%但GMV会停止下滑。"

一个不会"翻译"的分析师就是一个数据管道——把数据从数据库搬到Excel，再从Excel搬到PPT。这门课教了你翻译的方法和工具。从今天起，你的价值不在于"你会写SQL"，而在于"你能把业务问题翻译成数据问题，再把数据答案翻译成业务决策"。

### 4.2 你学会了"用框架思考"，而不仅仅是"用工具干活"

12周之前，面对"DAU跌了"，你可能：
- "我拉一下最近的数据看看"→然后面对一个巨大的表格无从下手

12周之后，面对"DAU跌了"，你自动：

- 结论+建议："根因是X，建议做Y，预计效果Z"

这就是"会用框架思考"和"只会拉数据"的区别。

### 4.3 你已经有了怀疑精神和审视能力

### 6.1 三门课程的关系

- **02analyst_thinking**给你"脑"——怎么思考、怎么框架化
- **03analyst_thinking**给你"实战"——把脑和手结合在一起解决真实问题
- **真实工作**给你"经验"——在项目中学到的永远比课程中多

### 6.2 这门课的独特价值

市面上的数据分析课程大多教"怎么用工具"——怎么画漏斗图、怎么做AB测试的t检验、怎么写SQL。但很少有人教"怎么思考"——什么时候该用漏斗、什么时候该用同期群、什么时候该做实验、什么时候该做因果推断。

这门课教了你的，不只是"怎么做"，更是"怎么想"：
- 不是"怎么用SQL写留存率"——而是"为什么留存率重要、怎么分析留存变化、怎么根据留存做决策"
- 不是"怎么做t检验"——而是"为什么不能每天做t检验、什么样的实验结论才是可信的"
- 不是"怎么写PPT"——而是"怎么把这个数据发现翻译成决策者想要的答案"

这些"怎么想"的能力——没有任何AI工具可以替代，也不会随着工具更新而过时。

### 6.3 02analyst_thinking——按专题的深度参考库

`02analyst_thinking/` 下的内容（按周组织，7天×12周）是一份结构化的深度参考库。当你需要深入某个特定主题时——比如"怎么做一份渠道归因分析"——你不需要从头回忆，直接找到W3D1的参考笔记，里面有：练习+场景+检验+连接。

这是你的"分析外脑"——不需要全记住，但需要知道怎么找。

### 三条"毕业箴言"带回家

1. **"先别拉数据"**——这可能是你在这门课里听到最多的一句话。它值得成为你职业生涯中默认的第一反应。因为框架对了，答案就在框架里。框架错了，再多的数据也只是在错误的路上越走越远。

2. **"分析不落地 = 没做"**——你的SQL写得再好、p值算得再精确、图表画得再漂亮，如果没有人因为你的分析做出了更好的决策——你的工作就没有产生价值。做分析时永远记住：你的终点不是"分析完"，是"别人懂了并且行动了"。

3. **"带着框架去实践"**——课程给了你框架，框架给了你方法论，方法论让你在任何陌生问题面前都有章法可循。但框架必须与实践结合——每一个你在课程中"知道"的工具，都需要在真实项目中至少使用过3次，才能真正内化为你的肌肉记忆。
