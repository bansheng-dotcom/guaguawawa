## 一、核心概念与第一性原理

### 1.1 OSM 是什么

OSM 是一个将"模糊目标"转化为"可执行策略+可度量指标"的框架：

```
O — Objective（目标）：我们要达成什么？
S — Strategy（策略）：我们打算怎么做？
M — Measurement（度量）：怎么知道策略有没有用？
```

这三个字母不是孤立的——它们之间有一条**因果链**：
- O 决定了 S 的方向（如果目标是"提升留存"，策略就不能是"花钱买量"）
- S 决定了 M 的选择（每条策略必须有对应的度量，否则无法判断策略是否有效）
- M 反馈给 O（度量结果告诉你目标达成了多少）

### 1.2 从第一性原理推导：为什么需要 OSM

组织协作的第一性原理：**多人协作时，如果没有共同的目标和衡量标准，每个人会朝自己认为"对"的方向努力——结果就是团队在朝不同方向前进。**

业务方说："我们要提升用户活跃度。"

没有 OSM 的后果：
- 产品觉得要加新功能（直播、话题）
- 运营觉得要发 push、做签到
- 技术觉得要优化加载速度
- 三个月后，每个人都做了很多事，但没人说得清活跃度到底有没有提升

有了 OSM：
```
O：提升新用户次日留存率（从15%到20%，Q3完成）  ← 可量化的目标
S：
  策略1：注册后引导用户关注3个感兴趣的话题    ← 基于"新用户注册后不知道看什么"的洞察
  策略2：首次发布内容后1分钟内给一个赞+评论    ← 基于"新用户需要正反馈"的洞察
  策略3：次日push提醒"有X人看了你的内容"       ← 基于"召回可以提升次留"的洞察
M：
  度量1：注册引导完成率 → 关注话题的用户比例
  度量2：首次内容发布率 → 发布后次日回访率
  度量3：push点击率 → 召回用户次日回访率
```

**OSM 的本质：强迫你在做事之前想清楚——"做了这件事，看哪个数字能判断它有没有用？"**

> **今天最重要的认知**：OSM 强制你在做事之前先想清楚因果链——"做了这件事，看哪个数字能判断它有没有用？"如果回答不了，说明策略和目标之间的因果链是断的，这条策略就是空洞的。

## 二、知识点详细拆解

### 2.1 O（目标）怎么定——SMART 原则

**为什么重要**："提升用户体验"这种目标无法指引行动。SMART 把模糊变成精确。

| 字母 | 含义 | 坏例子 | 好例子 |
|------|------|--------|--------|
| S | 具体(Specific) | "提升用户体验" | "提升新用户次日留存率" |
| M | 可量化(Measurable) | "变得更好" | "从15%提升到20%" |
| A | 可达成(Achievable) | "从15%提升到90%" | "从15%提升到20%" |
| R | 与业务相关(Relevant) | 新闻App提升游戏时长 | 新闻App提升阅读完成率 |
| T | 有时间边界(Time-bound) | "尽快" | "在Q3结束前（9月30日）" |

**定目标的常见错误**：目标和度量混淆。O="把DAU做到50万"——DAU是度量，不是目标。目标应该是"成为用户每日必备的学习工具"，DAU只是衡量这个目标的一个数字。

### 2.2 S（策略）怎么出——三个来源

**为什么重要**：策略不是"想到什么做什么"，而是基于洞察的假设。

策略的三个来源：
1. **数据洞察**：留存漏斗中哪一步掉得最厉害？（→策略应该针对这个环节）
2. **竞品分析**：同行怎么做新用户引导？（→借鉴但不照搬）
3. **用户调研**：用户为什么不回来？（→策略应该解决用户真实痛点）

**好策略的三个特征**：
- 有清晰的因果假设："因为XX，所以我们做YY，预期会导致ZZ"
- 可被验证：策略的效果可以通过M来检验
- 有优先级：资源有限时，先做最可能有效的1-2条

### 2.3 M（度量）怎么选——过程度量 vs 结果度量

**为什么重要**：很多人只看"最终结果"，不看"中间过程"。如果策略没效果，你不知道是"策略本身错了"还是"策略根本没被执行"。

每条策略至少配两个度量：
- **过程度量**（Process）：策略被执行了吗？如push发送量、引导完成率
- **结果度量**（Outcome）：策略达到了预期效果吗？如次留率变化

**两者的关系**：过程度量是"必要条件"，结果度量是"充分条件"。过程都没发生→结果肯定不会有。过程发生了但结果没有→策略本身无效。

### 2.4 OSM 的四个常见错误

| 错误 | 例子 | 为什么错 | 怎么改 |
|------|------|---------|--------|
| 目标=度量 | O="提升DAU到50万" | DAU是衡量工具，不是目标 | O="成为用户每日必备工具" |
| 策略太多 | 列了15条策略 | 资源分散，无法判断哪条有效 | 聚焦3-5条，排优先级 |
| 策略-度量脱节 | S="优化推送文案", M=只看DAU | DAU涨了也未必是推送的功劳 | M应直接衡量推送效果 |
| 只看结果不看过程 | 留存没涨→撤策略 | 可能推送都没发出去，不是策略问题 | 先看过程度量，再看结果度量 |

## 三、动手练习

### 练习1：补全一个不完整的 OSM（基础）

````python
# ===== 练习1：补全 OSM =====

print("=" * 50)
print("练习1：某内容社区App想要提升用户发帖量")
print("=" * 50)
print()

# 用字典结构来组织 OSM（配对01 W1D3：字典的使用）
osm_plan = {
    "目标(O)": {
        "总目标": "提升社区UGC活跃度",
        "SMART目标": "将周活跃发帖率从8%提升到12%，Q2完成",
        "当前基线": "周活跃发帖率 = 8%，约4000人/周",
    },
    "策略(S)": [
        {
            "策略": "降低发布门槛：支持短图文（140字以内）",
            "假设": "用户不发帖是因为'觉得要写很长'",
            "过程度量": "短图文发布功能的完成率",
            "结果度量": "新用户首次发帖率"
        },
        {
            "策略": "冷启动内容供给：新用户注册后自动推荐热门话题",
            "假设": "用户不发帖是因为'不知道发什么'",
            "过程度量": "话题推荐曝光率 + 点击率",
            "结果度量": "看到推荐话题用户的发帖率 vs 没看到"
        },
        {
            "策略": "发布即反馈：发帖后5分钟内给10个曝光+1个赞",
            "假设": "用户不发帖是因为'发了也没人看'",
            "过程度量": "首次发帖5分钟内的曝光量和互动量",
            "结果度量": "首次发帖用户的次日留存率"
        },
    ],
}

print(f"O: {osm_plan['目标(O)']['总目标']}")
print(f"SMART: {osm_plan['目标(O)']['SMART目标']}")
print(f"基线: {osm_plan['目标(O)']['当前基线']}")
print()
for i, s in enumerate(osm_plan["策略(S)"], 1):
    print(f"策略{i}: {s['策略']}")
    print(f"  假设: {s['假设']}")
    print(f"  过程度量: {s['过程度量']}")
    print(f"  结果度量: {s['结果度量']}")
    print()

print("【预期输出】3条策略，每条都有清晰的假设+过程度量+结果度量")
````

## 四、业务场景应用

### 场景1：两个团队用同一个目标，策略完全不同——OSM 让协作有了一致语言

某出行App的目标(O)：**"将高峰期完单率从85%提升到92%"**

同一个 O，三个团队的 S 和 M 完全不同：

| 团队 | 策略(S) | 假设 | 度量(M) |
|------|--------|------|--------|
| 算法 | 优化派单算法，减少司机空驶 | "派单效率提升能提高完单率" | 司机接单时长、空驶率 |
| 运营 | 高峰期涨价1.2x，抑制部分需求 | "价格杠杆能让供需更平衡" | 价格弹性、用户取消率 |
| 产品 | 优化下单流程，减少'犹豫'导致的取消 | "流程简化能减少用户取消" | 下单完成率、取消原因分布 |

**OSM 的好处**：不是"你的方案好还是我的方案好"，而是"你的策略对哪个度量产生了什么影响，跟我的策略的效果有什么差异"。这让跨团队协作从"互相说服"变成了"数据说话"。

**用vs不用的对比**：
- 不用OSM：每个团队说自己做得好，没有统一的M来评判
- 用OSM：每个团队的S都有对应的M，效果可量化、可对比

### 场景2：补贴ROI分析——"发券花了100万，赚回来多少？"

运营发了一批优惠券，花了100万。GMV涨了500万。运营说："ROI=5，券有效！"

你作为分析师，需要拆解这500万里有多少是真正的增量：
- 领券用户中，有多少是"本来就要买"的？（→券是纯成本）
- 有多少是"因为券才买"的？（→这才是券的增量价值）
- 有多少是"本来下周买，因为券提前到这周买了"？（→时间转移，非真增量）

**OSM 视角**：
- O：提升补贴的增量ROI（增量毛利/补贴成本 > 1）
- S：对不同用户群发不同面额的券（高价值用户发小额、沉睡用户发大额）
- M：按用户群分别计算增量ROI

**用vs不用的对比**：
- 不用OSM+增量思维："ROI=5，券有效！"（其实大部分GMV本来就会发生）
- 用OSM+增量思维：只有40%是增量→实际ROI=2，但活动值得继续，需要优化券的面额和人群

### 场景3：新功能上线——怎么判断它"有没有用"？

产品上线了"智能推荐"功能。PM说："上线后人均观看时长涨了12%，功能有效！"

**OSM 帮你冷静分析**：
- O：功能上线时的目标是什么？提升留存？提升时长？提升付费？
- S：推荐算法具体做了什么改动？
- M：有没有对照组？12%里有多少是"功能的效果"vs"自然增长"vs"季节因素"？

**如果没有预先定义 OSM**，任何事后看到的指标变化都可能被归因为"功能的效果"——这叫"事后合理化"。

**用vs不用的对比**：
- 不用OSM+实验设计：看到的任何变化都归因于功能（幸存者偏差+确认偏误）
- 用OSM+实验设计：能区分"功能效果"和"自然波动"（W7-W8会学到怎么做）

## 五、分级检验清单

### 初级检验
- [ ] 能写出 O、S、M 各自的全称和含义
- [ ] 能识别"模糊目标"和"SMART目标"的区别，并能把前者改成后者
- [ ] 能为一个给定的策略写出过程度量和结果度量

**答案提示**：O=目标(要达成什么)，S=策略(怎么做)，M=度量(怎么知道有没有用)。SMART=具体+可量化+可达成+相关+有时间边界。

### 中级检验
- [ ] 能独立为一个业务场景写出完整的 OSM（至少3条策略，每条有假设）
- [ ] 每条策略能说清背后的数据假设（"因为数据发现XX，所以做YY，预期会导致ZZ"）
- [ ] 能区分"策略没被正确执行"（过程度量出问题）和"策略本身无效"（结果度量不达预期）

**答案提示**：过程度量看"执行了没"，结果度量看"有效果没"。先查过程，再评结果。

### 高级检验
- [ ] 能为同一个目标设计不同团队的差异化策略和度量（如算法vs运营vs产品）
- [ ] 能设计 OSM 的"验证计划"——怎么判断每个策略的独立效果（需要AB实验）
- [ ] 当策略无效时，能判断是该换策略还是目标本身定错了

**答案提示**：多团队OSM的关键是各团队的M要独立可衡量；验证独立效果需要AB实验(W7-W8)；策略连续3次无效→可能是假设有问题或目标不切实际。

## 七、当日小结

> **今天最重要的一个认知**：策略与度量是一一对应的。好的 OSM 应该做到——每一条策略，都有一个数字能告诉你"它有没有用"。如果你的策略写出来但找不到对应的度量，说明这条策略是空洞的。

**三条金句带回家**：
1. OSM 强迫你在做事之前先想清楚因果链——"做了A，看哪个数字能判断A有没有用？"
2. 每条策略配两个度量：过程度量看"做了没"，结果度量看"有用没"
3. 策略不是越多越好——聚焦3-5条最可能有效的，每一条都有数据假设支撑
