# 第三部分：概念网络展开

## 🔴 核心深化：逐一替换法的三种顺序——同样的数据，不同的顺序，不同的结论

> 主课教你逐一替换法能算出每个因子的贡献。但有一个坑主课没展开：**替换顺序不同，贡献值不同**。如果你不知道这件事，你可能把同一个变化归给不同的因子。

### 🔴 认知冲突

快达 Q2 的完单量从 8 万涨到 10 万（+25%）。你把它拆成三个因子：

```
完单量 = 派单量 × 接单率 × 完单率
Q1: 9.5万次 × 91% × 93% ≈ 8.0万
Q2: 11.2万次 × 89% × 91% ≈ 9.1万（接近 10万，简化）
```

现在用逐一替换法算三个因子的贡献。但你发现——**先替派单量再替接单率**，和**先替接单率再替派单量**——算出来的贡献值不一样。

**先替派单量**：
- 仅派单量变：11.2万 × 91% × 93% = 9.5万 → 贡献 +1.5 万
- 再替接单率：11.2万 × 89% × 93% = 9.3万 → 贡献 -0.2 万
- 最后替完单率：11.2万 × 89% × 91% = 9.1万 → 贡献 -0.2 万
- 总贡献：+1.5 -0.2 -0.2 = **+1.1 万（派单量是绝对主角）**

**先替接单率**：
- 仅接单率变：9.5万 × 89% × 93% = 7.9万 → 贡献 -0.1 万
- 再替派单量：11.2万 × 89% × 93% = 9.3万 → 贡献 +1.4 万
- 最后替完单率：11.2万 × 89% × 91% = 9.1万 → 贡献 -0.2 万
- 总贡献：-0.1 +1.4 -0.2 = **+1.1 万（还是派单量主角，但接单率的「责任」从 -0.2 变成了 -0.1）**

**问题**：同一个 Q1→Q2 的数据，两种顺序给出了不同的贡献分配。运营王姐说「完单量涨全是派单量涨的功劳——接单率还拖了后腿」。产品李哥说「不对——如果先算接单率，我拖的后腿没那么大」。

谁对？

> 都对——也都不完全对。逐一替换法的「顺序敏感」是一个已知的数学缺陷。关键不是「选哪个顺序」，而是「你知道这个缺陷存在，并且根据业务场景选最合理的顺序」。

### [必修] 三种替换顺序——什么时候用哪种

**① 一句话白话**：逐一替换法就像切蛋糕——先切哪一刀，会影响后面每刀的起点位置。三种顺序各有道理，选哪种取决于你的业务场景。

**② 三种顺序**：

| 顺序 | 怎么替 | 每个因子用的是「变化前还是变化后」的基准 | 什么时候用 | 快达完单量案例 |
|------|--------|----------------------------------|----------|-------------|
| **上游优先** | 按转化链从上到下替——先替最上游的（派单量），再替中游的（接单率），最后替下游的（完单率） | 替上游时用 Q1 的下游值，替下游时用 Q2 的上游值 | 你关心「上游的改善对下游产生了多少连带影响」——派单量涨了，即使接单率和完单率没变，总完单也会涨 | 派单量贡献 +1.5 万——这是增长团队拉新/投放带来的「纯粹增量」。接单率贡献 -0.2 万——但注意，这个 -0.2 是在「派单量已经涨了」的基础上算的——如果派单质量也下降了（新渠道来的乘客取消率高），接单率的跌可能部分原因是派单量涨了 |
| **下游优先** | 从下往上替——先替下游的（完单率），再替中游的，最后替上游的 | 替下游时用 Q1 的上游值，替上游时用 Q2 的下游值 | 你关心「如果下游没变化，上游的变化有多大效果」——把下游的变化先「锁定」，单独看上中游 | 先锁完单率不变→再锁接单率不变→最后看派单量「单独」贡献了多少。这种算法下，中下游的变化被固定为 Q2 的值——所以上游的贡献会变小 |
| **结构分解法** | 不按转化链，按「结构」→「效率」→「纯度」的顺序。结构 = 最容易变的（流量结构、渠道结构），效率 = 需要优化的（转化率），纯度 = 最难变的（客单价、毛利率） | 先替结构因子（最可控），再替效率因子，最后替纯度因子 | 你在做资源分配决策——先把最容易变的动了，看看还剩多少缺口需要靠「难但持久」的改动来填 | 先把「派单量的渠道结构」变了（停了低质渠道→派单量可能降但质量升），再看剩下的完单量缺口有多大——如果缺口不大，就不动接单率和完单率 |

**③ 核心原则**：不要纠结「哪种算法更正确」——它们都是正确的近似。你该纠结的是「我的业务场景是什么」——如果你在做渠道优化（上游是杠杆），用上游优先；如果你在评估产品改版（中下游是杠杆），用下游优先。

**④ 与核心概念的关系**：这是贡献度分析的「升级版」——主课教你怎么做逐一替换，拓展告诉你**替换顺序有陷阱**以及**怎么根据场景选顺序**。

### 练习：选择替换顺序

以下 3 个场景，分别应该用哪种替换顺序？为什么？

**场景 A**：快达 Q3 做了渠道优化——砍掉了 3 个低质投放渠道，新增了 2 个高质渠道。派单量降了 10% 但完单量涨了 5%（因为来的都是高质量乘客，取消率低）。你需要向 CEO 汇报「完单量涨 5% 主要是渠道优化的功劳还是接单率/完单率改善的功劳」。

**场景 B**：快达 Q3 做了产品大改版——重新设计了下单流程（影响完单率）和派单算法（影响接单率）。派单量没变。你需要判断「产品改版的哪个部分贡献最大」。

**场景 C**：快达 Q3 同时发生了三件事——竞争对手涨价（快达乘客增加→派单量↑）、快达自己也涨了价（客单价↑）、同时在成都做了司机拉新（接单率↑）。完单量和收入都涨了。你需要分清楚「多少增长是竞品送来的（不可持续）、多少是我们自己做的（可持续）」。

---

### 参考分析

| 场景 | 推荐顺序 | 为什么 |
|------|--------|------|
| A | **上游优先** | 渠道优化直接影响派单量（上游）。派单量降了但质量升了——下游的接单率和完单率自动改善。用上游优先可以区分「渠道优化的直接效果（派单量 -10%）」和「质量提升带来的间接效果（接单率/完单率 +X%）」 |
| B | **下游优先** | 产品改版影响的是中下游（完单率、接单率），上游（派单量）没变。用下游优先可以先「锁定」产品改版的效果（完单率+接单率的变化），再看派单量有没有间接被影响 |
| C | **结构分解法** | 三个因子同时动了，而且有些是外部原因（竞品涨价），有些是内部原因（自己涨价、司机拉新）。先用结构分解法区分「外部因素」→「内部策略」→「定价变化」，再分别归因——这样 CEO 知道「+X% 是竞品送的不持久，+Y% 是我们做出来的可持续」 |

## 概念群 A：[防护] 因子相关性陷阱——乘法拆解假设它们是独立的，但现实里不是

> **与核心概念的关系**：主课讲乘法拆解的时候，默认三个因子是「互相独立的」。但现实中——流量涨了，转化率通常会跌（因为来的大多是低意向用户）。如果你忽略这个相关性，你会高估流量的贡献、低估转化率的重要性。

### 🔴 认知冲突（错误暴露型）

快达 Q2 运营团队加大了投放——日派单量从 9.5 万涨到 11.2 万（+18%）。但同期接单率从 91% 跌到 89%（-2pp）。

运营王姐说：「派单量贡献了 +1.5 万单，接单率只拖了 -0.2 万单——净贡献 +1.3 万。投放效果很好！」

产品李哥反驳：「你说接单率只拖了 -0.2 万——但**接单率跌的一部分原因就是你的投放**。新渠道来的乘客取消率高、新渠道覆盖的城市司机少——这些因素让接单率跌了。你把接单率的跌算成『和我无关』，但它恰好是你造成的。」

**问题**：李哥的质疑有道理吗？逐一替换法在这种情况下有没有系统性的偏差？

> 有。逐一替换法假设「我变流量的时候，转化率不变」。但现实中，流量和转化率通常是**负相关**的——因为边际流量质量递减。忽略这个相关性，会系统性地**高估上游因子的贡献、低估下游因子的贡献**。

### [必修] 因子相关性——怎么识别、怎么应对

**① 一句话白话**：乘法拆解 = 把一根绳子剪成三段，分别量长度。现实中你剪的时候，剪第一刀的位置会影响第二刀能剪多少——三段的「长度」不是独立的。

**② 三种常见的因子相关性**：

| 因子对 | 通常的关系 | 为什么会相关 | 对逐一替换法的影响 | 快达案例 |
|--------|----------|-----------|----------------|--------|
| 流量 ↑ vs 转化率 ↓ | 负相关 | 边际流量质量递减——每多花一块钱投放，带来的用户质量比上一块钱差 | 逐一替换法会**高估流量的正贡献、低估转化率的负贡献**——因为替流量时「假设转化率不变」，但现实中转化率会跟着跌 | 派单量 +18%，但新增派单中有 40% 来自低质渠道（转化率只有 70%，远低于平均 91%）→ 流量的「真实贡献」要打折 |
| 客单价 ↑ vs 转化率 ↓ | 负相关 | 涨价 → 价格敏感用户放弃购买 → 转化率跌 | 逐一替换法会**高估涨价的收入贡献**——因为替客单价时「假设转化率不变」，但现实中转化率会跌 | 快达高峰涨价 10% → 单均收入涨了，但完单量跌了 8%（价格弹性 ≈ -0.8）→ 净收入 +2% 而不是 +10% |
| 客单价 ↑ vs 毛利率 ↑ | 正相关 | 涨价直接提升毛利 | 正相关不会导致逐一替换法偏差——两个因子方向一致 | 涨价→单均收入↑且毛利率↑，两者同向，逐一替换法不会系统性高估或低估 |

**③ 应对方法——敏感度标注**：

在逐一替换法的结果旁边，加一列「如果两个因子相关，结论会差多少」。不需要精确算相关度——只需要做**情景推演**：

```
「派单量的贡献 = +1.5 万」← 这个数字假设接单率不变
但如果派单量涨的同时接单率自然跌 1pp（因为边际流量质量下降），
真实贡献大概只有 +1.2 万而不是 +1.5 万——差了 0.3 万。
```

**④ 与核心概念的关系**：这是逐一替换法的 [防护] 概念——不是否定它，而是告诉你**它的假设前提**是什么，以及**当假设不成立时结论会偏多少**。

## 概念群 B：[防护] 贡献度陷阱——「涨了 50%」不等于「很重要」

> **与核心概念的关系**：加法拆解后，你最容易犯的错误就是盯着「变化率最大」的那个维度追。但真正该追的是「贡献度最大」的维度——它可能变化率只有 5%，但因为基数大，对整体的影响远超那个「涨了 50%」的小渠道。

### 🔴 认知冲突

快达 Q2 的完单量按城市拆解后，出现了以下数据。运营王姐一眼看到了「银川 +50%」，兴奋地说：「银川模式大获成功！应该在全国推广！」

| 城市 | Q1 完单 | Q2 完单 | 变化量 | 变化率 | 占总完单的比例（Q1） | 贡献度 |
|------|---------|---------|--------|--------|-----------------|--------|
| 苏州 | 32,000 | 35,200 | +3,200 | +10% | 40% | ? |
| 合肥 | 24,000 | 24,000 | 0 | 0% | 30% | ? |
| 成都 | 16,000 | 12,800 | -3,200 | -20% | 20% | ? |
| 银川 | 800 | 1,200 | +400 | **+50%** | 1% | ? |
| 武汉 | 7,200 | 7,500 | +300 | +4% | 9% | ? |
| **总计** | **80,000** | **80,700** | **+700** | **+0.9%** | 100% | 100% |

**问题**：银川的变化率最惊人（+50%），苏州的变化量最大（+3,200 单），成都跌得最惨（-3,200 单，-20%）。整体完单量几乎没变（+700 单）。你会建议 CEO 「推广银川模式」吗？

> 银川只占总体的 1%。它涨 50% = 多了 400 单。苏州涨 10% = 多了 3,200 单。苏州的贡献度是 3,200/700 = **457%**（它不但自己涨了，还抵消了成都的暴跌）。银川的贡献度是 400/700 = **57%**。追银川是追错了方向——你该追的是「成都为什么暴跌」，以及「苏州做了什么让它涨了 10%」。

### [必修] 贡献度 vs 变化率——永远先看贡献度

**① 一句话白话**：「变化率」像身高——银川长高了 50%，但它本来就是个小孩（基数 1%）。「贡献度」像体重——苏州只长高了 10%，但它是成年人（基数 40%），对总重量的影响远超银川。

**② 判断优先级的三步法**：

| 步骤 | 操作 | 快达案例 |
|------|------|--------|
| **第 1 步**：看贡献度排序——哪个维度对整体变化的解释力最强？ | 苏州贡献了 +457%（因为它的正增长抵消了成都的负增长，所以贡献度超过 100%），成都贡献了 -457%。银川只贡献了 57% |
| **第 2 步**：对贡献度前 3 的维度，再看变化率——判断「这个变化是异常信号还是正常波动」 | 成都 -20%→大概率是异常信号（一个城市跌了 1/5，不是正常波动）。苏州 +10%→可能是投放增加或竞品出问题。银川 +50%→因为基数太小，400 单的变动可能是正常波动（小分母陷阱！） |
| **第 3 步**：区分「可复制的成功」和「不可复制的运气」 | 银川的 +400 单——涨了 50% 听起来很厉害，但如果银川是因为「刚好有一个大企业签了年度用车合同，增加了 300 单」，这个 +50% 就不可复制。苏州的 +3,200 单——如果是因为「优化了苏州的派单算法，接单率从 88%→93%」，这个 +10% 可以复制到其他城市 |

**③ 与核心概念的关系**：这是加法拆解的 [防护] 概念——拆完之后不是看谁变化率大就追谁，而是**先看贡献度（对整体的解释力），再看变化率（是信号还是噪音），最后看可复制性（值不值得推广）**。

### 练习：贡献度判断

快达 Q3 的 GMV 按品类拆解——CEO 只看了变化率那一列，决定「加大对零食品类的投入」。请用贡献度三步法重新判断这个决策是否合理。

| 品类 | Q2 GMV | Q3 GMV | 变化率 | 权重 |
|------|--------|--------|--------|------|
| 正餐 | 500 万 | 480 万 | -4% | 50% |
| 下午茶 | 200 万 | 210 万 | +5% | 20% |
| 零食 | 50 万 | 75 万 | **+50%** | 5% |
| 饮品 | 250 万 | 235 万 | -6% | 25% |
| 总 GMV | 1000 万 | 1000 万 | 0% | 100% |

---

### 参考分析

| 步骤 | 分析 |
|------|------|
| **贡献度** | 零食变化量 = +25 万。正餐变化量 = -20 万。总变化 = 0（正负抵消）。零食的贡献度无法正常算（分母≈0），但它只占总 GMV 的 5%。即使它涨 50%，绝对值也只有 25 万——正餐只要涨 5% 就是 25 万。**零食不是 GMV 增长的杠杆——正餐才是** |
| **变化率判断** | 零食 +50%——但基数 50 万，+25 万可能是「上了一个爆款零食」或「做了大促」。正餐 -4%（-20 万）——如果这 20 万跌是因为「用户转向了零食」→零食是抢了正餐的份额（品类间替代）。如果 20 万跌是因为「正餐整体需求下降」→零食的涨和正餐的跌无关 |
| **可复制性** | 如果零食的 +25 万是因为上了一个爆款→不可复制（爆款不能月月有）。如果是因为优化了推荐算法→可复制（算法可以用在所有品类） |

**结论**：CEO 的「加大对零食投入」不一定错，但论据不能是「零食涨了 50%」。应该先搞清楚——正餐的 -20 万是不是被零食抢走了？如果是→不需要加大投入，需要调整品类间的推荐策略。如果不是→再判断零食的增长是否可持续。**结论背后是分析逻辑，不是变化率。**

## 概念群 C：[落地] 从因子到责任人——拆到最底层后，必须能指给一个人

> **与核心概念的关系**：主课说「拆到可行动的层次」。但什么叫「可行动」？本概念给一个简单标准——**你拆出来的最底层因子，能不能在前面加上一个谁的名字**。

### [必修] 责任人测试

**① 一句话白话**：一个好的拆解长这样——「GMV 跌了 → 因为转化率跌了 → 因为支付页加载时长涨了 → 技术部张工，支付页加载时长涨了 2 秒，预期修好后 GMV 回升 3%。」拆到最后，有人名，有归因，有预期。

**② 责任人测试——三步走**：

| 步骤 | 做什么 | 例子（快达完单量拆解） |
|------|--------|---------------------|
| **第 1 步**：拆到因子层 | 完单量 = 派单量 × 接单率 × 完单率 | ✅ 这一步主课已经教了 |
| **第 2 步**：每个因子前面加一个谁 | 派单量→增长团队（负责渠道和投放）。接单率→算法团队（负责派单匹配）。完单率→产品团队（负责下单流程和取消率） | ⚠️ 这一步是大多数分析跳过的——你拆开了，但没有指给「谁」 |
| **第 3 步**：给每个谁一个行动方向 | 增长团队：「派单量涨了但质量降了，把预算从低质渠道往高质渠道挪 30%」。算法团队：「接单率跌了，成都的派单距离太远导致司机拒单——缩短成都的最大派单半径」。产品团队：「完单率跌了，取消率高的用户在支付页平均停留 8 秒——支付页的加载需要优化」 | ✅ 这一步才是分析的价值——不是「接单率跌了 2pp」，而是「因为成都派单半径太大，建议缩短到 3km 以内」 |

**③ 「不可行动」因子识别表**：

| 因子 | 可行动吗？ | 为什么 | 怎么改 |
|------|:--:|------|------|
| 天数 | ❌ | 没人能改变「一个月有多少天」 | 如果是「用户月均活跃天数」→可行动（通过push/内容/功能增加活跃天数） |
| 市场大盘 | ❌ | 你控制不了整个市场 | 如果是「市场份额」→可行动（你的份额 = 你的完单 / 市场总完单。优化你的产品→你的完单涨→份额涨） |
| 天气 | ❌ | 下雨天完单量涨，但你不能控制天气 | 不需要改变天气——但可以做「雨天运力预案」：下雨天提前多派司机上线 |
| 客单价 | ✅ | 可以通过调整定价策略/补贴/会员体系来影响 | — |
| 转化率 | ✅ | 可以通过优化页面/简化流程/改善加载来影响 | — |

**原则**：如果一个因子你自己不能改变，要么把它拆到可以改变的层次，要么接受它不可控——把它单独标注为「外部因素」，不纳入内部 KPI。

### 练习：给下面的拆解结果加上责任人和行动方向

快达的乘客等待时长从 5 分钟恶化到 7 分钟。你拆解后发现：

```
等待时长 = 接单时长 + 接驾时长
接单时长 = 系统派单到司机接单的时间（从 2.5 分钟 → 4 分钟——主要恶化来源）
接驾时长 = 司机接单后到乘客上车的时间（从 2.5 分钟 → 3 分钟——轻微恶化）
再拆接单时长：因为成都的「司机拒单率」从 9% 暴涨到 27%（司机看到是短途单→拒单→系统重新派单→等待变长）
```

请为最底层的「成都司机拒单率 27%」加一个责任人和一个行动方向。

---

### 参考分析

| 步骤 | 内容 |
|------|------|
| 责任人 | **成都城市运营经理 + 派单算法负责人**。双责任人——运营经理解决「短途单的司机激励」（让短途单也有利可图→减少拒单动机），算法负责人解决「拒单的惩罚机制」（连续拒单 3 次→降低派单优先级） |
| 行动方向 | **短期（1 周内上线）**：对短途单（≤3km）提供阶梯补贴——0-1km 补 3 元、1-3km 补 1 元。让短途单的司机收入不低于长途单。**中期（2-4 周）**：算法加入「拒单惩罚」——1 小时内拒单超过 3 次的司机，接下来 30 分钟的派单优先级降低 50%。**验证**：上线 2 周后看成都的接单时长是否从 4 分钟回落。如果回落 < 0.5 分钟→说明「拒单不是主因」→需要回溯拆解，找下一个候选因子 |
| **关键** | 行动方向里有时间线（1 周/2-4 周）、有验证标准（回落 < 0.5 分钟→回溯）。这是「可行动的拆解」和「纸面上的拆解」的区别——后者只告诉你「接单时长涨了因为拒单率涨了」，前者告诉你「谁、什么时候、做什么、怎么判断有没有用」 |

## 概念群 D：[防护] 什么时候不该拆——辛普森悖论场景下，拆得越细越容易被骗

> **与核心概念的关系**：主课暗示「拆解总是好的」。但有一个场景拆解得越细越危险——当整体趋势和各子组趋势方向相反的时候（辛普森悖论）。

### [必修] 拆解前必做的方向检查

**① 一句话白话**：拆之前先看——整体和各子组的方向一致吗？如果不一致，你看到「整体涨了」就拆开去追「哪个子组贡献最大」，可能会得出完全相反的结论。

**案例**：快达 Q2 的「整体乘客次留率」从 58% 涨到 61%（+3pp）。你按「新老乘客」拆开：
- 新乘客次留：Q1 35% → Q2 30%（**跌了 5pp**）
- 老乘客次留：Q1 72% → Q2 73%（+1pp）

整体涨了，但每个子组都在跌或微涨。怎么回事？因为新老乘客的权重变了——Q1 新乘客占 20%（剩下 80% 老乘客），Q2 新乘客占 35%（投放加大，大量新乘客涌入）。新乘客留存低（30%），但因为权重从 20%→35%，把整体拉高了。

如果你拆开看到「整体涨了 3pp」就去追「老乘客留存涨了 1pp 的功劳」——你就完全搞错了方向。**真正的问题是新乘客留存跌了 5pp**——投放拉来的新乘客质量在急剧下降。

**② 拆解前的方向检查（三秒原则）**：

1. 整体涨了→拆开看子组——有没有子组在跌？
2. 如果有→算标准化（控制权重不变，子组的趋势是什么方向）。
3. 如果标准化方向和整体方向相反→**辛普森悖论**。此时贡献度分析会误导你——因为它会高估「权重变大的子组」的贡献。

**③ 与核心概念的关系**：这不是说「不要拆」，而是「拆之前先做方向检查」。3 秒的检查可以避免 30 分钟的误导。

# 第四部分：综合沙盘演练

## 综合练习 1：拆解快达的毛利率——从 C 层指标到责任人

快达 C 层「毛利率」从 18% 跌到 9%。请完成完整的拆解+责任人链。

**第 1 步**：乘法拆解。毛利率可以拆成什么？（提示：毛利率 = 平台佣金 / GMV。但更实用的拆法是按「单位经济」拆——每单赚的钱 / 每单收的钱）

**第 2 步**：加法拆解。按什么维度拆？哪些城市/时段的毛利率最差？

**第 3 步**：逐一替换。用上游优先法计算——哪个因子对毛利率下跌的贡献最大？

**第 4 步**：加责任人。最底层因子前面加上谁的名字和行动方向。

---

### 参考方案

**乘法拆解**（单均视角）：
```
单均毛利率 = (单均佣金 - 单均补贴) / 单均GMV
Q1: (5.0 - 1.4) / 25.0 = 14.4%（简化数据）
Q2: (5.2 - 3.1) / 26.0 = 8.1%
→ 佣金微涨(+0.2)，但补贴暴涨(+1.7)，GMV也涨了(+1.0)
→ 逐一替换：补贴暴涨是主因（拉低了 4.2pp），GMV涨反而「稀释」了毛利率（分母变大→比率下降 1.1pp），佣金涨改善了 1.0pp
```

**加法拆解**（按城市）：成都毛利率 -3%、合肥 12%、苏州 17%。成都完单量占 10%，但亏损绝对值最大。

**责任人**：补贴暴涨 → 运营王姐（检查 Q2 新增了哪些补贴项目、哪些是不必要的）+ 财务陈姐（设补贴预算上限，超预算自动暂停）。

## 综合练习 2：检验因子相关性

上面逐一替换法算出来「补贴暴涨拉低了 4.2pp」。但这个计算假设「补贴涨了、GMV 不变」。

**问题**：在现实中，补贴涨了通常会带来什么连锁反应？这个反应会怎么影响「补贴拉低 4.2pp」这个结论？

---

### 参考分析

补贴涨了→完单量通常会涨（因为更便宜了→更多人打车）。所以「补贴涨了」和「GMV 涨了」是正相关的——两者方向相同（都在涨）。逐一替换法在这里不会系统性高估或低估——因为两个因子对毛利率都是「向下拉」的（补贴涨→毛利率跌，GMV 涨→分母变大→毛利率跌），方向一致。

但如果「佣金涨了」导致「完单量跌了」（因为平台涨价→乘客减少）——那逐一替换法会系统性高估佣金涨的正贡献。因为替佣金时「假设 GMV 不变」，但现实中佣金涨会导致 GMV 跌。

**关键**：不是所有因子都有相关性。你需要识别「有明显互相依赖关系」的因子对——它们通常是同一件事的两面（如价格和销量）。对于这类因子对，逐一替换法的结论需要打折。

## 综合练习 3：识别「拆解过度」

快达的产品李哥把完单量拆成了 7 层——最后一层的因子包括了「Push 文案的字号」「App 图标颜色」「注册页按钮的圆角半径」。

**问题**：这个拆解有什么问题？拆到哪一层算「拆够了」？

---

### 参考分析

**问题**：最后三层的因子和完单量的关系太远了。「按钮圆角半径」→「点击率」→「下单率」→「完单量」——每一层都有大量的噪音和混淆变量。你根本没法证明「圆角半径变了导致完单量变了」——AB 实验都测不出来这个效应量。

**「拆够了」的标准**——两句话：
1. 最底层因子可以被 AB 实验**独立验证**（「改了按钮半径→完单量变了吗？」——你跑 100 万用户可能都跑不出显著差异 → 拆过头了）
2. 最底层因子有**独立的责任人**（如果「按钮半径」和「点击率」属于同一个人的职责范围——不需要拆两层，合并成「页面体验」就够了）

**快达的合理层数**：完单量（1）→ 派单量 × 接单率 × 完单率（2）→ 完单率 = 下单率 × 支付率（3）→ 到这里够了。再往下拆（「支付按钮颜色」）就是过度拆解。
