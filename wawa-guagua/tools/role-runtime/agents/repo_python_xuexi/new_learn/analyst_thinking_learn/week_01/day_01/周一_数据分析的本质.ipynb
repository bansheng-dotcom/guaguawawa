## 一、核心概念与第一性原理

### 1.1 数据分析不是"看数字"

很多人以为数据分析就是：打开一个表 → 看看数字 → 画个图 → 交报告。

这不叫分析，这叫**报数**。

真正的数据分析是：**有人提出一个问题，你用数据找到答案，并且这个答案能指导下一步行动。**

三个要素缺一不可：
- **问题**：不是"帮我看看数据"，而是"为什么上个月GMV跌了15%？"
- **答案**：不是"因为客单价降了"，而是"客单价降了是因为新用户占比从20%涨到35%，新用户客单价只有老用户的40%"
- **行动**：不是"建议继续观察"，而是"建议对新用户做首单补贴，预计可挽回GMV跌幅的60%"

### 1.2 分析师的本质角色：翻译官

分析师 = 翻译官。把业务问题翻译成数据问题，把数据结果翻译成业务决策。

```
业务方："为什么用户最近不买了？"          ← 模糊的业务问题
    ↓ 翻译成数据问题
分析师："过去30天的复购率是多少？与30-60天前比变化了多少？
         变化来自新用户还是老用户？哪个品类跌得最厉害？"  ← 可量化的问题
    ↓ 分析
分析师："复购率从23%降到18%。拆解发现：老用户复购稳定，
         新用户的7日复购从15%跌到8%，主要集中在服装品类。"  ← 数据发现
    ↓ 翻译成业务决策
业务方："那我们对服装品类新用户做首单后7天内发优惠券。"  ← 可落地的行动
```

这个翻译过程，看起来很简单。但现实中绝大多数"分析"死在第一步——没搞清楚问题就开工了。

### 1.3 从第一性原理推导：为什么需要"五步循环"

分析的本质是"消除不确定性"。业务方有一个模糊的担忧（"生意好像变差了"），你的工作是把这个担忧变成可验证的事实。

这需要五个步骤，每一步解决一个特定的不确定性：

```
提问 → 拆解 → 取数 → 分析 → 落地
  ↑______________________________|
```

| 步骤 | 消除的不确定性 | 做什么 | 常见错误 |
|------|--------------|--------|---------|
| **提问** | "到底要回答什么问题？" | 把模糊需求变成可量化的问题 | 没搞清楚就开工 |
| **拆解** | "这个问题可以从哪几个角度回答？" | 把大问题拆成小问题，确定分析框架 | 一上来就拉数据 |
| **取数** | "需要哪些数据来回答这些问题？" | 从数据库/文件拿到需要的数据 | 拉太多无关数据 |
| **分析** | "数据告诉了我们什么？" | 计算、对比、找模式、验证假设 | 只描述不解释 |
| **落地** | "所以呢？该做什么？" | 把发现翻译成业务方听得懂、做得了的建议 | "建议继续观察" |

**为什么是循环？** 因为落地之后会产生新的问题——这是一个持续优化的过程。

### 1.4 分析师的三个层次

| 层次 | 角色 | 典型行为 | 价值 |
|------|------|---------|------|
| 第一层 | 取数工具人 | "你要什么数据我帮你拉" | 低——可被工具替代 |
| 第二层 | 分析执行者 | "数据告诉我们三件事：..." | 中——能发现问题 |
| 第三层 | 业务伙伴 | "我建议你做X，因为数据表明Y，预计带来Z" | 高——能推动决策 |

绝大多数分析师卡在第一层和第二层之间。这门课的目标是把你推到第三层。

> **今天最重要的认知**：分析始于问题，终于行动。一个好的分析师80%的时间花在"定义问题"和"拆解问题"上——剩下的20%才是拉数据和计算。如果你拿到需求就开工，你已经输了。

## 二、知识点详细拆解

### 2.1 好问题 vs 坏问题

**为什么重要**：问题是分析的起点。起点错了，后面的分析再精细也没用。

| 坏问题 | 为什么坏 | 改造后的好问题 |
|--------|---------|---------------|
| "帮我看看数据" | 没有目标，什么都看等于什么都没看 | "上周DAU跌了，帮我定位原因" |
| "用户喜欢我们的新产品吗？" | 不可量化，"喜欢"怎么定义？ | "新产品上线后，使用过的人数、次日留存、NPS评分分别是多少？" |
| "为什么GMV跌了？" | 太宽泛 | "GMV跌了10%，是流量跌了还是转化率跌了？哪个品类/渠道变化最大？" |
| "做个分析报告" | 不知道给谁看、做什么决策 | "给运营总监看，帮ta决定下个月的优惠券预算该怎么分配" |

**识别坏问题的三个信号**：
1. 问题里没有数字（不可量化）
2. 问题里没有时间范围（"最近"是多近？）
3. 问题里没有决策意图（"看看"不等于"决定"）

### 2.2 拆解：分析师的第一个核心能力

**为什么重要**：大问题无法直接回答。"为什么GMV跌了"这个问题的答案不是藏在某一行数据里，而是需要拆成更小的问题逐一排查。

拆解的两种方式：

**乘法拆解**（按转化链路）：
```
GMV = 流量 × 转化率 × 客单价
    = (自然流量 + 付费流量) × (浏览→加购 × 加购→下单 × 下单→支付) × (件单价 × 人均件数)
```

**加法拆解**（按维度分类）：
```
GMV = 新用户GMV + 老用户GMV
    = 服装GMV + 数码GMV + 食品GMV + ...
    = 各渠道GMV之和
```

拆解之后，你不再面对"GMV为什么跌"这个庞然大物，而是逐个检查：
- 自然流量跌了吗？
- 加购→下单的转化率跌了吗？
- 服装品类的GMV跌了吗？

每一个小问题都可以用数据直接回答。**拆解的终极目的：把"回答不了的问题"变成"回答得了的问题"。**

### 2.3 假设驱动的分析 vs 数据驱动的分析

**为什么重要**：这是新手和老手的思维分水岭。

| | 数据驱动（新手） | 假设驱动（老手） |
|------|---------------|---------------|
| 方式 | 先拉数据 → 再找规律 | 先猜 → 再验证 |
| 效率 | 拉50张表，不知哪个有用 | 验证3-5个假设即可 |
| 风险 | 被数据中的噪音误导，发现假规律 | 假设本身可能偏（遗漏真因） |

**正确的做法**：假设驱动为主，数据探索为辅。先根据业务理解列出3-5个最可能的假设，逐一用数据验证。如果都不成立，再做探索性分析。

### 2.4 分析与行动的"最后一公里"

**为什么重要**：分析不落地 = 没做。

常见的"假落地"：
- "建议继续观察" → 其实是没有结论
- "需要进一步分析" → 其实是分析没做完
- "数据表明存在相关性" → 其实是不敢下判断

**好的落地建议应该满足**：
1. **可执行**：运营/产品拿到就能做，不需要再"翻译"
2. **有预期**：做了之后预计什么指标会怎么变化
3. **可验证**：做完之后能用数据判断是否达到了预期

## 四、业务场景应用

### 场景1：老板丢给你一句话——"DAU涨了但收入没涨"

> "小王，我们App的DAU最近涨了20%，但收入没涨。你去看看怎么回事。"

**初级分析师的做法**：打开数据后台，把DAU曲线和收入曲线截图发给老板，附一句"确实，DAU涨了收入没涨"。老板看了更困惑了。

**资深分析师的做法**：

**第1步 — 确认问题**："老板，你说的'最近'是指多长时间？跟什么时间段比？收入是指GMV还是毛利？"
→ 问清楚定义，避免后续返工。

**第2步 — 拆解**：
```
收入 = DAU × 付费转化率 × ARPU
DAU涨了20%但收入没涨 → 说明付费转化率或ARPU跌了
继续拆：DAU = 新增 + 留存 + 回流
        付费转化率 = 新用户付费率 vs 老用户付费率
        ARPU = 付费用户人均消费
```

**第3步 — 先提假设，再验证**："我猜是新用户涨了（因为最近加大了投放），但新用户付费转化率远低于老用户，所以多出来的DAU没贡献多少收入。"

**第4步 — 验证假设**：拉数据——新增用户涨了40%，新用户付费率只有3%（老用户15%），确认假设成立。

**第5步 — 给出行动建议**："DAU增长来自投放拉新，但新用户付费率太低。建议对新用户做首单立减，预计能把新用户付费率从3%提到8%，带来增量收入约X万/月。"

**用vs不用的对比**：
- 不用五步法：花了2小时拉了50张表，发给老板，老板不回。
- 用五步法：1小时完成分析，老板回复"好的，去执行"。

### 场景2：运营说"我感觉用户质量变差了"——怎么把"感觉"变成分析

运营总监在周会上说："我感觉最近用户质量变差了。"

**分析师的第一反应不是"我去拉数据"，而是问**：
- "你说的'用户质量'具体指什么？新用户留存？付费转化？LTV？"
- "'最近'是多近？跟哪段时间比？"

**然后做假设驱动的分析**：
- 假设1：新用户次留在降 → 拉新用户留存趋势
- 假设2：新增用户中低质渠道占比变高 → 拉各渠道新用户占比和留存
- 假设3：核心老用户的消费频率在降 → 拉核心用户的购买频次趋势

**1小时后汇报**：
"你的感觉是对的，用户质量在下降。主要原因是新增中信息流渠道占比从30%涨到55%，但这个渠道的次留只有8%（整体平均20%）。建议控制信息流渠道占比在40%以下，或优化投放素材提高用户匹配度。"

**用vs不用的对比**：
- 没有假设驱动：花了3天拉了50张表，发现"确实有些指标降了"——但说不清哪个是主要原因。
- 用假设驱动：1小时验证了3个假设，锁定根因，给出了具体建议。

### 场景3：你朋友开了一家奶茶店——最接地气的分析案例

你朋友说："最近生意好像变差了。"

**初级朋友**："是吗？我最近去的时候人还挺多的啊。"（用印象代替数据）

**你（分析师朋友）**：
1. 先确认："差了多少？上周日均收入多少？这周呢？"
2. 再拆解："日均收入 = 客流量 × 客单价。哪个变了？"
3. 取数据：看了收银记录——客流量没变，客单价从28降到了18。
4. 再拆客单价："客单价 = 杯均价格 × 人均杯数。"→ 人均杯数从1.8降到了1.1。
5. 找原因：发现"第二杯半价"活动上周到期了，没续。
6. 给建议："恢复活动，或者试一下'买二送一'看看效果。"

**用vs不用的对比**：
- 不用分析思维：凭感觉说"可能是天气热了大家不想喝奶茶"，然后不做任何改变。
- 用分析思维：数据驱动地找到了根因（活动到期），精确地给出了解决方案（恢复活动）。

## 五、分级检验清单

### 初级检验（必须能独立完成的）
- [ ] 能用自己的话讲出数据分析的五步循环
- [ ] 能把一个模糊的业务问题改造成可量化的问题（至少改3个）
- [ ] 能区分"报数"和"分析"

**答案提示**：
- 五步循环：提问→拆解→取数→分析→落地，是一个持续循环
- 量化改造：加入时间范围、具体指标、对比基准
- 报数 vs 分析：报数只描述发生了什么，分析还解释为什么+建议做什么

### 中级检验（能解释给同事听的原理）
- [ ] 能把一个复合指标（如GMV、DAU）拆解到第二层
- [ ] 能识别一个分析需求中缺失的关键信息（时间范围、对比基准、定义口径、决策意图）
- [ ] 能把数据发现翻译成业务建议（至少给一个"你可以做什么"）

**答案提示**：
- GMV拆解：GMV = 流量 × 转化率 × 客单价，每个因子可以继续拆
- 关键信息检查：时间多长？跟什么比？谁用这个分析做决策？
- 业务建议模板："建议____，因为数据表明____，预计带来____的变化"

### 高级检验（能举一反三的迁移能力）
- [ ] 拆解后能判断每个子指标的优先级（先看哪个、哪个影响最大）
- [ ] 能在分析前先写假设（"我猜问题出在XX环节"）再验证，而非漫无目的地探索
- [ ] 向非技术同事解释分析结果时，对方能听懂并在30秒内做出决策

**答案提示**：
- 优先级判断：看权重（哪个因子占比最大）和变化幅度（哪个因子波动最大）
- 假设驱动：基于业务理解先猜最可能的2-3个原因，逐一验证
- 30秒测试：如果你的解释超过30秒对方还没点头→你的表达需要简化

## 七、当日小结

> **今天最重要的一个认知**：分析始于问题，终于行动。一个好的分析师80%的时间花在"定义问题"和"拆解问题"上——剩下的20%才是拉数据和计算。如果你拿到需求就开工，你已经输了。

**三条金句带回家**：
1. 数据分析不是"看数字"，是"把业务问题翻译成数据问题，再把数据结果翻译成业务决策"
2. 五步循环（提问→拆解→取数→分析→落地）是你面对任何分析需求时的默认流程
3. "建议继续观察"五个字是分析师最贵的五个字——它意味着你的分析没有产生任何价值

### 课后拓展

1. **练习**：找一个你自己生活中的决策（比如"最近花钱太多了"），用五步法重新审视它。你会怎么提问？怎么拆解？需要什么数据？

2. **思考**：为什么"先提假设再验证"比"先拉数据再找规律"更高效？前者有什么风险？

3. **预习**：明天要学的"指标的第一性原理"会告诉你，不是所有数字都叫指标——怎么区分"好指标"和"虚荣指标"。

4. **行动**：下次有人找你"帮我看看数据"，先别开工。回一句"你想通过这个数据做什么决策？"——这是你从第一层到第三层的第一步。
