# 拓展：好指标，到底好在哪？——从「背5条标准」到「能用它做判断」

> **本拓展讲什么**：主课告诉你好指标有5条标准。本拓展带你用这5条标准去**诊断真实的指标**——就像医生拿着检查清单给病人做体检，每条标准对应一项检查。
> **适合谁**：刚学完周二主课，知道5条标准的名字但不知道怎么用。
> **预计时长**：约 2h（不含代码，全部是阅读+思考+填空练习）

# 第一部分：沙盘设定

> 后续所有案例和练习都在这个沙盘里。先花 3 分钟读完，建立一个完整的场景。

## 公司画像

**快达出行**——一个网约车平台，你手机里的打车 App 差不多就是这个模式。

- **产品**：乘客叫车 → 系统派给附近的司机 → 司机接单 → 到达目的地 → 乘客付款
- **怎么赚钱**：每单抽取 20% 作为平台佣金（乘客付 25 元，平台拿 5 元，司机拿 20 元）
- **规模**：每天约 120 万人打开 App（乘客端），平台上活跃着约 8 万个司机
- **阶段**：B 轮融资（已经证明了产品有人用，现在要证明能赚钱）

## 当前发生了什么

CEO 在周一早会上发了火：

> 「Q2 的完单量涨了 25%，从每天 8 万单涨到 10 万单。但毛利率从 18% 跌到了 9%。」
> 「运营跟我说完单量涨是好事。财务跟我说毛利跌了要砍预算。产品跟我说你们先统一个标准。」
> 「我就问一句：到底什么是『好』？完单量涨算好，还是毛利高算好？」

**关键术语速查**（本拓展第一次出现的行业词，后面不再解释）：
- **完单量**：每天乘客打到车、司机送到目的地、双方都确认了的订单数。打车 App 里最核心的交易指标。
- **毛利率**：(收入 - 直接成本) / 收入。比如乘客付 25 元，司机拿 20 元，平台剩 5 元，毛利率 = 5/25 = 20%。毛利率越高，平台越赚钱。
- **DAU**（Daily Active Users）：每天至少打开一次 App 的用户数。
- **CAC**（Customer Acquisition Cost）：获取一个新用户要花多少钱（投放广告费 ÷ 带来的新用户数）。

## 你的角色

你是快达出行唯一的数据分析师。CEO 让你在下周五的管理层会上给出答案：
1. 快达现在盯的那些数字里，哪些是「看着好看但没用」的虚荣指标？
2. 快达该用什么作为北极星指标？怎么防止它像完单量一样被「玩坏」？
3. 怎么确保全公司所有人算出来的「完单量」是同一个数字？

| 人物 | 角色 | 最近在盯什么数字 | 为什么盯这个 |
|------|------|----------------|------------|
| 张总 | CEO | 完单量、毛利率 | 想知道公司到底健不健康 |
| 王姐 | 运营 VP | 完单量、新增乘客数 | 她管投放预算，需要证明钱花得值 |

**核心矛盾**：四个人盯着四个不同的数字。完单量涨的时候王姐说「投放有效」，陈姐说「毛利在跌」，李哥说「等待时长涨了说明体验在变差」。CEO 不知道该听谁的——因为没有一个统一的标准来判断「什么是好」。

## 🔴 核心深化：好指标 5 条标准——不只是「背出来」，而是「用起来」

### 🔴 认知冲突

快达运营 VP 王姐在周会上非常兴奋：

### [必修] 5 条标准的诊断检查表

主课给了 5 条标准的定义。但定义是死的——你需要一张「检查表」，拿着它去一条一条对照。

**这里的关键**：不是说「我感觉它满足第 1 条」，而是要找到**证据**。证据可以是数据、可以是逻辑推理、可以是不同部门的态度对比。

下面就是这张检查表——以快达的「日均完单量」为例，走一遍完整的诊断：

| 标准 | 诊断问题（这样问自己） | 快达完单量的实际表现 | 通过？ |
|------|-------------------|-----------------|-------|
| **① 反映用户价值** | 「这个数字涨了，用户是更开心了还是更烦了？」司机奖励让完单量涨了，但乘客等得更久了——说明完单量在跟着「平台多赚钱」走，不是跟着「乘客少等一分钟」走。 | 奖励驱动下，完单量↑ 但等待时长也↑——用户价值没有同步提升 | ❌ 不通过 |
| **② 可指引长期战略** | 「如果连续 3 个月完单量都涨，能说明公司在往对的方向走吗？」不一定——因为完单量可以被补贴「买」来。补贴一停，完单量大概率跌回去。真正重要的是「没有补贴的情况下用户还打不打车」。 | 完单量增长包含大量补贴水分，不能独立指引长期方向 | ❌ 不通过 |
| **③ 可拆解** | 「完单量跌了，你能分别告诉增长团队、算法团队、运营团队各自该做什么吗？」能——完单量 = 派单量 × 接单率 × 完单率。派单量归增长团队（负责拉新），接单率归算法团队（负责匹配效率），完单率归产品+运营（负责下单流程和司机激励）。 | 可拆成 3 层，每层有明确的责任人 | ✅ 通过 |
| **④ 抗短期波动** | 「如果今天完单量比昨天跌了 15%，你会紧张吗？」如果你知道「周一本来就比周五低」或者「下雨天完单量会暴涨」，那 15% 的日波动可能只是正常周期——但如果每次都紧张，说明这个数太敏感了。更好的做法是看「周均完单量」而非「日均完单量」。完单量的日波动确实有规律（工作日高、周末低），看周均就稳定得多。 | 日粒度波动有周期性，周粒度更稳定 | ✅ 通过（但强烈建议用周均而非日均） |
| **⑤ 跨部门共识** | 「四个 VP 都认可『完单量涨了就是好』吗？」运营王姐给完单量打 5 分（最重要），财务陈姐只给 2 分（她更关心毛利率），产品李哥给 3 分（他觉得等待时长更反映产品好坏）。三个人给了三个完全不同的分数——这说明完单量没有跨部门共识。 | 运营 5 分 / 产品 3 分 / 财务 2 分——分歧巨大 | ❌ 不通过 |

**诊断结论**：完单量在 5 条标准中通过了 2 条（可拆解、抗短期波动），不通过 3 条（用户价值、长期战略、跨部门共识）。它不是一个足够好的北极星指标——至少，在没有护栏的情况下不是。

| 标准 | 分析 | 通过？ |

## 概念群 A：[并行] 虚荣指标三问识别法——不用背定义，问三个问题就够

> **与核心概念的关系**：主课给了虚荣指标的定义和对照表。本概念给你一个更简单的判断工具——三个问题。任何一个数字，问完这三个问题，你就知道它是不是虚荣的。

### 🔴 认知冲突

快达的 BI 看板上有 35 个指标。CEO 说「把里面虚荣的拿掉」。

如果你翻出主课的虚荣指标定义——「让你自我感觉良好，但不能帮你做决策的数字」——你会发现这个定义太模糊了。什么叫「自我感觉良好」？什么叫「不能帮你做决策」？面对 35 个指标，你需要更精确的工具。

以下是快达看板上的 6 个数字。试试用你的直觉判断哪些是虚荣的：

| 数字 | 过去 12 个月的表现 | 你的直觉 |
|------|-----------------|--------|
| 累计注册乘客数：500 万 | 月月涨 | ? |
| 日均完单量：10.2 万 | 有涨有跌 | ? |
| App 累计下载量：800 万 | 月月涨 | ? |
| 司机周活跃率：68% | 在 63%-72% 之间波动 | ? |
| 月 GMV：3200 万 | 有涨有跌 | ? |
| 微博粉丝数：3.5 万 | 月月涨 | ? |

> 下面给一个更简单的判断方法——只需要问三个问题。

### [必修] 虚荣指标三问识别法

**① 一句话白话**：判断一个数字是不是虚荣，只需要问三句话。触发任何一句，它就值得怀疑。触发两句以上，基本就是虚荣的。

**② 三问**：

### 第一问：这个数字会跌吗？

> 如果一个数字在过去 12 个月里几乎从不下跌（比如 12 个月中只跌过 1-2 次），说明它不是「反映变化」，而是「单纯在累积」。

在快达的例子里：
- 「累计注册乘客数」——永远不会跌。注册了就是注册了，没人能「取消注册」。→ ⚠️ 触发
- 「日均完单量」——下雨天涨、晴天跌、周一低、周五高。是活的数字。→ ✅ 不触发
- 「App 累计下载量」——永远不会跌。→ ⚠️ 触发

### 第二问：这个数字涨了，你知道该做什么吗？

> 如果答案是「不确定」「好像没什么可做的」「继续做现在的事就行」，那它不能帮你做决策。

在快达的例子里：
- 「累计注册 500 万→510 万」——然后呢？你该做什么？不知道。→ ⚠️ 触发
- 「完单量 9.8 万→10.2 万」——查是哪个城市涨的、哪个时段涨的、要不要加运力。知道该做什么。→ ✅ 不触发
- 「微博粉丝 3.5 万→3.6 万」——然后呢？微博粉丝涨了跟打车有什么关系？→ ⚠️ 触发

### 第三问：这个数字代表的，是用户得到了好处，还是公司得到了好处？

> 好指标优先衡量「用户得到了什么」。虚荣指标通常衡量「公司得到了什么」。

在快达的例子里：
- 「完单量」——用户得到了「打到车」，公司得到了「佣金」。**两边都有**。→ ✅ 不触发
- 「GMV」——衡量的是「乘客付了多少钱（公司的收入）」。用户得到了「坐车」，但 GMV 这个数字本身不反映用户体验。→ ⚠️ 部分触发
- 「累计注册」——公司得到了「注册用户数」，用户什么都没得到（注册≠打到车）。→ ⚠️ 触发

### 判定规则

触发 0 问 → ✅ 大概率不是虚荣指标
触发 1 问 → ⚠️ 可疑，需要进一步审视
触发 ≥ 2 问 → 🚨 强烈怀疑是虚荣指标，应该从核心看板移除或降级为辅助参考

**③ 与核心概念的关系**：三问识别法是核心概念第①④条（反映用户价值+抗短期波动）的简化操作版——让主课的定义变得可以直接使用。

## 概念群 B：[防护] 小分母陷阱——为什么「拉萨次留 78%」不可信

> **与核心概念的关系**：好指标第④条说「抗短期波动」。但有一种波动不是业务造成的，纯粹是数学造成的——分母太小。这个概念让你以后看到比率时，第一反应是「分母多大？」

### 🔴 认知冲突

快达最近新开了 3 个城市。运营王姐非常兴奋，在周会上展示了这张表：

| 城市 | 上线多久 | 每天多少单 | 乘客次日留存率 | 运营的判断 |
|------|---------|----------|-------------|----------|
| 苏州 | 30 天 | 2500 单 | **62%** | 「表现最好，继续保持」 |
| 合肥 | 30 天 | 1800 单 | **58%** | 「也不错」 |
| 拉萨 | 14 天 | 85 单 | **78%** | 「拉萨最牛！要加大投入！」 |

你盯着「拉萨 78%」看了 3 秒——直觉告诉你这个数字不对劲。但你能说清楚它为什么不对劲吗？

> 下面是答案：分母太小，导致这个 78% 可能根本不是真实的。

### [必修] 小分母陷阱——手工推演版

**① 一句话白话**：看比率（留存率、转化率、好评率）的时候，如果分母太小，这个比率就像抛硬币——今天抛 10 次得到 8 次正面，不代表你的硬币比别人厉害，只是因为你抛的次数太少。

**② 手工推演——用抛硬币来理解**：

假设拉萨的「真实次日留存率」是 50%（这个真实值我们永远看不到，只能通过抽样估计）。

现在拉薩每天大约 85 单，意味着「昨天新增了大约 85 个乘客」。这 85 个人里，明天有多少人会回来？——这本质上是一个**随机抽样**的问题。

就像你从一个装着 50% 红球、50% 蓝球的袋子里摸球。摸 2300 次（苏州），红球比例会很接近 50%。摸 85 次（拉萨），红球比例可能在 35%~65% 之间任意位置。

**真实的拉萨留存率可能根本不是 78%**。它可能只有 50%，而你看到的 78% 纯粹是因为「今天运气好，碰巧昨天那 85 个新用户里愿意回来的比较多」。

**③ 手工推算工具——标准误（Standard Error）**：

统计学里有一个公式可以帮你估算「抽这么少的人，结果大概偏多远」：

$$标准误(SE) = \sqrt{\frac{p \times (1-p)}{n}}$$

其中 p 是你看到的比率，n 是分母（样本量）。

我们来手算三个城市：

| 城市 | 分母 n | 看到 p | 手算 SE | SE 的含义 |
|------|--------|--------|---------|---------|
| 苏州 | 约 2300 | 62% | √(0.62×0.38/2300) ≈ 1.0% | 如果你「重抽」2300 人再测一次，结果大概会偏 ±1% |
| 合肥 | 约 1700 | 58% | √(0.58×0.42/1700) ≈ 1.2% | 重测一次大概偏 ±1.2% |
| 拉萨 | 约 80 | 78% | √(0.78×0.22/80) ≈ 4.6% | 重测一次大概偏 ±4.6%！ |

**95% 置信区间（约等于 p ± 2×SE）**：
- 苏州：62% ± 2.0% = [60%, 64%] — 很窄，可信
- 合肥：58% ± 2.4% = [56%, 60%] — 较窄，基本可信
- 拉萨：78% ± 9.2% = [69%, 87%] — **宽达 18 个百分点！几乎不可用！**

这意味着什么？拉萨的真实次留率可能在 69% 到 87% 之间——如果真实值是 69%，那它和苏州的 62% 差距并不大（7pp）。你因为看到了 78% 就判定「拉萨最牛」，是被小分母骗了。

**④ 三条铁律（不用背，理解了自然记住）**：

| 分母 n | 这个比率能信吗？ | 怎么做 |
|--------|----------------|--------|
| n < 30 | 基本没意义 | 别看比率，只看绝对值。如「拉萨本周新增 85 人」就够了 |
| n 30~100 | 可以参考方向，但不能精确对比 | 必须附带 ±范围（如「拉萨次留约 78% ± 9%」）。不做城市间排名对比 |
| n > 100 | 可以做对比 | 但 ±5pp 以内的差距可能是噪音——别为「62% vs 65%」的差别做重大决策 |

**⑤ 与核心概念的关系**：直接对应好指标第④条「抗短期波动」——小分母造成的波动不是业务问题，是数学问题。你不懂这个，就会对噪音做决策。

### 练习：判断以下快达的数据陈述哪些受了小分母的骗

## 概念群 C：[防护] 指标腐烂——一个 KPI 被设为目标后，会经历什么

> **与核心概念的关系**：主课用一句话提到了古德哈特定律（「当一个指标成为目标，它就不再是好指标」）。本概念把这个定律展开——告诉你好指标是如何一步一步变坏的，以及在哪个阶段还能救回来。

### 🔴 认知冲突

回过头来看快达 CEO 的困惑。Q1 的时候，「日均完单量」被设为公司的北极星指标。

**Q1 第一个月**：完单量从 8 万涨到 9.5 万（+19%）。团队做了什么？算法优化了派单效率，产品简化了下单流程——这些是真正的改进。完单量涨了，业务真的更好了。

**Q1 第三个月**：完单量继续涨到 10.5 万。但开始出现奇怪的现象——司机接单后取消率从 2% 涨到了 8%。深挖发现：司机发现接单后如果乘客目的地太近（短途单），取消后等系统重新派单，可能派到一个更远的。完单量还在涨，但乘客等得更久了。

**Q2 第二个月**：完单量涨到 11 万。但毛利率从 15% 跌到了 7%。怎么回事？运营为了冲完单量，开始对短途单做补贴——「你接这单，平台补你 3 元」。完单量继续涨，但每一单都在亏钱。

**问题**：完单量在 Q1 第一个月是「好指标」吗？是。到了 Q2 还是吗？不是。它是从什么时候开始变坏的？你能提前预判吗？

> 指标不会一夜之间腐烂。它经历五个可预测的阶段。如果你知道每个阶段的前兆信号，你可以在阶段 2-3 就换指标——而不是等到阶段 4（利润崩盘）才发现。

### [必修] 指标腐烂的五阶段模型

**① 一句话白话**：好指标被设为 KPI 后，就像鲜肉放在室温下——它不会瞬间腐烂，而是经过几个阶段慢慢变质。每个阶段都有信号——你的任务是学会识别这些信号。

**② 五阶段详表**：

| 阶段 | 叫什么 | 发生了什么 | 快达完单量的表现 | 这个阶段还能救吗？ |
|------|--------|---------|---------------|----------------|
| **1** | 蜜月期 | 指标涨了，业务真的更好了。所有人都在做正确的事——优化产品、提升效率。指标和业务利益一致。 | Q1 月：完单 +19%。派单优化→匹配更快→真增长。毛利率稳定。 | ✅ 指标健康，正常使用 |
| **2** | 试探期 | 少数人发现「有一种方式能让指标涨，但对业务未必好」。边缘行为开始出现，但还没有扩散。 | 个别司机发现：接单后取消短途单→重新派可能派到长途→自己的完单/小时更好看。取消率从 2%→4%。 | ⚠️ 还能救。立即设护栏（如取消率上限），并调查边缘行为的规模 |
| **3** | 适应期 | 边缘行为从少数扩散到多数。「这个指标可以被操纵」成了公开的秘密。大部分人不再做正确的事，而是做「让指标好看」的事。 | 取消率从 4%→8%→15%。乘客投诉增加。运营开始接受「取消是正常的」。 | ⚠️ 需要严肃干预。护栏升级为强制红线（触发则自动报警），同时启动寻找替代指标 |
| **4** | 腐烂期 | 指标仍在涨，但业务实质上在恶化。团队花更多精力「维护指标」而不是「服务用户」。护栏已经被习惯性忽视。 | 运营用补贴冲量→完单量↑但毛利率从 15%→7%。每多一单都在亏更多钱。 | ❌ 必须立即换指标。当前指标已经反向（完单越涨利润越跌） |
| **5** | 僵尸期 | 所有人都知道这个指标没用了。但没人敢换——因为它已经嵌入了所有报表、KPI 考核、奖金计算。大家默契地「继续报这个数」，然后私下看别的数字做决策。 | 完单量成为「面子指标」——CEO 看到 11 万觉得还行，实际上公司靠借钱撑着。 | ❌ 组织问题大于数据问题。需要 CEO 级别的政治意愿才能更换 |

**关键的认知转换**：大多数公司在阶段 4-5 才意识到问题。但最好的分析师在**阶段 2** 就能发现并发出预警——因为他们不仅看北极星本身，还看「护栏指标」和「边缘行为信号」。

**③ 与核心概念的关系**：这是好指标第②⑤条的[防护]版。第②条「可指引长期战略」要求在腐烂发生前换指标。第⑤条「跨部门共识」在阶段 5 变成假共识——大家表面同意但实际上不看这个指标了。

## 概念群 D：[落地] 指标身份证——让「完单量」三个字变成一个精确的数字

> **与核心概念的关系**：主课好指标第①条「可衡量」说指标必须有明确的定义和计算方式。但这个「明确的定义」到底该写多细？本概念给你一个 10 字段模板——填完它，任何人用任何工具算出来的都是同一个数。

### 🔴 认知冲突

快达的周会上。运营王姐说「上周完单量 9.8 万」，财务陈姐说「明明是 9.2 万」。两个人吵了 10 分钟，最后发现：

- 王姐的「完单」= 司机在 App 上点了「到达目的地」，就算完单。
- 陈姐的「完单」= 乘客付款成功 + 没有退款，才算完单。

**同一个「完单量」，两个部门算出了两个不同的数字。** 这就是「可衡量」的反面——你用的是一把橡皮尺，每个人拉出来的长度不一样。

> 指标身份证就是把橡皮尺换成钢尺。

### [必修] 指标身份证——10 字段模板

**① 一句话白话**：指标身份证就是这个数字的「户口本」——名字叫什么、怎么算的、数据从哪来、谁对这个数字负责。全部写死。以后任何人要算这个数字，拿着户口本就能跑出完全一样的结果。

**② 10 字段模板（以「日均完单量」为例）**：

| # | 字段 | 填什么 | 快达的「日均完单量」怎么填 | 这一栏有什么用 |
|----|------|--------|------------------------|-------------|
| 1 | **唯一英文名** | 一个不会被误解的英文变量名 | `daily_completed_rides` | 写 SQL 和代码时用，保证引用一致 |
| 2 | **中文名** | 面向业务人员的名字 | 日均完单量 | 开会和报表里用 |
| 3 | **一句话业务定义** | 用白话说这是什么 | 「每天乘客打到车、司机送到目的地、双方都确认了的行程数」 | 让非数据同事 5 秒理解 |
| 4 | **精确计算公式** | 分子是什么，分母是什么（如果有分母的话） | 分子 = 状态为「已完成」且「已支付」且「未退款」的订单数。分母 = 无（这是计数指标，不需要分母） | 消除模糊空间——「已完成」到底是什么意思 |
| 5 | **数据来源** | 从哪张表的哪个字段取 | `ride_order.order_id`，条件：`status='completed' AND pay_status='paid' AND refund_status IS NULL` | 防止从错误的表取数据 |
| 6 | **排除条件** | 什么情况不算 | (1) 测试账号的订单 (2) 同一设备 1 小时内超过 3 单（疑似刷单）(3) 内部员工账号 | 清除脏数据，保证数字干净 |
| 7 | **计算时间窗口** | 算哪段时间的数据 | 自然日（00:00:00 ~ 23:59:59），按订单创建时间 `create_time` 归属日期 | 防止「1 月的数据在 2 月算」这种时间错位 |
| 8 | **刷新频率** | 多久算一次，什么时候出数 | 每天凌晨 3:00 计算前一天的完单量，T+1 上午 8:00 前产出 | 让使用者知道什么时候能看到最新数据 |
| 9 | **责任人** | 谁对这个数字的准确性负责 | 数据平台组 / 张三 | 数字出问题知道找谁 |
| 10 | **已知坑位** | 这个数字有什么历史遗留问题 | (1) 跨城长单的完单确认可能延迟 1-2 小时 (2) 系统故障期间的订单可能重复计算 | 让后来的人不用再踩一遍坑 |

**③ 换人测试**：你休年假了。一个新同事拿着这 10 个字段，能不能算出跟你完全一模一样的「日均完单量」？如果能 → 指标身份证合格。如果不能 → 哪里没写死，就补哪里。

## 🔮 [预览] HEART 框架——如果北极星「太公司视角」了

> **与核心概念的关系**：主课讲北极星是从「公司怎么增长」出发。但 Google 的 HEART 框架提供了另一个视角——从「用户有没有得到价值」出发。当你发现选出来的北极星全是公司视角的数字时（GMV、收入、DAU），HEART 帮你切换视角。⏳ W2D1 会深入 AARRR。

| 视角 | AARRR（下周学） | HEART（今天预览） |
|------|------|------|
| 核心问题 | 「用户从哪来，在哪流失，怎么变现」 | 「用户有没有成功完成他们想做的事」 |
| 五个维度 | 获取→激活→留存→变现→推荐 | 满意度(Happiness)→参与度(Engagement)→采用率(Adoption)→留存率(Retention)→任务成功率(Task Success) |
| 适合 | 有明确增长目标的 C 端产品 | 工具型、SaaS、内容型产品 |
| 在快达场景 | 乘客：下载→注册→首单→复购→推荐 | 乘客：打到车了吗？(Task Success) 等得久吗？(Engagement) 满意吗？(Happiness) |

HEART 的核心价值：它强迫你从「用户想完成什么任务」出发，而不是从「公司想从用户身上赚多少钱」出发。快达的 HEART：
- **Task Success**：乘客叫车后，3 分钟内被接单的比例 → 这是用户来快达要完成的核心任务
- **Happiness**：乘客评分 ≥ 4 星的比例 → 用户满意吗
- **Engagement**：人均每周打车次数 → 用户依赖快达吗
- **Adoption**：新用户首周完成 ≥ 3 单的比例 → 新用户真的用起来了吗
- **Retention**：本月活跃乘客中，下月仍活跃的比例 → 常规留存

## 六个概念怎么协同使用

在快达沙盘中，你面对「完单量 +25% 但毛利腰斩」的困局时：

1. **三问法**先扫一遍现有的 35 个指标——把累计注册、累计下载这些虚荣的移除。
2. **5 条检查表**诊断剩下的候选指标——完单量在 5 条中只通过 2 条。
3. **小分母检查**确保护栏指标的分母够大——新城市的比率指标全部标注可信度。
4. **腐烂模型**解释完单量为什么从好指标变成了坏指标——它在 Q2 进入了阶段 4。
5. **指标身份证**为新方案中的北极星 + 护栏写死口径——确保全公司算同一个数。
