## 三、本周核心收获——如果你只能带走三句话

1. **分析始于问题，终于行动** — 好的分析师80%的时间花在"定义问题"和"拆解问题"上。拿到需求就开工=已经输了。

2. **不是所有数字都叫指标** — 好指标让你看到数字就知道该做什么，虚荣指标只涨不跌但毫无价值。每个北极星配2个护栏。

3. **拆解是分析师的核心手艺** — 乘法找环节，加法找群体。贡献度>变化率。拆到可行动的层级为止。
