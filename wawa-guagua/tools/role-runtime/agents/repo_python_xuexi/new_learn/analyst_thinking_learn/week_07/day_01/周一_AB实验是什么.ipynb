## 📘 区块1：核心概念与第一性原理

### 1.1 一个让你无法回答的问题

运营说："我们上周发了100万优惠券，GMV涨了15%，券有效果！"

你心里清楚：这15%里，有多少是券的效果？有多少是上周刚好有个节日？有多少是自然增长？有多少是拿了券但本来就要买的人？

你无法回答——因为你没有**对照组**。这就是AB实验存在的根本原因。

### 1.2 AB实验的核心逻辑：随机化 = 因果推断

```
把用户随机分成两组：
  A组（对照组）：看到旧版/不领券
  B组（实验组）：看到新版/领券

因为分组是随机的 → 两组在实验前"平均来说"完全一样
→ 如果实验后两组有差异 → 这个差异只能是"新版/券"导致的
→ 这就是因果推断。
```

随机化 = 消除了所有混淆变量。不需要统计模型、不需要机器学习——**随机化本身就是最强大的因果推断工具。**

### 1.3 从第一性原理推导：为什么随机化能推断因果？

因果推断的核心挑战是**混淆变量**。假设我们想判断"领券"是否导致"购买增加"：

- **混淆变量示例**：活跃用户既更可能领券，也更可能购买。所以即使观察到"领券用户购买更多"，也不能说"券导致了购买"——可能是"活跃"同时导致了"领券"和"购买"。

- **随机化如何解决**：随机分组确保了"活跃用户"在两个组中的比例（在期望上）完全相同。同样，所有其他特征——年龄、性别、收入、偏好——在两组中都是均衡的。

- **数学表达**：随机化使得处理变量T与所有协变量X独立：T ⊥ X。因此E[Y|T=1] - E[Y|T=0] = 平均处理效应(ATE)，不需要控制任何X。

这就是为什么一个简单的t检验比一个复杂的机器学习模型更能回答"这个策略有没有用"——因为随机化消除了机器学习试图建模的所有混淆。

### 1.4 AB实验的"黄金标准"地位——从医学到互联网

医学用RCT（随机对照试验）判断新药是否有效——已经用了70多年。互联网的AB实验本质上就是RCT的数字版本。

**历史脉络**：
- 1948年：英国医学研究委员会开展第一个现代RCT（链霉素治疗肺结核）
- 1990年代：直接营销行业开始用AB测试优化邮件标题
- 2000年：Google开始系统化使用AB实验（每年跑数万个实验）
- 2010年代：Microsoft、Amazon、Netflix等全面AB驱动决策
- 现在：AB实验已成为互联网公司的标准决策工具

### 1.5 AB实验与机器学习的关系

一个重要但常被忽视的关系：
- **机器学习**擅长预测（"这个用户会买吗？"）
- **AB实验**擅长因果（"这个策略能让人买吗？"）
- 两者的结合：ML模型提出假设 → AB实验验证假设 → ML模型吸收结论持续优化

> **今天最重要的认知**：随机化是免费午餐。把用户随机分成两组，自然规律就帮你控制了所有你知道和不知道的混淆因素。这就是为什么AB实验是因果推断的黄金标准。

### 1.6 面试高频考点

**面试官常问**：
- "为什么AB实验能推断因果？"
- "AB实验和观察性分析有什么区别？"
- "什么情况下不能做AB实验？"
- "随机化单元怎么选？用户级还是设备级？"

**回答要点**：
- 随机化消除了混淆变量，使得两组在实验前完全可比
- 观察性分析只能发现相关性，AB实验能发现因果性
- 网络效应场景、长期效果评估、全量干预场景不能做AB实验
- 用户级是最常用的随机化单元，但需考虑登录态和跨设备问题

## 🧩 区块2：知识点详细拆解

### 2.1 实验 vs 观察——因果推断的两个世界

**为什么重要**：混淆两者是企业决策失误的主要原因之一。

| 维度 | 观察性分析 | AB实验 |
|------|-----------|--------|
| 怎么做的 | 看已经发生的数据 | 主动干预+随机分组 |
| 能回答 | "发券的用户比没发的买得多" | "发券**导致**用户买得更多" |
| 因果性 | 弱（反向因果/第三因素/选择偏误） | **强** |
| 成本 | 低（数据已经有了） | 高（需要设计和运行） |
| 时间 | 即时（查数据就行） | 需要等待（几天到几周） |
| 典型陷阱 | 幸存者偏差、辛普森悖论 | Peeking、SRM、指标偷换 |

**经典案例：幸存者偏差**
- 观察性分析："付费用户留存更高，所以引导用户付费能提升留存"
- 可能真相：留存高的用户才更可能付费（反向因果）
- AB实验才能验证：随机让一部分用户看到付费引导 → 看留存是否真的提升

### 2.2 AB实验的三个基本要素

**为什么重要**：三个要素缺一不可，设计错了任何一个都会导致实验无效。

**要素1：随机化单元（Randomization Unit）**

以什么粒度进行随机分组？

| 单元类型 | 优点 | 缺点 | 适用场景 |
|---------|------|------|----------|
| 用户级 | 最干净，用户体验一致 | 需要登录态 | 大多数实验 |
| 设备级 | 不需要登录 | 同一用户可能被分到两组 | 未登录场景 |
| 会话级 | 快速迭代 | 用户体验不一致 | 短期UI测试 |
| 城市/地区级 | 适合线下场景 | 样本量小，统计功效低 | 骑手端、司机端 |
| 店铺级 | 适合B端实验 | 店铺间差异大 | 商家端功能 |

**要素2：处理（Treatment）**

实验组接受什么干预？
- 单一处理：对照组 vs 一个实验组（A/B测试最简形式）
- 多处理：对照组 vs 多个实验组（A/B/n测试）
- 因子设计：多个因素交叉（如：颜色×文案，4个实验组）

**要素3：指标（Metrics）**

用什么来衡量效果？（W1的指标体系在这里派上用场）
- 主要指标（Primary Metric）：判断实验是否成功的核心指标（通常1个）
- 次要指标（Secondary Metrics）：辅助判断的其他指标
- 护栏指标（Guardrail Metrics）：必须不能变差的指标（如崩溃率、卸载率）

### 2.3 随机化的三种实现方式

**为什么重要**：不同的实现方式有不同的工程复杂度和随机性保证。

**方式1：简单随机（Simple Randomization）**
```python
group = 'treatment' if random() < 0.5 else 'control'
```
- 优点：最简单
- 缺点：小样本下两组比例可能不均衡（如48:52）

**方式2：哈希分流（Hash-based Assignment）**
```python
bucket = hash(user_id + experiment_id + salt) % 100
group = 'treatment' if bucket < 50 else 'control'
```
- 优点：确定性（同一用户永远在同一组）、可复现
- 缺点：需要好的哈希函数
- 这是业界最常用的方式

**方式3：分层随机（Stratified Randomization）**
- 先按重要特征分层（如城市、平台），再在每层内随机分组
- 优点：确保关键维度上两组均衡
- 缺点：工程复杂度高

### 2.4 SUTVA——稳定单元处理值假设

**为什么重要**：SUTVA是AB实验有效性的理论前提，违反SUTVA会导致估计偏差。

SUTVA包含两部分：
1. **无干扰（No Interference）**：一个用户的处理不会影响其他用户的结果
2. **一致性（Consistency）**：同一处理对同一用户的效果是稳定的

**违反SUTVA的经典场景**：
- 社交产品：实验组用户的新功能影响了对照组用户（因为他们在互动）
- 双边市场：给一部分司机加补贴 → 他们接单更多 → 其他司机的单变少
- 广告竞价：给一部分广告主降出价 → 其他广告主的竞争变小 → 广告成本下降

**应对策略**：
- 集群随机化：以城市/社群为单元随机
- 网络效应模型：用图模型估计溢出效应
- 两阶段实验：第一阶段随机城市，第二阶段在城市内随机用户

### 2.5 AB实验不能做什么——局限性与边界

**为什么重要**：了解边界才能正确使用工具，避免过度推广实验结论。

AB实验不能：
1. **告诉你"为什么"**：实验告诉你"新版提升了转化率"，但不能告诉你"因为按钮更大还是颜色更鲜艳"。需要用户研究、眼动实验等补充。
2. **测长期效果**：大部分实验只跑2-4周。用户体验的长期影响（如推送疲劳、广告耐受）需要更长时间的观察。
3. **测网络效应**：如前所述，社交产品中的用户体验受其他用户影响。
4. **测全量干预**：有些策略必须对所有人生效才有意义（如修改隐私政策、修改定价体系）。
5. **测罕见事件**：如果事件的基线概率极低（如欺诈、重大安全事故），需要巨大的样本量。

### 2.6 灰度发布 vs AB实验——最常被混淆的概念

**为什么重要**：很多团队用灰度替代AB实验，这是错误的。

| 维度 | 灰度发布 | AB实验 |
|------|---------|--------|
| 目的 | 安全性验证（有没有致命bug） | 效果评估（好不好用） |
| 分组 | 先小后大，逐步放量 | 随机分成两组，同时对比 |
| 对照组 | 通常没有同期对照组 | 必须有同期对照组 |
| 决策依据 | 系统监控指标（错误率、崩溃率） | 业务指标（转化率、留存率） |
| 典型问题 | 早期用户不代表全体 | 样本量不够可能不显著 |

**灰度不能替代AB实验的核心原因**：
1. 早期用户通常更活跃、更容忍bug（选择偏误）
2. 没有同期对照组，无法排除时间因素（节假日、竞对动作）
3. 5%的用户数据不能代表全量效果（异质性）

### 2.7 AB实验平台的演进

**为什么重要**：理解实验平台的架构有助于设计更好的实验。

**第一代（2010前）**：手工分流
- 工程师在代码里写if-else分流
- 分析师手动拉数据分析
- 问题：容易出错、不可复现、无法审计

**第二代（2010-2015）**：配置化平台
- 分流逻辑统一管理
- 自动数据采集和基础报表
- 代表：Google Analytics Experiments、Optimizely

**第三代（2015-至今）**：智能实验平台
- 自动样本量计算和MDE估计
- 内置SRM检测和AA验证
- 序贯检验和提前停止规则
- CUPED等方差缩减技术
- 代表：Google's internal platform、Microsoft ExP、Uber's XP

### 2.8 AB实验的组织文化

**为什么重要**：工具再好，文化不支持也白搭。

**实验驱动的组织特征**：
1. **数据说话**：决策基于实验结果，而非HiPPO（最高薪酬者的意见）
2. **拥抱失败**：大部分实验不会显著提升指标——这是正常的，不是失败
3. **快速迭代**：从"想一个完美的方案"→"提出假设→实验验证→学习→迭代"
4. **标准化流程**：每个实验都经过相同的设计和审查流程

**常见阻力**：
- "我们已经知道答案了，不需要实验" → 回答：亚马逊改了按钮颜色，多赚了3亿
- "实验太慢了" → 回答：一个错误的决策比不做决策代价更大
- "实验组用户体验差了怎么办" → 回答：护栏指标+小流量起步+随时可停止

## 💻 区块3：动手练习

### 步骤0：理解随机化的力量——模拟实验

我们生成包含混淆变量的数据，对比"观察性分析"和"AB实验"的结论差异。

````python
import numpy as np
np.random.seed(42)

# 步骤0：生成带有混淆变量的模拟数据
n = 10000

# 用户活跃度（混淆变量）：活跃用户既更可能领券，也更可能购买
activeness = np.random.beta(2, 5, n)  # 大部分用户不太活跃

# 观察性数据：领券行为受活跃度影响（活跃用户更容易看到券并领取）
coupon_prob = 0.1 + 0.5 * activeness  # 活跃用户领券概率高
received_coupon = np.random.binomial(1, coupon_prob)

# 购买行为：也受活跃度影响（活跃用户本身就更可能购买）
purchase_base = 0.05 + 0.3 * activeness  # 基础购买概率
purchase = np.random.binomial(1, purchase_base)

# —— 注意：这里券对购买没有任何因果效应！——
# 但观察性分析会错误地认为"券有效果"

print("观察性分析（存在混淆）:")
coupon_users = purchase[received_coupon == 1]
no_coupon_users = purchase[received_coupon == 0]
print(f"  领券用户购买率: {coupon_users.mean():.1%}")
print(f"  未领券用户购买率: {no_coupon_users.mean():.1%}")
print(f"  差异: {(coupon_users.mean() - no_coupon_users.mean())*100:.1f} 个百分点")
print(f"  → 看起来券'有效'，但其实是活跃度的混淆！")

print()
print("AB实验模拟（随机分组，无混淆）:")
# 随机分成两组
group = np.random.binomial(1, 0.5, n)
# 实验组获得券（处理效应=0，即券真的无效）
purchase_exp = np.random.binomial(1, purchase_base)
control_purchase = purchase_exp[group == 0]
treatment_purchase = purchase_exp[group == 1]
print(f"  对照组购买率: {control_purchase.mean():.1%}")
print(f"  实验组购买率: {treatment_purchase.mean():.1%}")
print(f"  差异: {(treatment_purchase.mean() - control_purchase.mean())*100:.1f} 个百分点")
print(f"  → 正确地发现券没有效果！随机化消除了混淆")
````

### 步骤1：基础练习——AB实验模拟

1.1 模拟一个有真实效果的AB实验
1.2 理解样本量对检测效果的影响

````python
# 步骤1.1：模拟有真实效果的AB实验
n = 1000
baseline_rate = 0.10  # 对照组转化率10%
true_lift = 0.02       # 真实提升2个百分点

control = np.random.binomial(1, baseline_rate, n)
experiment = np.random.binomial(1, baseline_rate + true_lift, n)

print("AB实验模拟：真实提升2个百分点")
print(f"  对照组转化率: {control.mean():.1%}")
print(f"  实验组转化率: {experiment.mean():.1%}")
print(f"  观察到的差值: {(experiment.mean() - control.mean())*100:.1f} 个百分点")
print(f"  真实差值: {true_lift*100:.1f} 个百分点")
print(f"  采样误差: {abs(experiment.mean() - control.mean() - true_lift)*100:.2f} 个百分点")
print()
print("关键洞察：由于随机性，观察到的差值和真实差值不会完全相等")
print("这就是为什么我们需要假设检验（明天学）来判断差异是否'显著'")
````

````python
# 步骤1.2：样本量对估计精度的影响
true_lift = 0.02
print("样本量对估计精度的影响（真实效应=2pp）:")
print(f"{'样本量':<10} {'对照组':<10} {'实验组':<10} {'观察差值':<12} {'误差':<10}")
print("-" * 52)
for n in [100, 500, 2000, 10000, 50000]:
    c = np.random.binomial(1, 0.10, n)
    e = np.random.binomial(1, 0.10 + true_lift, n)
    obs_diff = e.mean() - c.mean()
    error = abs(obs_diff - true_lift)
    print(f"{n:<10} {c.mean():<10.1%} {e.mean():<10.1%} {obs_diff*100:<12.2f}pp {error*100:<10.2f}pp")
print()
print("关键洞察：样本量越大，观察差值越接近真实差值（大数定律）")
print("但样本量越大，成本也越高 → 需要样本量计算（周三学）")
````

### 步骤2：进阶练习——哈希分流模拟

2.1 实现一个简单的哈希分流系统
2.2 验证分流的均匀性

````python
import hashlib

# 步骤2.1：实现哈希分流
def hash_assignment(user_id, experiment_id, salt="default", total_buckets=100):
    """基于哈希的确定性分流"""
    hash_input = f"{user_id}_{experiment_id}_{salt}"
    hash_value = int(hashlib.md5(hash_input.encode()).hexdigest(), 16)
    return hash_value % total_buckets

# 步骤2.2：验证分流的均匀性
n_users = 10000
buckets = [hash_assignment(i, "exp_001") for i in range(n_users)]

from collections import Counter
bucket_counts = Counter(buckets)

# 对照组 = bucket 0-49, 实验组 = bucket 50-99
control_count = sum(bucket_counts[i] for i in range(50))
treatment_count = sum(bucket_counts[i] for i in range(50, 100))

print(f"哈希分流验证（{n_users}个用户，100个桶）:")
print(f"  对照组(bucket 0-49): {control_count} 用户 ({control_count/n_users:.1%})")
print(f"  实验组(bucket 50-99): {treatment_count} 用户 ({treatment_count/n_users:.1%})")
print(f"  每组预期: {n_users//2}")

# 验证确定性：同一用户永远在同一组
user_42_1 = hash_assignment(42, "exp_001")
user_42_2 = hash_assignment(42, "exp_001")
print(f"\n确定性验证: user_id=42 第一次={user_42_1}, 第二次={user_42_2} → {'✓ 一致' if user_42_1 == user_42_2 else '✗ 不一致！'}")

# 验证不同实验的独立性
bucket_exp1 = hash_assignment(42, "exp_001")
bucket_exp2 = hash_assignment(42, "exp_002")
print(f"不同实验独立性: exp_001={bucket_exp1}, exp_002={bucket_exp2}")
print(f"→ 同一用户在不同实验中可能在不同组")
````

### 步骤3：挑战练习——混淆变量的可视化

3.1 用可视化展示为什么观察性分析会得出错误结论
3.2 展示随机化如何消除混淆

````python
# 步骤3：展示随机化消除混淆的神奇效果
np.random.seed(123)
n = 5000

# 生成两个混淆维度：活跃度和价格敏感度
activeness = np.random.beta(2, 5, n)
price_sensitivity = np.random.normal(0.5, 0.15, n)

# 观察性场景：领券受两个混淆因素影响
coupon_prob_obs = 0.05 + 0.4 * activeness + 0.2 * (1 - price_sensitivity)
coupon_obs = np.random.binomial(1, np.clip(coupon_prob_obs, 0, 1))

# 购买也受两个混淆因素影响（券本身没有因果效应）
purchase_prob = 0.05 + 0.3 * activeness + 0.1 * (1 - price_sensitivity)
purchase = np.random.binomial(1, np.clip(purchase_prob, 0, 1))

# 随机分组（AB实验）
group = np.random.binomial(1, 0.5, n)  # 0=对照, 1=实验

print("="*60)
print("场景：券对购买的真实因果效应 = 0")
print("但存在两个混淆变量：活跃度、价格敏感度")
print("="*60)

print("\n【观察性分析】——直接对比领券vs未领券")
obs_control = purchase[coupon_obs == 0]
obs_treatment = purchase[coupon_obs == 1]
print(f"  未领券用户购买率: {obs_control.mean():.2%}")
print(f"  领券用户购买率:   {obs_treatment.mean():.2%}")
print(f"  差异: {(obs_treatment.mean() - obs_control.mean())*100:.1f}pp")
print(f"  → 错误结论：券'有效'！实际是活跃用户既领券又购买")

print("\n【AB实验】——随机分组")
ab_control = purchase[group == 0]
ab_treatment = purchase[group == 1]
print(f"  对照组购买率: {ab_control.mean():.2%}")
print(f"  实验组购买率: {ab_treatment.mean():.2%}")
print(f"  差异: {(ab_treatment.mean() - ab_control.mean())*100:.1f}pp")
print(f"  → 正确结论：券没有效果！（差异接近0）")

print("\n【检验随机化是否均衡了混淆变量】")
print(f"  对照组平均活跃度: {activeness[group==0].mean():.4f}")
print(f"  实验组平均活跃度: {activeness[group==1].mean():.4f}")
print(f"  对照组价格敏感度: {price_sensitivity[group==0].mean():.4f}")
print(f"  实验组价格敏感度: {price_sensitivity[group==1].mean():.4f}")
print(f"  → 两组在混淆变量上几乎相同！这就是随机化的力量")
````

## 📊 区块4：业务场景应用

### 场景1：为什么"灰度发布"不等同于AB实验

**问题**：某产品经理说："我们做了AB实验——先给5%的用户上灰度，看数据没跌就全量了。"

**分析**：这不是AB实验，这是"灰度验证"。

| 区别 | 灰度发布 | AB实验 |
|------|---------|--------|
| 目的 | 安全性验证（有没有致命bug） | 效果评估（好不好用） |
| 对照组 | 通常没有同期对照组 | 必须有同期对照组 |
| 时间 | 逐步放量，可能跨越多天 | 两组同时运行 |
| 用户特征 | 早期用户≠全体用户 | 随机分组，两组可比 |

**灰度不能替代AB实验的原因**：
1. 5%的早期用户通常更活跃（选择偏误）
2. 没有同期对照组，无法排除时间因素
3. 灰度只看"有没有跌"，不看"有没有涨"

**正确做法**：灰度+AB实验同时进行。灰度验证安全性，AB实验评估效果。

### 场景2：一个"看起来很美"的观察性分析

**问题**：数据分析师发现"使用搜索功能的用户，留存率是不使用的3倍"。PM决定强推搜索功能。

**分析**：这是典型的观察性分析陷阱。

**可能的混淆因素**：
- 高意愿用户更可能使用搜索（他们本来就更有目的性）
- 熟悉产品的用户更可能使用搜索（他们知道有这个功能）
- 网络好的用户更可能使用搜索（加载速度快）

**AB实验才能回答**：
随机让一部分新用户看到更突出的搜索入口 → 强制增加搜索使用率 → 看留存是否真的提升

**决策**：推搜索入口之前先做AB实验，而不是基于观察性数据做决策。

**教训**："使用X功能的用户表现更好"是最常见的观察性分析陷阱——几乎总是存在自选择偏误。

### 场景3：社交产品的AB实验困境

**问题**：某社交App想测试"新消息提醒方式"是否能提升用户活跃度。

**分析**：社交产品存在网络效应——用户A的体验受用户B的行为影响。

**SUTVA违反场景**：
- 如果实验组用户因为新提醒更活跃 → 他们发更多消息 → 对照组用户也收到更多消息 → 对照组也更活跃了
- 结果：两组看起来差不多 → 错误结论"新提醒没效果"

**解决方案**：
1. **集群随机化**：以城市/社群为单元随机。整个城市的人都在同一组。
2. **两阶段设计**：先随机城市，再在城市内随机用户。
3. **网络效应建模**：用社交图谱估计溢出效应。

**决策**：社交产品做AB实验需要特别小心，用户级随机化可能严重低估效果。

### 场景4：推送疲劳——短期效果 vs 长期效果

**问题**：某App测试"增加推送频率"策略，AB实验显示DAU上涨5%。全量上线后第3个月，DAU反而下降了。

**分析**：
- AB实验跑了2周，看到的是短期效果（用户被更多推送唤醒）
- 但长期效果是相反的——用户感到厌烦，关闭推送权限甚至卸载App
- 实验时间太短，没有捕捉到长期衰减效应

**应对策略**：
1. 延长实验时间（4-8周），观察效果是否衰减
2. 用早期用户（"老实验组"）的长期数据做补充分析
3. 监控推送关闭率作为护栏指标
4. 全量上线后持续监控，设计"回滚触发条件"

**决策**：对于高频率触达类策略，短期AB实验的结论可能严重高估效果。上线后需要长期监控。

## 📝 区块7：当日小结

> **AB实验是因果推断的黄金标准。核心逻辑：随机分组→两组在实验前完全可比→实验后的差异只能是处理导致的→这就是因果。随机化消除了所有混淆变量，不需要复杂的模型。观察性分析只能发现相关性，AB实验才能发现因果性。AB实验的三个基本要素：随机化单元（以什么粒度分组）、处理（什么干预）、指标（什么衡量标准）。**

**三条金句带回家**：
1. 随机化是因果推断的免费午餐——不需要模型，自然规律帮你控制一切混淆
2. "使用X的用户表现更好"是最常见的观察性分析陷阱——几乎总是自选择偏误
3. 灰度发布验证安全性，AB实验评估效果——两者不可互相替代
