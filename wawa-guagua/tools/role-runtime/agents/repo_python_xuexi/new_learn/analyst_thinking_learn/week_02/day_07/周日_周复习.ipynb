## 一、本周知识地图——6个框架，6个核心问题，1条诊断流水线

本周每一天都在学习一个独立的增长分析框架。它们各自回答一个核心问题，但真实分析中需要组合成一条完整的诊断流水线：

```
D1 AARRR海盗模型  → 看全局：增长的5个环节(获取→激活→留存→变现→推荐)
   核心问题：「产品主要卡在增长的哪个环节？」
   类比：AARRR是「增长GPS」——帮你看清车队坏在哪段路
   输入：产品名称+各环节转化数据 → 输出：5环节漏斗+最薄弱环节诊断

D2 同期群分析     → 看纵向：不同批次新用户的留存质量变化趋势
   核心问题：「新用户质量是在变好还是变差？」
   类比：同期群是「层析CT」——扫描不同批次的内在结构
   输入：按注册月份的留存数据 → 输出：三角矩阵+纵向W1/W2列趋势判断

D3 漏斗分析       → 看横向：用户在转化路径上每一步的流失率
   核心问题：「哪个步骤流失最严重？哪个步骤转化率最低？」
   类比：漏斗分析是「内窥镜」——伸进产品转化路径里看哪个环节在疯狂漏水
   输入：各步骤人数 → 输出：最大漏水点+分群漏斗对比(新老用户/渠道)

D4 留存分析与LTV  → 看长期：留存曲线的形态和用户终身价值
   核心问题：「产品有PMF吗？每个用户值多少钱？获客成本合理吗？」
   类比：留存+LTV是「财务透视」——剥开增长表面看每个用户到底翻出多少钱
   输入：留存曲线+ARPU → 输出：留存形态诊断+LTV估算+LTV/CAC判断

D5 RFM用户分群    → 看用户：按活跃度/忠诚度/贡献度把用户分成8个群体
   核心问题：「哪些用户是核心资产？哪些值得花重金挽回？哪些不值得花钱？」
   类比：RFM是「分诊台」——把用户按价值分层，告诉运营该抢救谁、放弃谁
   输入：用户R/F/M数据 → 输出：8群体标签+差异化策略+预算优先级

D6 PDS日活拆解    → 看结构：DAU内部由哪三部分组成
   核心问题：「DAU增长是真增长还是虚胖？Retained底盘在变大还是萎缩？」
   类比：PDS是「骨架X光」——透过DAU的皮肉看内部的结构健康度
   输入：DAU+新增数据 → 输出：PDS三元占比+增长模式诊断+Retained趋势
```

**六者配合使用的标准诊断流程**：
1. AARRR定位大方向（「产品卡在激活→变现之间」）
2. 同期群+漏斗精确定位（「新用户质量恶化，恶化发生在浏览→加购」）
3. 留存+LTV量化影响（「次留从25%→18%=每月损失约X万毛利」）
4. RFM指导运营（「重要价值维持、重要挽留加码、流失不花钱」）
5. PDS日常监控（「每周看Retained绝对值趋势→连续3周降就告警」）

**本周关键公式汇总**：
| 框架 | 核心公式 | 说明 |
|------|---------|------|
| AARRR | 端到端转化率 = 获取率 × 激活率 × 留存率 × 付费率 × K因子 | 五个环节乘法串联 |
| 同期群 | 纵向对比 W1_col: cohort_1_W1 vs cohort_2_W1 vs ... vs cohort_N_W1 | 趋势=列的变化方向 |
| 漏斗 | 步骤转化率 = 本步人数 / 上步人数；绝对流失 = 上步 - 本步 | 两个视角互补 |
| 留存+LTV | LTV = Σ(day_i_retention × day_i_avg_revenue)；LTV/CAC > 3 健康 | 积分法+铁律 |
| RFM | R≤中位=1(高)；F≥中位=1(高)；M≥中位=1(高)→2×2×2=8群体 | 乘法交叉正交 |
| PDS | DAU = New + Retained + Resurrected；贡献度 = ΔPart / ΔDAU | 加法拆解无遗漏 |

**六个框架的协同关系**：AARRR是「宏观GPS」先定位大方向，同期群是「纵向CT」看批次趋势，漏斗是「微观内窥镜」找精准漏点，留存+LTV是「财务X光」量化损失，RFM是「分诊台」分配运营资源，PDS是「日常体检」监控结构健康。六个框架从宏观到微观、从诊断到行动、从一次性分析到持续监控，构成完整的增长诊断体系。

## 二、周考核：生鲜电商增长诊断——综合运用6大框架

### 背景
你是生鲜电商的第一位数据分析师。CEO给你过去6个月的核心数据：

| 指标 | 6个月前 | 现在 | 变化 |
|------|--------|------|------|
| MAU | 50万 | 65万 | +30% |
| 月GMV | 800万 | 750万 | -6.25% |
| 月新增用户 | 8万 | 15万 | +87.5% |
| 新用户7日留存 | 25% | 18% | -28% |
| 月下单用户数 | 12万 | 13万 | +8.3% |

CEO问：**「MAU涨了30%但GMV反而跌了。人多了，钱怎么少了？我们的增长出了什么问题？」**

````python
import numpy as np
import pandas as pd
np.random.seed(42)

print("=" * 65)
print("生鲜电商增长诊断报告")
print("=" * 65)
print()

# ============================================================
# Part A: 数据准备
# ============================================================
print("━" * 65)
print("Part A: 数据准备")
print("━" * 65)
print()

months = np.arange(1, 7)
mau = np.array([50, 53, 56, 59, 62, 65])
gmv = np.array([800, 790, 780, 770, 760, 750])
new_users = np.array([8, 10, 12, 14, 15, 15])
d7_retention = np.array([25, 22, 20, 19, 18, 18])
order_users = np.array([12.0, 12.2, 12.4, 12.6, 12.8, 13.0])

print("月度核心指标:")
print(f"{'月份':>6} {'MAU(万)':>8} {'GMV(万)':>9} {'新增(万)':>8} {'7留%':>7} {'下单(万)':>8}")
print("-" * 55)
for i in range(6):
    print(f"第{i+1}月: {mau[i]:>8.0f} {gmv[i]:>9.0f} {new_users[i]:>8.0f} {d7_retention[i]:>7.1f} {order_users[i]:>8.1f}")

retained_m = np.array([35, 35, 34.5, 34, 34, 34])
resurrected_m = np.array([7, 8, 9.5, 11, 13, 16])

print("\nPDS结构(月均,万):")
print(f"New: {new_users[0]}->{new_users[-1]} (+{(new_users[-1]/new_users[0]-1)*100:.0f}%)")
print(f"Retained: {retained_m[0]}->{retained_m[-1]} ({(retained_m[-1]/retained_m[0]-1)*100:.1f}%)")
print(f"Resurrected: {resurrected_m[0]}->{resurrected_m[-1]} (+{(resurrected_m[-1]/resurrected_m[0]-1)*100:.0f}%)")

print("\n渠道结构变化:")
print(f"  信息流占比: 20% -> 55% (低质渠道权重滚大)")
print(f"  SEO占比: 35% -> 15% (高质渠道被稀释)")
print(f"  自然量占比: 45% -> 30%")

cohort_w1 = np.array([30, 28, 25, 22, 18, 15])
print("\n同期群W1留存(按注册月份):")
for i in range(6):
    print(f"  第{i+1}月新用户W1留存: {cohort_w1[i]}%")
print(f"  -> 纵向持续恶化 (30%->15%, -50%相对恶化)")

print("\n漏斗步骤转化率:")
funnel_steps = ["打开->浏览", "浏览->加购", "加购->下单", "下单->支付"]
funnel_before = [60, 30, 50, 80]
funnel_after  = [55, 18, 56, 80]
for i, step in enumerate(funnel_steps):
    change = funnel_after[i] - funnel_before[i]
    flag = " << 最大恶化!" if i == 1 else ""
    print(f"  {step}: {funnel_before[i]}%->{funnel_after[i]}% ({change:+d}pp){flag}")
print()

# ============================================================
# Part B: AARRR诊断
# ============================================================
print("━" * 65)
print("Part B: AARRR海盗模型诊断")
print("━" * 65)
print()
print("AARRR各环节数据解读:")
print(f"  获取(A): 新增+{((new_users[-1]/new_users[0])-1)*100:.0f}% ({new_users[0]}万->{new_users[-1]}万)")
print(f"  激活(A): 7留{d7_retention[0]}%->{d7_retention[-1]}%({d7_retention[-1]-d7_retention[0]:+.0f}pp)")
print(f"  变现(R): 下单用户+{((order_users[-1]/order_users[0])-1)*100:.1f}%——远低于MAU和新增增速")
print()
print("诊断：卡在【激活→变现】之间")
print("  拉新在加速(+87.5%),但新用户质量差->激活率降->付费转化低")
print("  -> 你在「往漏水的桶里倒水」——拉新越快,浪费越多")
print()

# ============================================================
# Part C: 漏斗分析
# ============================================================
print("━" * 65)
print("Part C: 漏斗分析")
print("━" * 65)
print()
print("核心漏斗对比:")
print(f"  {'步骤':<20} {'半年前':>8} {'现在':>8} {'变化':>8}")
print(f"  {'-'*44}")
for i, step in enumerate(funnel_steps):
    change = funnel_after[i] - funnel_before[i]
    print(f"  {step:<20} {funnel_before[i]:>7}% {funnel_after[i]:>7}% {change:+>7}pp")
print()
print("诊断：核心恶化在【浏览→加购】环节(-12pp, -40%相对恶化)")
print("  分新老用户看：新用户28%->12%(-16pp暴跌)，老用户35%->32%(-3pp基本稳)")
print("  -> 问题集中在新用户——不信任/价格敏感/商品不吸引")
print()

# ============================================================
# Part D: 同期群分析
# ============================================================
print("━" * 65)
print("Part D: 同期群分析")
print("━" * 65)
print()
print("按注册月份看新用户W1/W2/W4留存（纵向对比）:")
cohort_w2 = np.array([20, 18, 16, 14, 11, 9])
cohort_w4 = np.array([14, 13, 11, 9, 7, 5])
print(f"  {'月份':<8} {'W1':>6} {'W2':>6} {'W4':>6}")
print(f"  {'-'*30}")
for i in range(6):
    print(f"  第{i+1}月:   {cohort_w1[i]:>5}% {cohort_w2[i]:>5}% {cohort_w4[i]:>5}%")
print()
print("纵向趋势判断:")
print(f"  W1列: 30%->15% (-15pp) —— 新用户首周体验在恶化")
print(f"  W2列: 20%->9%  (-11pp) —— 第二周也在降")
print(f"  W4列: 14%->5%  (-9pp)  —— 长期留存也在恶化")
print("  -> 三列全部恶化——不是季节性波动,是结构性恶化")
print()
print("进一步按渠道拆:")
print(f"  信息流W1留存: 15%->3%  (持续暴跌)")
print(f"  SEO W1留存:    32%->29% (基本稳定)")
print("  -> 不是产品变差了——是投放渠道结构变了(低质渠道权重滚大)")
print()

# ============================================================
# Part E: PDS拆解
# ============================================================
print("━" * 65)
print("Part E: PDS日活拆解")
print("━" * 65)
print()
print("DAU(MAU近似) = New + Retained + Resurrected")
print(f"  半年前(MAU=50万): New={new_users[0]}万({new_users[0]/50*100:.0f}%) Retained={retained_m[0]}万({retained_m[0]/50*100:.0f}%) Resurrected={resurrected_m[0]}万({resurrected_m[0]/50*100:.0f}%)")
print(f"  现在(MAU=65万):    New={new_users[-1]}万({new_users[-1]/65*100:.0f}%) Retained={retained_m[-1]}万({retained_m[-1]/65*100:.0f}%) Resurrected={resurrected_m[-1]}万({resurrected_m[-1]/65*100:.0f}%)")
print()
print("PDS四维诊断:")
print(f"  (1) Retained绝对值: {retained_m[0]}->{retained_m[-1]}万(-{(1-retained_m[-1]/retained_m[0])*100:.0f}%) <- 底盘在萎缩!")
print(f"  (2) Retained占比: {retained_m[0]/50*100:.0f}%->{retained_m[-1]/65*100:.0f}% <- 从留存驱动降级")
print(f"  (3) New: +{((new_users[-1]/new_users[0])-1)*100:.0f}% <- 砸钱买量")
print(f"  (4) Resurrected: +{((resurrected_m[-1]/resurrected_m[0])-1)*100:.0f}% <- 频繁push催活")
print("-> 典型【虚胖DAU】——表面+30%但底盘在萎缩")
print()

# 贡献度分析
dau_change = 65 - 50
new_contrib = new_users[-1] - new_users[0]
ret_contrib = retained_m[-1] - retained_m[0]
res_contrib = resurrected_m[-1] - resurrected_m[0]
print("DAU变化的贡献度分析 (DAU +15万):")
print(f"  New贡献: +{new_contrib}万 ({new_contrib/dau_change*100:.0f}%)")
print(f"  Retained贡献: {ret_contrib}万 ({ret_contrib/dau_change*100:.0f}%)  <- 负贡献!")
print(f"  Resurrected贡献: +{res_contrib}万 ({res_contrib/dau_change*100:.0f}%)")
print(f"  -> DAU涨了15万全靠New和Resurrected撑着——Retained在拖后腿")
print()

# ============================================================
# Part F: LTV/CAC分析
# ============================================================
print("━" * 65)
print("Part F: LTV/CAC分析")
print("━" * 65)
print()

print("按渠道LTV/CAC分析:")
ch_ltv_cac = {
    "信息流": {"cac": 30, "ltv": 18, "ratio": 0.6, "v": "亏本!大量投放=加速亏损"},
    "SEO": {"cac": 55, "ltv": 280, "ratio": 5.1, "v": "健康!虽贵但值,应加大投入"},
    "自然量": {"cac": 5, "ltv": 120, "ratio": 24.0, "v": "极健康!最好的用户"},
}
for ch, d in ch_ltv_cac.items():
    print(f"  {ch}: CAC={d['cac']}元, LTV={d['ltv']}元, LTV/CAC={d['ratio']} -> {d['v']}")
print()
print(f"结论: 信息流LTV/CAC=0.6——每获取一个用户亏12元")
print(f"  信息流月新增约{new_users[-1]*0.55:.0f}万用户->月亏约{new_users[-1]*0.55*(18-30):.0f}万元")
print()

# ============================================================
# Part G: RFM分群
# ============================================================
print("━" * 65)
print("Part G: RFM用户分群")
print("━" * 65)
print()

np.random.seed(123)
n_sample = 10000
r = np.random.exponential(20, n_sample).astype(int)
f = np.random.poisson(4, n_sample)
m = np.round(np.random.lognormal(4.2, 0.7, n_sample), 2)

rm_m, fm_m, mm_m = np.median(r), np.median(f), np.median(m)
r_high = (r <= rm_m).astype(int)
f_high = (f >= fm_m).astype(int)
m_high = (m >= mm_m).astype(int)

def rfm_label(ri, fi, mi):
    if ri==1 and fi==1 and mi==1: return '重要价值'
    if ri==0 and fi==1 and mi==1: return '重要挽留'
    if ri==1 and fi==0 and mi==1: return '重要发展'
    if ri==0 and fi==0 and mi==1: return '流失高价值'
    if ri==1 and fi==1 and mi==0: return '一般价值'
    if ri==0 and fi==1 and mi==0: return '一般挽留'
    if ri==1 and fi==0 and mi==0: return '新用户/低价值'
    return '流失用户'

segments = [rfm_label(r_high[i], f_high[i], m_high[i]) for i in range(n_sample)]
seg_counts = pd.Series(segments).value_counts()

print("RFM用户分群结果:")
print(f"{'群体':<12} {'人数':>6} {'占比':>7}")
print("-" * 28)
for seg, cnt in seg_counts.items():
    print(f"{seg:<12} {cnt:>6} {cnt/n_sample*100:>6.1f}%")
print()

strategies = {
    '重要价值': ('VIP客服+新品优先+专属社群', '高', '保护核心资产'),
    '重要挽留': ('大额券50元+电话/个性化短信召回', '高', '召回一个可能带来2000元消费'),
    '重要发展': ('推关联品类+7天内第二单优惠', '中', '趁活跃扩大消费'),
    '流失高价值': ('电话+大额券+诚恳询问离开原因', '高', '挽回一个就回本'),
    '一般价值': ('推高价商品+套餐组合', '低', '客单价提升空间大'),
    '一般挽留': ('小额券10元+push提醒', '低', '性价比不高'),
    '新用户/低价值': ('首单优惠+新手引导', '中', '引导升级'),
    '流失用户': ('不发券——省下的预算给重要挽留', '0', '不值得花钱'),
}
print("差异化策略:")
for seg, (s, cost, effect) in strategies.items():
    cnt = seg_counts.get(seg, 0)
    print(f"  {seg}({cnt}人): {s} | 投入={cost} | {effect}")
print()

# ============================================================
# Part H: 综合诊断结论
# ============================================================
print("━" * 65)
print("Part H: 综合诊断结论")
print("━" * 65)
print()
print("CEO问「MAU涨了GMV为什么跌了」的完整答案:")
print()
print("  1. MAU涨是因为花了更多钱买量(New+87.5%)+push催活(Resurrected+128%)")
print("  2. GMV跌是因为买来的新用户质量太差->大量新用户来了不买->付费转化率被拉低")
print("  3. 更深层：低质用户(信息流)权重从20%滚到55%->整体指标被持续拉低")
print("  4. 虽然MAU在涨——但「有效MAU」(真正活跃+付费的用户)实际上在降")
print("  5. Retained绝对值35->34万(-3%)——唯一免费且可持续的增长动力在减弱")
print()

print("━" * 65)
print("【行动建议——三级优先级】")
print("━" * 65)
print()
print("Short-term 短期止血（本周）:")
print("  (1) 暂停次留<15%的渠道(信息流),立即止血")
print("  (2) 省下预算(~40万/月)转做新用户首单补贴+激活引导")
print("  (3) 重新审核所有投放素材——检查是否标题党骗点击")
print()
print("Mid-term 中期修复（本月）:")
print("  (1) 优化浏览->加购: 完善商品详情页(更多实拍/评价/销量)")
print("  (2) 建立渠道质量门槛: W1留存<20%自动停投")
print("  (3) 新用户首单优惠+激活引导(目标: 加购率12%->20%)")
print()
print("Long-term 长期建设（本季）:")
print("  (1) PDS+同期群监控看板(周级自动刷新)")
print("  (2) RFM分群运营体系(差异化触达)")
print("  (3) 向CEO汇报「Retained绝对值」而非只看DAU总量")
print("  (4) 建立渠道质量评分体系(CAC+LTV+次留三维评分)")
print()

print("预期效果量化:")
print("  停止低质渠道->月省40万投放预算")
print("  浏览->加购优化: 从18%->25%(+7pp)->月增约3500付费用户")
print("  3个月后: MAU可能短期微降->但GMV预期回升至~820万(+9%)")
print("  6个月后: Retained止跌回升->可适度加大拉新->进入健康增长循环")
print()
print("=" * 65)
print("【诊断完成】6大框架联合诊断: AARRR+同期群+漏斗+LTV+PDS+RFM")
print("=" * 65)
````

## 三、本周核心收获——六个框架各一句话

1. **AARRR找杠杆点**：不是所有环节都优化——找到最漏水的环节（激活？变现？），集中资源堵上。所有环节都做=所有环节都做不好。
2. **同期群看趋势**：整体留存会骗你——纵向对比一列（新批次vs旧批次）才能看到「新用户质量在变好还是变差」。
3. **漏斗找漏水点**：绝对流失和步骤转化率两个视角指向的优化方向可能完全不同——两者结合判断最优目标。永远做分群漏斗。
4. **留存+LTV看长期价值**：留存曲线的「拉平点」比次日留存重要100倍。LTV/CAC > 3是铁律——<1则商业模式不成立。
5. **RFM做用户分层**：「我们的用户」不存在——R/F/M三维正交出8个群体，让运营对不同的人说不同的话。
6. **PDS看骨架**：DAU涨了≠增长健康。Retained在涨=真增长；Retained在跌但New/Resurrected在涨=虚胖应先止血。

### W2面试常见题汇总

**Q1: 如果老板让你提升DAU，你会先看什么？**
A: 先不看DAU总量，先做PDS拆解看结构。如果Retained占比>60%且绝对值在涨->留存驱动型->可以加大拉新。如果Retained在跌但New在涨->虚胖型->先止血修产品再拉新。核心逻辑：留存底盘不稳时拉新=往漏水的桶里倒水。

**Q2: 怎么判断一个产品是在变好还是变差？**
A: 用同期群的纵向对比——看最近几批新用户的W1/W2/W4留存是否持续高于更早的批次。如果W1/W2/W4三列都在涨->产品在改善。如果W1涨了但W4没涨->改版只改善了第一印象，没改善长期粘性。要排除季节性影响——跟去年同期比而非跟上个月比。

**Q3: 整体留存25%看起来很稳定，能说产品没问题吗？**
A: 不能。整体25%可能掩盖了「1月用户留存35%，6月用户留存15%」的持续恶化——这是辛普森悖论：各组内部都在变差，但因为权重变化(6月新用户数量暴增)，整体平均数看起来没变。必须用同期群纵向对比才能发现。

**Q4: LTV/CAC=2.5，能继续投放吗？**
A: LTV/CAC=2.5处于临界区。铁律是>3才健康。2.5意味着获客成本占40%，运营成本可能再占30-40%，净利润空间很薄。如果公司在快速扩张期可以暂时容忍(期待规模效应降CAC或网络效应提LTV)，但需要密切监控趋势——如果从3.0降到2.5再降到2.0->说明投放效率在恶化。

**Q5: 一个新用户来了不看商品就走了，问题出在哪？**
A: 用漏斗定位：打开->浏览转化率是多少？如果这一步就很低(<50%)->可能是落地页体验差或渠道不匹配。如果打开->浏览正常但浏览->加购低->商品详情页信息不足或价格不透明。永远分渠道看——不同渠道的用户在漏斗上的表现可能完全不同。
