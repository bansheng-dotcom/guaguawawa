### 1.4 一个合格的分析师对AB实验的态度

不是"实验平台说显著，我就写显著"——而是：
- 先看SRM对不对（地基）
- 再看效应量大不大（有没有业务意义）
- 再看护栏指标有没有恶化（不能拆东墙补西墙）
- 再看Peeking有没有被控制（没控制的p值不可信）
- 最后看p值和CI是否支撑结论（统计证据）

**五步审查法**：SRM → 效应量 → 护栏 → Peeking → p值/CI

## 🧩 区块2：本周难点深度再解

### 2.1 最难的概念1：反向MDE

**为什么难**：教科书中都是"给定MDE→算样本量"，但实际工作中"样本量是固定的，你能检测到什么？"——这个方向教科书很少教。

**核心理解**：反向MDE是用数值方法（二分法）求解"给定n, α, power, 基线值，能检测到的最小δ"。因为δ在分母的平方位置且p2=f(p1, δ)，反解δ没有闭式解。

**检验**：你能不能跟产品经理解释"以我们的流量，2周能检测到至少X%的提升"？

### 2.2 最难的概念2：CUPED的θ为什么只能在对照组算

**为什么难**：直觉上"用全量数据更稳定"——但用全量数据会引入偏差。

**核心理解**：如果实验有效果，实验组的Y比对照组高。把实验组数据混入θ的估计→θ既反映了"X-Y的自然相关性"，又反映了"处理效应"→θ被高估或低估→调整后的Y_adj有偏。只用对照组算θ保证θ反映的是"纯粹的自然相关性"。

### 2.3 最难的概念3：Cluster实验的Design Effect

**为什么难**：习惯了"10000个用户就是10000个独立观测"——但集群实验中不是。

**核心理解**：集群内的用户互相相似（同一城市、同一公司）→10000个用户其实只相当于n_effective个"独立信息"。Design Effect = 1+(m-1)×ICC，其中ICC是集群内相关系数。ICC=0.05（很小），m=500→Design Effect≈26→10000个用户≈只有385个"有效用户"。

### 2.4 最难的概念4：为什么Peeking的α膨胀不能用Bonferroni简单校正

**为什么难**：Bonferroni校正用于"独立的多次检验"——但Peeking中的每天p值是高度相关的（数据累积）。

**核心理解**：第1-7天，每天的数据有大量重叠——第7天的数据包含了第1天的数据。因此Bonferroni（假设独立检验）太保守了。序贯检验（如O'Brien-Fleming边界）专门为这个场景设计——在"重叠数据"下精确控制总体α。

````python
print("=" * 60)
print("W8 综合自测")
print("=" * 60)

questions = [
    ("Q1: MDE从10%降到5%，样本量需要增加多少倍？", "约4倍（MDE减半→n翻4倍）"),
    ("Q2: CUPED的方差缩减比例等于什么？", "ρ²（实验前后指标相关系数的平方）"),
    ("Q3: CUPED的θ应该在哪组数据上计算？", "只在对照组（避免实验组数据污染θ）"),
    ("Q4: SUTVA的'无干扰'假设是什么？", "用户A接受处理不影响用户B的结果"),
    ("Q5: 滴滴给部分司机补贴的实验应该用什么设计？", "Switchback（交替时段，避免组间溢出）"),
    ("Q6: 分层分流的两个核心原则是什么？", "层内互斥、层间正交"),
    ("Q7: Peeking 14天（每天看p值），α从5%膨胀到约多少？", "约51%（1-(1-0.05)^14≈0.512）"),
    ("Q8: 实验太多了，看了20个指标，预期有多少个假阳性？", "约1个（20×0.05=1）"),
    ("Q9: AB实验的8个步骤是什么？", "需求→设计→样本量→监控→分析→决策→报告→回顾"),
    ("Q10: 看实验结果时，5个检查点是什么？", "SRM→主指标→效应量→护栏→Peeking"),
]

score = 0
for q, a in questions:
    print(f"\n{q}")
    input("按Enter查看答案...")
    print(f"  答案: {a}")

print("\n" + "=" * 60)
print("自评：每答对1题得1分")
print("  8-10分：本周掌握扎实，可以进入W9因果推断")
print("  5-7分：核心概念理解了，需要复习细节")
print("  0-4分：建议回看本周的重点章节（D1/D2/D5）")
print("=" * 60)

import numpy as np
from scipy import stats

print("\n\n" + "=" * 60)
print("练习2：综合技能检验——独立完成一次实验分析")
print("=" * 60)

# 场景：给你两组数据，完成完整的分析
np.random.seed(123)
n = 5000
control = np.random.poisson(2.0, n)
experiment = np.random.poisson(2.15, n)  # 真实+7.5%

print("\n场景：新搜索算法AB实验")
print("  主要指标：人均搜索次数")
print("  护栏指标：搜索结果点击率（模拟数据）")
print("  实验周期：2周，每组5000人")

# 你的任务
print("\n请完成以下分析：")
print("1. SRM检查（假设两组都是5000人）")
print("2. 主指标t检验（p值、效应量、95%CI）")
print("3. 护栏指标检查（模拟点击率数据）")
print("4. 决策建议")

input("\n按Enter查看分析结果...")

# 答案
print("\n" + "-" * 40)
print("参考答案：")
print("-" * 40)

# SRM
print("\n1. SRM: 对照组5000人 vs 实验组5000人 → 50:50 ✓")

# 主指标
t, p = stats.ttest_ind(control, experiment)
effect = experiment.mean() - control.mean()
effect_rel = (experiment.mean() / control.mean() - 1) * 100
se = np.sqrt(control.var()/n + experiment.var()/n)
ci = stats.t.interval(0.95, 2*n-2, loc=effect, scale=se)
print(f"\n2. 主指标(人均搜索): 对照{control.mean():.2f}, 实验{experiment.mean():.2f}")
print(f"   提升{effect_rel:.1f}%, p={p:.4f}, 95%CI=[{ci[0]:.2f},{ci[1]:.2f}]")
print(f"   → {'显著' if p < 0.05 else '不显著'}")

# 护栏指标
c_ctr = np.random.beta(10, 90, n)
e_ctr = np.random.beta(11, 89, n)
t2, p2 = stats.ttest_ind(c_ctr, e_ctr)
print(f"\n3. 护栏(点击率): 对照{c_ctr.mean():.3f}, 实验{e_ctr.mean():.3f}, p={p2:.4f}")
print(f"   → {'⚠ 护栏指标有差异，需要调查' if p2 < 0.05 else '✓ 护栏指标正常'}")

# 决策
print(f"\n4. 决策:")
if p < 0.05 and p2 > 0.05:
    print(f"   ✅ 建议上线——主指标显著提升且护栏无恶化")
    print(f"   预期效果：人均搜索+{effect_rel:.1f}%（95%CI=[{ci[0]:.2f},{ci[1]:.2f}]）")
elif p < 0.05 and p2 < 0.05:
    print(f"   ❌ 不上线——护栏指标恶化")
elif p > 0.05:
    mde = np.sqrt(2)*(stats.norm.ppf(0.975)+stats.norm.ppf(0.80))*np.sqrt(control.var())/np.sqrt(n)
    print(f"   ⏸ 不显著，MDE={mde:.2f}次")
    if effect < mde:
        print(f"   效应量{effect:.2f} < MDE{mde:.2f} → 样本不够，建议延长实验")
    else:
        print(f"   效应量{effect:.2f} > MDE{mde:.2f}但n够了 → 策略可能真的没效果")

print("\n\n💡 以上就是一个完整的AB实验分析流程。如果你能独立完成，恭喜你——你已经具备了独立主导AB实验的能力！")
````

## 📊 区块4：本周知识整合——5个核心框架

### 框架1：实验设计的决策树

```
实验能做标准AB吗？
├── 是 → 标准AB实验
│   ├── 有实验前数据吗？
│   │   ├── 是 → 用CUPED
│   │   └── 否 → 用标准t检验
│   └── 实验设计
│       ├── 正向MDE（如果我们想检测X%，需要多少样本？）
│       └── 反向MDE（以我们的流量，能检测到多少？）
└── 否 → 为什么？
    ├── 网络效应 → Cluster/Switchback
    ├── 长期效应 → Holdout
    └── 供需耦合 → Switchback/双边市场设计
```

### 框架2：实验分析的5步审查法

```
① SRM → ② 主指标(p+效应量+CI) → ③ 效应量vs MDE → ④ 护栏 → ⑤ Peeking
```

### 框架3："不显著"的诊断流程

```
实验不显著(p>0.05)
├── 实际n ≥ MDE要求的n → 策略真的没效果（或效果<MDE）
├── 实际n < MDE要求的n → 样本不够→延长实验
├── SRM异常 → 实验无效，先修bug
└── 数据质量异常 → 先修数据
```

### 框架4：实验决策矩阵

| | 主指标显著 | 主指标不显著 |
|------|---------|---------|
| 护栏OK | ✅ 上线 | ⏸ n够→放弃; n不够→延长 |
| 护栏恶化 | ❌ 不上线 | ❌ 不上线 |

### 框架5：分析师的核心竞争力

```
Level 1: 会做t检验（会看p值）
Level 2: 会算样本量/MDE（会设计实验）
Level 3: 会识别陷阱（Peeking/多重检验/指标偷换）
Level 4: 会选实验设计（标准AB/Cluster/Switchback/Holdout）
Level 5: 会用实验推动业务决策（从"p显著"到"这个策略值得做"）
```

## 🔗 区块6：下周预告与知识衔接

### W9 因果推断——当AB实验不可行时

**为什么需要因果推断**：
- 不能做实验：不能随机分配用户"是否上大学"来测试教育对收入的影响
- 不想做实验：改变定价策略→可能伤害用户体验→不敢做AB实验
- 做不了实验：政策变化影响所有人→没有对照组

**W9核心内容预览**：
| 方法 | 核心思想 | 适用场景 |
|------|---------|---------|
| DID（双重差分） | 对比"处理组变化"和"对照组自然变化" | 政策评估、产品改版 |
| PSM（倾向得分匹配） | 为每个处理组用户找一个"最相似"的对照 | 用户分群对比 |
| IV（工具变量） | 找一个"只影响处理分配不影响结果"的变量 | 内生性问题 |
| RDD（断点回归） | 在某个阈值附近，两边的用户"几乎相同" | 补贴门槛、分数线 |

**W9和W8的关系**：
- AB实验是因果推断的"金标准"（随机化消除了所有混杂）
- W9的方法是"银标准"——在没有随机化时，尽量逼近随机化
- 理解W8的实验设计，才能理解为什么W9的方法需要"额外的假设"
