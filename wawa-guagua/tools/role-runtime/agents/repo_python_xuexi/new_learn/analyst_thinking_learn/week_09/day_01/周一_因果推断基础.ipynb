# 第9周 · 周一：因果推断基础——相关 ≠ 因果\n\n> 难度：⭐⭐⭐⭐ | 定位：理解因果推断的第一性原理，建立反事实思维框架\n> 前置依赖：假设检验(W6)、AB实验基础(W7-W8) | 后续用途：DID/PSM/IV/RDD(W9D2-D6)、整合应用(W10-W12)\n> 面试地位：大厂数据科学面试必考，尤其是"如何评估一个策略/功能的效果"类问题\n\n---\n\n## 📘 区块1：核心概念与第一性原理\n\n### 1.1 因果推断的统一框架：反事实思维\n\n因果推断的所有方法，本质上都在回答同一个问题：**如果没有接受处理，结果会怎样？**\n\n这被称为"反事实"（Counterfactual）——与已发生的事实相反的状态。因果效应定义为：\n\n```\n因果效应 = 事实结果 - 反事实结果\n     = Y(处理=1) - Y(处理=0)\n```\n\n但反事实永远无法被直接观测。你不可能同时让同一个用户既领券又不领券，既上大学又不上大学。这是因果推断的"根本问题"（Fundamental Problem of Causal Inference）。\n\n所有因果推断方法的区别，仅在于**如何构建一个可信的反事实**：\n- AB实验：随机分组 → 对照组就是反事实\n- DID：利用对照组的时间趋势来"模拟"实验组的反事实趋势\n- PSM：找一个"其他特征相同但没被处理"的人作为反事实\n- IV：利用工具变量创造"局部的随机化"\n- RDD：利用断点附近的局部随机性\n\n### 1.2 为什么"相关不等于因果"——从数学到直觉\n\n相关性衡量的是两个变量"一起变"的程度，而因果性衡量的是"一个变量导致另一个变量变化"的关系。两者的本质区别在于：\n\n**相关性是对称的，因果性是不对称的。**\n\n```\ncorr(X, Y) = corr(Y, X)  ← 相关性对称\n但 "X导致Y" ≠ "Y导致X"     ← 因果性不对称\n```\n\n相关性可以由以下任何一种原因产生，而只有第一种是真正的因果关系：\n\n| 关系类型 | 数学表达 | 经典例子 |\n|---------|---------|----------|\n| X导致Y（因果） | X→Y | 学习时间↑→考试成绩↑ |\n| Y导致X（反向因果） | Y→X | 销售额↑→广告投放↑（其实是看到销售好才追加投放）|\n| Z同时导致X和Y（混淆） | X←Z→Y | 夏天→冰淇淋销量↑ + 溺水死亡↑ |\n| X和Y纯属巧合 | X  Y | 某股票涨跌和某足球队胜负 |\n| 选择偏差 | X←S→Y | 只分析"存活下来"的用户→幸存者偏差 |\n\n### 1.3 混淆变量的第一性原理\n\n混淆变量（Confounder）是同时影响"处理分配"和"结果"的变量。它之所以危险，是因为它制造了"处理和结果之间的虚假关联"。\n\n**混淆变量的三个条件**（满足全部三个才是混淆变量）：\n1. 与处理变量相关（影响是否被处理）\n2. 与结果变量相关（影响最终结果）\n3. 不在处理到结果的因果路径上（不是中介变量）\n\n**判断是否控制某个变量的准则**：\n- 混淆变量 → 必须控制（否则有偏）\n- 中介变量 → 不能控制（否则阻断因果路径，低估效应）\n- 对撞变量 → 不能控制（否则引入虚假关联）\n- 工具变量 → 不需要控制（它本就是用来替代处理的）\n\n### 1.4 因果推断方法全景图\n\n在数据分析的实际工作中，你面对的场景决定了你该用什么方法：\n\n| 方法 | 反事实构建方式 | 核心假设 | 难度 | 可信度 |\n|------|--------------|---------|------|-------|\n| **AB实验(RCT)** | 随机分组→对照组 | 随机化成功 | ⭐ | ⭐⭐⭐⭐⭐ |\n| **DID** | 对照组趋势外推 | 平行趋势 | ⭐⭐ | ⭐⭐⭐⭐ |\n| **PSM** | 倾向得分匹配相似者 | 无未观测混淆 | ⭐⭐⭐ | ⭐⭐⭐ |\n| **IV** | 工具变量的局部随机 | 外生性(不可检验!) | ⭐⭐⭐⭐ | ⭐⭐⭐⭐ |\n| **RDD** | 断点附近的局部随机 | 断点附近连续 | ⭐⭐⭐ | ⭐⭐⭐⭐ |\n| **合成控制** | 合成一个"虚拟对照" | 处理前拟合良好 | ⭐⭐⭐⭐ | ⭐⭐⭐ |\n\n**选择原则**：能做实验优先做实验。不能做实验时，DID是第一备选（最直观），PSM是第二备选（数据要求高）。IV和RDD是"老天给的实验"，有就用，没有不能强求。\n\n### 1.5 面试高频考点\n\n**面试官常问**：\n- "相关性和因果性有什么区别？举三个例子"\n- "什么是混淆变量？如何识别和控制？"\n- "不能做AB实验时，如何评估一个策略的效果？"\n- "什么是反事实？为什么它观测不到？"\n- "辛普森悖论是什么？它如何说明'相关≠因果'？"\n\n**回答要点**：\n- 相关是对称的统计度量，因果是不对称的机制关系\n- 混淆变量同时影响处理分配和结果，是假相关的根源\n- 不能做实验时用DID/PSM/IV/RDD，每种方法有其假设\n- 反事实是"如果没被处理会怎样"，观测不到因为同一个人只能处于一种状态\n- 辛普森悖论：分组看趋势和总体看趋势相反，说明混淆变量在作祟

## 🧩 区块2：知识点详细拆解\n\n### 2.1 反事实框架的数学定义\n\n**为什么重要**：这是所有因果推断方法的数学基础，理解了它，所有方法都是同一框架的不同实现。\n\n**潜在结果框架（Rubin Causal Model）**：\n- $Y_i(1)$：个体i如果接受处理的结果\n- $Y_i(0)$：个体i如果不接受处理的结果\n- $T_i$：个体i实际是否接受处理（0或1）\n- $Y_i^{obs} = T_i \cdot Y_i(1) + (1-T_i) \cdot Y_i(0)$：实际观测到的结果\n\n**个体因果效应**：$\tau_i = Y_i(1) - Y_i(0)$\n\n**平均处理效应（ATE）**：$ATE = E[Y(1) - Y(0)]$\n\n**处理组平均处理效应（ATT）**：$ATT = E[Y(1) - Y(0) | T=1]$\n\n这三者的区别非常重要：\n- ATE：对所有人平均（包括永远不会被处理的人）\n- ATT：只对实际被处理的人（政策评估中最常用）\n- ATU（对照组平均处理效应）：只对实际没被处理的人\n\n**为什么ATT和ATE可能不同**：如果处理分配不是随机的（比如只有"有能力的人"才上大学），那么上过大学的人（T=1）和整体人群的因果效应可能不同。PSM通常估计ATT。\n\n### 2.2 混淆变量的DAG表示\n\n**为什么重要**：DAG（有向无环图）是可视化因果关系的标准工具，面试和论文中都会用到。\n\n**DAG的基本元素**：\n- 节点：变量\n- 有向边：因果方向（X→Y表示X导致Y）\n- 路径：连接两个节点的边的序列\n\n**三种基本结构**：\n\n```\n1. 链式结构: X → Z → Y\n   X和Y相关，但控制Z后，X和Y独立（Z是中介变量）\n\n2. 分叉结构: X ← Z → Y\n   X和Y相关（因为共同的因Z），控制Z后，X和Y独立（Z是混淆变量）\n\n3. 对撞结构: X → Z ← Y\n   X和Y独立，但控制Z后，X和Y变得相关！（Z是对撞变量）\n   经典例子：能力(X)和颜值(Y)独立，但控制"是否成为明星(Z)"后，\n   你发现"不是明星的人中，能力强但颜值低"——引入了虚假关联！\n```\n\n**DAG的核心价值**：帮助你判断该控制哪些变量、不该控制哪些变量。这在复杂项目中极其关键——控制错了变量反而引入偏误。\n\n### 2.3 选择偏差的四种经典形式\n\n**为什么重要**：选择偏差是观测数据中最常见的偏误来源，识别它是因果推断的第一步。\n\n| 类型 | 机制 | 经典案例 | 如何修正 |\n|------|------|---------|----------|\n| **自选择偏差** | 个体自己决定是否接受处理 | 更健康的人主动参加健身计划 → 高估效果 | PSM/IV |\n| **样本选择偏差** | 分析样本不能代表目标总体 | 只用App用户分析→忽略了非App用户 | 重新抽样/加权 |\n| **幸存者偏差** | 只看到"活下来"的样本 | 只分析还在活跃的用户→忽略流失用户 | 纳入流失用户数据 |\n| **志愿者偏差** | 参与者是自愿的，可能系统性不同 | 用户调研只收到满意用户的回复 | 主动联系未回复者 |\n\n### 2.4 辛普森悖论\n\n**为什么重要**：这是面试中最经典的"相关≠因果"案例，也是数据分析中最容易犯的错误。\n\n**辛普森悖论的定义**：在分组数据中呈现的趋势，与在聚合数据中呈现的趋势相反。\n\n**经典例子**：某医院两种肾结石治疗方案\n- 总体看：方案A治愈率78%，方案B治愈率83% → B更好\n- 分组看（按结石大小）：\n  - 小结石：方案A治愈率93%（方案B 87%）→ A更好\n  - 大结石：方案A治愈率73%（方案B 69%）→ A更好\n  - → 两种情况下都是A更好，但总体数据却是B更好！\n\n**为什么会这样**：医生倾向于给严重的大结石患者用B方案（B方案是"侵入性"疗法，适合重症），所以B方案处理的"重症患者占比更高"——这拉低了B的总体治愈率。但分组比较时这个偏差被消除了。\n\n**辛普森悖论的教训**：\n1. 永远不要只看总体数据就下结论\n2. 先画分组图（按可能的混淆变量分组）\n3. 如果分组结论和总体结论不一致，一定有混淆\n4. 这时候要相信分组结论（控制混淆后的）\n\n### 2.5 实验数据 vs 观测数据的思维差异\n\n**为什么重要**：很多分析师习惯了"跑个相关就跑个相关"，但不理解实验数据和观测数据在分析思路上有本质区别。\n\n| 维度 | 实验数据(AB实验) | 观测数据(历史数据) |\n|------|-----------------|-------------------|\n| 处理分配 | 随机化→无混淆 | 非随机→可能存在混淆 |\n| 分析重点 | 直接对比两组 | 先处理混淆再对比 |\n| 主要威胁 | 违反SUTVA/随机化失败 | 选择偏差/混淆变量 |\n| 可信度 | 高（如果设计好） | 取决于混淆控制程度 |\n| 所需技能 | 实验设计+假设检验 | 因果推断方法+因果图 |\n\n**关键**：很多人学了AB实验后，直接用t检验分析观测数据——这是非常危险的。观测数据需要先处理混淆变量，才能进行比较。\n\n### 2.6 因果推断在互联网业务中的应用全景\n\n**为什么重要**：知道"哪里可以用因果推断"比知道"因果推断是什么"更重要。\n\n| 业务场景 | 典型问题 | 推荐方法 |\n|---------|---------|----------|\n| 策略评估 | 新券种/新定价/新权益带来了多少增量？ | DID/PSM |\n| 渠道归因 | 某次广告投放带来了多少转化？ | PSM/DID |\n| 产品功能 | 新功能上线对留存的影响？ | DID |\n| 运营活动 | 某次大促带来了多少GMV增量（而不是透支）？ | DID+合成控制 |\n| 用户分群 | 高价值用户的"成因"是什么？ | PSM+异质性分析 |\n| 算法迭代 | 新推荐模型对用户时长的真实提升？ | AB实验(DID作为补充) |\n\n### 2.7 随机化为什么是金标准\n\n**为什么重要**：理解随机化的数学本质，才能理解为什么观测方法"只能逼近"而不能"证明"。\n\n**随机化的核心**：在随机化下，处理分配$T$独立于潜在结果$(Y(0), Y(1))$。\n\n这意味着：\n```\nE[Y(0)|T=1] = E[Y(0)|T=0] = E[Y(0)]  ← 处理组的"反事实"等于对照组的"事实"\n```\n\n所以观测到的组间差异 = 因果效应：\n```\nE[Y|T=1] - E[Y|T=0] = E[Y(1)|T=1] - E[Y(0)|T=0]  ← 随机化下\n                     = E[Y(1)] - E[Y(0)]        ← 独立于T\n                     = ATE                       ← 就是因果效应！\n```\n\n**非随机化下**（观测数据），这个等式不成立：\n```\nE[Y|T=1] - E[Y|T=0] = ATE + 选择偏差\n选择偏差 = E[Y(0)|T=1] - E[Y(0)|T=0]  ← 处理组和对照组即使都不被处理，\n                                           结果也不同（这就是混淆！）\n```\n\n### 2.8 可忽略性假设\n\n**为什么重要**：这是所有观测因果推断方法的核心假设——如果不能"忽略"处理分配机制，就无法识别因果效应。\n\n**可忽略性假设（Ignorability / Unconfoundedness）**：\n在给定协变量X的条件下，处理分配T独立于潜在结果。\n```\n(Y(0), Y(1)) ⊥ T | X\n```\n\n用白话：给定我能观测到的所有变量X，是否接受处理就和"如果没被处理的结果"无关了。\n\nPSM和DID（加了控制变量时）都依赖这个假设。但问题是：你永远无法从数据中检验它——因为Y(0)对处理组观测不到。这就是为什么因果推断永远带有一个"信念"成分。

### 步骤3：挑战练习——构建一个完整的因果推断工作流\n\n3.1 用真实商业场景（新功能上线评估）模拟完整的因果分析流程\n3.2 对比朴素估计、回归调整、PSM的差异

````python
# 步骤3：完整的因果分析工作流\nprint('\\n【步骤3】完整工作流：评估新功能的真实效果')\nprint('='* 60)\n\n# 场景：APP上线了"深色模式"功能，想知道它对用户时长的真实影响\n# 背景：用户可以选择是否开启，开启的用户可能"本来就对这个APP更忠诚"\n\nnp.random.seed(2024)\nn = 3000\n\n# 用户特征\nage = np.random.uniform(18, 55, n)              # 年龄\nloyalty = np.random.normal(0, 1, n)             # 忠诚度（未观测！）\ntech_savvy = np.random.normal(0, 1, n)          # 技术熟练度\nusage_days = np.random.poisson(20, n)           # 已使用天数\n\n# 是否开启深色模式：跟忠诚度+技术熟练度+年龄（年轻人更可能开启）有关\nprob_dark = 1 / (1 + np.exp(-(-0.5 + 0.6 * loyalty + 0.5 * tech_savvy - 0.3 * (age - 35)/10)))\ndark_mode = np.random.binomial(1, prob_dark)\n\n# 日均使用时长（分钟）：真实因果效应=15分钟（深色模式确实能提升时长）\ntrue_dark_effect = 15\ndaily_minutes = (60 + 5 * loyalty + 3 * tech_savvy + 0.2 * usage_days \n                 + true_dark_effect * dark_mode + np.random.normal(0, 10, n))\n\n# 构建数据框\ndf = pd.DataFrame({\n    'age': age, 'tech_savvy': tech_savvy, 'usage_days': usage_days,\n    'dark_mode': dark_mode, 'daily_minutes': daily_minutes, 'loyalty': loyalty\n})\n\n# --- 方法1：朴素对比（有偏）---\nnaive = df[df.dark_mode == 1]['daily_minutes'].mean() - df[df.dark_mode == 0]['daily_minutes'].mean()\nprint(f'方法1 - 朴素对比:        {naive:.1f}分钟 (真实效应={true_dark_effect}) 偏误={naive - true_dark_effect:.1f}')\n\n# --- 方法2：回归调整（控制了可观测变量）---\nfrom sklearn.linear_model import LinearRegression\nX_reg = df[['age', 'tech_savvy', 'usage_days', 'dark_mode']]\nreg = LinearRegression().fit(X_reg, df['daily_minutes'])\nreg_estimate = reg.coef_[3]  # dark_mode的系数\nprint(f'方法2 - 回归调整:        {reg_estimate:.1f}分钟 (真实效应={true_dark_effect}) 偏误={reg_estimate - true_dark_effect:.1f}')\n\n# --- 方法3：PSM简易版（用倾向得分匹配）---\nfrom sklearn.linear_model import LogisticRegression\nlr_ps = LogisticRegression()\nlr_ps.fit(df[['age', 'tech_savvy', 'usage_days']], df['dark_mode'])\ndf['ps'] = lr_ps.predict_proba(df[['age', 'tech_savvy', 'usage_days']])[:, 1]\n\n# 最近邻匹配（不放回）\ntreated_df = df[df.dark_mode == 1].copy()\ncontrol_df = df[df.dark_mode == 0].copy()\nmatched_controls = []\nfor _, row in treated_df.iterrows():\n    idx = (control_df['ps'] - row['ps']).abs().idxmin()\n    matched_controls.append(control_df.loc[idx, 'daily_minutes'])\n    control_df = control_df.drop(idx)\n\npsm_estimate = treated_df['daily_minutes'].mean() - np.mean(matched_controls)\nprint(f'方法3 - PSM(倾向得分匹配): {psm_estimate:.1f}分钟 (真实效应={true_dark_effect}) 偏误={psm_estimate - true_dark_effect:.1f}')\nprint()\n\nprint('结论：')\nprint('  1. 朴素对比高估了效果（因为开启深色模式的用户本身更忠诚）')\nprint('  2. 回归调整改善了估计（但依赖线性假设）')\nprint('  3. PSM进一步改善了估计（匹配了特征相似的用户）')\nprint(f'  4. 真实效应={true_dark_effect}分钟，各种方法都在向它逼近')\nprint('  5. 如果loyalty（忠诚度）不可观测（现实！），所有方法都还有偏——这是因果推断的根本困境')
````

## 📊 区块4：业务场景应用\n\n### 场景1：运营说"发了优惠券的GMV更高"——如何质疑和纠正\n\n**背景**：运营团队发了一批优惠券，发现领券用户的平均GMV是300元，未领券用户的平均GMV只有180元。运营据此要求扩大发券预算。\n\n**分析过程**：\n\n第一步——检查是否只是"选择偏差"。查发券记录发现：这批券是通过"高价值用户画像"定向发放的——只发给了过去3个月累计消费>1000元的用户。这就是典型的选择偏差。\n\n第二步——分层对比。按用户过去3个月消费分层：\n- 历史消费>1000元（发券组）vs 历史消费>1000元（未发券组，但用户画像相似）\n- 发现：两组GMV差异从120元缩小到25元\n\n第三步——结论：券的真实效应约25元，而非120元。300-180=120中，大部分是"这两组人本来就不一样"带来的差异。\n\n**决策**：如果券成本<25元，扩量可行；如果券成本>25元，目前策略不经济。建议做一次真正的AB测试来验证。\n\n**教训**：永远不要直接对比"被处理"和"没被处理"的原始均值——除非是随机化实验。\n\n### 场景2：产品经理说"新功能用户留存更高"——如何用因果思维审视\n\n**背景**：PM发现使用新功能的用户7日留存率是65%，未使用的用户只有40%。PM建议加大推广力度。\n\n**因果思维审视**：\n\n1. **反向因果**：是"用了功能所以留存高"，还是"留存意愿强的用户更可能去尝试新功能"？——后者的可能性很大！\n\n2. **选择偏差**：新功能是可选开启的，愿意尝鲜的用户本来就对产品更忠诚。\n\n3. **混淆变量**：\n   - 用户注册时间（老用户更可能尝试+留存更高）\n   - 用户活跃度（高活跃用户更可能尝试+留存更高）\n   - 设备类型（高端设备用户更可能尝试+体验更好→留存高）\n\n4. **正确做法**：\n   - 优先做AB实验（随机分组，控制组不展示新功能入口）\n   - 如果没法做实验：PSM匹配（匹配注册时间、活跃度、设备等）后再比较\n   - 或者用DID（如果新功能是某天全量上线的）\n\n**结论**：65% vs 40%的差异中，大部分可能是选择偏差，新功能的真实效果可能只有5-10个百分点的提升。需要实验验证。\n\n### 场景3：数据科学面试中的因果推断——如何回答"评估X策略效果"\n\n**典型面试问题**："老板让你评估我们的VIP会员体系是否有效提升了用户消费，你会怎么分析？"\n\n**标准回答框架**：\n\n1. **先确认能不能做实验**：\n   - "如果可以，我会设计一个AB实验——随机选取一部分用户给VIP邀请，另一部分不给，跟踪3个月后对比两组消费。随机化可以消除选择偏差。"\n\n2. **如果不能做实验，选因果推断方法**：\n   - "如果VIP是按某个规则统一推出的，可以用DID——对比推出前后VIP用户和非VIP用户的消费变化趋势差异。"\n   - "如果VIP是用户自己选择的，可以用PSM——用注册时长、历史消费、活跃度等变量做倾向得分匹配，找到'和VIP用户特征相似但没开VIP的人'，对比两组消费差异。"\n\n3. **主动说明方法的局限**：\n   - "PSM只控制了可观测的变量，如果'用户的内在消费意愿'（不可观测）同时影响'是否开VIP'和'消费金额'，估计仍然有偏。我的结论会标注这个局限性。"\n\n**加分项**：主动提到"即使是最好的观测因果推断方法，也不如一个设计良好的随机化实验"，展示你对因果推断局限性的清醒认知。\n\n### 场景4：CEO问"渠道A的用户质量比渠道B好吗"——混淆变量的识别实战\n\n**背景**：数据分析师拉了一份报告：渠道A来的用户人均消费500元，渠道B来的用户人均消费300元。CEO想砍掉渠道B的预算，全部投到A。\n\n**深入排查**：\n\n你不能只看人均消费，需要排查"渠道之间用户的初始特征本来就不同"这个可能性。\n\n1. **按城市分层**：渠道A主要在一线城市投放（用户消费力高），渠道B主要在低线城市投放（用户消费力低）。在同一个城市内，两个渠道的用户消费差异大幅缩小。\n\n2. **按设备分层**：渠道A主要来自iOS端（高端用户多），渠道B主要来自Android端。同一设备类型内，两渠道差异缩小。\n\n3. **结论**：渠道A的"高消费"主要是用户群体本身更优质，而非渠道带来的增量价值。砍B扩A可能不会带来预期的GMV提升——因为A渠道覆盖的人群快饱和了。\n\n4. **正确做法**：在同一个城市+同一设备类型内，对比A渠道和B渠道用户的长期LTV差异，或者做渠道的增量实验。\n\n**教训**：当两个群体在某指标上差异很大时，先想"他们是否本来就不同"，而不是"处理导致了差异".

## 📝 区块7：当日小结\n\n> **因果推断的所有方法本质上是同一件事：构建可信的反事实。AB实验用随机化构建反事实，DID用趋势外推，PSM用相似人群，IV用工具变量的局部随机化，RDD用断点附近的局部随机化。"相关不等于因果"不是一句口号——它有清晰的数学定义：相关可能来自混淆、反向因果、选择偏差或纯属巧合。理解DAG（因果图）是判断"该控制什么变量"的基础工具。辛普森悖论提醒我们：永远先分组再看总体，而不是相反。在实际工作中，能做实验优先做实验；不能做实验时，诚实地说清楚方法的假设和局限，比假装证明了因果更重要。**\n\n**三条金句带回家**：\n1. 所有因果推断方法都在构建反事实——"如果没被处理会怎样"\n2. 混淆变量是"相关≠因果"的根源——控制可观测的混淆，诚实面对不可观测的混淆\n3. 辛普森悖论的准则：分组结论比总体结论更可信——先分层，再比较\n\n### 🔖 课后拓展\n\n1. **练习**：找一个你工作中遇到的"相关但可能非因果"的例子，画出它的DAG图，标注混淆变量和可能的因果路径\n\n2. **思考**：如果用户的历史行为数据中有100个特征，你怎么判断哪些应该放进倾向得分模型、哪些不应该？（提示：因果知识>统计选择→基于领域知识画DAG，先判断每个变量是混淆/中介/对撞/无关）\n\n3. **预习**：明天要学的DID会用"对照组的时间趋势"来构建反事实。思考：为什么DID不要求所有混淆变量都可观测（而PSM要求）？\n\n4. **面试准备**：背诵以下面试题答案：\n   - "相关性和因果性有什么区别？举三个例子"\n   - "什么是反事实？为什么它观测不到？"\n   - "不能做AB实验时，如何评估策略效果？"\n   - "辛普森悖论是什么？它如何说明'相关≠因果'？"\n\n---\n\n*03analyst_thinking · Week 9 Day 1 · 2026-05-21*
